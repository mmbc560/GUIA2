```python
# ==================================================================================================
# 🚀 SUPPLY CHAIN DECISION LAB
# INVENTORY & GLOBAL PROCUREMENT ANALYTICS
# ==================================================================================================
#
# EXAMPLE CASE:
# Industrial Spare Parts Importer
#
# BUSINESS CONTEXT:
# A company imports industrial spare parts from suppliers located in different countries.
# The company faces:
#
# - Long international lead times
# - Demand variability
# - Different supplier reliability levels
# - High inventory investment
# - Risk of stockouts
# - Different levels of product criticality
#
# MANAGEMENT QUESTION:
# How can the company improve inventory availability and reduce supply risk while considering
# cost, service level, supplier reliability and implementation feasibility?
#
# --------------------------------------------------------------------------------------------------
# THIS NOTEBOOK INTEGRATES:
#
# 1. Data analysis
# 2. Inventory diagnosis
# 3. Supply risk analysis
# 4. Pareto analysis
# 5. Sensitivity analysis
# 6. Scenario analysis
# 7. Strategic alternatives
# 8. AHP - Analytic Hierarchy Process
# 9. AHP consistency analysis
# 10. Strategy ranking
# 11. Human vs Model comparison
# 12. Executive decision summary
#
# --------------------------------------------------------------------------------------------------
# IMPORTANT FOR STUDENTS
#
# This is a REFERENCE MODEL.
#
# You DO NOT need to build everything from zero.
#
# Adapt the following sections to your own business problem:
#
# 🟨 SECTION A → BUSINESS DATA
# 🟨 SECTION B → VARIABLES / RISK LOGIC
# 🟨 SECTION C → SCENARIOS
# 🟨 SECTION D → STRATEGIES
# 🟨 SECTION E → AHP CRITERIA
# 🟨 SECTION F → MANAGERIAL DECISION
#
# The purpose is NOT to demonstrate programming skills.
#
# The purpose is to transform a business problem into:
#
# DATA → ANALYSIS → RISK → SCENARIOS → ALTERNATIVES → DECISION
#
# ==================================================================================================


# ==================================================================================================
# 0. IMPORT LIBRARIES
# ==================================================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure pandas to display all columns
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 150)

# Configure chart style
sns.set_theme(style="whitegrid")

print("=" * 100)
print("🚀 SUPPLY CHAIN DECISION LAB")
print("Inventory & Global Procurement Analytics")
print("=" * 100)
print("\n✅ Libraries loaded successfully.")


# ==================================================================================================
# 🟨 SECTION A - BUSINESS DATA
# ==================================================================================================
#
# STUDENTS:
#
# THIS IS ONE OF THE MAIN SECTIONS YOU SHOULD MODIFY.
#
# Replace these example data with variables related to YOUR problem.
#
# You may use:
#
# - Real company data
# - Historical data
# - Data obtained from interviews
# - Supplier information
# - Justified simulated data
#
# IMPORTANT:
# If simulated data are used, explain why the values are reasonable.
#
# ==================================================================================================


data = {

    # Unique product identifier
    "SKU": [
        "SP001", "SP002", "SP003", "SP004", "SP005",
        "SP006", "SP007", "SP008", "SP009", "SP010"
    ],

    # Product description
    "Product": [
        "Hydraulic Pump",
        "Industrial Bearing",
        "Pressure Valve",
        "Electric Motor",
        "Control Sensor",
        "Drive Belt",
        "Gearbox",
        "Cooling Fan",
        "PLC Module",
        "Mechanical Seal"
    ],

    # Current inventory in units
    "Inventory": [
        90, 180, 45, 70, 25,
        300, 35, 160, 20, 110
    ],

    # Average monthly demand
    "Monthly_Demand": [
        60, 75, 50, 40, 35,
        80, 30, 55, 25, 45
    ],

    # International supplier lead time in days
    "Lead_Time": [
        90, 45, 120, 75, 100,
        40, 130, 50, 110, 65
    ],

    # Supplier reliability
    #
    # Example:
    # 0.95 means approximately 95% reliability
    #
    "Supplier_Reliability": [
        0.88, 0.96, 0.82, 0.91, 0.85,
        0.97, 0.78, 0.94, 0.80, 0.92
    ],

    # Unit purchase cost in USD
    "Unit_Cost": [
        450, 80, 220, 650, 310,
        45, 900, 120, 750, 160
    ],

    # Operational criticality
    #
    # Scale:
    # 1 = Very low criticality
    # 2 = Low
    # 3 = Medium
    # 4 = High
    # 5 = Very high
    #
    "Criticality": [
        5, 3, 4, 5, 4,
        2, 5, 3, 5, 4
    ],

    # Main supplier
    "Supplier": [
        "Asia Parts",
        "Euro Bearings",
        "Asia Parts",
        "Global Motors",
        "Asia Electronics",
        "Local Industrial",
        "Asia Mechanical",
        "Euro Components",
        "Asia Electronics",
        "Global Parts"
    ]
}


# Convert dictionary into DataFrame
df = pd.DataFrame(data)


print("\n")
print("=" * 100)
print("📦 1. ORIGINAL BUSINESS DATABASE")
print("=" * 100)

display(df)


print("""
INTERPRETATION:
This table represents the information currently available to the supply chain manager.

The objective is NOT simply to observe the data.

The objective is to transform these data into information that supports inventory
and global procurement decisions.
""")


# ==================================================================================================
# 2. INVENTORY DIAGNOSIS
# ==================================================================================================
#
# We now transform raw data into supply chain indicators.
#
# DAILY DEMAND:
#
# Monthly Demand / 30
#
#
# DAYS OF COVERAGE:
#
# Inventory / Daily Demand
#
# This estimates how many days current inventory can support expected demand.
#
#
# INVENTORY VALUE:
#
# Inventory × Unit Cost
#
#
# SUPPLY GAP:
#
# Days of Coverage - Lead Time
#
# Positive:
# Inventory coverage exceeds supplier lead time.
#
# Negative:
# Inventory may be exhausted before replenishment arrives.
#
# ==================================================================================================


df["Daily_Demand"] = df["Monthly_Demand"] / 30

df["Days_Coverage"] = df["Inventory"] / df["Daily_Demand"]

df["Inventory_Value"] = df["Inventory"] * df["Unit_Cost"]

df["Supply_Gap"] = df["Days_Coverage"] - df["Lead_Time"]


print("\n")
print("=" * 100)
print("📊 2. INVENTORY DIAGNOSIS")
print("=" * 100)


display(
    df[
        [
            "SKU",
            "Product",
            "Inventory",
            "Monthly_Demand",
            "Days_Coverage",
            "Lead_Time",
            "Supply_Gap",
            "Inventory_Value"
        ]
    ].round(2)
)


print("""
INTERPRETATION:

Days_Coverage estimates how long current inventory could support expected demand.

Supply_Gap compares inventory coverage with supplier lead time.

• Positive Supply Gap:
  Current inventory could theoretically cover demand until replenishment arrives.

• Negative Supply Gap:
  The product could face a supply exposure before replenishment arrives.

A negative gap does NOT automatically mean a stockout will occur.
It is a warning indicator that requires managerial analysis.
""")


# ==================================================================================================
# 🟨 SECTION B - SUPPLY RISK MODEL
# ==================================================================================================
#
# We create a simple academic Supply Risk Score.
#
# IMPORTANT:
#
# THIS IS NOT A UNIVERSAL INDUSTRY FORMULA.
#
# It is a decision-support model designed for this exercise.
#
# Students should modify the variables and/or weights according to their problem.
#
# In this example:
#
# Coverage Risk       = 35%
# Lead Time Risk      = 25%
# Supplier Risk       = 20%
# Criticality Risk    = 20%
#
# Total               = 100%
#
# ==================================================================================================


# Coverage risk
#
# If coverage is below lead time → risk = 1
# Otherwise → risk = 0

df["Coverage_Risk"] = np.where(
    df["Days_Coverage"] < df["Lead_Time"],
    1,
    0
)


# Supplier risk
#
# Reliability 95% → Risk 5%

df["Supplier_Risk"] = 1 - df["Supplier_Reliability"]


# Normalize Lead Time between 0 and 1

df["Lead_Time_Risk"] = (
    df["Lead_Time"] /
    df["Lead_Time"].max()
)


# Normalize criticality between 0 and 1

df["Criticality_Risk"] = (
    df["Criticality"] / 5
)


# Composite Supply Risk Score

df["Risk_Score"] = (

    df["Coverage_Risk"] * 0.35 +

    df["Lead_Time_Risk"] * 0.25 +

    df["Supplier_Risk"] * 0.20 +

    df["Criticality_Risk"] * 0.20

) * 100


# Risk classification function

def classify_risk(score):

    if score >= 65:
        return "CRITICAL"

    elif score >= 50:
        return "HIGH"

    elif score >= 35:
        return "MODERATE"

    else:
        return "LOW"


df["Risk_Level"] = df["Risk_Score"].apply(classify_risk)


print("\n")
print("=" * 100)
print("🚨 3. SUPPLY RISK ANALYSIS")
print("=" * 100)


risk_table = df[
    [
        "SKU",
        "Product",
        "Days_Coverage",
        "Lead_Time",
        "Supplier_Reliability",
        "Criticality",
        "Risk_Score",
        "Risk_Level"
    ]
].sort_values(
    "Risk_Score",
    ascending=False
)


display(risk_table.round(2))


print("""
INTERPRETATION:

The Supply Risk Score combines several dimensions of the problem.

A product may become important not only because inventory is low,
but because it combines:

• Insufficient coverage
• Long lead time
• Low supplier reliability
• High operational criticality

The purpose of the model is to PRIORITIZE managerial attention.

IMPORTANT QUESTION:

Does the quantitative classification make sense when compared with
what managers know about the real business?
""")


# ==================================================================================================
# 4. GRAPH - SUPPLY CHAIN RISK MAP
# ==================================================================================================
#
# X axis  = Supplier Lead Time
# Y axis  = Supply Risk Score
# Bubble  = Inventory Value
# Color   = Risk Level
#
# ==================================================================================================


plt.figure(figsize=(13, 7))


risk_colors = {
    "LOW": "#2ecc71",
    "MODERATE": "#f1c40f",
    "HIGH": "#e67e22",
    "CRITICAL": "#e74c3c"
}


for level in ["LOW", "MODERATE", "HIGH", "CRITICAL"]:

    subset = df[df["Risk_Level"] == level]

    if len(subset) > 0:

        plt.scatter(
            subset["Lead_Time"],
            subset["Risk_Score"],
            s=(subset["Inventory_Value"] / 15) + 100,
            color=risk_colors[level],
            label=level,
            alpha=0.75,
            edgecolors="black"
        )


# Add SKU labels

for _, row in df.iterrows():

    plt.text(
        row["Lead_Time"] + 1,
        row["Risk_Score"] + 0.5,
        row["SKU"],
        fontsize=9
    )


plt.axhline(
    50,
    linestyle="--",
    color="gray",
    alpha=0.7
)


plt.xlabel(
    "Supplier Lead Time (days)",
    fontsize=11
)

plt.ylabel(
    "Supply Risk Score",
    fontsize=11
)

plt.title(
    "SUPPLY CHAIN RISK MAP\nBubble Size Represents Inventory Value",
    fontsize=15,
    fontweight="bold"
)

plt.legend(
    title="Risk Level"
)

plt.tight_layout()

plt.show()


print("""
GRAPH INTERPRETATION:

Products located higher on the graph have greater estimated supply risk.

Products located further to the right have longer replenishment lead times.

Large bubbles represent greater inventory investment.

Therefore, a product with:

HIGH RISK + LONG LEAD TIME + HIGH INVENTORY VALUE

should receive special managerial attention.

This does NOT necessarily mean that more inventory should be purchased.
The appropriate response could involve inventory, suppliers, sourcing,
contracts or another procurement strategy.
""")


# ==================================================================================================
# 5. PARETO ANALYSIS - INVENTORY INVESTMENT
# ==================================================================================================


pareto = df.sort_values(
    "Inventory_Value",
    ascending=False
).copy()


pareto["Cumulative_Value"] = (
    pareto["Inventory_Value"].cumsum()
)


pareto["Cumulative_Percentage"] = (

    pareto["Cumulative_Value"] /
    pareto["Inventory_Value"].sum()

) * 100


fig, ax1 = plt.subplots(
    figsize=(13, 6)
)


# Inventory value bars

ax1.bar(
    pareto["SKU"],
    pareto["Inventory_Value"],
    color="#3498db",
    alpha=0.8
)

ax1.set_xlabel(
    "SKU"
)

ax1.set_ylabel(
    "Inventory Value (USD)",
    color="#3498db"
)


# Second axis for cumulative percentage

ax2 = ax1.twinx()


ax2.plot(
    pareto["SKU"],
    pareto["Cumulative_Percentage"],
    color="#e74c3c",
    marker="o",
    linewidth=3
)


ax2.axhline(
    80,
    color="#f39c12",
    linestyle="--",
    linewidth=2
)


ax2.set_ylabel(
    "Cumulative Inventory Value (%)",
    color="#e74c3c"
)


plt.title(
    "INVENTORY INVESTMENT PARETO ANALYSIS",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.show()


print("""
GRAPH INTERPRETATION:

The Pareto analysis shows where inventory capital is concentrated.

Management should compare this graph with the Supply Risk Map.

IMPORTANT MANAGERIAL QUESTION:

Are the products receiving the largest inventory investment also the products
that require the greatest protection?

A company may simultaneously have:

• Excess inventory in some products
• Stockout risk in other products

Therefore, total inventory alone does not explain inventory performance.
Inventory composition is also important.
""")


# ==================================================================================================
# 6. IDENTIFY THE HIGHEST-RISK SKU
# ==================================================================================================


critical_sku = df.loc[
    df["Risk_Score"].idxmax()
]


print("\n")
print("=" * 100)
print("🎯 4. PRODUCT SELECTED FOR DEEP ANALYSIS")
print("=" * 100)


print(
    f"""
SKU: {critical_sku['SKU']}

Product: {critical_sku['Product']}

Current Inventory: {critical_sku['Inventory']:.0f} units

Monthly Demand: {critical_sku['Monthly_Demand']:.0f} units

Days of Coverage: {critical_sku['Days_Coverage']:.2f} days

Supplier Lead Time: {critical_sku['Lead_Time']:.0f} days

Supplier Reliability: {critical_sku['Supplier_Reliability']:.1%}

Risk Score: {critical_sku['Risk_Score']:.2f}

Risk Level: {critical_sku['Risk_Level']}
"""
)


# ==================================================================================================
# 7. SENSITIVITY ANALYSIS
# ==================================================================================================
#
# QUESTION:
#
# What happens to inventory coverage if demand changes?
#
#
# IMPORTANT CONCEPT:
#
# Sensitivity analysis changes ONE variable while keeping the others constant.
#
# Demand changes from -20% to +50%.
#
# ==================================================================================================


demand_changes = np.arange(
    -0.20,
    0.55,
    0.05
)


sensitivity_results = []


for change in demand_changes:

    projected_demand = (
        critical_sku["Monthly_Demand"] *
        (1 + change)
    )

    projected_daily_demand = (
        projected_demand / 30
    )

    projected_coverage = (
        critical_sku["Inventory"] /
        projected_daily_demand
    )

    sensitivity_results.append(
        [
            change,
            projected_demand,
            projected_coverage
        ]
    )


sensitivity = pd.DataFrame(
    sensitivity_results,
    columns=[
        "Demand_Change",
        "Projected_Demand",
        "Projected_Coverage"
    ]
)


print("\n")
print("=" * 100)
print("🎛️ 5. DEMAND SENSITIVITY ANALYSIS")
print("=" * 100)


display(
    sensitivity.round(2)
)


# Sensitivity chart

plt.figure(
    figsize=(12, 6)
)


plt.plot(
    sensitivity["Demand_Change"] * 100,
    sensitivity["Projected_Coverage"],
    marker="o",
    linewidth=3,
    markersize=7,
    color="#8e44ad"
)


plt.axhline(
    critical_sku["Lead_Time"],
    linestyle="--",
    linewidth=2,
    color="#e74c3c",
    label="Current Supplier Lead Time"
)


plt.xlabel(
    "Demand Change (%)"
)

plt.ylabel(
    "Projected Days of Coverage"
)

plt.title(
    f"DEMAND SENSITIVITY ANALYSIS - {critical_sku['SKU']}\n{critical_sku['Product']}",
    fontsize=14,
    fontweight="bold"
)

plt.legend()

plt.tight_layout()

plt.show()


print("""
SENSITIVITY INTERPRETATION:

The graph shows how inventory coverage changes when demand changes.

As demand increases, the same inventory is consumed faster and coverage decreases.

The horizontal red line represents supplier lead time.

If projected coverage remains below lead time, the company may face supply exposure
before the next replenishment arrives.

Sensitivity analysis helps answer:

WHICH VARIABLE HAS THE POWER TO CHANGE THE RESULT?
""")


# ==================================================================================================
# 🟨 SECTION C - SCENARIO ANALYSIS
# ==================================================================================================
#
# Unlike sensitivity analysis, scenario analysis changes MULTIPLE variables.
#
#
# BASE
# Normal business conditions.
#
#
# DEMAND SHOCK
# Demand increases by 30%.
#
#
# SUPPLY DISRUPTION
# Lead time increases by 35%.
# Supplier reliability falls by 15 percentage points.
#
#
# PERFECT STORM
# Demand increases +30%.
# Lead time increases +35%.
# Reliability falls by 20 percentage points.
#
#
# STUDENTS:
# Replace these scenarios with situations that make sense for YOUR problem.
#
# ==================================================================================================


scenarios = {

    "BASE": {
        "Demand": 0.00,
        "Lead_Time": 0.00,
        "Reliability": 0.00
    },

    "DEMAND SHOCK": {
        "Demand": 0.30,
        "Lead_Time": 0.00,
        "Reliability": 0.00
    },

    "SUPPLY DISRUPTION": {
        "Demand": 0.00,
        "Lead_Time": 0.35,
        "Reliability": -0.15
    },

    "PERFECT STORM": {
        "Demand": 0.30,
        "Lead_Time": 0.35,
        "Reliability": -0.20
    }
}


scenario_results = []


for scenario_name, values in scenarios.items():

    projected_demand = (
        critical_sku["Monthly_Demand"] *
        (1 + values["Demand"])
    )

    projected_lead_time = (
        critical_sku["Lead_Time"] *
        (1 + values["Lead_Time"])
    )

    projected_reliability = (
        critical_sku["Supplier_Reliability"] +
        values["Reliability"]
    )

    # Reliability cannot be lower than 0 or higher than 1

    projected_reliability = max(
        0,
        min(projected_reliability, 1)
    )

    projected_daily_demand = (
        projected_demand / 30
    )

    projected_coverage = (
        critical_sku["Inventory"] /
        projected_daily_demand
    )

    projected_supply_gap = (
        projected_coverage -
        projected_lead_time
    )

    scenario_results.append(
        [
            scenario_name,
            projected_demand,
            projected_lead_time,
            projected_reliability,
            projected_coverage,
            projected_supply_gap
        ]
    )


scenario_df = pd.DataFrame(
    scenario_results,
    columns=[
        "Scenario",
        "Projected_Demand",
        "Projected_Lead_Time",
        "Projected_Reliability",
        "Projected_Coverage",
        "Supply_Gap"
    ]
)


print("\n")
print("=" * 100)
print("🌎 6. SUPPLY CHAIN SCENARIO ANALYSIS")
print("=" * 100)


display(
    scenario_df.round(2)
)


# ==================================================================================================
# 8. SCENARIO CHART
# ==================================================================================================


plt.figure(
    figsize=(12, 6)
)


scenario_colors = [
    "#2ecc71",
    "#f1c40f",
    "#e67e22",
    "#e74c3c"
]


bars = plt.bar(
    scenario_df["Scenario"],
    scenario_df["Supply_Gap"],
    color=scenario_colors,
    edgecolor="black"
)


plt.axhline(
    0,
    color="black",
    linewidth=1.5
)


plt.ylabel(
    "Supply Gap = Coverage - Lead Time"
)


plt.title(
    "SUPPLY CHAIN STRESS TEST",
    fontsize=15,
    fontweight="bold"
)


plt.xticks(
    rotation=10
)


for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + (2 if height >= 0 else -2),
        f"{height:.1f}",
        ha="center",
        va="bottom" if height >= 0 else "top",
        fontweight="bold"
    )


plt.tight_layout()

plt.show()


print("""
SCENARIO INTERPRETATION:

Supply Gap = Projected Coverage - Projected Lead Time

• Positive Supply Gap:
  Coverage exceeds lead time.

• Negative Supply Gap:
  The company may consume available inventory before replenishment arrives.

The purpose of scenario analysis is NOT to predict the future perfectly.

The purpose is to ask:

WHAT WOULD WE DO IF CONDITIONS CHANGE?
""")


# ==================================================================================================
# 🟨 SECTION D - STRATEGIC ALTERNATIVES
# ==================================================================================================
#
# The company now needs possible solutions.
#
# Example alternatives:
#
# 1. Increase Safety Stock
# 2. Dual Sourcing
# 3. Nearshoring
# 4. Current Policy
#
#
# Each alternative is evaluated from 1 to 10.
#
# IMPORTANT:
#
# Higher score = more attractive performance.
#
# Therefore:
#
# Cost = 10 means economically very attractive.
# Cost = 1 means economically unattractive.
#
#
# STUDENTS:
#
# Replace these strategies with solutions related to YOUR problem.
#
# Scores should ideally be supported by:
#
# - Data
# - Literature
# - Experts
# - Suppliers
# - Previous analysis
# - Simulation results
#
# ==================================================================================================


strategies = pd.DataFrame({

    "Strategy": [
        "Increase Safety Stock",
        "Dual Sourcing",
        "Nearshoring",
        "Current Policy"
    ],

    # Expected ability to improve service level
    "Service_Level": [
        8,
        9,
        8,
        5
    ],

    # Cost attractiveness
    # Higher = better
    "Cost": [
        5,
        7,
        6,
        9
    ],

    # Ability to reduce supply risk
    "Risk_Reduction": [
        6,
        9,
        8,
        3
    ],

    # Ease / feasibility of implementation
    "Feasibility": [
        9,
        7,
        6,
        10
    ]
})


print("\n")
print("=" * 100)
print("💡 7. STRATEGIC ALTERNATIVES")
print("=" * 100)


display(
    strategies
)


print("""
INTERPRETATION:

The alternatives represent different ways to respond to the identified supply problem.

Notice that no strategy is perfect.

For example:

• Increasing Safety Stock may improve service but increase inventory investment.

• Dual Sourcing may reduce supplier dependency but require supplier development.

• Nearshoring may reduce lead time but could increase purchase cost.

• Maintaining the Current Policy may be easy and inexpensive but may preserve existing risk.

Therefore, the problem is MULTICRITERIA.

This is why AHP will now be used.
""")


# ==================================================================================================
# 🟨 SECTION E - AHP: ANALYTIC HIERARCHY PROCESS
# ==================================================================================================
#
# DECISION GOAL:
#
# Select the most appropriate inventory / global procurement strategy.
#
#
# CRITERIA:
#
# 1. Service Level
# 2. Cost
# 3. Risk Reduction
# 4. Feasibility
#
#
# SAATY SCALE:
#
# 1 = Equal importance
# 3 = Moderate importance
# 5 = Strong importance
# 7 = Very strong importance
# 9 = Extreme importance
#
# 2,4,6,8 = Intermediate judgments
#
#
# Reciprocal values express the opposite relationship.
#
# Example:
#
# If A is 3 times more important than B:
#
# A vs B = 3
# B vs A = 1/3
#
#
# STUDENTS:
# Construct your own matrix according to YOUR decision problem.
#
# ==================================================================================================


criteria = [
    "Service Level",
    "Cost",
    "Risk Reduction",
    "Feasibility"
]


# Pairwise Comparison Matrix

A = np.array([

    [1,   3,   1/2, 4],

    [1/3, 1,   1/4, 2],

    [2,   4,   1,   5],

    [1/4, 1/2, 1/5, 1]

])


ahp_matrix = pd.DataFrame(
    A,
    index=criteria,
    columns=criteria
)


print("\n")
print("=" * 100)
print("⚖️ 8. AHP PAIRWISE COMPARISON MATRIX")
print("=" * 100)


display(
    ahp_matrix.round(2)
)


# ==================================================================================================
# 10. CALCULATE AHP CRITERIA WEIGHTS
# ==================================================================================================


# Sum each column

column_sums = A.sum(
    axis=0
)


# Normalize the matrix

normalized_matrix = (
    A / column_sums
)


# Calculate average value of each row

weights = (
    normalized_matrix.mean(
        axis=1
    )
)


weights_df = pd.DataFrame({

    "Criterion": criteria,

    "Weight": weights

})


weights_df["Weight_Percentage"] = (
    weights_df["Weight"] * 100
)


print("\n")
print("=" * 100)
print("🧠 9. AHP CRITERIA WEIGHTS")
print("=" * 100)


display(
    weights_df.round(3)
)


# ==================================================================================================
# 11. AHP WEIGHTS GRAPH
# ==================================================================================================


plt.figure(
    figsize=(10, 6)
)


colors_ahp = [
    "#3498db",
    "#f39c12",
    "#e74c3c",
    "#2ecc71"
]


bars = plt.bar(
    weights_df["Criterion"],
    weights_df["Weight_Percentage"],
    color=colors_ahp,
    edgecolor="black"
)


plt.ylabel(
    "Weight (%)"
)


plt.title(
    "AHP DECISION CRITERIA IMPORTANCE",
    fontsize=15,
    fontweight="bold"
)


for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 0.5,
        f"{height:.1f}%",
        ha="center",
        fontweight="bold"
    )


plt.tight_layout()

plt.show()


print("""
AHP INTERPRETATION:

The graph represents the relative importance assigned to each decision criterion.

This means that the final decision does not treat all criteria equally.

The priorities reflect managerial judgment.

IMPORTANT:

The weights should NOT be copied automatically from this example.

Students should explain WHY certain criteria are more important in their own problem.
""")


# ==================================================================================================
# 12. AHP CONSISTENCY ANALYSIS
# ==================================================================================================
#
# AHP includes a consistency test.
#
# Consistency Ratio (CR):
#
# CR < 0.10 → generally considered acceptable
#
# CR >= 0.10 → pairwise comparisons should be reviewed
#
# ==================================================================================================


n = len(criteria)


# Weighted sum vector

weighted_sum = (
    A @ weights
)


# Consistency vector

consistency_vector = (
    weighted_sum / weights
)


# Approximation of maximum eigenvalue

lambda_max = (
    consistency_vector.mean()
)


# Consistency Index

CI = (
    (lambda_max - n) /
    (n - 1)
)


# Random Index for a matrix of size 4

RI = 0.90


# Consistency Ratio

CR = (
    CI / RI
)


print("\n")
print("=" * 100)
print("🔎 10. AHP CONSISTENCY CHECK")
print("=" * 100)


print(
    f"Lambda Max: {lambda_max:.3f}"
)

print(
    f"Consistency Index (CI): {CI:.3f}"
)

print(
    f"Consistency Ratio (CR): {CR:.3f}"
)


if CR < 0.10:

    print(
        "\n✅ ACCEPTABLE CONSISTENCY"
    )

    print(
        "The pairwise judgments show a reasonable level of consistency."
    )

else:

    print(
        "\n⚠️ REVIEW REQUIRED"
    )

    print(
        "The pairwise comparisons should be reviewed because the Consistency Ratio is above 0.10."
    )


# ==================================================================================================
# 13. STRATEGY PERFORMANCE MATRIX - HEATMAP
# ==================================================================================================


heatmap_data = strategies.set_index(
    "Strategy"
)[
    [
        "Service_Level",
        "Cost",
        "Risk_Reduction",
        "Feasibility"
    ]
]


plt.figure(
    figsize=(11, 6)
)


sns.heatmap(
    heatmap_data,
    annot=True,
    cmap="YlGnBu",
    linewidths=0.8,
    fmt=".0f",
    vmin=1,
    vmax=10
)


plt.title(
    "STRATEGIC ALTERNATIVES PERFORMANCE MATRIX",
    fontsize=15,
    fontweight="bold"
)


plt.xlabel(
    "Decision Criteria"
)


plt.ylabel(
    "Strategy"
)


plt.tight_layout()

plt.show()


print("""
HEATMAP INTERPRETATION:

The heatmap allows managers to compare the strengths and weaknesses of each strategy.

A strategy may perform very well in one criterion and poorly in another.

This illustrates an important supply chain principle:

MANAGERIAL DECISIONS INVOLVE TRADE-OFFS.

The purpose of AHP is to combine those trade-offs according to managerial priorities.
""")


# ==================================================================================================
# 14. EVALUATE STRATEGIES USING AHP WEIGHTS
# ==================================================================================================
#
# We now combine:
#
# STRATEGY PERFORMANCE
#
# +
#
# AHP CRITERIA WEIGHTS
#
# =
#
# FINAL STRATEGIC SCORE
#
# ==================================================================================================


evaluation = strategies.copy()


# Normalize strategy performance scores
#
# Divide each score by the maximum score for that criterion

for criterion in [
    "Service_Level",
    "Cost",
    "Risk_Reduction",
    "Feasibility"
]:

    evaluation[
        criterion + "_Normalized"
    ] = (

        evaluation[criterion] /
        evaluation[criterion].max()

    )


# Calculate weighted AHP score

evaluation["AHP_Score"] = (

    evaluation["Service_Level_Normalized"] * weights[0]

    +

    evaluation["Cost_Normalized"] * weights[1]

    +

    evaluation["Risk_Reduction_Normalized"] * weights[2]

    +

    evaluation["Feasibility_Normalized"] * weights[3]

)


# Ranking

evaluation["Ranking"] = (

    evaluation["AHP_Score"]
    .rank(
        ascending=False,
        method="min"
    )
    .astype(int)

)


# Sort from best to worst

evaluation = evaluation.sort_values(
    "AHP_Score",
    ascending=False
).reset_index(drop=True)


print("\n")
print("=" * 100)
print("🏆 11. AHP STRATEGY RANKING")
print("=" * 100)


display(

    evaluation[
        [
            "Strategy",
            "Service_Level",
            "Cost",
            "Risk_Reduction",
            "Feasibility",
            "AHP_Score",
            "Ranking"
        ]
    ].round(3)

)


# ==================================================================================================
# 15. FINAL STRATEGY RANKING GRAPH
# ==================================================================================================


plt.figure(
    figsize=(11, 6)
)


ranking_colors = [
    "#2ecc71",
    "#3498db",
    "#f39c12",
    "#e74c3c"
]


bars = plt.barh(
    evaluation["Strategy"],
    evaluation["AHP_Score"],
    color=ranking_colors,
    edgecolor="black"
)


plt.xlabel(
    "AHP Strategic Score"
)


plt.title(
    "STRATEGIC ALTERNATIVES RANKING\nInventory & Global Procurement Decision",
    fontsize=15,
    fontweight="bold"
)


plt.gca().invert_yaxis()


for bar in bars:

    width = bar.get_width()

    plt.text(
        width + 0.005,
        bar.get_y() + bar.get_height()/2,
        f"{width:.3f}",
        va="center",
        fontweight="bold"
    )


plt.xlim(
    0,
    max(evaluation["AHP_Score"]) * 1.15
)


plt.tight_layout()

plt.show()


print("""
STRATEGY RANKING INTERPRETATION:

The strategy with the highest AHP score provides the strongest overall performance
according to:

1. The alternatives evaluated
2. Their performance scores
3. The criteria selected
4. The AHP managerial priorities

IMPORTANT:

The highest-ranked strategy is a MODEL RECOMMENDATION.

It is NOT automatically the final managerial decision.
""")


# ==================================================================================================
# 16. MODEL RECOMMENDATION
# ==================================================================================================


best_strategy = evaluation.iloc[0]


model_decision = (
    best_strategy["Strategy"]
)


model_score = (
    best_strategy["AHP_Score"]
)


print("\n")
print("=" * 100)
print("🤖 12. MODEL RECOMMENDATION")
print("=" * 100)


print(
    f"""
Recommended Strategy:

🏆 {model_decision}

AHP Score:

{model_score:.3f}

The model recommends this strategy because it achieved the highest
weighted performance according to the AHP criteria.
"""
)


# ==================================================================================================
# 🟨 SECTION F - MANAGERIAL DECISION
# ==================================================================================================
#
# THIS IS A VERY IMPORTANT PART OF THE EXERCISE.
#
# Analytics does NOT replace managerial judgment.
#
# The student must now select a strategy.
#
#
# AVAILABLE OPTIONS:
#
# Increase Safety Stock
# Dual Sourcing
# Nearshoring
# Current Policy
#
#
# STUDENT:
#
# Change manager_decision below.
#
# ==================================================================================================


manager_decision = "Dual Sourcing"


print("\n")
print("=" * 100)
print("👤 13. MANAGER FINAL DECISION")
print("=" * 100)


print(
    f"Manager selected strategy: {manager_decision}"
)


# ==================================================================================================
# 18. HUMAN-MODEL ALIGNMENT
# ==================================================================================================


print("\n")
print("=" * 100)
print("🧠 14. HUMAN–MODEL ALIGNMENT")
print("=" * 100)


print(
    f"Manager Decision:       {manager_decision}"
)

print(
    f"Model Recommendation:   {model_decision}"
)


if manager_decision == model_decision:

    alignment = "ALIGNED DECISION"

    print(
        "\n✅ ALIGNED DECISION"
    )

    print(
        """
The manager selected the same strategy recommended by the quantitative model.

The student should still explain WHY this strategy is appropriate.
"""
    )

else:

    alignment = "DIFFERENT DECISION"

    print(
        "\n⚠️ DIFFERENT DECISION"
    )

    print(
        """
The manager selected a different strategy from the model recommendation.

THIS IS NOT AUTOMATICALLY AN ERROR.

The manager may have additional information that is not represented
in the quantitative model.

Examples:

• Contract restrictions
• Supplier availability
• Implementation time
• Cash-flow limitations
• Regulatory constraints
• Operational experience
• Strategic relationships
• Negotiation conditions
• Information obtained from experts

The student must explain the difference.
"""
    )


# ==================================================================================================
# 19. DECISION GAP
# ==================================================================================================
#
# We compare the AHP score of:
#
# MODEL STRATEGY
#
# versus
#
# MANAGER STRATEGY
#
# ==================================================================================================


manager_score = evaluation.loc[
    evaluation["Strategy"] == manager_decision,
    "AHP_Score"
].iloc[0]


decision_gap = abs(
    model_score -
    manager_score
)


# Simple academic interpretation

if decision_gap < 0.05:

    gap_interpretation = "LOW GAP"

elif decision_gap < 0.15:

    gap_interpretation = "MODERATE GAP"

else:

    gap_interpretation = "HIGH GAP"


print("\n")
print("=" * 100)
print("📏 15. DECISION GAP")
print("=" * 100)


print(
    f"""
Manager Strategy Score: {manager_score:.3f}

Model Strategy Score:   {model_score:.3f}

Decision Gap:           {decision_gap:.3f}

Interpretation:         {gap_interpretation}
"""
)


print("""
DECISION GAP INTERPRETATION:

The Decision Gap does NOT measure whether the manager is right or wrong.

It measures how different the manager's selected alternative is from
the highest-scoring quantitative alternative.

A larger gap requires stronger managerial justification.
""")


# ==================================================================================================
# 20. MANAGERIAL JUSTIFICATION
# ==================================================================================================
#
# STUDENT:
#
# Replace the text below with YOUR justification.
#
# ==================================================================================================


manager_justification = """

Although the quantitative model recommends another alternative,
Dual Sourcing is selected because the company already has a potential
secondary supplier that could be activated relatively quickly.

This alternative could reduce dependence on the current supplier
without immediately increasing inventory investment substantially.

The decision therefore prioritizes supply continuity, supplier
diversification and implementation speed.

"""


print("\n")
print("=" * 100)
print("💬 16. MANAGERIAL JUSTIFICATION")
print("=" * 100)


print(
    manager_justification
)


# ==================================================================================================
# 21. FINAL DECISION COMPARISON GRAPH
# ==================================================================================================


comparison = evaluation[
    ["Strategy", "AHP_Score"]
].copy()


comparison["Decision_Type"] = "Alternative"


comparison.loc[
    comparison["Strategy"] == model_decision,
    "Decision_Type"
] = "Model Recommendation"


comparison.loc[
    comparison["Strategy"] == manager_decision,
    "Decision_Type"
] = "Manager Decision"


# If both decisions are equal

if manager_decision == model_decision:

    comparison.loc[
        comparison["Strategy"] == manager_decision,
        "Decision_Type"
    ] = "Manager + Model"


decision_colors = []

for strategy in comparison["Strategy"]:

    if strategy == manager_decision and strategy == model_decision:

        decision_colors.append("#9b59b6")

    elif strategy == model_decision:

        decision_colors.append("#2ecc71")

    elif strategy == manager_decision:

        decision_colors.append("#3498db")

    else:

        decision_colors.append("#bdc3c7")


plt.figure(
    figsize=(11, 6)
)


bars = plt.barh(
    comparison["Strategy"],
    comparison["AHP_Score"],
    color=decision_colors,
    edgecolor="black"
)


plt.xlabel(
    "AHP Score"
)


plt.title(
    "HUMAN vs MODEL DECISION",
    fontsize=15,
    fontweight="bold"
)


plt.gca().invert_yaxis()


for bar in bars:

    width = bar.get_width()

    plt.text(
        width + 0.005,
        bar.get_y() + bar.get_height()/2,
        f"{width:.3f}",
        va="center",
        fontweight="bold"
    )


plt.xlim(
    0,
    max(comparison["AHP_Score"]) * 1.15
)


plt.tight_layout()

plt.show()


print("""
GRAPH INTERPRETATION:

The chart compares the quantitative ranking with the final managerial decision.

The objective is NOT to force the manager to follow the model.

The objective is to make the decision transparent and explainable.

A strong managerial decision should combine:

DATA + ANALYTICS + BUSINESS KNOWLEDGE + JUDGMENT
""")


# ==================================================================================================
# 22. AUTOMATIC EXECUTIVE SUMMARY
# ==================================================================================================


highest_risk_product = (

    df.sort_values(
        "Risk_Score",
        ascending=False
    ).iloc[0]

)


perfect_storm = scenario_df.loc[
    scenario_df["Scenario"] == "PERFECT STORM"
].iloc[0]


print("\n")
print("=" * 100)
print("🚀 17. SUPPLY CHAIN DECISION LAB - EXECUTIVE SUMMARY")
print("=" * 100)


print(
    f"""

BUSINESS PROBLEM
----------------------------------------------------------------------------------------------------

The company faces inventory availability challenges associated with demand variability,
international lead times and supplier uncertainty.


MOST CRITICAL PRODUCT
----------------------------------------------------------------------------------------------------

SKU:
{highest_risk_product['SKU']}

Product:
{highest_risk_product['Product']}

Risk Score:
{highest_risk_product['Risk_Score']:.2f}

Risk Level:
{highest_risk_product['Risk_Level']}

Current Days of Coverage:
{highest_risk_product['Days_Coverage']:.2f}

Supplier Lead Time:
{highest_risk_product['Lead_Time']:.2f} days


PERFECT STORM SCENARIO
----------------------------------------------------------------------------------------------------

Projected Demand:
{perfect_storm['Projected_Demand']:.2f}

Projected Lead Time:
{perfect_storm['Projected_Lead_Time']:.2f} days

Projected Supplier Reliability:
{perfect_storm['Projected_Reliability']:.2%}

Projected Coverage:
{perfect_storm['Projected_Coverage']:.2f} days

Projected Supply Gap:
{perfect_storm['Supply_Gap']:.2f} days


AHP ANALYSIS
----------------------------------------------------------------------------------------------------

Consistency Ratio:
{CR:.3f}

Model Recommended Strategy:
{model_decision}

Model Strategy Score:
{model_score:.3f}


MANAGERIAL DECISION
----------------------------------------------------------------------------------------------------

Manager Selected Strategy:
{manager_decision}

Manager Strategy Score:
{manager_score:.3f}

Decision Gap:
{decision_gap:.3f}

Decision Gap Interpretation:
{gap_interpretation}

Human-Model Alignment:
{alignment}


FINAL DECISION PRINCIPLE
----------------------------------------------------------------------------------------------------

DATA provides evidence.

INVENTORY ANALYSIS identifies current exposure.

SENSITIVITY ANALYSIS identifies influential variables.

SCENARIO ANALYSIS tests uncertainty.

AHP structures multiple decision criteria.

THE MODEL recommends.

THE MANAGER decides.

"""
)


# ==================================================================================================
# 23. FINAL LEARNING QUESTIONS
# ==================================================================================================


print("\n")
print("=" * 100)
print("🎓 FINAL MANAGERIAL QUESTIONS")
print("=" * 100)


print(
    """
1. What is the main inventory or procurement problem identified by the analysis?

2. Which product, material, supplier or process requires priority attention?

3. Which variables have the greatest influence on the problem?

4. What did the sensitivity analysis reveal?

5. Which scenario creates the greatest supply chain exposure?

6. What inventory or procurement strategies could address the problem?

7. Why were the AHP criteria selected?

8. What does the AHP Consistency Ratio indicate?

9. Which strategy does the quantitative model recommend?

10. Does the management team agree with the model?

11. If not, what business information explains the difference?

12. What decision would you implement tomorrow if you were the Supply Chain Manager?
"""
)


# ==================================================================================================
# END OF NOTEBOOK
# ==================================================================================================


print("\n")
print("=" * 100)
print("✅ ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 100)


print(
    """
SUPPLY CHAIN DECISION JOURNEY COMPLETED:

📚 BUSINESS PROBLEM
        ↓
📊 DATA
        ↓
📦 INVENTORY DIAGNOSIS
        ↓
🚨 RISK ANALYSIS
        ↓
🎛️ SENSITIVITY ANALYSIS
        ↓
🌎 SCENARIO ANALYSIS
        ↓
💡 STRATEGIC ALTERNATIVES
        ↓
⚖️ AHP
        ↓
🤖 MODEL RECOMMENDATION
        ↓
👤 MANAGERIAL DECISION


Remember:

"The model does not make the decision.
It makes the decision more informed, transparent and defensible."
"""
)
```
