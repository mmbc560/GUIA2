```python
# =====================================================
# TEST GRUPAL DE AFINIDAD SECTORIAL + INFERENCIAS
# Trend Tech
# Máximo 6 integrantes
# =====================================================

import matplotlib.pyplot as plt

# =====================================================
# 1. Sectores, interpretaciones e inferencias
# =====================================================

sectores = {
    "a": "Tecnología y transformación digital",
    "b": "Manufactura, logística y operaciones",
    "c": "Salud, educación y sector social",
    "d": "Ambiente, energía y sostenibilidad",
    "e": "Comercio, retail y servicios"
}

interpretaciones = {
    "a": "El grupo tiene afinidad por soluciones digitales, inteligencia artificial, datos, automatización y transformación tecnológica.",
    "b": "El grupo se orienta hacia la mejora de procesos, eficiencia operativa, producción, logística e inventarios.",
    "c": "El grupo muestra interés por proyectos enfocados en personas, bienestar, educación, salud o impacto social.",
    "d": "El grupo tiene una orientación hacia sostenibilidad, economía circular, energías limpias e impacto ambiental.",
    "e": "El grupo se enfoca en negocios, experiencia del cliente, comercio, servicios y generación de valor en el mercado."
}

inferencias = {
    "a": [
        "Si la empresa tiene app, pueden analizar experiencia digital.",
        "Si tiene chatbot, pueden analizar uso de IA en servicio al cliente.",
        "Si la página web es lenta o poco clara, puede existir un problema tecnológico.",
        "Si maneja muchos datos, puede haber oportunidad de analítica avanzada.",
        "Si muchos procesos son manuales, puede haber oportunidad de automatización."
    ],
    "b": [
        "Si venden productos físicos, hay procesos logísticos por analizar.",
        "Si tienen muchos productos, hay gestión de inventarios.",
        "Si tienen varias sedes, hay redes de distribución o abastecimiento.",
        "Si hay demoras, puede existir un problema de eficiencia operativa.",
        "Si producen algo físico, pueden analizar manufactura y mejora de procesos."
    ],
    "c": [
        "Si atienden muchas personas, pueden analizar experiencia del usuario.",
        "Si hay filas, citas o tiempos largos, puede haber problemas de servicio.",
        "Si manejan información de usuarios, puede haber retos de datos.",
        "Si prestan servicios educativos o de salud, pueden analizar digitalización del servicio.",
        "Si el usuario necesita acompañamiento, puede haber oportunidad de apps o plataformas."
    ],
    "d": [
        "Si generan residuos, pueden analizar economía circular.",
        "Si consumen mucha energía, pueden analizar eficiencia energética.",
        "Si usan empaques, pueden analizar sostenibilidad de materiales.",
        "Si transportan productos, pueden analizar huella de carbono.",
        "Si mencionan sostenibilidad, pueden revisar si realmente aplican acciones medibles."
    ],
    "e": [
        "Si venden online, hay e-commerce, logística y experiencia del cliente.",
        "Si tienen muchos clientes, puede haber oportunidad de personalización con IA.",
        "Si usan redes sociales, pueden analizar marketing digital.",
        "Si tienen tiendas físicas y virtuales, pueden analizar omnicanalidad.",
        "Si hay promociones o ventas frecuentes, pueden analizar datos de clientes."
    ]
}

# =====================================================
# 2. Preguntas del test
# =====================================================

preguntas = [
    "1. ¿Qué tipo de temas te llaman más la atención?\n a) Tecnología e innovación\n b) Procesos y producción\n c) Salud y bienestar\n d) Medio ambiente\n e) Negocios y clientes",
    "2. ¿Qué tipo de problemas te gustaría resolver?\n a) Uso de tecnología y datos\n b) Eficiencia de procesos\n c) Problemas sociales\n d) Problemas ambientales\n e) Mercado y clientes",
    "3. ¿Qué tecnología te interesa más?\n a) Inteligencia Artificial\n b) Automatización\n c) Apps sociales\n d) Energías limpias\n e) Plataformas digitales",
    "4. ¿En qué te sientes más cómodo?\n a) Analizar datos\n b) Organizar procesos\n c) Trabajar con personas\n d) Impacto ambiental\n e) Ideas de negocio",
    "5. ¿Qué tipo de empresa prefieres?\n a) Startup tecnológica\n b) Empresa industrial o logística\n c) Institución social\n d) Empresa ambiental\n e) Empresa comercial",
    "6. ¿Qué impacto quieres generar?\n a) Transformación digital\n b) Eficiencia operativa\n c) Bienestar social\n d) Sostenibilidad\n e) Crecimiento económico",
    "7. ¿Qué entorno prefieres?\n a) Innovador\n b) Estructurado\n c) Social\n d) Ambiental\n e) Comercial",
    "8. ¿Qué te motiva más?\n a) Crear tecnología\n b) Mejorar procesos\n c) Ayudar personas\n d) Cuidar el ambiente\n e) Generar negocio"
]

# =====================================================
# 3. Entrada de datos
# =====================================================

print("TEST GRUPAL DE AFINIDAD SECTORIAL - TREND TECH")
print("Este ejercicio identifica el sector económico más afín al equipo.\n")

while True:
    try:
        num_integrantes = int(input("Ingrese el número de integrantes del equipo (máximo 6): "))
        if 1 <= num_integrantes <= 6:
            break
        else:
            print("Debe ingresar un número entre 1 y 6.")
    except ValueError:
        print("Ingrese un número válido.")

conteo_grupal = {"a": 0, "b": 0, "c": 0, "d": 0, "e": 0}
resultados_individuales = {}

for i in range(1, num_integrantes + 1):
    print("\n" + "="*60)
    nombre = input(f"Ingrese el nombre del integrante {i}: ")
    print(f"\nRespuestas de {nombre}")

    respuestas_integrante = []

    for pregunta in preguntas:
        while True:
            print("\n" + pregunta)
            respuesta = input("Tu respuesta (a, b, c, d o e): ").lower().strip()

            if respuesta in ["a", "b", "c", "d", "e"]:
                respuestas_integrante.append(respuesta)
                conteo_grupal[respuesta] += 1
                break
            else:
                print("Respuesta inválida. Escriba solo a, b, c, d o e.")

    resultados_individuales[nombre] = respuestas_integrante

# =====================================================
# 4. Análisis de resultados
# =====================================================

max_valor = max(conteo_grupal.values())
opciones_ganadoras = [op for op, val in conteo_grupal.items() if val == max_valor]

# =====================================================
# 5. Resultados grupales
# =====================================================

print("\n" + "="*70)
print("RESULTADO GENERAL DEL EQUIPO")
print("="*70)

print("\nResumen de respuestas del grupo:")
for opcion, cantidad in conteo_grupal.items():
    print(f"Opción {opcion.upper()}: {cantidad} respuesta(s) → {sectores[opcion]}")

print("\nFactor común principal del grupo:")

if len(opciones_ganadoras) == 1:
    opcion_ganadora = opciones_ganadoras[0]
    print(f"El sector más afín para el equipo es: {sectores[opcion_ganadora]}")

    print("\nInterpretación:")
    print(interpretaciones[opcion_ganadora])

else:
    print("Se presentó un empate entre estos sectores:")
    for op in opciones_ganadoras:
        print(f"- {sectores[op]}")

    print("\nInterpretación:")
    print("El grupo tiene intereses diversos. Puede elegir una empresa que combine estos sectores o discutir cuál ofrece mejores oportunidades para aplicar tecnología.")

# =====================================================
# 6. Inferencias para buscar empresa y problema
# =====================================================

print("\n" + "="*70)
print("INFERENCIAS PARA BUSCAR EMPRESAS EN COLOMBIA")
print("="*70)

if len(opciones_ganadoras) == 1:
    print(f"\nComo el equipo tiene afinidad con {sectores[opcion_ganadora]}, deberían buscar empresas donde puedan observar lo siguiente:\n")

    for item in inferencias[opcion_ganadora]:
        print(f"- {item}")

else:
    print("\nComo hay empate, revisen las pistas de cada sector ganador:\n")

    for op in opciones_ganadoras:
        print(f"\nSector: {sectores[op]}")
        for item in inferencias[op]:
            print(f"- {item}")

# =====================================================
# 7. Recomendación estratégica final
# =====================================================

print("\n" + "="*70)
print("RECOMENDACIÓN PARA EL PROYECTO TREND TECH")
print("="*70)

print("""
Ahora el equipo debe hacer lo siguiente:

1. Buscar 10 empresas en Colombia relacionadas con el sector ganador.
2. Revisar la página web de cada empresa.
3. Identificar señales de problemas o necesidades.
4. Elegir una empresa con información disponible.
5. Definir un problema inicial.
6. Proponer una posible solución tecnológica.

Pregunta clave:
¿Dónde está perdiendo valor esta empresa por no usar bien la tecnología?
""")

# =====================================================
# 8. Gráfica de resultados grupales
# =====================================================

nombres_sectores = [sectores[op] for op in conteo_grupal.keys()]
valores = list(conteo_grupal.values())

plt.figure(figsize=(11, 5))
plt.bar(nombres_sectores, valores)
plt.title("Afinidad sectorial del equipo - Trend Tech")
plt.xlabel("Sectores económicos")
plt.ylabel("Cantidad de respuestas")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# =====================================================
# 9. Resumen individual
# =====================================================

print("\n" + "="*70)
print("RESUMEN INDIVIDUAL")
print("="*70)

for nombre, respuestas in resultados_individuales.items():
    conteo_individual = {op: respuestas.count(op) for op in ["a", "b", "c", "d", "e"]}
    max_individual = max(conteo_individual.values())
    opciones_individuales = [op for op, val in conteo_individual.items() if val == max_individual]

    print(f"\n{nombre}:")
    if len(opciones_individuales) == 1:
        op_ind = opciones_individuales[0]
        print(f"Sector más afín individual: {sectores[op_ind]}")
    else:
        print("Tiene afinidad mixta entre:")
        for op in opciones_individuales:
            print(f"- {sectores[op]}")
```
