```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ==================================================
# 1. ESCRIBA AQUÍ EL NOMBRE EXACTO DEL ARCHIVO
# ==================================================

nombre_archivo = "Base_Datos_Dashboard_TrendTech_SmartRouteAI.xlsx"

# ==================================================
# 2. CARGAR HOJAS DEL EXCEL
# ==================================================

datos = pd.read_excel(nombre_archivo, sheet_name="Datos_KPI")
procesos = pd.read_excel(nombre_archivo, sheet_name="Procesos")
riesgos = pd.read_excel(nombre_archivo, sheet_name="Riesgos")
beneficios = pd.read_excel(nombre_archivo, sheet_name="Beneficios")

# ==================================================
# 3. ZONA EDITABLE POR LOS ESTUDIANTES
# ==================================================

nombre_proyecto = "Nombre del proyecto"

indicador_positivo = "Entregas a tiempo"
indicador_negativo = "Retrasos"
indicador_costo = "Costo promedio por entrega"

orden_meses = ["Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio"]

# ==================================================
# 4. PREPARAR DATOS
# ==================================================

kpi = datos.pivot_table(
    index="Mes",
    columns="Indicador",
    values=["Estado actual", "Estado proyectado con tecnología"],
    aggfunc="mean"
).reset_index()

kpi["Mes"] = pd.Categorical(
    kpi["Mes"],
    categories=orden_meses,
    ordered=True
)

kpi = kpi.sort_values("Mes")
meses = kpi["Mes"].astype(str)

def obtener_columna(escenario, indicador):
    return kpi[(escenario, indicador)]

# ==================================================
# 5. CREAR DASHBOARD COMPARATIVO
# ==================================================

plt.figure(figsize=(22, 13))

plt.suptitle(
    f"Dashboard Comparativo: Estado Actual vs Proyectado - {nombre_proyecto}",
    fontsize=22,
    fontweight="bold"
)

# Gráfica 1: Indicador positivo
plt.subplot(2, 3, 1)
plt.plot(
    meses,
    obtener_columna("Estado actual", indicador_positivo) * 100,
    marker="o",
    linewidth=3,
    color="#E74C3C",
    label="Estado actual"
)
plt.plot(
    meses,
    obtener_columna("Estado proyectado con tecnología", indicador_positivo) * 100,
    marker="o",
    linewidth=3,
    color="#27AE60",
    label="Proyectado con tecnología"
)
plt.title(f"Mejora en {indicador_positivo}", fontweight="bold")
plt.ylabel("%")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.legend()

# Gráfica 2: Indicador negativo
plt.subplot(2, 3, 2)
x = np.arange(len(meses))
ancho = 0.35

plt.bar(
    x - ancho/2,
    obtener_columna("Estado actual", indicador_negativo) * 100,
    width=ancho,
    color="#C0392B",
    label="Estado actual"
)

plt.bar(
    x + ancho/2,
    obtener_columna("Estado proyectado con tecnología", indicador_negativo) * 100,
    width=ancho,
    color="#2ECC71",
    label="Proyectado con tecnología"
)

plt.title(f"Reducción de {indicador_negativo}", fontweight="bold")
plt.ylabel("%")
plt.xticks(x, meses, rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.legend()

# Gráfica 3: Costo, tiempo u otro indicador cuantitativo
plt.subplot(2, 3, 3)
plt.plot(
    meses,
    obtener_columna("Estado actual", indicador_costo),
    marker="o",
    linewidth=3,
    color="#D35400",
    label="Estado actual"
)
plt.plot(
    meses,
    obtener_columna("Estado proyectado con tecnología", indicador_costo),
    marker="o",
    linewidth=3,
    color="#1ABC9C",
    label="Proyectado con tecnología"
)
plt.title(f"Comparación de {indicador_costo}", fontweight="bold")
plt.ylabel("Valor")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.legend()

# Gráfica 4: Procesos antes y después
plt.subplot(2, 3, 4)
y = np.arange(len(procesos["Proceso"]))

plt.barh(
    y - ancho/2,
    procesos["Nivel actual (1-5)"],
    height=ancho,
    color="#95A5A6",
    label="Estado actual"
)

plt.barh(
    y + ancho/2,
    procesos["Nivel proyectado (1-5)"],
    height=ancho,
    color="#3498DB",
    label="Proyectado con tecnología"
)

plt.yticks(y, procesos["Proceso"])
plt.title("Comparación del nivel de procesos", fontweight="bold")
plt.xlabel("Nivel 1 a 5")
plt.xlim(0, 5)
plt.grid(axis="x", alpha=0.3)
plt.legend()

# Gráfica 5: Riesgos antes y después
plt.subplot(2, 3, 5)

riesgos_ordenados = riesgos.sort_values(
    "Nivel riesgo actual",
    ascending=True
)

y = np.arange(len(riesgos_ordenados["Riesgo/Causa crítica"]))

plt.barh(
    y - ancho/2,
    riesgos_ordenados["Nivel riesgo actual"],
    height=ancho,
    color="#E67E22",
    label="Estado actual"
)

plt.barh(
    y + ancho/2,
    riesgos_ordenados["Nivel riesgo proyectado"],
    height=ancho,
    color="#27AE60",
    label="Proyectado con tecnología"
)

plt.yticks(y, riesgos_ordenados["Riesgo/Causa crítica"])
plt.title("Reducción de riesgos críticos", fontweight="bold")
plt.xlabel("Probabilidad x Impacto")
plt.grid(axis="x", alpha=0.3)
plt.legend()

# Gráfica 6: Beneficios esperados
plt.subplot(2, 3, 6)

x = np.arange(len(beneficios["Beneficio esperado"]))

plt.bar(
    x - ancho/2,
    beneficios["Valor actual"],
    width=ancho,
    color="#7F8C8D",
    label="Estado actual"
)

plt.bar(
    x + ancho/2,
    beneficios["Valor proyectado"],
    width=ancho,
    color="#9B59B6",
    label="Proyectado con tecnología"
)

plt.title("Beneficios actuales vs proyectados", fontweight="bold")
plt.ylabel("Valor")
plt.xticks(
    x,
    beneficios["Beneficio esperado"],
    rotation=45,
    ha="right"
)
plt.grid(axis="y", alpha=0.3)
plt.legend()

plt.tight_layout(rect=[0, 0, 1, 0.95])

plt.savefig(
    f"Dashboard_Comparativo_{nombre_proyecto}.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
