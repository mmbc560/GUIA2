```python

# ==============================================================
# MBA - OPERATIONAL EXCELLENCE LAB 2.0
# DIAGNÓSTICO GERENCIAL LEAN SIX SIGMA
# Google Colab | Curso de Excelencia Operacional
#
# Incluye:
# - Contexto empresarial
# - 32 preguntas de diagnóstico
# - Muda, Mura, Muri, Lean y Six Sigma
# - Ejemplos y evidencias
# - 6 visualizaciones con diferentes colores
# - Interpretaciones específicas
# - Recomendaciones gerenciales
# - Ruta DMAIC
# - Reflexión estratégica
# - Referencias bibliográficas
# - Exportación de resultados
#
# IMPORTANTE:
# Este es un instrumento académico exploratorio.

# ==============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import ipywidgets as widgets
import textwrap
import re

from IPython.display import display, clear_output, HTML
from google.colab import files
from datetime import datetime


# ==============================================================
# 1. CONFIGURACIÓN VISUAL
# ==============================================================

plt.rcParams["figure.dpi"] = 110
plt.rcParams["font.size"] = 10

AZUL = "#123B70"

COLORES = {
    "MUDA - Transporte": "#E74C3C",
    "MUDA - Inventario": "#F39C12",
    "MUDA - Movimiento": "#3498DB",
    "MUDA - Espera": "#8E44AD",
    "MUDA - Sobreproducción": "#16A085",
    "MUDA - Sobreprocesamiento": "#D35400",
    "MUDA - Defectos": "#C0392B",
    "MUDA - Talento no utilizado": "#2980B9",
    "MURA - Irregularidad": "#F1C40F",
    "MURI - Sobrecarga": "#E67E22",
    "LEAN - Flujo y valor": "#27AE60",
    "SIX SIGMA - Calidad y variabilidad": "#6C3483"
}


def titulo(texto):
    display(HTML(
        "<h2 style='color:#123B70'>" +
        texto +
        "</h2>"
    ))


def parrafo(texto):
    display(HTML(
        "<div style='padding:10px;"
        "background:#F2F6FA;"
        "border-left:4px solid #2474A6;"
        "margin:8px 0'>"
        + texto +
        "</div>"
    ))


titulo("MBA | OPERATIONAL EXCELLENCE LAB 2.0")

parrafo(
    "<b>Objetivo:</b> identificar desperdicios, "
    "irregularidades, sobrecargas, problemas de flujo "
    "y oportunidades de mejora mediante un diagnóstico "
    "gerencial exploratorio."
)


# ==============================================================
# 2. DATOS DE LA EMPRESA
# ==============================================================

empresa = widgets.Text(
    description="Empresa:",
    placeholder="Nombre o seudónimo",
    style={"description_width": "140px"},
    layout=widgets.Layout(width="700px")
)

sector = widgets.Dropdown(
    options=[
        "Servicios financieros",
        "Salud",
        "Logística y transporte",
        "Manufactura",
        "Educación",
        "Comercio",
        "Tecnología",
        "Sector público",
        "Otro"
    ],
    description="Sector:",
    style={"description_width": "140px"},
    layout=widgets.Layout(width="700px")
)

proceso = widgets.Text(
    description="Proceso:",
    placeholder="Ejemplo: gestión de solicitudes",
    style={"description_width": "140px"},
    layout=widgets.Layout(width="700px")
)

problema = widgets.Textarea(
    description="Problema:",
    placeholder="Describa el problema operacional",
    style={"description_width": "140px"},
    layout=widgets.Layout(width="800px", height="110px")
)

estrategia = widgets.Textarea(
    description="Estrategia:",
    placeholder="¿Qué objetivo estratégico se ve afectado?",
    style={"description_width": "140px"},
    layout=widgets.Layout(width="800px", height="90px")
)

titulo("FASE 1. CONTEXTO ORGANIZACIONAL")

display(
    empresa,
    sector,
    proceso,
    problema,
    estrategia
)


# ==============================================================
# 3. BANCO DE PREGUNTAS
# ==============================================================

PREGUNTAS = {

    "MUDA - Transporte": [
        "¿Se trasladan documentos, materiales o solicitudes innecesariamente?",
        "¿Los traspasos entre áreas generan demoras?"
    ],

    "MUDA - Inventario": [
        "¿Se acumulan solicitudes, pedidos o trabajos pendientes?",
        "¿Existen acumulaciones superiores a la capacidad de respuesta?"
    ],

    "MUDA - Movimiento": [
        "¿Los empleados buscan repetidamente información o herramientas?",
        "¿Se realizan movimientos o acciones innecesarias?"
    ],

    "MUDA - Espera": [
        "¿Los clientes esperan aprobaciones o respuestas?",
        "¿Los procesos se detienen por falta de información o recursos?"
    ],

    "MUDA - Sobreproducción": [
        "¿Se generan informes o documentos que nadie utiliza?",
        "¿Se realiza trabajo antes de que sea necesario?"
    ],

    "MUDA - Sobreprocesamiento": [
        "¿Existen aprobaciones o verificaciones duplicadas?",
        "¿Se solicita varias veces la misma información?"
    ],

    "MUDA - Defectos": [
        "¿Se presentan errores, reclamaciones o reprocesos?",
        "¿Las correcciones consumen recursos importantes?"
    ],

    "MUDA - Talento no utilizado": [
        "¿Se desaprovechan las ideas de los empleados?",
        "¿Las personas realizan tareas por debajo de sus capacidades?"
    ],

    "MURA - Irregularidad": [
        "¿La demanda cambia significativamente durante el día o mes?",
        "¿Los tiempos de ejecución varían entre casos similares?",
        "¿Se presentan cambios inesperados en la programación?",
        "¿La información llega de manera irregular?"
    ],

    "MURI - Sobrecarga": [
        "¿Los empleados trabajan frecuentemente bajo presión excesiva?",
        "¿Los equipos operan cerca de sus límites de capacidad?",
        "¿Se necesitan horas extra para cumplir las metas?",
        "¿Las metas superan la capacidad disponible?"
    ],

    "LEAN - Flujo y valor": [
        "¿Existen actividades que no agregan valor al cliente?",
        "¿Se presentan cuellos de botella recurrentes?",
        "¿Faltan procedimientos estandarizados?",
        "¿Se desconoce el tiempo total del proceso?"
    ],

    "SIX SIGMA - Calidad y variabilidad": [
        "¿Los resultados varían considerablemente entre casos?",
        "¿Existen defectos o incumplimientos frecuentes?",
        "¿Se toman decisiones sin datos suficientes?",
        "¿Faltan indicadores para controlar calidad y variabilidad?"
    ]
}


# ==============================================================
# 4. BASE DE CONOCIMIENTO GERENCIAL
#
# Cada dimensión tiene:
# - Significado
# - Riesgo empresarial
# - Herramientas
# - Datos que deben investigarse
# - Recomendación gerencial
# - Referencia académica
# ==============================================================

CONOCIMIENTO = {

"MUDA - Transporte": {
"concepto":
"Movimientos o transferencias innecesarias de materiales, "
"documentos o información entre etapas.",
"riesgo":
"Incrementa costos de transferencia, tiempos de ciclo "
"y posibilidades de pérdida de información.",
"herramientas":
"Value Stream Mapping (VSM), diagrama spaghetti "
"y rediseño de interfaces entre áreas.",
"datos":
"Número de transferencias, distancia, tiempo por traspaso "
"y frecuencia de devoluciones.",
"accion":
"Mapear las transferencias y evaluar cuáles pueden "
"eliminarse, combinarse o simplificarse.",
"referencia":"Womack & Jones (2003); Ohno (1988)"
},

"MUDA - Inventario": {
"concepto":
"Acumulación de trabajo en proceso, solicitudes, "
"expedientes, productos o tareas pendientes.",
"riesgo":
"Puede aumentar los tiempos de entrega, ocultar cuellos "
"de botella y generar costos de almacenamiento o gestión.",
"herramientas":
"Kanban, límites WIP, análisis de colas, VSM "
"y Ley de Little cuando se cumplen sus supuestos.",
"datos":
"Solicitudes pendientes por día, entradas, salidas, "
"antigüedad de casos y tiempo de ciclo.",
"accion":
"Identificar dónde se acumula el trabajo y contrastar "
"la tasa de llegada con la capacidad de procesamiento.",
"referencia":"Little (1961); Womack & Jones (2003)"
},

"MUDA - Movimiento": {
"concepto":
"Desplazamientos físicos o acciones digitales innecesarias "
"realizadas por los trabajadores.",
"riesgo":
"Consume tiempo, reduce productividad y puede incrementar "
"fatiga o errores.",
"herramientas":
"5S, estudio de métodos, diagrama spaghetti "
"y análisis de tareas digitales.",
"datos":
"Tiempo de búsqueda, número de movimientos, clics "
"y desplazamientos por transacción.",
"accion":
"Observar directamente el trabajo y rediseñar "
"la ubicación de recursos y el acceso a información.",
"referencia":"Liker (2004); Ohno (1988)"
},

"MUDA - Espera": {
"concepto":
"Tiempo improductivo mientras personas, solicitudes "
"o recursos esperan la siguiente actividad.",
"riesgo":
"Deteriora la experiencia del cliente, alarga el lead time "
"y puede elevar costos de servicio.",
"herramientas":
"VSM, análisis de colas, estudio de tiempos "
"y análisis de cuellos de botella.",
"datos":
"Tiempo de espera por etapa, tiempo de servicio, "
"colas y cumplimiento de acuerdos de servicio.",
"accion":
"Separar el tiempo de procesamiento del tiempo de espera "
"y localizar las principales restricciones.",
"referencia":"Womack & Jones (2003); Liker (2004)"
},

"MUDA - Sobreproducción": {
"concepto":
"Producción de bienes, información o trabajo antes "
"de necesitarse o en cantidades superiores a la demanda.",
"riesgo":
"Consume capacidad, genera acumulaciones y puede "
"desviar recursos de actividades prioritarias.",
"herramientas":
"Pull systems, Kanban, revisión de reportes "
"y análisis de demanda real.",
"datos":
"Entregables no utilizados, frecuencia de generación, "
"volumen producido y demanda efectiva.",
"accion":
"Determinar quién utiliza cada resultado y eliminar "
"trabajos que carecen de un destinatario o necesidad.",
"referencia":"Ohno (1988); Womack & Jones (2003)"
},

"MUDA - Sobreprocesamiento": {
"concepto":
"Controles, revisiones, aprobaciones o actividades "
"más complejas de lo necesario para crear valor.",
"riesgo":
"Incrementa costos administrativos, tiempo de ciclo "
"y complejidad organizacional.",
"herramientas":
"VSM, análisis de valor agregado, simplificación "
"de procesos y ECRS.",
"datos":
"Número de aprobaciones, revisiones repetidas, "
"horas de procesamiento y devoluciones.",
"accion":
"Evaluar el propósito de cada control y simplificar "
"sin eliminar requisitos legales o controles necesarios.",
"referencia":"Womack & Jones (2003); Liker (2004)"
},

"MUDA - Defectos": {
"concepto":
"Errores, incumplimientos o resultados que requieren "
"corrección, devolución o reproceso.",
"riesgo":
"Eleva costos de no calidad, reduce confiabilidad "
"y afecta satisfacción del cliente.",
"herramientas":
"Pareto, 5 porqués, Ishikawa, poka-yoke "
"y análisis de capacidad del proceso.",
"datos":
"Tasa de defectos, frecuencia de reprocesos, "
"costo de no calidad y causas registradas.",
"accion":
"Definir qué constituye un defecto, medir su frecuencia "
"y validar sus causas antes de intervenir.",
"referencia":"Montgomery (2020); ASQ (s. f.)"
},

"MUDA - Talento no utilizado": {
"concepto":
"Desaprovechamiento de habilidades, experiencia, "
"conocimiento y propuestas de mejora del personal.",
"riesgo":
"Limita innovación, aprendizaje organizacional "
"y capacidad de mejora continua.",
"herramientas":
"Kaizen, círculos de mejora, matriz de habilidades "
"y participación del personal.",
"datos":
"Propuestas recibidas, porcentaje implementado, "
"habilidades disponibles y participación.",
"accion":
"Involucrar a quienes ejecutan el proceso "
"en la identificación y experimentación de mejoras.",
"referencia":"Liker (2004); Womack & Jones (2003)"
},

"MURA - Irregularidad": {
"concepto":
"Variaciones y desequilibrios en demanda, carga "
"de trabajo, programación o flujo operacional.",
"riesgo":
"Puede producir picos de congestión, tiempos "
"impredecibles y dificultades de coordinación.",
"herramientas":
"Heijunka, análisis temporal, pronósticos, "
"balanceo de carga y gestión de capacidad.",
"datos":
"Demanda por hora o día, variabilidad del tiempo "
"de ciclo, distribución de carga y estacionalidad.",
"accion":
"Identificar patrones de demanda y distinguir "
"la variabilidad externa de la creada por la organización.",
"referencia":"Lean Enterprise Institute (s. f.); Liker (2004)"
},

"MURI - Sobrecarga": {
"concepto":
"Exigencia excesiva sobre personas, equipos "
"o procesos frente a su capacidad sostenible.",
"riesgo":
"Puede incrementar errores, fatiga, averías, "
"horas extra y vulnerabilidad operacional.",
"herramientas":
"Análisis capacidad-demanda, balanceo de trabajo, "
"ergonomía y mantenimiento preventivo.",
"datos":
"Utilización por recurso, horas extra, demanda pico, "
"ausentismo y fallas de equipos.",
"accion":
"Revisar cargas y metas, identificar recursos críticos "
"y evaluar capacidad de respuesta en picos.",
"referencia":"Lean Enterprise Institute (s. f.); Liker (2004)"
},

"LEAN - Flujo y valor": {
"concepto":
"Dificultades para entregar valor mediante "
"un flujo continuo, claro y coordinado.",
"riesgo":
"Genera procesos fragmentados, tiempos elevados "
"y pérdida de capacidad competitiva.",
"herramientas":
"VSM, SIPOC, trabajo estandarizado, "
"Kaizen y análisis de valor agregado.",
"datos":
"Lead time, tiempo de proceso, actividades sin valor "
"y desempeño de cada etapa.",
"accion":
"Construir un mapa del estado actual, identificar "
"restricciones y diseñar un estado futuro.",
"referencia":"Womack & Jones (2003); Rother & Shook (2003)"
},

"SIX SIGMA - Calidad y variabilidad": {
"concepto":
"Variación de resultados, defectos y debilidades "
"en la medición o control del proceso.",
"riesgo":
"Reduce predictibilidad, cumplimiento "
"y confiabilidad de la promesa de servicio.",
"herramientas":
"DMAIC, Pareto, histogramas, cartas de control, "
"análisis de capacidad y MSA cuando aplique.",
"datos":
"Observaciones individuales, límites de especificación, "
"defectos, oportunidades y confiabilidad de medición.",
"accion":
"Definir características críticas para la calidad "
"(CTQ), validar datos y estudiar las fuentes de variación.",
"referencia":"Montgomery (2020); ASQ (s. f.)"
}
}


# ==============================================================
# 5. ESCALA DE CALIFICACIÓN
# ==============================================================

ESCALA = [
    ("Seleccione una respuesta", None),
    ("0 - No ocurre", 0),
    ("1 - Rara vez", 1),
    ("2 - Algunas veces", 2),
    ("3 - Frecuentemente", 3),
    ("4 - Críticamente", 4)
]

def nivel(valor):
    if valor >= 3:
        return "ALTO"
    elif valor >= 2:
        return "MODERADO"
    elif valor >= 1:
        return "BAJO"
    return "NO REPORTADO"


# ==============================================================
# 6. CUESTIONARIO INTERACTIVO
# ==============================================================

titulo("FASE 2. DIAGNÓSTICO OPERACIONAL")

parrafo(
    "Califique cada situación entre 0 y 4. "
    "Escriba ejemplos concretos. "
    "Las puntuaciones representan percepciones, "
    "no causas raíz confirmadas."
)

CONTROLES = []

for dimension, lista in PREGUNTAS.items():

    elementos = [
        widgets.HTML(
            f"<h3 style='color:{COLORES[dimension]}'>"
            f"{dimension}</h3>"
        )
    ]

    for numero, pregunta in enumerate(lista, 1):

        selector = widgets.Dropdown(
            options=ESCALA,
            value=None,
            layout=widgets.Layout(width="350px")
        )

        evidencia = widgets.Textarea(
            placeholder=(
                "Describa un ejemplo real, "
                "una observación o un dato disponible"
            ),
            layout=widgets.Layout(
                width="750px",
                height="65px"
            )
        )

        CONTROLES.append({
            "dimension": dimension,
            "pregunta": pregunta,
            "selector": selector,
            "evidencia": evidencia
        })

        elementos.extend([
            widgets.HTML(
                f"<b>{numero}. {pregunta}</b>"
            ),
            selector,
            evidencia
        ])

    display(
        widgets.VBox(
            elementos,
            layout=widgets.Layout(
                border="1px solid #D7DFE8",
                padding="15px",
                margin="8px 0"
            )
        )
    )


# ==============================================================
# 7. PREGUNTAS DE REFLEXIÓN GERENCIAL
# ==============================================================

titulo("FASE 3. REFLEXIÓN GERENCIAL")

REFLEXIONES = {}

preguntas_reflexion = {
    "Causas":
        "¿Qué causas del Ishikawa considera prioritarias?",
    "Datos":
        "¿Qué información necesita para validar esas causas?",
    "Alternativas":
        "¿Qué dos alternativas de mejora podría evaluar?",
    "Trade-offs":
        "¿Qué costos, riesgos o sacrificios implican?",
    "Estrategia":
        "¿Cómo se relacionan con la estrategia empresarial?"
}

for clave, pregunta in preguntas_reflexion.items():

    display(HTML(f"<b>{pregunta}</b>"))

    campo = widgets.Textarea(
        layout=widgets.Layout(
            width="800px",
            height="75px"
        )
    )

    REFLEXIONES[clave] = campo
    display(campo)


# ==============================================================
# 8. FUNCIONES DE GRÁFICOS
# ==============================================================

def grafico_dimensiones(resumen):

    # GRÁFICO 1: Barras de las 12 dimensiones

    datos = resumen.sort_values()

    fig, ax = plt.subplots(figsize=(12, 7))

    colores = [
        COLORES[d] for d in datos.index
    ]

    ax.barh(
        datos.index,
        datos.values,
        color=colores
    )

    ax.set_xlim(0, 4.3)
    ax.set_xlabel("Severidad percibida (0-4)")
    ax.set_title(
        "1. Perfil de diagnóstico operacional",
        fontweight="bold"
    )

    for i, valor in enumerate(datos.values):
        ax.text(
            valor + 0.04,
            i,
            f"{valor:.2f}",
            va="center"
        )

    plt.tight_layout()
    plt.show()
    plt.close(fig)


def grafico_radar(resumen):

    # GRÁFICO 2: Radar de dimensiones

    etiquetas = list(resumen.index)
    valores = resumen.values.tolist()

    n = len(valores)

    angulos = np.linspace(
        0,
        2 * np.pi,
        n,
        endpoint=False
    ).tolist()

    valores += valores[:1]
    angulos += angulos[:1]

    fig, ax = plt.subplots(
        figsize=(11, 9),
        subplot_kw={"polar": True}
    )

    ax.plot(
        angulos,
        valores,
        color="#7D3C98",
        linewidth=2
    )

    ax.fill(
        angulos,
        valores,
        color="#BB8FCE",
        alpha=0.25
    )

    ax.scatter(
        angulos[:-1],
        valores[:-1],
        c=[
            COLORES[d] for d in etiquetas
        ],
        s=65,
        zorder=3
    )

    ax.set_xticks(angulos[:-1])

    ax.set_xticklabels(
        [
            d.replace(" - ", "\n")
            for d in etiquetas
        ],
        fontsize=8
    )

    ax.set_ylim(0, 4)

    ax.set_title(
        "2. Radar de riesgos percibidos",
        pad=35,
        fontweight="bold"
    )

    plt.tight_layout()
    plt.show()
    plt.close(fig)


def grafico_tres_m(resumen):

    # GRÁFICO 3: Comparación Muda-Mura-Muri

    muda = resumen[
        resumen.index.str.startswith("MUDA")
    ].mean()

    mura = resumen["MURA - Irregularidad"]
    muri = resumen["MURI - Sobrecarga"]

    nombres = ["MUDA", "MURA", "MURI"]
    valores = [muda, mura, muri]

    fig, ax = plt.subplots(figsize=(8, 5))

    barras = ax.bar(
        nombres,
        valores,
        color=[
            "#E74C3C",
            "#F1C40F",
            "#3498DB"
        ],
        width=0.55
    )

    ax.set_ylim(0, 4.5)
    ax.set_ylabel("Severidad percibida")
    ax.set_title(
        "3. Comparación de las tres M",
        fontweight="bold"
    )

    for barra, valor in zip(barras, valores):

        ax.text(
            barra.get_x() + barra.get_width() / 2,
            valor + 0.08,
            f"{valor:.2f}",
            ha="center",
            fontweight="bold"
        )

    plt.tight_layout()
    plt.show()
    plt.close(fig)


def grafico_muda(resumen):

    # GRÁFICO 4: Perfil de los 8 desperdicios

    datos = resumen[
        resumen.index.str.startswith("MUDA")
    ]

    etiquetas = [
        d.replace("MUDA - ", "")
        for d in datos.index
    ]

    fig, ax = plt.subplots(figsize=(12, 5))

    colores = [
        COLORES[d] for d in datos.index
    ]

    ax.bar(
        etiquetas,
        datos.values,
        color=colores
    )

    ax.set_ylim(0, 4.5)

    ax.set_ylabel("Severidad percibida")

    ax.set_title(
        "4. Los ocho desperdicios Lean",
        fontweight="bold"
    )

    plt.xticks(
        rotation=35,
        ha="right"
    )

    for i, valor in enumerate(datos.values):
        ax.text(
            i,
            valor + 0.07,
            f"{valor:.2f}",
            ha="center"
        )

    plt.tight_layout()
    plt.show()
    plt.close(fig)


def grafico_calor(df):

    # GRÁFICO 5: Mapa de calor de preguntas
    # Las dimensiones con 2 preguntas dejan celdas vacías.

    matriz = []

    for dimension, grupo in df.groupby(
        "Dimension",
        sort=False
    ):

        valores = grupo["Calificacion"].tolist()

        valores += [
            np.nan
        ] * (4 - len(valores))

        matriz.append(valores)

    matriz = np.array(
        matriz,
        dtype=float
    )

    cmap = plt.get_cmap("YlOrRd").copy()
    cmap.set_bad("#ECEFF1")

    fig, ax = plt.subplots(figsize=(9, 8))

    im = ax.imshow(
        matriz,
        cmap=cmap,
        vmin=0,
        vmax=4,
        aspect="auto"
    )

    ax.set_yticks(
        range(len(PREGUNTAS))
    )

    ax.set_yticklabels(
        list(PREGUNTAS.keys()),
        fontsize=9
    )

    ax.set_xticks(range(4))

    ax.set_xticklabels([
        "Pregunta 1",
        "Pregunta 2",
        "Pregunta 3",
        "Pregunta 4"
    ])

    for i in range(matriz.shape[0]):
        for j in range(matriz.shape[1]):

            if not np.isnan(matriz[i, j]):

                valor = matriz[i, j]

                ax.text(
                    j,
                    i,
                    str(int(valor)),
                    ha="center",
                    va="center",
                    color=(
                        "white" if valor >= 3
                        else "black"
                    )
                )

    fig.colorbar(
        im,
        ax=ax,
        label="Severidad percibida"
    )

    ax.set_title(
        "5. Mapa de calor de las 32 preguntas",
        fontweight="bold"
    )

    plt.tight_layout()
    plt.show()
    plt.close(fig)


def grafico_prioridades(resumen):

    # GRÁFICO 6: Dimensiones que requieren investigación
    # El orden usa severidad y, para empates,
    # conserva el orden del cuestionario.
    # No es una medición de impacto económico.

    datos = resumen.sort_values(
        ascending=False,
        kind="stable"
    ).head(6)

    fig, ax = plt.subplots(figsize=(11, 5))

    colores = [
        COLORES[d] for d in datos.index
    ]

    ax.barh(
        datos.index[::-1],
        datos.values[::-1],
        color=colores[::-1]
    )

    ax.set_xlim(0, 4.4)

    ax.set_title(
        "6. Primeras dimensiones para investigar",
        fontweight="bold"
    )

    ax.set_xlabel("Severidad percibida")

    for i, valor in enumerate(datos.values[::-1]):
        ax.text(
            valor + 0.05,
            i,
            f"{valor:.2f}",
            va="center"
        )

    plt.tight_layout()
    plt.show()
    plt.close(fig)


# ==============================================================
# 9. INTERPRETACIONES ESPECÍFICAS
# ==============================================================

def interpretar_dimension(dimension, valor):

    info = CONOCIMIENTO[dimension]

    print("\n" + "=" * 65)
    print(dimension.upper())
    print("=" * 65)

    print(
        f"CALIFICACIÓN: {valor:.2f}/4 "
        f"| NIVEL: {nivel(valor)}"
    )

    print("\nSIGNIFICADO:")
    print(textwrap.fill(info["concepto"], 95))

    if valor >= 3:
        print(
            "\nINTERPRETACIÓN: La percepción del equipo "
            "indica presencia frecuente o crítica "
            "de esta situación."
        )

    elif valor >= 2:
        print(
            "\nINTERPRETACIÓN: Se reporta una situación "
            "intermitente que merece revisión."
        )

    elif valor >= 1:
        print(
            "\nINTERPRETACIÓN: Se reporta una situación "
            "ocasional que conviene monitorear."
        )

    else:
        print(
            "\nINTERPRETACIÓN: El equipo no reporta "
            "presencia de esta situación. "
            "Esto no demuestra su ausencia."
        )

    print("\nRIESGO GERENCIAL:")
    print(textwrap.fill(info["riesgo"], 95))

    print("\nHERRAMIENTAS RECOMENDADAS:")
    print(textwrap.fill(info["herramientas"], 95))

    print("\nDATOS PARA VALIDAR:")
    print(textwrap.fill(info["datos"], 95))

    print("\nACCIÓN SUGERIDA:")
    print(textwrap.fill(info["accion"], 95))

    print("\nFUNDAMENTO:")
    print(info["referencia"])


# ==============================================================
# 10. ANÁLISIS INTEGRADO DE MUDA, MURA Y MURI
# ==============================================================

def interpretar_tres_m(resumen):

    muda = resumen[
        resumen.index.str.startswith("MUDA")
    ].mean()

    mura = resumen["MURA - Irregularidad"]
    muri = resumen["MURI - Sobrecarga"]

    titulo("ANÁLISIS INTEGRADO: MUDA, MURA Y MURI")

    print(f"MUDA: {muda:.2f}")
    print(f"MURA: {mura:.2f}")
    print(f"MURI: {muri:.2f}")

    print("\nINTERPRETACIÓN GERENCIAL:")

    if mura >= 3 and muri >= 3:

        print(textwrap.fill(
            "El equipo reporta simultáneamente "
            "irregularidad elevada y sobrecarga elevada. "
            "Una hipótesis es que los picos de demanda "
            "o la programación desequilibrada generan "
            "presión sobre personas y equipos. "
            "Se debe comprobar comparando demanda, "
            "capacidad y carga de trabajo por periodo.",
            95
        ))

    elif mura >= 3:

        print(textwrap.fill(
            "La irregularidad destaca como una posible "
            "condición estructural. Conviene investigar "
            "fluctuaciones, programación y distribución "
            "de recursos antes de proponer aumentos "
            "permanentes de capacidad.",
            95
        ))

    elif muri >= 3:

        print(textwrap.fill(
            "La sobrecarga destaca como riesgo. "
            "Conviene revisar capacidad efectiva, "
            "metas, utilización, fallas y horas extra. "
            "La causa podría ser insuficiencia de "
            "recursos, diseño del proceso u otros factores.",
            95
        ))

    else:

        print(textwrap.fill(
            "No se observa una percepción dominante "
            "de irregularidad o sobrecarga. "
            "Conviene revisar los desperdicios específicos "
            "y validar si el diagnóstico coincide "
            "con los registros operacionales.",
            95
        ))

    if muda >= 3:

        print("\nMUDA:")
        print(textwrap.fill(
            "Los desperdicios reportados son elevados. "
            "Antes de eliminar actividades, distinga "
            "las que no agregan valor de aquellas "
            "que son necesarias por seguridad, "
            "regulación o control.",
            95
        ))

    print("\nRELACIÓN TEÓRICA:")

    print(textwrap.fill(
        "Mura puede contribuir a Muri y ambos pueden "
        "generar Muda. Esta secuencia es una hipótesis "
        "de funcionamiento, no una relación causal "
        "demostrada por el cuestionario.",
        95
    ))


# ==============================================================
# 11. PLAN DMAIC
# ==============================================================

def mostrar_dmaic():

    titulo("RUTA DMAIC PARA EL PROYECTO")

    fases = {

        "DEFINE":
        "Precisar el problema, el cliente afectado, "
        "el alcance, el objetivo y los requisitos críticos.",

        "MEASURE":
        "Recolectar datos de tiempos, defectos, "
        "demanda, capacidad y desempeño. "
        "Validar la confiabilidad de la medición.",

        "ANALYZE":
        "Contrastar las causas del Ishikawa con datos. "
        "Utilizar Pareto, estratificación y análisis "
        "estadístico cuando sea pertinente.",

        "IMPROVE":
        "Evaluar alternativas de rediseño, "
        "estandarización, balanceo, automatización "
        "o mantenimiento. Realizar una prueba piloto.",

        "CONTROL":
        "Definir responsables, indicadores, "
        "estándares, frecuencia de seguimiento "
        "y planes de reacción."
    }

    for fase, explicacion in fases.items():

        print(f"\n{fase}")
        print(textwrap.fill(explicacion, 95))


# ==============================================================
# 12. REFERENCIAS BIBLIOGRÁFICAS
# ==============================================================

REFERENCIAS = [
    "American Society for Quality. (s. f.). "
    "DMAIC process: Define, measure, analyze, improve, control. "
    "https://asq.org/quality-resources/dmaic",

    "Lean Enterprise Institute. (s. f.). "
    "Muda, mura, muri. "
    "https://www.lean.org/lexicon-terms/muda-mura-muri/",

    "Liker, J. K. (2004). "
    "The Toyota way: 14 management principles "
    "from the world's greatest manufacturer. McGraw-Hill.",

    "Little, J. D. C. (1961). "
    "A proof for the queuing formula: L = lambda W. "
    "Operations Research, 9(3), 383-387. "
    "https://doi.org/10.1287/opre.9.3.383",

    "Montgomery, D. C. (2020). "
    "Introduction to statistical quality control "
    "(8th ed.). Wiley.",

    "Ohno, T. (1988). "
    "Toyota production system: Beyond large-scale production. "
    "Productivity Press.",

    "Rother, M., & Shook, J. (2003). "
    "Learning to see: Value-stream mapping to create "
    "value and eliminate muda. Lean Enterprise Institute.",

    "Womack, J. P., & Jones, D. T. (2003). "
    "Lean thinking: Banish waste and create wealth "
    "in your corporation. Free Press."
]


# ==============================================================
# 13. BOTÓN PARA GENERAR EL DIAGNÓSTICO
# ==============================================================

boton = widgets.Button(
    description="GENERAR DIAGNÓSTICO AVANZADO",
    button_style="success",
    layout=widgets.Layout(
        width="350px",
        height="50px"
    )
)

salida = widgets.Output()


# ==============================================================
# 14. FUNCIÓN PRINCIPAL
# ==============================================================

def generar_diagnostico(b):

    with salida:

        clear_output(wait=True)

        if not problema.value.strip():

            print(
                "ERROR: Primero describa "
                "el problema operacional."
            )
            return

        registros = []

        for item in CONTROLES:

            registros.append({
                "Dimension": item["dimension"],
                "Pregunta": item["pregunta"],
                "Calificacion": item["selector"].value,
                "Evidencia": item["evidencia"].value
            })

        df = pd.DataFrame(registros)

        pendientes = df["Calificacion"].isna().sum()

        if pendientes > 0:

            print(
                f"Faltan {pendientes} preguntas."
            )
            return

        resumen = (
            df.groupby("Dimension", sort=False)
            ["Calificacion"]
            .mean()
        )

        titulo("INFORME GERENCIAL DE EXCELENCIA OPERACIONAL")

        print("Empresa:", empresa.value)
        print("Sector:", sector.value)
        print("Proceso:", proceso.value)
        print("Problema:", problema.value)
        print("Estrategia:", estrategia.value)

        print("\nRESULTADOS GENERALES")

        print(
            resumen.sort_values(
                ascending=False,
                kind="stable"
            ).round(2).to_string()
        )

        # ------------------------------------------------------
        # 15. VISUALIZACIONES
        # ------------------------------------------------------

        titulo("DASHBOARD DE DIAGNÓSTICO")

        grafico_dimensiones(resumen)
        grafico_radar(resumen)
        grafico_tres_m(resumen)
        grafico_muda(resumen)
        grafico_calor(df)
        grafico_prioridades(resumen)

        # ------------------------------------------------------
        # 16. INTERPRETACIONES INDIVIDUALES
        # ------------------------------------------------------

        titulo("INTERPRETACIÓN POR DIMENSIÓN")

        for dimension, valor in resumen.items():

            interpretar_dimension(
                dimension,
                valor
            )

        # ------------------------------------------------------
        # 17. INTERPRETACIÓN INTEGRADA
        # ------------------------------------------------------

        interpretar_tres_m(resumen)

        # ------------------------------------------------------
        # 18. SITUACIONES CON MAYOR PUNTUACIÓN
        # ------------------------------------------------------

        titulo("SITUACIONES PARA INVESTIGAR")

        criticas = df.sort_values(
            "Calificacion",
            ascending=False,
            kind="stable"
        ).head(8)

        for _, fila in criticas.iterrows():

            print("\nDimensión:", fila["Dimension"])
            print("Situación:", fila["Pregunta"])
            print("Calificación:", fila["Calificacion"])
            print("Ejemplo:", fila["Evidencia"])

        print(
            "\nNOTA: Las preguntas empatadas tienen "
            "la misma severidad percibida. "
            "El orden mostrado no representa "
            "mayor impacto económico ni causalidad."
        )

        # ------------------------------------------------------
        # 19. RECOMENDACIONES GERENCIALES
        # ------------------------------------------------------

        titulo("RECOMENDACIONES PARA LA DIRECCIÓN")

        print("""
1. VALIDAR EL PROBLEMA:
Contrastar las percepciones con registros reales.

2. IDENTIFICAR VALOR:
Distinguir actividades que agregan valor,
actividades necesarias y desperdicios eliminables.

3. ESTUDIAR VARIABILIDAD:
Analizar diferencias por periodo, cliente,
turno, equipo, producto o servicio.

4. REVISAR CAPACIDAD:
Comparar carga de trabajo con capacidad efectiva.

5. PRIORIZAR CON EVIDENCIA:
Evaluar frecuencia, impacto económico,
experiencia del cliente y viabilidad.

6. EVITAR AUTOMATIZAR DESPERDICIOS:
Simplificar primero el proceso cuando corresponda.

7. REALIZAR PILOTOS:
Probar alternativas antes de inversiones importantes.

8. CONECTAR CON LA ESTRATEGIA:
Explicar cómo la mejora contribuiría a costo,
calidad, rapidez, confiabilidad o diferenciación.
""")

        # ------------------------------------------------------
        # 20. DMAIC
        # ------------------------------------------------------

        mostrar_dmaic()

        # ------------------------------------------------------
        # 21. REFLEXIÓN GERENCIAL
        # ------------------------------------------------------

        titulo("REFLEXIÓN ESTRATÉGICA DEL EQUIPO")

        for clave, campo in REFLEXIONES.items():

            print(
                f"\n{clave.upper()}:"
            )

            print(
                campo.value or "Sin respuesta"
            )

        # ------------------------------------------------------
        # 22. REFERENCIAS
        # ------------------------------------------------------

        titulo("REFERENCIAS BIBLIOGRÁFICAS")

        for referencia in REFERENCIAS:
            print("\n" + referencia)

        # ------------------------------------------------------
        # 23. EXPORTACIÓN CSV
        # ------------------------------------------------------

        nombre = "diagnostico_mba_lean_six_sigma_2.csv"

        df.to_csv(
            nombre,
            index=False,
            encoding="utf-8-sig"
        )

        print(
            "\nArchivo CSV generado correctamente."
        )

        files.download(nombre)


# ==============================================================
# 24. EJECUCIÓN DEL BOTÓN
# ==============================================================

boton.on_click(generar_diagnostico)

titulo("FASE 4. GENERAR INFORME")

display(boton, salida)

print("""
INSTRUCCIONES FINALES

1. Complete los datos de la empresa.
2. Responda las 32 preguntas.
3. Justifique sus respuestas con ejemplos.
4. Responda las preguntas gerenciales.
5. Presione GENERAR DIAGNÓSTICO AVANZADO.
6. Analice las seis visualizaciones.
7. Revise las interpretaciones.
8. Descargue el archivo CSV.

Las puntuaciones no representan niveles sigma,
capacidad estadística ni causas raíz confirmadas.
""")
```
