```python

# ================================================================
# UNIVERSIDAD EAN
# UNIDAD DE ESTUDIO: TREND TECH
# LABORATORIO: ISHIKAWA CON SEMÁFOROS
#
# TEMA:
# DEL PROBLEMA EMPRESARIAL A UNA SOLUCIÓN TECNOLÓGICA
#
# DOCENTE: MARIA MERCEDES BERNAL CERQUERA, PhD.
#
# OBJETIVO:
# Identificar las causas de un problema empresarial,
# evaluar su impacto y relacionar las causas prioritarias
# con posibles funcionalidades de una aplicación.
#
# MODALIDADES:
# 1. Ejemplo agrícola: AgroFresco
# 2. Proyecto propio de cada equipo
#
# RESULTADOS:
# - Tabla de diagnóstico
# - Diagrama de Ishikawa con semáforos
# - Ranking de causas
# - Comparación de categorías
# - Matriz de funcionalidades
# - Interpretaciones automáticas
# - Archivo Excel con resultados
#
# NOTA:
# Las calificaciones son valoraciones iniciales.
# Deben verificarse con información de la empresa.
# ================================================================


# ================================================================
# 1. INSTALAR E IMPORTAR LIBRERÍAS
# ================================================================

# En Google Colab normalmente están disponibles.
# Si alguna falta, ejecutar:
# !pip install pandas numpy matplotlib openpyxl

import os
import textwrap
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.lines import Line2D
from matplotlib.patches import FancyBboxPatch
from IPython.display import display

# Configuración visual
plt.rcParams["font.family"] = "DejaVu Sans"
plt.rcParams["figure.dpi"] = 130

# Colores
ROJO = "#E53935"
AMARILLO = "#F9A825"
VERDE = "#43A047"
AZUL = "#1565C0"
OSCURO = "#263238"
GRIS = "#607D8B"
FONDO = "#F5F8FC"

# Carpeta para resultados
CARPETA = "Resultados_Trend_Tech"
os.makedirs(CARPETA, exist_ok=True)


# ================================================================
# 2. FUNCIONES GENERALES
# ================================================================

def titulo(texto):
    """Muestra títulos para organizar el laboratorio."""
    print("\n" + "=" * 70)
    print(texto)
    print("=" * 70)


def ajustar_texto(texto, ancho=30):
    """Divide textos largos para facilitar su visualización."""
    return "\n".join(
        textwrap.wrap(str(texto), width=ancho)
    )


def clasificar_semaforo(impacto):
    """
    Clasificación de impacto:

    1-2: Verde = Bajo
    3: Amarillo = Medio
    4-5: Rojo = Alto
    """
    if impacto >= 4:
        return "ROJO", ROJO, "ALTA"
    elif impacto == 3:
        return "AMARILLO", AMARILLO, "MEDIA"
    else:
        return "VERDE", VERDE, "BAJA"


def pedir_numero(mensaje):
    """Solicita un número entero entre 1 y 5."""
    while True:
        try:
            numero = int(input(mensaje))

            if 1 <= numero <= 5:
                return numero

            print("Escribe un número entre 1 y 5.")

        except ValueError:
            print("Debes ingresar un número entero.")


def pedir_texto(mensaje):
    """Evita respuestas vacías."""
    while True:
        respuesta = input(mensaje).strip()

        if respuesta:
            return respuesta

        print("Debes escribir una respuesta.")


def guardar_grafica(nombre):
    """Guarda y muestra una gráfica."""
    ruta = os.path.join(CARPETA, nombre)

    plt.savefig(
        ruta,
        dpi=180,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.show()
    plt.close()


# ================================================================
# 3. INTRODUCCIÓN PARA LOS ESTUDIANTES
# ================================================================

titulo("UNIVERSIDAD EAN | TREND TECH")

print("""
LABORATORIO:
DEL PROBLEMA A UNA SOLUCIÓN TECNOLÓGICA

En este ejercicio aprenderemos a:

1. Identificar un problema empresarial.
2. Reconocer sus posibles causas.
3. Organizar las causas mediante Ishikawa.
4. Evaluar su impacto utilizando semáforos.
5. Identificar las causas prioritarias.
6. Proponer funcionalidades para una aplicación.

ESCALA DE IMPACTO:

1 = Muy bajo
2 = Bajo
3 = Medio
4 = Alto
5 = Muy alto

SEMÁFOROS:

ROJO:     Impacto alto (4-5)
AMARILLO: Impacto medio (3)
VERDE:    Impacto bajo (1-2)

IMPORTANTE:

Los semáforos muestran prioridades estimadas.
No prueban automáticamente las causas del problema.
""")


# ================================================================
# 4. SELECCIONAR MODALIDAD
# ================================================================

print("\nSELECCIONA UNA OPCIÓN:\n")

print("1. Ejecutar ejemplo de AgroFresco")
print("2. Analizar el proyecto de mi equipo")

opcion = input("\nEscribe 1 o 2: ").strip()

while opcion not in ["1", "2"]:
    opcion = input("Selecciona 1 o 2: ").strip()


# ================================================================
# 5. DEFINIR CATEGORÍAS DEL ISHIKAWA
# ================================================================

# Estas categorías son sencillas y sirven para
# diferentes tipos de empresas y servicios.

categorias = [
    "Clientes",
    "Comunicación",
    "Procesos",
    "Información",
    "Tecnología",
    "Experiencia"
]

preguntas = {
    "Clientes":
        "¿Qué dificultades tienen los clientes?",

    "Comunicación":
        "¿Qué problemas existen al comunicarse?",

    "Procesos":
        "¿Qué actividades generan errores o demoras?",

    "Información":
        "¿Qué datos faltan o están desorganizados?",

    "Tecnología":
        "¿Qué herramientas faltan o son insuficientes?",

    "Experiencia":
        "¿Qué situaciones generan insatisfacción?"
}


# ================================================================
# 6. MODALIDAD 1: EJEMPLO AGRÍCOLA
# ================================================================

if opcion == "1":

    titulo("CASO DE EJEMPLO: AGROFRESCO")

    empresa = "AgroFresco"

    problema = (
        "Pérdida de oportunidades de venta "
        "y dificultades en la gestión de pedidos"
    )

    solucion = (
        "Aplicación para consultar productos, "
        "realizar pedidos y programar entregas"
    )

    print("""
AgroFresco es una empresa agrícola ficticia
que cultiva y comercializa hortalizas.

Sus clientes son restaurantes, tiendas
y familias.

Actualmente presenta dificultades para
informar sobre productos disponibles,
registrar pedidos y coordinar entregas.

La empresa desea explorar una aplicación
para mejorar la atención a sus clientes.
""")

    # Causas y valoraciones ilustrativas.
    # No corresponden a datos empresariales reales.

    datos = [
        ["Clientes",
         "Dificultad para realizar pedidos", 5],

        ["Clientes",
         "Desconocimiento de productos disponibles", 3],

        ["Comunicación",
         "Respuestas tardías a los clientes", 5],

        ["Comunicación",
         "Cambios de pedido sin confirmar", 4],

        ["Procesos",
         "Pedidos registrados manualmente", 4],

        ["Procesos",
         "Entregas sin programación clara", 3],

        ["Información",
         "Inventario de productos desactualizado", 5],

        ["Información",
         "Precios dispersos en diferentes registros", 3],

        ["Tecnología",
         "No existe un canal digital de pedidos", 4],

        ["Tecnología",
         "Registros digitales no integrados", 2],

        ["Experiencia",
         "Clientes desconocen el estado del pedido", 4],

        ["Experiencia",
         "Poca retroalimentación de los clientes", 2]
    ]

    df = pd.DataFrame(
        datos,
        columns=["Categoria", "Causa", "Impacto"]
    )


# ================================================================
# 7. MODALIDAD 2: PROYECTO DE CADA EQUIPO
# ================================================================

else:

    titulo("PROYECTO DEL EQUIPO")

    empresa = pedir_texto(
        "Nombre de la empresa: "
    )

    problema = pedir_texto(
        "¿Cuál es el problema principal?: "
    )

    solucion = pedir_texto(
        "¿Qué aplicación o solución digital proponen?: "
    )

    datos = []

    print("""
Ahora identificarán dos causas por categoría.

EJEMPLO DE CAUSA:
Los clientes no reciben respuestas oportunas.

EJEMPLO INCORRECTO:
Crear una aplicación.

Recuerden:

Primero se identifica qué está fallando.
Después se propone una solución.
""")

    for categoria in categorias:

        titulo("CATEGORÍA: " + categoria.upper())

        print("\nPREGUNTA ORIENTADORA:")
        print(preguntas[categoria])

        for numero in range(1, 3):

            print(f"\nCAUSA {numero}")

            causa = pedir_texto(
                "Describe una posible causa: "
            )

            print("""
¿Qué impacto estiman que tiene esta causa
sobre el problema principal?

1 = Muy bajo
2 = Bajo
3 = Medio
4 = Alto
5 = Muy alto
""")

            impacto = pedir_numero(
                "Calificación (1-5): "
            )

            datos.append([
                categoria,
                causa,
                impacto
            ])

    df = pd.DataFrame(
        datos,
        columns=["Categoria", "Causa", "Impacto"]
    )


# ================================================================
# 8. CLASIFICACIÓN AUTOMÁTICA
# ================================================================

df["Semaforo"] = df["Impacto"].apply(
    lambda x: clasificar_semaforo(x)[0]
)

df["Color"] = df["Impacto"].apply(
    lambda x: clasificar_semaforo(x)[1]
)

df["Prioridad"] = df["Impacto"].apply(
    lambda x: clasificar_semaforo(x)[2]
)

# Ordenar causas por impacto.
ranking = df.sort_values(
    "Impacto",
    ascending=False,
    kind="stable"
).reset_index(drop=True)

titulo("DIAGNÓSTICO INICIAL")

print("\nEMPRESA:", empresa)
print("PROBLEMA:", problema)
print("SOLUCIÓN PROPUESTA:", solucion)

display(
    df[
        [
            "Categoria",
            "Causa",
            "Impacto",
            "Semaforo",
            "Prioridad"
        ]
    ]
)


# ================================================================
# 9. GRÁFICA 1: DIAGRAMA DE ISHIKAWA CON SEMÁFOROS
# ================================================================

def grafica_ishikawa(df, problema):

    fig, ax = plt.subplots(figsize=(20, 12))

    fig.patch.set_facecolor("white")
    ax.set_facecolor("white")

    # Espina principal del diagrama.
    ax.plot(
        [1, 12],
        [0, 0],
        color=OSCURO,
        linewidth=3.5
    )

    # Cabeza del pez: problema central.
    cabeza = FancyBboxPatch(
        (12, -1.0),
        4.5, 2.0,
        boxstyle="round,pad=0.08,rounding_size=0.25",
        facecolor="#E3F2FD",
        edgecolor=AZUL,
        linewidth=2.5
    )

    ax.add_patch(cabeza)

    ax.text(
        14.25,
        0,
        ajustar_texto(problema, 30),
        ha="center",
        va="center",
        fontsize=10,
        fontweight="bold",
        color=OSCURO
    )

    # Tres categorías arriba y tres abajo.
    posiciones_x = [
        2.7, 6.0, 9.3,
        2.7, 6.0, 9.3
    ]

    for i, categoria in enumerate(categorias):

        x = posiciones_x[i]

        # Categorías superiores e inferiores.
        lado = 1 if i < 3 else -1

        y_extremo = 4.0 * lado

        # Espina de categoría.
        ax.plot(
            [x, x - 1.15],
            [0, y_extremo],
            color=GRIS,
            linewidth=2.5
        )

        # Título de la categoría.
        ax.text(
            x - 1.15,
            y_extremo + 0.55 * lado,
            categoria.upper(),
            ha="center",
            va="center",
            fontsize=12,
            fontweight="bold",
            color=AZUL
        )

        causas_categoria = df[
            df["Categoria"] == categoria
        ].reset_index(drop=True)

        for j, fila in causas_categoria.iterrows():

            fraccion = 0.37 + j * 0.32

            punto_x = x - 1.15 * fraccion
            punto_y = y_extremo * fraccion

            color = fila["Color"]

            # Línea horizontal de causa.
            ax.plot(
                [punto_x, punto_x + 0.55],
                [punto_y, punto_y],
                color=color,
                linewidth=2
            )

            # Semáforo.
            ax.scatter(
                punto_x + 0.55,
                punto_y,
                s=420,
                color=color,
                edgecolor="white",
                linewidth=2,
                zorder=5
            )

            # Calificación dentro del círculo.
            ax.text(
                punto_x + 0.55,
                punto_y,
                str(fila["Impacto"]),
                ha="center",
                va="center",
                fontsize=10,
                fontweight="bold",
                color="white",
                zorder=6
            )

            # Texto de la causa.
            ax.text(
                punto_x + 0.75,
                punto_y + 0.15 * lado,
                ajustar_texto(fila["Causa"], 20),
                fontsize=9,
                ha="left",
                va="bottom" if lado == 1 else "top",
                color=OSCURO
            )

    # Leyenda.
    leyenda = [
        Line2D(
            [0], [0],
            marker="o",
            color="w",
            markerfacecolor=ROJO,
            markersize=12,
            label="Alto (4-5)"
        ),
        Line2D(
            [0], [0],
            marker="o",
            color="w",
            markerfacecolor=AMARILLO,
            markersize=12,
            label="Medio (3)"
        ),
        Line2D(
            [0], [0],
            marker="o",
            color="w",
            markerfacecolor=VERDE,
            markersize=12,
            label="Bajo (1-2)"
        )
    ]

    ax.legend(
        handles=leyenda,
        loc="lower center",
        bbox_to_anchor=(0.5, -0.07),
        ncol=3,
        frameon=False,
        fontsize=11
    )

    ax.set_xlim(0, 17)
    ax.set_ylim(-6.6, 6.6)
    ax.axis("off")

    ax.set_title(
        "UNIVERSIDAD EAN | TREND TECH\n"
        "DIAGRAMA DE ISHIKAWA CON SEMÁFOROS",
        fontsize=18,
        fontweight="bold",
        color=OSCURO,
        pad=25
    )

    plt.tight_layout()
    guardar_grafica("01_Ishikawa.png")


grafica_ishikawa(df, problema)


# ================================================================
# 10. INTERPRETACIÓN AUTOMÁTICA DEL ISHIKAWA
# ================================================================

titulo("INTERPRETACIÓN DEL ISHIKAWA")

altas = df[df["Impacto"] >= 4]
medias = df[df["Impacto"] == 3]
bajas = df[df["Impacto"] <= 2]

print(f"""
Se identificaron {len(df)} posibles causas.

CAUSAS DE IMPACTO ALTO: {len(altas)}
CAUSAS DE IMPACTO MEDIO: {len(medias)}
CAUSAS DE IMPACTO BAJO: {len(bajas)}

INTERPRETACIÓN:

El diagrama permite observar que el problema
puede estar relacionado con diferentes causas.

Las causas rojas tienen mayor impacto estimado
y requieren una revisión prioritaria.

Las causas amarillas deben analizarse.

Las causas verdes presentan menor impacto
estimado y podrían atenderse posteriormente.

CONCLUSIÓN:

Antes de desarrollar la aplicación, es
necesario identificar cuáles de estas
causas pueden resolverse mediante tecnología.
""")


# ================================================================
# 11. GRÁFICA 2: RANKING DE CAUSAS
# ================================================================

fig, ax = plt.subplots(figsize=(13, 9))

orden = ranking.iloc[::-1]

ax.barh(
    range(len(orden)),
    orden["Impacto"],
    color=orden["Color"],
    height=0.7
)

ax.set_yticks(range(len(orden)))

ax.set_yticklabels(
    [
        ajustar_texto(causa, 40)
        for causa in orden["Causa"]
    ],
    fontsize=9
)

for i, valor in enumerate(orden["Impacto"]):

    ax.text(
        valor + 0.08,
        i,
        f"{valor}/5",
        va="center",
        fontsize=10,
        fontweight="bold"
    )

ax.set_xlim(0, 5.7)

ax.set_xlabel("Impacto estimado")

ax.set_title(
    "TREND TECH | RANKING DE CAUSAS\n"
    "¿Qué dificultades requieren mayor atención?",
    fontsize=15,
    fontweight="bold",
    pad=18
)

ax.grid(axis="x", alpha=0.15)
ax.set_axisbelow(True)

for borde in ax.spines.values():
    borde.set_visible(False)

plt.tight_layout()
guardar_grafica("02_Ranking.png")


# ================================================================
# 12. INTERPRETACIÓN DEL RANKING
# ================================================================

titulo("INTERPRETACIÓN DEL RANKING")

print("""
Esta gráfica permite comparar las causas
según su impacto estimado.

Las causas con calificaciones más altas
son candidatas para orientar el diseño
de la solución tecnológica.

Cuando varias causas tienen la misma
calificación, comparten el mismo nivel
de prioridad estimada.
""")

print("\nTRES PRIMERAS CAUSAS DEL RANKING:\n")

for i, fila in ranking.head(3).iterrows():

    print(f"{i+1}. {fila['Causa']}")
    print(f"   Categoría: {fila['Categoria']}")
    print(f"   Impacto: {fila['Impacto']}/5")
    print(f"   Semáforo: {fila['Semaforo']}\n")


# ================================================================
# 13. GRÁFICA 3: COMPARACIÓN POR CATEGORÍAS
# ================================================================

resumen = (
    df.groupby("Categoria")["Impacto"]
    .mean()
    .reindex(categorias)
    .sort_values(ascending=False)
)

# Color de acuerdo con el promedio.
def color_promedio(valor):
    if valor >= 4:
        return ROJO
    elif valor >= 3:
        return AMARILLO
    return VERDE


colores = [
    color_promedio(valor)
    for valor in resumen.values
]

fig, ax = plt.subplots(figsize=(12, 6))

barras = ax.bar(
    resumen.index,
    resumen.values,
    color=colores,
    width=0.65
)

for barra, valor in zip(barras, resumen.values):

    ax.text(
        barra.get_x() + barra.get_width()/2,
        valor + 0.08,
        f"{valor:.1f}",
        ha="center",
        fontsize=12,
        fontweight="bold"
    )

ax.set_ylim(0, 5.6)
ax.set_ylabel("Impacto promedio")

ax.set_title(
    "TREND TECH | IMPACTO POR CATEGORÍA\n"
    "¿Dónde se concentran las dificultades?",
    fontsize=15,
    fontweight="bold",
    pad=18
)

ax.grid(axis="y", alpha=0.15)
ax.set_axisbelow(True)

for borde in ax.spines.values():
    borde.set_visible(False)

plt.xticks(rotation=15)
plt.tight_layout()

guardar_grafica("03_Categorias.png")


# ================================================================
# 14. INTERPRETACIÓN POR CATEGORÍAS
# ================================================================

titulo("INTERPRETACIÓN POR CATEGORÍAS")

categoria_critica = resumen.idxmax()
promedio_critico = resumen.max()

print(f"""
La categoría con mayor impacto promedio es:

{categoria_critica.upper()}

PROMEDIO: {promedio_critico:.2f}/5

INTERPRETACIÓN:

Las causas de esta categoría recibieron,
en promedio, las calificaciones más altas.

Esto sugiere que el equipo debería revisar
esta área con mayor detalle.

No significa que todas las causas de esta
categoría estén comprobadas ni que una
aplicación pueda resolverlas por completo.
""")


# ================================================================
# 15. DEFINIR FUNCIONALIDADES DE LA APLICACIÓN
# ================================================================

titulo("FUNCIONALIDADES DE LA APLICACIÓN")

# Las funcionalidades se pueden modificar.
# Para el ejemplo agrícola se utilizan
# herramientas relacionadas con pedidos y clientes.

if opcion == "1":

    funciones = [
        "Catálogo de productos",
        "Registro de pedidos",
        "Mensajes automáticos",
        "Programación de entregas",
        "Consulta de disponibilidad",
        "Seguimiento del pedido"
    ]

else:

    print("""
Escribe seis funcionalidades que podría
tener la aplicación de tu equipo.

Ejemplos:

- Registro de clientes
- Agenda de citas
- Catálogo digital
- Notificaciones
- Seguimiento de solicitudes
- Encuestas de satisfacción
""")

    funciones = []

    for i in range(6):

        funcion = pedir_texto(
            f"Funcionalidad {i+1}: "
        )

        funciones.append(funcion)


# ================================================================
# 16. RELACIONAR CAUSAS Y FUNCIONALIDADES
# ================================================================

# Se seleccionan las seis causas del ranking.
# En caso de empate se conserva el orden original.
#
# Para cada causa y función se valora:
#
# 0 = No existe relación identificada
# 1 = Relación baja
# 2 = Relación media
# 3 = Relación alta
#
# Estas relaciones son hipótesis de diseño.
# No son resultados de una prueba técnica.

top_causas = ranking.head(6).copy()

if opcion == "1":

    # Ejemplo ilustrativo de relaciones para AgroFresco.
    #
    # Cada fila representa una causa prioritaria.
    # Cada columna representa una funcionalidad.
    #
    # Las relaciones se asignan por significado,
    # no por la posición accidental en el ranking.

    relaciones_ejemplo = {
        "Dificultad para realizar pedidos":
            [1, 3, 1, 0, 1, 2],

        "Respuestas tardías a los clientes":
            [1, 1, 3, 0, 1, 2],

        "Inventario de productos desactualizado":
            [2, 1, 0, 0, 3, 1],

        "Cambios de pedido sin confirmar":
            [0, 3, 3, 1, 0, 2],

        "Pedidos registrados manualmente":
            [0, 3, 1, 1, 0, 2],

        "No existe un canal digital de pedidos":
            [2, 3, 2, 0, 2, 2],

        "Clientes desconocen el estado del pedido":
            [0, 1, 2, 2, 0, 3]
    }

    relaciones = np.array([
        relaciones_ejemplo.get(
            causa,
            [0] * len(funciones)
        )
        for causa in top_causas["Causa"]
    ])

else:

    print("""
AHORA VAMOS A RELACIONAR CAUSAS Y FUNCIONES.

Para cada causa, califica cuánto podría
contribuir una funcionalidad a atenderla.

0 = Ninguna relación
1 = Relación baja
2 = Relación media
3 = Relación alta

No estamos midiendo beneficios reales.
Estamos evaluando ideas para el diseño.
""")

    relaciones = []

    for _, fila in top_causas.iterrows():

        titulo("CAUSA: " + fila["Causa"])

        fila_relaciones = []

        for funcion in funciones:

            while True:

                try:
                    valor = int(input(
                        f"Relación con '{funcion}' (0-3): "
                    ))

                    if 0 <= valor <= 3:
                        break

                    print("Escribe 0, 1, 2 o 3.")

                except ValueError:
                    print("Debes ingresar un número.")

            fila_relaciones.append(valor)

        relaciones.append(fila_relaciones)

    relaciones = np.array(relaciones)


# ================================================================
# 17. GRÁFICA 4: MATRIZ DE FUNCIONALIDADES
# ================================================================

# La puntuación combina:
#
# Impacto de la causa (1-5)
# multiplicado por
# Relación con la funcionalidad (0-3)
#
# Valor máximo posible por celda: 15.
#
# Se utiliza únicamente para comparar
# prioridades dentro del ejercicio.

impactos = top_causas["Impacto"].to_numpy()

matriz = relaciones * impactos[:, np.newaxis]

fig, ax = plt.subplots(figsize=(14, 8))

imagen = ax.imshow(
    matriz,
    cmap="YlGnBu",
    vmin=0,
    vmax=15,
    aspect="auto"
)

ax.set_xticks(range(len(funciones)))

ax.set_xticklabels(
    [ajustar_texto(x, 20) for x in funciones],
    fontsize=9
)

ax.set_yticks(range(len(top_causas)))

ax.set_yticklabels(
    [
        ajustar_texto(x, 35)
        for x in top_causas["Causa"]
    ],
    fontsize=9
)

# Mostrar puntajes dentro de la matriz.
for i in range(matriz.shape[0]):

    for j in range(matriz.shape[1]):

        valor = matriz[i, j]

        ax.text(
            j,
            i,
            str(valor) if valor > 0 else "-",
            ha="center",
            va="center",
            fontsize=11,
            fontweight="bold",
            color="white" if valor >= 9 else OSCURO
        )

ax.set_title(
    "TREND TECH | MATRIZ DE FUNCIONALIDADES\n"
    "¿Qué funciones podrían atender las causas prioritarias?",
    fontsize=14,
    fontweight="bold",
    pad=20
)

barra = plt.colorbar(imagen, ax=ax)

barra.set_label(
    "Puntaje orientativo: impacto × relación"
)

plt.tight_layout()

guardar_grafica("04_Matriz_Funcionalidades.png")


# ================================================================
# 18. INTERPRETACIÓN DE LA MATRIZ
# ================================================================

titulo("INTERPRETACIÓN DE FUNCIONALIDADES")

print("""
La matriz permite relacionar las causas
prioritarias con posibles funcionalidades
de la aplicación.

Los valores altos representan combinaciones
de causas importantes y funciones que
podrían contribuir a atenderlas.

Los valores bajos indican menor relación
o menor impacto estimado.

IMPORTANTE:

Estos puntajes NO representan porcentajes
de éxito ni eficacia comprobada.

Se utilizan para orientar la discusión
y seleccionar funcionalidades.
""")

puntajes_funciones = matriz.sum(axis=0)

orden_funciones = np.argsort(
    -puntajes_funciones,
    kind="stable"
)

print("\nFUNCIONALIDADES CON MAYOR PUNTAJE:\n")

for posicion in orden_funciones[:3]:

    print(
        f"- {funciones[posicion]}: "
        f"{puntajes_funciones[posicion]} puntos"
    )


# ================================================================
# 19. GRÁFICA 5: PRIORIZACIÓN DE FUNCIONALIDADES
# ================================================================

# Esta gráfica adicional facilita la decisión
# sobre qué incluir primero en la aplicación.

resumen_funciones = pd.DataFrame({
    "Funcionalidad": funciones,
    "Puntaje": puntajes_funciones
})

resumen_funciones = resumen_funciones.sort_values(
    "Puntaje",
    ascending=True
)

fig, ax = plt.subplots(figsize=(12, 6))

barras = ax.barh(
    resumen_funciones["Funcionalidad"],
    resumen_funciones["Puntaje"],
    color=AZUL,
    height=0.65
)

for barra in barras:

    valor = barra.get_width()

    ax.text(
        valor + 0.3,
        barra.get_y() + barra.get_height()/2,
        f"{valor:.0f}",
        va="center",
        fontweight="bold"
    )

ax.set_xlim(
    0,
    max(5, resumen_funciones["Puntaje"].max() * 1.2)
)

ax.set_title(
    "TREND TECH | PRIORIZACIÓN DE FUNCIONALIDADES\n"
    "¿Qué debería incluir primero la aplicación?",
    fontsize=15,
    fontweight="bold",
    pad=18
)

ax.set_xlabel("Puntaje orientativo acumulado")

ax.grid(axis="x", alpha=0.15)
ax.set_axisbelow(True)

for borde in ax.spines.values():
    borde.set_visible(False)

plt.tight_layout()

guardar_grafica("05_Priorizacion_Funciones.png")


# ================================================================
# 20. INTERPRETACIÓN DE FUNCIONALIDADES
# ================================================================

titulo("DECISIONES PARA LA APLICACIÓN")

print("""
La gráfica compara las funcionalidades
según los puntajes obtenidos en la matriz.

Las funciones mejor posicionadas pueden
considerarse candidatas para una primera
versión de la aplicación.

Sin embargo, antes de seleccionarlas,
también deben analizarse:

- Utilidad para los clientes.
- Facilidad de implementación.
- Disponibilidad de información.
- Tiempo para construir el prototipo.
- Opinión de la empresa.
""")

print("\nTRES FUNCIONES CANDIDATAS:\n")

for i, posicion in enumerate(
    orden_funciones[:3], start=1
):

    print(
        f"{i}. {funciones[posicion]} "
        f"({puntajes_funciones[posicion]} puntos)"
    )


# ================================================================
# 21. CONECTAR CAUSAS CON SOLUCIONES
# ================================================================

titulo("DEL DIAGNÓSTICO A LA SOLUCIÓN")

print("""
Una aplicación no debe construirse solamente
porque la tecnología está disponible.

Debe responder a un problema identificado.

Ahora revisaremos tres causas prioritarias
y las funciones que podrían ayudar.
""")

for i, fila in top_causas.head(3).iterrows():

    print("\n" + "-" * 60)

    print("CAUSA:", fila["Causa"])
    print("IMPACTO:", fila["Impacto"])
    print("CATEGORÍA:", fila["Categoria"])

    # Encontrar la posición de la causa en la matriz.
    posicion_causa = top_causas.index.get_loc(i)

    fila_puntajes = matriz[posicion_causa]

    mejor_funcion = np.argmax(fila_puntajes)

    if fila_puntajes[mejor_funcion] > 0:

        print(
            "FUNCIÓN QUE PODRÍA AYUDAR:",
            funciones[mejor_funcion]
        )

    else:

        print(
            "No se identificó una funcionalidad "
            "relacionada con esta causa."
        )

    print("""
PREGUNTAS DE REFLEXIÓN:

1. ¿Qué dificultad experimenta el cliente?
2. ¿Cómo podría ayudar la aplicación?
3. ¿Qué información necesita la función?
4. ¿Cómo comprobaríamos su utilidad?
""")


# ================================================================
# 22. RESUMEN EJECUTIVO AUTOMÁTICO
# ================================================================

titulo("RESUMEN EJECUTIVO DEL PROYECTO")

print(f"""
UNIVERSIDAD EAN
UNIDAD DE ESTUDIO: TREND TECH

EMPRESA:
{empresa}

PROBLEMA:
{problema}

SOLUCIÓN TECNOLÓGICA:
{solucion}

DIAGNÓSTICO:

Se identificaron {len(df)} posibles causas
distribuidas en seis categorías.

Causas de impacto alto: {len(altas)}
Causas de impacto medio: {len(medias)}
Causas de impacto bajo: {len(bajas)}

CATEGORÍA DE MAYOR IMPACTO PROMEDIO:
{categoria_critica}

INTERPRETACIÓN GENERAL:

El análisis permite organizar posibles
causas del problema empresarial y
establecer prioridades iniciales.

Las funcionalidades de la aplicación
deben relacionarse con necesidades
concretas de los clientes y la empresa.

Antes de desarrollar el prototipo,
se recomienda validar las causas
y las funcionalidades seleccionadas.
""")


# ================================================================
# 23. EXPORTAR RESULTADOS A EXCEL
# ================================================================

archivo_excel = os.path.join(
    CARPETA,
    "Diagnostico_Trend_Tech.xlsx"
)

# Preparar tabla de funcionalidades
tabla_matriz = pd.DataFrame(
    matriz,
    index=top_causas["Causa"],
    columns=funciones
)

with pd.ExcelWriter(
    archivo_excel,
    engine="openpyxl"
) as writer:

    df.drop(
        columns=["Color"]
    ).to_excel(
        writer,
        sheet_name="Causas",
        index=False
    )

    ranking.drop(
        columns=["Color"]
    ).to_excel(
        writer,
        sheet_name="Ranking",
        index=False
    )

    resumen.reset_index().to_excel(
        writer,
        sheet_name="Categorias",
        index=False
    )

    tabla_matriz.to_excel(
        writer,
        sheet_name="Matriz_Funciones"
    )

    resumen_funciones.to_excel(
        writer,
        sheet_name="Prioridades",
        index=False
    )


# ================================================================
# 24. ACTIVIDAD FINAL PARA LOS ESTUDIANTES
# ================================================================

titulo("ACTIVIDAD FINAL | TREND TECH")

print("""
Cada equipo debe responder:

1. ¿Cuál es el problema principal de la empresa?

2. ¿Cuáles son las tres causas que recibieron
   las mayores calificaciones?

3. ¿Qué información tienen para justificar
   que esas causas son importantes?

4. ¿Cuáles son las tres funcionalidades
   que incluirían primero en su aplicación?

5. ¿Cómo contribuirían esas funciones
   a mejorar la experiencia del cliente?

6. ¿Cómo comprobarían que la aplicación
   realmente ayuda a resolver el problema?
""")


# ================================================================
# 25. CIERRE DEL LABORATORIO
# ================================================================

titulo("LABORATORIO FINALIZADO")

print(f"""
EMPRESA ANALIZADA:
{empresa}

RESULTADOS GENERADOS:

1. Diagrama de Ishikawa con semáforos.
2. Ranking de causas.
3. Comparación por categorías.
4. Matriz de causas y funcionalidades.
5. Priorización de funcionalidades.
6. Interpretaciones automáticas.
7. Archivo Excel con resultados.

CARPETA:
{CARPETA}

ARCHIVOS:

01_Ishikawa.png
02_Ranking.png
03_Categorias.png
04_Matriz_Funcionalidades.png
05_Priorizacion_Funciones.png
Diagnostico_Trend_Tech.xlsx

REFLEXIÓN FINAL:

Una buena solución tecnológica no comienza
con la programación de una aplicación.

Comienza con la comprensión del problema,
las necesidades de los clientes y las
causas que se desea atender.

UNIVERSIDAD EAN | TREND TECH
FIN DEL LABORATORIO
""")
```
