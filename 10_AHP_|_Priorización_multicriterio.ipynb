```python
# =============================================================================
# INVENTARIOS Y COMPRAS GLOBALES
# CASO INTEGRADOR: GLOBALBIKE COMPONENTS COLOMBIA S.A.S.
#
# ETAPA 10: AHP - ANALYTIC HIERARCHY PROCESS
#
# OBJETIVO:
# Priorizar tres alternativas de solución para reducir el riesgo de stockout
# del controlador E-500 utilizando múltiples criterios de decisión.
#
# =============================================================================
#
# PROBLEMA:
#
# En las etapas anteriores identificamos:
#
# - Variabilidad de la demanda.
# - Errores de pronóstico.
# - Variabilidad del lead time.
# - Dependencia del proveedor principal.
# - Riesgo de stockout.
#
# Después construimos tres alternativas:
#
# A1 = Ajustar el stock de seguridad.
#
# A2 = Implementar dual sourcing.
#
# A3 = Implementar reabastecimiento predictivo.
#
#
# EL PROBLEMA AHORA ES:
#
# Las tres alternativas tienen ventajas y desventajas.
#
# Por ejemplo:
#
# A1 puede implementarse rápidamente,
# pero aumenta el inventario.
#
# A2 reduce la dependencia del proveedor,
# pero puede aumentar costos y complejidad.
#
# A3 permite anticipar el riesgo,
# pero requiere datos y capacidad analítica.
#
#
# Por eso necesitamos responder:
#
# ¿CÓMO PRIORIZAR LAS ALTERNATIVAS
# CONSIDERANDO SIMULTÁNEAMENTE VARIOS CRITERIOS?
#
#
# Para eso utilizaremos:
#
# AHP = ANALYTIC HIERARCHY PROCESS
#
# desarrollado por Thomas L. Saaty.
#
# =============================================================================


# =============================================================================
# 1. IMPORTAR LIBRERÍAS
# =============================================================================

import numpy as np

import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns


# Configuración visual.

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 6)

plt.rcParams["font.size"] = 11


# =============================================================================
# 2. DEFINIR EL OBJETIVO DEL AHP
# =============================================================================
#
# Toda aplicación de AHP debe comenzar con un objetivo claro.
#
# Nuestro objetivo será:
#
# "PRIORIZAR LA ALTERNATIVA MÁS ADECUADA PARA REDUCIR
# EL RIESGO DE STOCKOUT DEL CONTROLADOR E-500"
#
#
# La estructura jerárquica será:
#
#
#                         OBJETIVO
#
#                            ↓
#
#                        CRITERIOS
#
#        ┌───────────┬───────────┬───────────┐
#        ↓           ↓           ↓           ↓
#
#      COSTO      SERVICIO     RIESGO      TIEMPO ...
#
#                            ↓
#
#                       ALTERNATIVAS
#
#                A1         A2         A3
#
#
# =============================================================================


objetivo = """

PRIORIZAR LA ALTERNATIVA DE SOLUCIÓN
PARA REDUCIR EL RIESGO DE STOCKOUT
DEL CONTROLADOR E-500

"""


print("\n" + "=" * 78)

print("OBJETIVO DEL MODELO AHP")

print("=" * 78)

print(objetivo)


# =============================================================================
# 3. DEFINIR LAS ALTERNATIVAS
# =============================================================================


alternativas = [

    "A1 - Stock de seguridad",

    "A2 - Dual sourcing",

    "A3 - Reabastecimiento predictivo"

]


# =============================================================================
# 4. DEFINIR LOS CRITERIOS
# =============================================================================
#
# Utilizaremos cinco criterios.
#
#
# C1 = COSTO TOTAL
#
# Considera:
#
# - inversión,
# - inventario adicional,
# - desarrollo de proveedor,
# - tecnología,
# - operación.
#
#
# C2 = NIVEL DE SERVICIO
#
# Evalúa cuánto contribuye la alternativa
# a mejorar la disponibilidad del E-500.
#
#
# C3 = RIESGO DE SUMINISTRO
#
# Evalúa cuánto contribuye la alternativa
# a reducir la exposición a interrupciones.
#
#
# C4 = TIEMPO DE IMPLEMENTACIÓN
#
# Evalúa qué tan rápidamente puede implementarse.
#
#
# C5 = FLEXIBILIDAD
#
# Evalúa la capacidad de responder
# ante cambios de demanda o abastecimiento.
#
# =============================================================================


criterios = [

    "Costo total",

    "Nivel de servicio",

    "Riesgo de suministro",

    "Tiempo de implementación",

    "Flexibilidad"

]


# =============================================================================
# 5. RECORDAR LA ESCALA FUNDAMENTAL DE SAATY
# =============================================================================
#
# AHP utiliza comparaciones PAREADAS.
#
# Esto significa que no preguntamos:
#
# "¿Cuánto pesa cada criterio?"
#
# directamente.
#
# Preguntamos:
#
# "Entre estos dos criterios,
# ¿cuál es más importante y cuánto más importante?"
#
#
# ESCALA FUNDAMENTAL:
#
# 1 = Igual importancia.
#
# 3 = Importancia moderada.
#
# 5 = Importancia fuerte.
#
# 7 = Importancia muy fuerte.
#
# 9 = Importancia extrema.
#
#
# También pueden utilizarse:
#
# 2, 4, 6 y 8
#
# como valores intermedios.
#
#
# Si A es 5 veces más importante que B:
#
# A / B = 5
#
# entonces:
#
# B / A = 1/5
#
#
# A esto lo llamamos:
#
# RECIPROCIDAD.
#
# =============================================================================


escala_saaty = pd.DataFrame({

    "Valor": [
        1, 2, 3, 4, 5, 6, 7, 8, 9
    ],

    "Interpretacion": [

        "Igual importancia",

        "Valor intermedio",

        "Importancia moderada",

        "Valor intermedio",

        "Importancia fuerte",

        "Valor intermedio",

        "Importancia muy fuerte",

        "Valor intermedio",

        "Importancia extrema"

    ]

})


print("\n" + "=" * 78)

print("ESCALA FUNDAMENTAL DE SAATY")

print("=" * 78)

print(
    escala_saaty.to_string(
        index=False
    )
)


# =============================================================================
# 6. CONSTRUIR LA MATRIZ DE COMPARACIÓN DE LOS CRITERIOS
# =============================================================================
#
# IMPORTANTE:
#
# Los valores siguientes son JUICIOS DEL EJEMPLO ACADÉMICO.
#
# NO son valores universales.
#
# En un proyecto real deberían provenir de:
#
# - expertos,
# - gerentes,
# - responsables de Supply Chain,
# - compradores,
# - responsables de inventarios,
# - comité de decisión.
#
#
# Para GlobalBike asumiremos que:
#
# NIVEL DE SERVICIO y RIESGO DE SUMINISTRO
#
# tienen especial importancia.
#
#
# La matriz debe cumplir:
#
# a(i,j) = 1 / a(j,i)
#
# y la diagonal siempre es:
#
# 1
#
# porque un criterio comparado consigo mismo
# tiene exactamente la misma importancia.
#
# =============================================================================


matriz_criterios = np.array([

#     Costo   Servicio   Riesgo   Tiempo   Flexibilidad

    [1,       1/3,       1/3,     2,       1/2],

    [3,       1,         1,       5,       3],

    [3,       1,         1,       5,       3],

    [1/2,     1/5,       1/5,     1,       1/3],

    [2,       1/3,       1/3,     3,       1]

], dtype=float)


# Convertimos la matriz en DataFrame para visualizarla mejor.


df_matriz_criterios = pd.DataFrame(

    matriz_criterios,

    index=criterios,

    columns=criterios

)


print("\n" + "=" * 78)

print("MATRIZ DE COMPARACIÓN PAREADA DE LOS CRITERIOS")

print("=" * 78)

print(
    df_matriz_criterios.round(3)
)


# =============================================================================
# 7. ¿CÓMO LEER LA MATRIZ?
# =============================================================================


print("""

EJEMPLO DE LECTURA:

En la matriz aparece:

Costo total vs Nivel de servicio = 1/3.

Esto significa que:

NIVEL DE SERVICIO se considera aproximadamente
3 veces más importante que COSTO TOTAL
para esta decisión particular.

Por reciprocidad:

Nivel de servicio vs Costo total = 3.

IMPORTANTE:

Esto NO significa que el costo no importe.

Significa que, bajo los juicios definidos para este ejemplo,
el nivel de servicio tiene mayor prioridad relativa.

""")


# =============================================================================
# 8. VISUALIZAR LA MATRIZ DE CRITERIOS
# =============================================================================


plt.figure(figsize=(10, 8))


sns.heatmap(

    df_matriz_criterios,

    annot=True,

    fmt=".2f",

    cmap="YlGnBu",

    linewidths=0.5

)


plt.title(
    "Matriz de comparación pareada de criterios",
    fontsize=16,
    fontweight="bold"
)


plt.tight_layout()

plt.show()


# =============================================================================
# 9. NORMALIZAR LA MATRIZ
# =============================================================================
#
# Una forma pedagógica de obtener los pesos AHP consiste en:
#
# PASO 1:
#
# sumar cada columna.
#
#
# PASO 2:
#
# dividir cada elemento por el total de su columna.
#
#
# Fórmula:
#
#
#              a(i,j)
# n(i,j) = ----------------
#           suma columna j
#
#
# Después:
#
# PASO 3:
#
# calculamos el promedio de cada fila.
#
#
# Ese promedio aproxima el VECTOR DE PRIORIDADES.
#
# =============================================================================


suma_columnas = (
    matriz_criterios.sum(axis=0)
)


print("\nSUMA DE CADA COLUMNA:")

for criterio, suma in zip(
    criterios,
    suma_columnas
):

    print(
        f"{criterio}: {suma:.3f}"
    )


# Normalizamos.


matriz_normalizada = (

    matriz_criterios

    /

    suma_columnas

)


df_normalizada = pd.DataFrame(

    matriz_normalizada,

    index=criterios,

    columns=criterios

)


print("\n" + "=" * 78)

print("MATRIZ NORMALIZADA")

print("=" * 78)

print(
    df_normalizada.round(3)
)


# =============================================================================
# 10. CALCULAR LOS PESOS DE LOS CRITERIOS
# =============================================================================
#
# Calculamos el promedio de cada fila.
#
#
# Peso_i =
#
# suma de valores normalizados de la fila
# -----------------------------------------
# número de criterios
#
#
# Todos los pesos deben sumar aproximadamente:
#
# 1
#
# o:
#
# 100 %
#
# =============================================================================


pesos_criterios = (
    matriz_normalizada.mean(axis=1)
)


df_pesos = pd.DataFrame({

    "Criterio": criterios,

    "Peso": pesos_criterios,

    "Peso_%": pesos_criterios * 100

})


df_pesos = df_pesos.sort_values(

    "Peso",

    ascending=False

)


print("\n" + "=" * 78)

print("PESOS DE LOS CRITERIOS")

print("=" * 78)


print(

    df_pesos.round(3).to_string(
        index=False
    )

)


print(
    f"\nSuma de los pesos = "
    f"{pesos_criterios.sum():.4f}"
)


# =============================================================================
# 11. GRÁFICA DE PESOS
# =============================================================================


plt.figure(figsize=(11, 6))


sns.barplot(

    data=df_pesos,

    x="Peso_%",

    y="Criterio",

    hue="Criterio",

    palette=[
        "#1565C0",
        "#00897B",
        "#43A047",
        "#F9A825",
        "#8E24AA"
    ],

    legend=False

)


plt.title(
    "Prioridad relativa de los criterios",
    fontsize=16,
    fontweight="bold"
)


plt.xlabel(
    "Peso (%)"
)


plt.ylabel(
    ""
)


plt.tight_layout()

plt.show()


print("""

INTERPRETACIÓN:

Los pesos representan la importancia relativa
de cada criterio dentro de ESTA decisión.

No deben interpretarse como valores universales.

Si otro comité de expertos realiza comparaciones diferentes,
los pesos pueden cambiar.

Esto es precisamente una fortaleza de AHP:

hace explícitos los juicios que están detrás
de una decisión.

""")


# =============================================================================
# 12. VERIFICAR LA CONSISTENCIA
# =============================================================================
#
# Esta es una de las partes MÁS IMPORTANTES de AHP.
#
# Supongamos que alguien dice:
#
# A es más importante que B.
#
# B es más importante que C.
#
# pero después afirma:
#
# C es muchísimo más importante que A.
#
# Existe una posible contradicción.
#
# AHP permite medir esa consistencia.
#
#
# Utilizaremos:
#
# λmax = Lambda máximo.
#
# CI = Consistency Index.
#
# CR = Consistency Ratio.
#
# =============================================================================


# =============================================================================
# 13. CALCULAR LAMBDA MÁXIMO
# =============================================================================
#
# Primero multiplicamos:
#
# MATRIZ ORIGINAL × VECTOR DE PESOS
#
# =============================================================================


vector_ponderado = (

    matriz_criterios

    @

    pesos_criterios

)


# Después dividimos cada valor
# entre el peso correspondiente.


lambda_individual = (

    vector_ponderado

    /

    pesos_criterios

)


# Lambda máximo es el promedio.


lambda_max = (
    lambda_individual.mean()
)


print(
    f"\nLambda máximo = "
    f"{lambda_max:.4f}"
)


# =============================================================================
# 14. CALCULAR CONSISTENCY INDEX - CI
# =============================================================================
#
# Fórmula:
#
#
#          λmax - n
# CI = ----------------
#            n - 1
#
#
# donde:
#
# n = número de criterios.
#
# =============================================================================


n = len(
    criterios
)


CI = (

    lambda_max
    -
    n

) / (

    n
    -
    1

)


print(
    f"Consistency Index (CI) = "
    f"{CI:.4f}"
)


# =============================================================================
# 15. RANDOM INDEX - RI
# =============================================================================
#
# Saaty propone valores de referencia denominados:
#
# RANDOM INDEX.
#
# Valores habituales:
#
# n = 1 → 0
# n = 2 → 0
# n = 3 → 0.58
# n = 4 → 0.90
# n = 5 → 1.12
# n = 6 → 1.24
# n = 7 → 1.32
# n = 8 → 1.41
# n = 9 → 1.45
# n =10 → 1.49
#
#
# Tenemos cinco criterios:
#
# RI = 1.12
#
# =============================================================================


RI_tabla = {

    1: 0.00,

    2: 0.00,

    3: 0.58,

    4: 0.90,

    5: 1.12,

    6: 1.24,

    7: 1.32,

    8: 1.41,

    9: 1.45,

    10: 1.49

}


RI = RI_tabla[n]


# =============================================================================
# 16. CALCULAR CONSISTENCY RATIO - CR
# =============================================================================
#
# Fórmula:
#
#
#          CI
# CR = -----------
#          RI
#
#
# Regla comúnmente utilizada:
#
# CR < 0.10
#
# indica un nivel de consistencia generalmente aceptable.
#
#
# Si CR es elevado:
#
# NO debemos simplemente continuar.
#
# Debemos revisar las comparaciones.
#
# =============================================================================


CR = (
    CI / RI
)


print(
    f"Random Index (RI) = "
    f"{RI:.2f}"
)


print(
    f"Consistency Ratio (CR) = "
    f"{CR:.4f}"
)


print(
    f"Consistency Ratio (%) = "
    f"{CR * 100:.2f}%"
)


# Interpretación automática.


if CR < 0.10:

    print("""

RESULTADO:

La matriz presenta una consistencia
generalmente aceptable bajo el umbral del 10 %.

Podemos continuar con el análisis.

""")

else:

    print("""

ATENCIÓN:

La matriz supera el umbral de consistencia del 10 %.

Antes de continuar sería recomendable revisar
las comparaciones pareadas.

""")


# =============================================================================
# 17. CREAR UNA FUNCIÓN AHP
# =============================================================================
#
# Hasta ahora hicimos todo paso por paso
# para entender el procedimiento.
#
# Pero tendremos que repetir el cálculo
# varias veces.
#
# Por eso crearemos una FUNCIÓN.
#
# La función recibirá una matriz
# y devolverá:
#
# - pesos,
# - lambda máximo,
# - CI,
# - CR.
#
# =============================================================================


def calcular_ahp(matriz):

    """
    Calcula prioridades y consistencia de una matriz AHP.

    Parámetro:
    ----------
    matriz : numpy array
        Matriz recíproca de comparación pareada.

    Retorna:
    --------
    pesos : numpy array
        Vector de prioridades.

    lambda_max : float
        Valor propio máximo aproximado.

    CI : float
        Índice de consistencia.

    CR : float
        Razón de consistencia.
    """

    # Número de elementos comparados.

    n = matriz.shape[0]


    # Suma de columnas.

    suma_columnas = matriz.sum(
        axis=0
    )


    # Normalización.

    normalizada = (

        matriz
        /
        suma_columnas

    )


    # Promedio de las filas.

    pesos = normalizada.mean(
        axis=1
    )


    # Vector ponderado.

    vector_ponderado = (

        matriz
        @
        pesos

    )


    # Lambda máximo.

    lambda_max = np.mean(

        vector_ponderado
        /
        pesos

    )


    # Índice de consistencia.

    if n > 1:

        CI = (

            lambda_max
            -
            n

        ) / (

            n
            -
            1

        )

    else:

        CI = 0


    # Tabla RI.

    tabla_RI = {

        1: 0.00,

        2: 0.00,

        3: 0.58,

        4: 0.90,

        5: 1.12,

        6: 1.24,

        7: 1.32,

        8: 1.41,

        9: 1.45,

        10: 1.49

    }


    RI = tabla_RI.get(
        n,
        1.49
    )


    # CR.

    if RI == 0:

        CR = 0

    else:

        CR = (
            CI / RI
        )


    return (
        pesos,
        lambda_max,
        CI,
        CR
    )


# =============================================================================
# 18. EVALUAR ALTERNATIVAS PARA CADA CRITERIO
# =============================================================================
#
# Ahora comienza la segunda parte del AHP.
#
# Ya sabemos cuánto pesa cada CRITERIO.
#
# Ahora necesitamos saber:
#
# ¿Cómo se comportan A1, A2 y A3
# respecto a cada criterio?
#
#
# IMPORTANTE:
#
# Nuevamente utilizaremos comparaciones pareadas.
#
# =============================================================================


# =============================================================================
# 19. CRITERIO 1: COSTO TOTAL
# =============================================================================
#
# Recordemos:
#
# A1 = Stock de seguridad.
#
# A2 = Dual sourcing.
#
# A3 = Reabastecimiento predictivo.
#
#
# Para COSTO queremos favorecer
# la alternativa que represente una carga económica
# relativamente menor.
#
#
# Para el ejemplo suponemos:
#
# A1 tiene ventaja moderada sobre A2.
#
# A1 tiene una ligera ventaja sobre A3.
#
# A3 tiene ventaja moderada sobre A2.
#
# =============================================================================


matriz_costo = np.array([

    [1,   3,   2],

    [1/3, 1,   1/3],

    [1/2, 3,   1]

], dtype=float)


# =============================================================================
# 20. CRITERIO 2: NIVEL DE SERVICIO
# =============================================================================
#
# Los escenarios del punto anterior indicaban
# que las tres alternativas podían mejorar el servicio.
#
# Para el ejemplo consideraremos:
#
# Dual sourcing con una ligera ventaja,
# seguido del stock de seguridad
# y del reabastecimiento predictivo.
#
# =============================================================================


matriz_servicio = np.array([

    [1,   1/2, 2],

    [2,   1,   3],

    [1/2, 1/3, 1]

], dtype=float)


# =============================================================================
# 21. CRITERIO 3: RIESGO DE SUMINISTRO
# =============================================================================
#
# Aquí Dual Sourcing tiene una ventaja importante.
#
# ¿Por qué?
#
# Porque modifica directamente la dependencia
# de una sola fuente de abastecimiento.
#
# =============================================================================


matriz_riesgo = np.array([

    [1,   1/5, 1],

    [5,   1,   4],

    [1,   1/4, 1]

], dtype=float)


# =============================================================================
# 22. CRITERIO 4: TIEMPO DE IMPLEMENTACIÓN
# =============================================================================
#
# A1 puede implementarse rápidamente.
#
# A3 requiere desarrollo analítico.
#
# A2 requiere búsqueda, evaluación,
# homologación y negociación con otro proveedor.
#
# =============================================================================


matriz_tiempo = np.array([

    [1,   5,   3],

    [1/5, 1,   1/2],

    [1/3, 2,   1]

], dtype=float)


# =============================================================================
# 23. CRITERIO 5: FLEXIBILIDAD
# =============================================================================
#
# Dual sourcing y reabastecimiento predictivo
# ofrecen mayor capacidad de adaptación.
#
# =============================================================================


matriz_flexibilidad = np.array([

    [1,   1/3, 1/3],

    [3,   1,   1],

    [3,   1,   1]

], dtype=float)


# =============================================================================
# 24. GUARDAR LAS MATRICES EN UN DICCIONARIO
# =============================================================================
#
# Esto facilita recorrerlas automáticamente.
#
# =============================================================================


matrices_alternativas = {

    "Costo total":
        matriz_costo,

    "Nivel de servicio":
        matriz_servicio,

    "Riesgo de suministro":
        matriz_riesgo,

    "Tiempo de implementación":
        matriz_tiempo,

    "Flexibilidad":
        matriz_flexibilidad

}


# =============================================================================
# 25. CALCULAR PRIORIDADES LOCALES
# =============================================================================
#
# PRIORIDAD LOCAL significa:
#
# ¿Qué prioridad tiene cada alternativa
# DENTRO DE UN CRITERIO?
#
#
# Ejemplo:
#
# Dentro del criterio RIESGO:
#
# A2 podría tener:
#
# 70 %
#
# A1:
#
# 15 %
#
# A3:
#
# 15 %
#
#
# Eso todavía NO es el resultado global.
#
# =============================================================================


prioridades_locales = []


consistencias = []


for criterio in criterios:

    matriz = (
        matrices_alternativas[
            criterio
        ]
    )


    pesos, lam, ci, cr = calcular_ahp(
        matriz
    )


    prioridades_locales.append(
        pesos
    )


    consistencias.append({

        "Criterio":
            criterio,

        "Lambda_max":
            lam,

        "CI":
            ci,

        "CR":
            cr,

        "CR_%":
            cr * 100

    })


# Convertimos a matriz.
#
# Actualmente:
#
# filas = criterios
# columnas = alternativas.


prioridades_locales = np.array(
    prioridades_locales
)


df_prioridades = pd.DataFrame(

    prioridades_locales,

    index=criterios,

    columns=alternativas

)


print("\n" + "=" * 78)

print("PRIORIDADES LOCALES DE LAS ALTERNATIVAS")

print("=" * 78)


print(
    df_prioridades.round(3)
)


# =============================================================================
# 26. VERIFICAR CONSISTENCIA DE TODAS LAS MATRICES
# =============================================================================


df_consistencias = pd.DataFrame(
    consistencias
)


print("\n" + "=" * 78)

print("CONSISTENCIA DE LAS MATRICES DE ALTERNATIVAS")

print("=" * 78)


print(

    df_consistencias.round(4).to_string(
        index=False
    )

)


# =============================================================================
# 27. GRÁFICA DE CONSISTENCIA
# =============================================================================


plt.figure(figsize=(11, 6))


plt.bar(

    df_consistencias["Criterio"],

    df_consistencias["CR_%"],

    color=[
        "#1565C0",
        "#00897B",
        "#43A047",
        "#F9A825",
        "#8E24AA"
    ]

)


plt.axhline(

    10,

    linestyle="--",

    color="#D32F2F",

    linewidth=2,

    label="Umbral de referencia = 10%"

)


plt.title(
    "Consistency Ratio de las matrices AHP",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel(
    "CR (%)"
)


plt.xticks(
    rotation=20
)


plt.legend()


plt.tight_layout()

plt.show()


print("""

INTERPRETACIÓN:

Cada matriz de comparación debería revisarse
antes de utilizar sus resultados.

Si una matriz presenta un CR elevado,
no deberíamos ignorarlo.

Debemos volver a las comparaciones
y revisar si nuestros juicios son coherentes.

""")


# =============================================================================
# 28. MAPA DE CALOR DE PRIORIDADES LOCALES
# =============================================================================


plt.figure(figsize=(11, 7))


sns.heatmap(

    df_prioridades * 100,

    annot=True,

    fmt=".1f",

    cmap="YlGnBu",

    linewidths=0.5,

    cbar_kws={
        "label": "Prioridad local (%)"
    }

)


plt.title(
    "Desempeño relativo de las alternativas por criterio",
    fontsize=16,
    fontweight="bold"
)


plt.xlabel(
    "Alternativa"
)


plt.ylabel(
    "Criterio"
)


plt.tight_layout()

plt.show()


print("""

CÓMO LEER ESTA GRÁFICA:

Cada FILA representa un criterio.

Por ejemplo:

En RIESGO DE SUMINISTRO podemos observar
qué alternativa recibe mayor prioridad.

Pero todavía no debemos seleccionar
una alternativa solamente mirando una fila.

Necesitamos combinar:

PESO DEL CRITERIO

×

PRIORIDAD LOCAL DE LA ALTERNATIVA.

""")


# =============================================================================
# 29. CALCULAR PRIORIDAD GLOBAL
# =============================================================================
#
# Llegamos al corazón de la decisión.
#
#
# Fórmula conceptual:
#
#
# PRIORIDAD GLOBAL DE UNA ALTERNATIVA =
#
# Σ
#
# (
#
# Peso del criterio
#
# ×
#
# Prioridad de la alternativa en ese criterio
#
# )
#
#
# Ejemplo:
#
# Si:
#
# Riesgo pesa 30 %
#
# y
#
# A2 tiene prioridad 70 % en Riesgo
#
#
# contribución:
#
# 0.30 × 0.70
#
# =
#
# 0.21
#
#
# Repetimos para todos los criterios
# y sumamos.
#
# =============================================================================


prioridad_global = (

    pesos_criterios

    @

    prioridades_locales

)


# Creamos tabla.


resultado_global = pd.DataFrame({

    "Alternativa":
        alternativas,

    "Prioridad":
        prioridad_global,

    "Prioridad_%":
        prioridad_global * 100

})


resultado_global = (

    resultado_global

    .sort_values(
        "Prioridad",
        ascending=False
    )

    .reset_index(drop=True)

)


print("\n" + "=" * 78)

print("RESULTADO GLOBAL DEL AHP")

print("=" * 78)


print(

    resultado_global.round(3).to_string(
        index=False
    )

)


# =============================================================================
# 30. GRÁFICA DEL RESULTADO GLOBAL
# =============================================================================


plt.figure(figsize=(11, 6))


colores_resultado = [

    "#43A047",
    "#1565C0",
    "#8E24AA"

]


plt.bar(

    resultado_global[
        "Alternativa"
    ],

    resultado_global[
        "Prioridad_%"
    ],

    color=colores_resultado

)


plt.title(
    "Priorización global de alternativas | AHP",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel(
    "Prioridad global (%)"
)


plt.xlabel(
    ""
)


plt.xticks(
    rotation=10
)


# Agregamos valores encima de las barras.


for i, valor in enumerate(
    resultado_global[
        "Prioridad_%"
    ]
):

    plt.text(

        i,

        valor + 1,

        f"{valor:.1f}%",

        ha="center",

        fontweight="bold",

        fontsize=12

    )


plt.tight_layout()

plt.show()


# =============================================================================
# 31. IDENTIFICAR LA ALTERNATIVA PRIORIZADA
# =============================================================================


alternativa_priorizada = (
    resultado_global.iloc[0][
        "Alternativa"
    ]
)


puntaje_priorizado = (
    resultado_global.iloc[0][
        "Prioridad_%"
    ]
)


print("\n" + "=" * 78)

print("RESULTADO DEL EJEMPLO")

print("=" * 78)


print(
    f"""

Bajo los criterios, pesos y comparaciones
definidos en ESTE ejercicio académico:

La alternativa con mayor prioridad es:

{alternativa_priorizada}

con una prioridad global aproximada de:

{puntaje_priorizado:.1f} %

IMPORTANTE:

Esto NO significa que esta alternativa
sea universalmente "la mejor".

Significa que resulta priorizada
BAJO LOS JUICIOS Y CRITERIOS
UTILIZADOS EN ESTE MODELO.

"""
)


# =============================================================================
# 32. ¿DE DÓNDE SALE EL RESULTADO?
# =============================================================================
#
# Queremos que los estudiantes puedan ver
# cuánto aporta cada criterio a cada alternativa.
#
#
# CONTRIBUCIÓN =
#
# peso criterio × prioridad local.
#
# =============================================================================


contribuciones = (

    prioridades_locales

    *

    pesos_criterios.reshape(-1, 1)

)


df_contribuciones = pd.DataFrame(

    contribuciones * 100,

    index=criterios,

    columns=alternativas

)


print("\n" + "=" * 78)

print("CONTRIBUCIÓN DE CADA CRITERIO AL RESULTADO GLOBAL")

print("=" * 78)


print(
    df_contribuciones.round(2)
)


# =============================================================================
# 33. GRÁFICA APILADA DE CONTRIBUCIONES
# =============================================================================


df_contribuciones.T.plot(

    kind="bar",

    stacked=True,

    figsize=(12, 7),

    color=[

        "#1565C0",

        "#43A047",

        "#E53935",

        "#F9A825",

        "#8E24AA"

    ]

)


plt.title(
    "¿Qué criterios explican la prioridad de cada alternativa?",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel(
    "Contribución a la prioridad global (%)"
)


plt.xlabel(
    ""
)


plt.xticks(
    rotation=10
)


plt.legend(

    title="Criterios",

    bbox_to_anchor=(1.05, 1),

    loc="upper left"

)


plt.tight_layout()

plt.show()


print("""

INTERPRETACIÓN:

Esta gráfica es especialmente importante.

No solamente muestra CUÁNTO obtuvo cada alternativa.

Muestra POR QUÉ obtuvo ese resultado.

Por ejemplo:

una alternativa puede recibir una contribución elevada
porque tiene buen desempeño precisamente en los criterios
que el comité considera más importantes.

Así AHP hace transparente la lógica de la decisión.

""")


# =============================================================================
# 34. RADAR DE LAS ALTERNATIVAS
# =============================================================================
#
# Ahora construiremos un radar utilizando
# las prioridades locales.
#
# Esto permite observar el perfil de cada alternativa.
#
# =============================================================================


categorias = criterios


N = len(
    categorias
)


angulos = np.linspace(

    0,

    2 * np.pi,

    N,

    endpoint=False

).tolist()


angulos += angulos[:1]


fig = plt.figure(
    figsize=(10, 8)
)


ax = plt.subplot(
    111,
    polar=True
)


colores_radar = [

    "#1565C0",

    "#43A047",

    "#8E24AA"

]


for j, alternativa in enumerate(
    alternativas
):

    valores = (

        prioridades_locales[:, j]
        * 100

    ).tolist()


    valores += valores[:1]


    ax.plot(

        angulos,

        valores,

        linewidth=2.5,

        label=alternativa,

        color=colores_radar[j]

    )


    ax.fill(

        angulos,

        valores,

        alpha=0.07,

        color=colores_radar[j]

    )


ax.set_xticks(
    angulos[:-1]
)


ax.set_xticklabels(
    categorias
)


plt.title(

    "Perfil multicriterio de las alternativas",

    fontsize=16,

    fontweight="bold",

    pad=25

)


plt.legend(

    bbox_to_anchor=(1.45, 1.10),

    loc="upper right"

)


plt.tight_layout()

plt.show()


print("""

INTERPRETACIÓN DEL RADAR:

El radar permite visualizar que ninguna alternativa
tiene necesariamente el mismo desempeño
en todos los criterios.

Una puede destacar en costo.

Otra en riesgo.

Otra en flexibilidad.

Esto muestra visualmente por qué
la decisión es MULTICRITERIO.

""")


# =============================================================================
# 35. ANÁLISIS DE SENSIBILIDAD
# =============================================================================
#
# Una buena decisión multicriterio NO debería terminar
# solamente con el resultado base.
#
# También debemos preguntar:
#
# ¿QUÉ PASA SI CAMBIAN LAS PRIORIDADES?
#
#
# Esto se denomina:
#
# ANÁLISIS DE SENSIBILIDAD.
#
#
# Vamos a construir tres escenarios:
#
# ESCENARIO 1:
# prioridad al costo.
#
# ESCENARIO 2:
# prioridad al servicio.
#
# ESCENARIO 3:
# prioridad al riesgo.
#
#
# IMPORTANTE:
#
# Estos pesos son escenarios hipotéticos.
#
# No reemplazan los pesos AHP originales.
#
# =============================================================================


escenarios_sensibilidad = {

    "Base AHP":
        pesos_criterios,

    "Prioridad Costo":
        np.array([
            0.45,
            0.20,
            0.15,
            0.10,
            0.10
        ]),

    "Prioridad Servicio":
        np.array([
            0.15,
            0.45,
            0.20,
            0.10,
            0.10
        ]),

    "Prioridad Riesgo":
        np.array([
            0.10,
            0.20,
            0.50,
            0.10,
            0.10
        ])

}


resultados_sensibilidad = []


for nombre, pesos in escenarios_sensibilidad.items():

    resultado = (

        pesos

        @

        prioridades_locales

    )


    for alternativa, valor in zip(
        alternativas,
        resultado
    ):

        resultados_sensibilidad.append({

            "Escenario":
                nombre,

            "Alternativa":
                alternativa,

            "Prioridad":
                valor * 100

        })


df_sensibilidad = pd.DataFrame(
    resultados_sensibilidad
)


print("\n" + "=" * 78)

print("ANÁLISIS DE SENSIBILIDAD")

print("=" * 78)


tabla_sensibilidad = (

    df_sensibilidad

    .pivot(

        index="Alternativa",

        columns="Escenario",

        values="Prioridad"

    )

)


print(
    tabla_sensibilidad.round(2)
)


# =============================================================================
# 36. GRÁFICA DEL ANÁLISIS DE SENSIBILIDAD
# =============================================================================


plt.figure(figsize=(13, 7))


sns.barplot(

    data=df_sensibilidad,

    x="Escenario",

    y="Prioridad",

    hue="Alternativa",

    palette=[
        "#1565C0",
        "#43A047",
        "#8E24AA"
    ]

)


plt.title(
    "Análisis de sensibilidad de la decisión",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel(
    "Prioridad (%)"
)


plt.xlabel(
    "Escenario"
)


plt.xticks(
    rotation=10
)


plt.legend(

    title="Alternativa",

    bbox_to_anchor=(1.02, 1),

    loc="upper left"

)


plt.tight_layout()

plt.show()


print("""

INTERPRETACIÓN DEL ANÁLISIS DE SENSIBILIDAD:

Aquí hacemos una pregunta fundamental:

¿LA DECISIÓN CAMBIA CUANDO CAMBIAN LAS PRIORIDADES?

Si la misma alternativa permanece arriba
en muchos escenarios diferentes,
podríamos decir que el resultado presenta
cierta robustez frente a esos cambios.

Si la alternativa cambia rápidamente,
significa que la decisión es sensible
a las prioridades del comité.

Eso NO es necesariamente un problema.

Es información importante para la gerencia.

""")


# =============================================================================
# 37. IDENTIFICAR LA ALTERNATIVA PRIORIZADA POR ESCENARIO
# =============================================================================


ganadores_escenarios = (

    df_sensibilidad

    .loc[
        df_sensibilidad
        .groupby("Escenario")[
            "Prioridad"
        ]
        .idxmax()
    ]

    .sort_values(
        "Escenario"
    )

)


print("\n" + "=" * 78)

print("ALTERNATIVA PRIORIZADA EN CADA ESCENARIO")

print("=" * 78)


print(

    ganadores_escenarios[[

        "Escenario",

        "Alternativa",

        "Prioridad"

    ]]

    .round(2)

    .to_string(
        index=False
    )

)


# =============================================================================
# 38. CONSTRUIR UN RESUMEN EJECUTIVO
# =============================================================================


resumen_ejecutivo = pd.DataFrame({

    "Elemento": [

        "Problema",

        "Alternativas",

        "Criterios",

        "Consistencia criterios",

        "Resultado base",

        "Siguiente paso"

    ],

    "Resultado": [

        "Riesgo de stockout del E-500",

        "Stock seguridad | Dual sourcing | Reabastecimiento predictivo",

        "Costo | Servicio | Riesgo | Tiempo | Flexibilidad",

        f"CR = {CR * 100:.2f}%",

        f"{alternativa_priorizada} ({puntaje_priorizado:.1f}%)",

        "Convertir la alternativa priorizada en un plan de implementación"

    ]

})


print("\n" + "=" * 78)

print("RESUMEN EJECUTIVO")

print("=" * 78)


print(

    resumen_ejecutivo.to_string(
        index=False
    )

)


# =============================================================================
# 39. GUARDAR LOS RESULTADOS
# =============================================================================


df_pesos.to_csv(

    "GlobalBike_AHP_Pesos_Criterios.csv",

    index=False,

    encoding="utf-8-sig"

)


df_prioridades.to_csv(

    "GlobalBike_AHP_Prioridades_Locales.csv",

    encoding="utf-8-sig"

)


resultado_global.to_csv(

    "GlobalBike_AHP_Resultado_Global.csv",

    index=False,

    encoding="utf-8-sig"

)


df_consistencias.to_csv(

    "GlobalBike_AHP_Consistencia.csv",

    index=False,

    encoding="utf-8-sig"

)


df_sensibilidad.to_csv(

    "GlobalBike_AHP_Sensibilidad.csv",

    index=False,

    encoding="utf-8-sig"

)


print("\nARCHIVOS GENERADOS:")


print("""

1. GlobalBike_AHP_Pesos_Criterios.csv

2. GlobalBike_AHP_Prioridades_Locales.csv

3. GlobalBike_AHP_Resultado_Global.csv

4. GlobalBike_AHP_Consistencia.csv

5. GlobalBike_AHP_Sensibilidad.csv

""")


# =============================================================================
# 40. CONCLUSIÓN FINAL AUTOMÁTICA
# =============================================================================


print("\n" + "=" * 78)

print("CONCLUSIÓN DEL AHP")

print("=" * 78)


print(
    f"""

El método AHP permitió estructurar la decisión
de GlobalBike Components utilizando cinco criterios:

1. Costo total.

2. Nivel de servicio.

3. Riesgo de suministro.

4. Tiempo de implementación.

5. Flexibilidad.


Las comparaciones pareadas permitieron transformar
los juicios del ejemplo en pesos cuantitativos.


La matriz de criterios obtuvo:

CR = {CR * 100:.2f}%


Bajo los juicios utilizados en este ejercicio,
la alternativa con mayor prioridad global fue:

{alternativa_priorizada}

con aproximadamente:

{puntaje_priorizado:.1f}%


IMPORTANTE:

AHP no demuestra que esta alternativa sea
universalmente superior.

El resultado significa:

"DADAS estas alternativas,

DADOS estos criterios,

DADOS estos juicios,

y DADOS estos pesos,

esta alternativa recibe la mayor prioridad."


Por eso el resultado debe acompañarse de:

- revisión de consistencia,
- análisis de sensibilidad,
- conocimiento del negocio,
- viabilidad operativa,
- validación de los supuestos.


El siguiente paso ya no consiste
en seguir comparando alternativas.

Ahora debemos preguntarnos:

¿CÓMO CONVERTIMOS LA DECISIÓN
EN UN PLAN DE ACCIÓN?


Esto nos conduce a:

PUNTO 11

INTEGRATION & DECISION MAKING

y posteriormente:

PUNTO 12

FROM ANALYSIS TO ACTION.

"""
)


# =============================================================================
# 41. MAPA FINAL DEL PROCESO
# =============================================================================


print("""

==========================================================================

               GLOBALBIKE COMPONENTS | PROCESO DE DECISIÓN

==========================================================================


                    PROBLEMA DE INVENTARIO

                             ↓

                    DATA ANALYTICS

                             ↓

                IDENTIFICACIÓN DE PATRONES

                             ↓

                   MACHINE LEARNING

                             ↓

                  ALERTA DE STOCKOUT

                             ↓

                 TRES ALTERNATIVAS

                             ↓

        ┌────────────────────┼────────────────────┐

        ↓                    ↓                    ↓

  STOCK SEGURIDAD      DUAL SOURCING      REABASTECIMIENTO
                                            PREDICTIVO

        └────────────────────┼────────────────────┘

                             ↓

                     CINCO CRITERIOS

                             ↓

                            AHP

                             ↓

                  VERIFICAR CONSISTENCIA

                             ↓

                  PRIORIDADES GLOBALES

                             ↓

                ANÁLISIS DE SENSIBILIDAD

                             ↓

                  DECISIÓN SUSTENTADA

                             ↓

                   PLAN DE IMPLEMENTACIÓN


==========================================================================

""")
```
