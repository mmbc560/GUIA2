```python
# -*- coding: utf-8 -*-
"""CASO 1 | Descubrimiento de patrones, alertas y decisiones operacionales.
Ejecutable en Google Colab o Python local. Datos ficticios, no evidencia empresarial.
"""
import warnings
warnings.filterwarnings('ignore', category=FutureWarning)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.ensemble import IsolationForest
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from pathlib import Path

# ================================================================
# 0. CONFIGURACIÓN EDITABLE POR EL EQUIPO
# ================================================================
SEMILLA = 42
N_INICIATIVAS = 720
MESES_HISTORICOS = 12
TOLERANCIA_RETRASO_PCT = 5.0
PRESUPUESTO_MAX = 18_000_000  # unidades monetarias ilustrativas
N_SIMULACIONES = 3000
CONTAMINACION = 0.06  # fracción aproximada de anomalías buscadas
K_MIN, K_MAX = 2, 6
GUARDAR_ARCHIVOS = True
CARPETA = Path('salidas_caso_1')
CARPETA.mkdir(exist_ok=True)
RNG = np.random.default_rng(SEMILLA)
sns.set_theme(style='whitegrid', palette='Set2')
plt.rcParams.update({'figure.facecolor':'white', 'axes.facecolor':'white', 'savefig.facecolor':'white', 'font.size':10})

# Para trabajar con datos reales, sustituir la generación sintética por:
# datos = pd.read_csv('datos_caso_1.csv', parse_dates=['fecha_inicio'])
# Se requieren las columnas que aparecen en VARIABLES_SEGMENTACION, además de
# fecha_inicio, duracion_plan_dias y retraso_pct (esta última solo para validar).

def titulo(txt):
    print('\n' + '='*86 + '\n' + txt + '\n' + '='*86)

def explicar(txt):
    print('\nINTERPRETACIÓN GERENCIAL\n' + txt.strip() + '\n')

def cerrar_grafica(nombre):
    plt.tight_layout()
    if GUARDAR_ARCHIVOS:
        plt.savefig(CARPETA / nombre, dpi=180, bbox_inches='tight', facecolor='white')
    plt.show()

# ================================================================
# 1. GENERACIÓN DE DATOS SINTÉTICOS
# ================================================================
# Cada registro representa una iniciativa. Los tres perfiles generadores
# se usan ÚNICAMENTE para crear datos de práctica; no se entregan a K-Means.
# Las variables tienen unidades interpretables y variación aleatoria.
titulo('1. DATOS HISTÓRICOS Y CONTEXTO OPERACIONAL')
n = N_INICIATIVAS
perfil_oculto = RNG.choice(3, n, p=[0.43, 0.35, 0.22])
fechas = pd.to_datetime('2025-01-01') + pd.to_timedelta(
    RNG.integers(0, MESES_HISTORICOS * 30, n), unit='D')

centros = {
    'areas': [3, 5, 7], 'aprobaciones': [3, 6, 9],
    'espera_aprobacion_dias': [3, 8, 13], 'reprocesos': [0.6, 1.8, 3.5],
    'carga_equipo': [2.0, 3.1, 4.2], 'coordinacion': [4.3, 3.0, 1.9],
    'cambios_alcance': [0.5, 1.3, 2.5]
}

def variable_normal(nombre, desviacion, minimo, maximo, entero=False):
    medias = np.array(centros[nombre])[perfil_oculto]
    valores = np.clip(RNG.normal(medias, desviacion, n), minimo, maximo)
    return np.rint(valores).astype(int) if entero else np.round(valores, 2)

datos = pd.DataFrame({
    'id_iniciativa': [f'I-{i+1:04d}' for i in range(n)],
    'fecha_inicio': fechas,
    'areas': variable_normal('areas', 0.9, 2, 10, True),
    'aprobaciones': variable_normal('aprobaciones', 1.4, 1, 14, True),
    'espera_aprobacion_dias': variable_normal('espera_aprobacion_dias', 2.7, 0.5, 20),
    'reprocesos': variable_normal('reprocesos', 0.9, 0, 7, True),
    'carga_equipo': variable_normal('carga_equipo', 0.65, 1, 5),
    'coordinacion': variable_normal('coordinacion', 0.7, 1, 5),
    'cambios_alcance': variable_normal('cambios_alcance', 0.75, 0, 6, True),
    'duracion_plan_dias': RNG.integers(45, 151, n)
})
# Se incorpora una presión temporal ilustrativa para practicar monitoreo.
meses_indice = (datos.fecha_inicio.dt.year - 2025)*12 + datos.fecha_inicio.dt.month - 1
presion_temporal = np.maximum(0, meses_indice - 6) * 0.32
# Ecuación ficticia del generador: NO debe interpretarse como relación causal
# descubierta por K-Means. En un caso real el retraso se mediría con fechas.
datos['retraso_pct'] = np.maximum(0, np.round(
    -5 + 0.75*datos.aprobaciones + 0.8*datos.espera_aprobacion_dias
    + 2.3*datos.reprocesos + 0.9*datos.areas
    + 1.8*datos.carga_equipo - 2.6*datos.coordinacion
    + 1.1*datos.cambios_alcance + presion_temporal
    + RNG.normal(0, 4.5, n), 2))
datos['duracion_real_dias'] = np.round(datos.duracion_plan_dias * (1 + datos.retraso_pct/100), 1)
datos['incumple'] = datos.retraso_pct > TOLERANCIA_RETRASO_PCT
datos['mes'] = datos.fecha_inicio.dt.to_period('M').astype(str)

VARIABLES_SEGMENTACION = [
    'areas', 'aprobaciones', 'espera_aprobacion_dias', 'reprocesos',
    'carga_equipo', 'coordinacion', 'cambios_alcance'
]
# Se excluye 'retraso_pct' del agrupamiento para evaluar el desempeño DESPUÉS.
# Si se quisiera alertar al inicio, solo se usarían variables disponibles
# en ese momento: por ejemplo, reprocesos futuros no serían admisibles.
print(datos.head(8).to_string(index=False))
print(f'\nIniciativas: {n} | Retraso promedio: {datos.retraso_pct.mean():.1f}% | '
      f'Incumplimiento: {datos.incumple.mean():.1%}')
explicar('La media general resume la magnitud del problema, pero no indica si todas las '
         'iniciativas se comportan de manera similar. Antes de formular una intervención '
         'general, conviene investigar la heterogeneidad: algunas iniciativas pueden '
         'necesitar menos aprobaciones, mientras otras requieren claridad de roles o '
         'prevención de reprocesos. La meta de retraso se usa como referencia gerencial '
         'y no como etiqueta para construir los grupos.')

# ================================================================
# 2. EXPLORACIÓN DESCRIPTIVA
# ================================================================
titulo('2. DISTRIBUCIÓN DEL RETRASO Y VARIABILIDAD')
fig, ax = plt.subplots(figsize=(10, 5))
sns.histplot(datos.retraso_pct, bins=30, color='#2675B8', ax=ax)
ax.axvline(TOLERANCIA_RETRASO_PCT, color='#D34E4E', ls='--', lw=2,
           label=f'Tolerance: {TOLERANCIA_RETRASO_PCT:.0f}%')
ax.axvline(datos.retraso_pct.mean(), color='#E39B32', ls='-', lw=2,
           label=f'Mean: {datos.retraso_pct.mean():.1f}%')
ax.set(title='Delay Distribution Across Strategic Initiatives', xlabel='Delay (%)', ylabel='Number of initiatives')
ax.legend(); cerrar_grafica('01_delay_distribution.png')
explicar('La dispersión permite verificar si el retraso promedio representa adecuadamente '
         'la operación. Una cola de retrasos elevados exige investigar los casos extremos '
         'y no limitarse a reducir el promedio. La decisión futura debe distinguir entre '
         'mejorar la experiencia general y controlar eventos severos. Para ello se '
         'compararán segmentos y anomalías, en vez de adoptar una solución única.')

# ================================================================
# 3. K-MEANS: ESCALAMIENTO Y SELECCIÓN DEL NÚMERO DE GRUPOS
# ================================================================
titulo('3. SEGMENTACIÓN NO SUPERVISADA: K-MEANS')
X = datos[VARIABLES_SEGMENTACION].copy()
scaler = StandardScaler()
X_std = scaler.fit_transform(X)
# Estandarizar evita que una variable en días domine a otra medida de 1 a 5.
ks = list(range(K_MIN, min(K_MAX, n-1)+1))
siluetas = []
for k in ks:
    m = KMeans(n_clusters=k, random_state=SEMILLA, n_init=15)
    etiquetas = m.fit_predict(X_std)
    siluetas.append(silhouette_score(X_std, etiquetas))
mejor_k = ks[int(np.argmax(siluetas))]
print('Silhouette por k:', dict(zip(ks, np.round(siluetas, 3))))
print('Número de clusters sugerido:', mejor_k)
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(ks, siluetas, '-o', color='#2675B8', lw=2)
ax.axvline(mejor_k, ls='--', color='#D67B31')
ax.set(title='Cluster Validation: Silhouette Score', xlabel='Number of clusters', ylabel='Silhouette score')
cerrar_grafica('02_silhouette.png')

modelo_kmeans = KMeans(n_clusters=mejor_k, random_state=SEMILLA, n_init=20)
datos['cluster'] = modelo_kmeans.fit_predict(X_std)
pca = PCA(n_components=2, random_state=SEMILLA)
coord = pca.fit_transform(X_std)
fig, ax = plt.subplots(figsize=(10, 6))
sns.scatterplot(x=coord[:,0], y=coord[:,1], hue=datos.cluster.astype(str),
                palette='tab10', s=42, alpha=0.72, ax=ax)
ax.set(title='Operational Clusters (PCA Visualization)', xlabel='Principal component 1', ylabel='Principal component 2')
ax.legend(title='Cluster'); cerrar_grafica('03_kmeans_clusters.png')
print('Varianza explicada por dos componentes:', round(pca.explained_variance_ratio_.sum(), 3))
explicar(f'La segmentación distingue {mejor_k} perfiles de iniciativas con condiciones '
         'operacionales similares. Los colores NO representan niveles de éxito o fracaso; '
         'su significado se establece al revisar los promedios y los resultados externos. '
         'La visualización PCA simplifica siete variables a dos e inevitablemente pierde '
         'parte de la información. Por ello, la interpretación de los grupos debe '
         'basarse en sus perfiles originales, no solamente en la distancia visual.')

# ================================================================
# 4. MAPA DE CALOR Y DESEMPEÑO POR SEGMENTO
# ================================================================
titulo('4. PERFIL OPERACIONAL Y DESEMPEÑO POR CLUSTER')
perfil = datos.groupby('cluster')[VARIABLES_SEGMENTACION].mean()
perfil_z = (perfil - X.mean()) / X.std(ddof=0)
fig, ax = plt.subplots(figsize=(12, max(3.8, mejor_k*0.9)))
sns.heatmap(perfil_z, annot=True, fmt='.2f', cmap='RdYlBu_r', center=0,
            linewidths=.5, ax=ax, cbar_kws={'label':'Standardized difference'})
ax.set(title='Operational Profile by Cluster (Standardized Means)', xlabel='Operational variables', ylabel='Cluster')
plt.xticks(rotation=30, ha='right'); cerrar_grafica('04_cluster_heatmap.png')

resumen = datos.groupby('cluster').agg(
    iniciativas=('id_iniciativa','count'),
    retraso_promedio=('retraso_pct','mean'),
    retraso_p90=('retraso_pct', lambda s: s.quantile(.9)),
    incumplimiento=('incumple','mean'),
    duracion_real=('duracion_real_dias','mean'))
print('\nRESUMEN POR SEGMENTO\n', resumen.round(2).to_string())
fig, ax = plt.subplots(figsize=(9, 5))
colores = sns.color_palette('Set2', mejor_k)
ax.bar(resumen.index.astype(str), resumen.retraso_promedio, color=colores)
ax.axhline(TOLERANCIA_RETRASO_PCT, ls='--', color='#D34E4E', label='Tolerance')
ax.set(title='Average Delay by Operational Cluster', xlabel='Cluster', ylabel='Average delay (%)')
ax.legend(); cerrar_grafica('05_cluster_performance.png')

cluster_prioritario = int(resumen.retraso_promedio.idxmax())
cluster_mas_estable = int(resumen.retraso_promedio.idxmin())
for c in resumen.index:
    diferencias = perfil_z.loc[c].sort_values(key=lambda s: s.abs(), ascending=False)
    rasgos = ', '.join(f'{v} ({valor:+.2f} DE)' for v,valor in diferencias.head(3).items())
    explicar(f'CLUSTER {c}: contiene {int(resumen.loc[c,"iniciativas"])} iniciativas; '
             f'retraso promedio {resumen.loc[c,"retraso_promedio"]:.1f}%; '
             f'incumplimiento {resumen.loc[c,"incumplimiento"]:.1%}. '
             f'Sus características más diferenciadoras son: {rasgos}. '
             'Estas diferencias son descriptivas, no pruebas causales. Se recomienda '
             'examinar casos representativos, reconstruir el flujo de aprobaciones y '
             'validar si los factores destacados explican esperas o reprocesos. '
             'Si el segmento presenta desempeño desfavorable, la intervención debería '
             'ser específica y su efectividad comprobarse mediante un piloto.')
explicar(f'PRIORIZACIÓN: el cluster {cluster_prioritario} presenta el mayor retraso '
         f'promedio ({resumen.loc[cluster_prioritario,"retraso_promedio"]:.1f}%). '
         f'El cluster {cluster_mas_estable} muestra el menor. Conviene comparar ambos '
         'para identificar prácticas transferibles y condiciones que no sean comparables. '
         'No se recomienda transferir una práctica sin evaluar complejidad, recursos y '
         'restricciones de cada grupo. El porcentaje de incumplimiento y el percentil '
         '90 deben complementar el promedio al decidir prioridades.')

# ================================================================
# 5. ISOLATION FOREST: ANOMALÍAS
# ================================================================
titulo('5. DETECCIÓN DE ANOMALÍAS: ISOLATION FOREST')
modelo_anomalias = IsolationForest(contamination=CONTAMINACION,
                                  random_state=SEMILLA, n_estimators=250)
datos['anomalia'] = modelo_anomalias.fit_predict(X_std) == -1
# score_samples: valores más negativos suelen indicar mayor anomalía.
datos['puntaje_anomalia'] = -modelo_anomalias.score_samples(X_std)
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(coord[~datos.anomalia,0], coord[~datos.anomalia,1],
           c='#2A7DBA', alpha=.4, s=27, label='Typical observations')
ax.scatter(coord[datos.anomalia,0], coord[datos.anomalia,1],
           c='#D84855', marker='X', s=75, label='Anomalies')
ax.set(title='Isolation Forest: Unusual Operational Conditions',
       xlabel='Principal component 1', ylabel='Principal component 2')
ax.legend(); cerrar_grafica('06_anomalies.png')

an = datos[datos.anomalia].sort_values('puntaje_anomalia', ascending=False)
print(f'Anomalías señaladas: {len(an)} de {n} ({len(an)/n:.1%})')
print(an[['id_iniciativa','cluster','aprobaciones','espera_aprobacion_dias',
          'reprocesos','coordinacion','retraso_pct','puntaje_anomalia']].head(10).to_string(index=False))
explicar(f'Se identificaron {len(an)} observaciones inusuales bajo el umbral '
         f'configurado ({CONTAMINACION:.0%}). Esta proporción depende en parte del '
         'parámetro de contaminación y NO es una tasa observada de fallas. '
         'Las anomalías deben revisarse una por una: pueden representar combinaciones '
         'inusuales de aprobaciones, cambios de alcance o coordinación, pero también '
         'registros incorrectos o proyectos excepcionalmente complejos. Una medida '
         'práctica sería instaurar una revisión preventiva de casos atípicos, con '
         'criterios de escalamiento y trazabilidad de decisiones.')

# ================================================================
# 6. EVOLUCIÓN TEMPORAL Y ALERTAS
# ================================================================
titulo('6. MONITOREO TEMPORAL DE LOS SEGMENTOS')
conteos = pd.crosstab(datos.mes, datos.cluster)
participacion = conteos.div(conteos.sum(axis=1), axis=0) * 100
fig, ax = plt.subplots(figsize=(11, 5.5))
for c in participacion.columns:
    ax.plot(participacion.index, participacion[c], marker='o', lw=2,
            label=f'Cluster {c}')
ax.set(title='Temporal Monitoring of Operational Clusters',
       xlabel='Month', ylabel='Share of initiatives (%)')
plt.xticks(rotation=40); ax.legend(ncol=3); cerrar_grafica('07_temporal_clusters.png')

serie_riesgo = participacion[cluster_prioritario]
ventana = min(3, len(serie_riesgo))
media_reciente = float(serie_riesgo.tail(ventana).mean())
media_inicial = float(serie_riesgo.head(ventana).mean())
cambio_pp = media_reciente - media_inicial
print(f'Participación inicial del segmento prioritario: {media_inicial:.1f}%')
print(f'Participación reciente del segmento prioritario: {media_reciente:.1f}%')
print(f'Cambio: {cambio_pp:+.1f} puntos porcentuales')
explicar(f'La participación del segmento con mayor retraso pasó de un promedio de '
         f'{media_inicial:.1f}% al inicio a {media_reciente:.1f}% en los últimos '
         f'{ventana} meses ({cambio_pp:+.1f} puntos). Una variación creciente '
         'sugeriría investigar presión sobre la capacidad, cambios en la cartera '
         'de iniciativas o deterioro de las interfaces entre áreas. Una variación '
         'decreciente podría ser favorable, pero también reflejar cambios de mezcla. '
         'No debe atribuirse automáticamente a una intervención. Es recomendable '
         'comparar periodos equivalentes y definir umbrales de alerta con más historia.')

# ================================================================
# 7. PERSPECTIVA PREDICTIVA SIN APRENDIZAJE SUPERVISADO
# ================================================================
titulo('7. PROYECCIÓN EXPLORATORIA DE TENDENCIAS')
# Se mantiene la decisión de usar únicamente ML no supervisado.
# La siguiente extrapolación lineal NO es un pronóstico validado ni usa RF.
# Sirve para ilustrar qué podría ocurrir si persistiera la tendencia reciente.
y = serie_riesgo.to_numpy(dtype=float)
periodos = min(6, len(y))
x_rec = np.arange(periodos)
pendiente, intercepto = np.polyfit(x_rec, y[-periodos:], 1)
proximos = np.arange(periodos, periodos+3)
proyeccion = np.clip(intercepto + pendiente*proximos, 0, 100)
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(np.arange(len(y)), y, '-o', color='#237BB4', label='Observed cluster share')
ax.plot(np.arange(len(y),len(y)+3), proyeccion, '--o',
        color='#DE8A30', label='Exploratory projection')
ax.set(title='Exploratory Outlook: High-Delay Cluster Share',
       xlabel='Month index', ylabel='Share of initiatives (%)')
ax.legend(); cerrar_grafica('08_exploratory_outlook.png')
print('Participación proyectada, próximos tres periodos:', np.round(proyeccion,1))
explicar('La proyección prolonga una tendencia reciente de la participación del '
         'segmento prioritario; no estima directamente el retraso de un proyecto '
         'nuevo. Puede utilizarse como escenario de vigilancia: si aumentara la '
         'concentración de iniciativas en ese segmento, convendría reforzar el '
         'seguimiento y comprobar capacidad disponible. No obstante, seis meses '
         'son insuficientes para validar estacionalidad o precisión predictiva. '
         'Para tomar decisiones reales se necesitan más observaciones temporales, '
         'comparación con un pronóstico ingenuo e intervalos de incertidumbre.')

# ================================================================
# 8. PRESCRIPCIÓN: SIMULACIÓN DE ESCENARIOS GERENCIALES
# ================================================================
titulo('8. EVALUACIÓN PRESCRIPTIVA DE ESCENARIOS')
# IMPORTANTE: se usan supuestos de intervención explícitos.
# No se consideran efectos causales estimados a partir del clustering.
# Las reducciones en retraso se calculan mediante una función de respuesta
# ilustrativa, que debería sustituirse por resultados de pilotos reales.
# La intervención se aplica al segmento prioritario, no a toda la organización.

ALTERNATIVAS = {
    'Baseline': {'costo':0, 'aprobaciones':0, 'espera':0, 'reprocesos':0, 'coordinacion':0},
    'Approval redesign': {'costo':7_000_000, 'aprobaciones':2, 'espera':2.5, 'reprocesos':0, 'coordinacion':0},
    'Coordination protocol': {'costo':11_000_000, 'aprobaciones':0, 'espera':1.0, 'reprocesos':0, 'coordinacion':1.2},
    'Rework prevention': {'costo':9_000_000, 'aprobaciones':0, 'espera':0, 'reprocesos':1.7, 'coordinacion':0},
    'Combined intervention': {'costo':24_000_000, 'aprobaciones':2, 'espera':3.0, 'reprocesos':1.7, 'coordinacion':1.2}
}
# Se simula una cartera futura con características similares a las del
# segmento prioritario, sin afirmar que los casos futuros serán idénticos.
base = datos.loc[datos.cluster==cluster_prioritario].copy()
indices = RNG.integers(0, len(base), size=N_SIMULACIONES)
cartera = base.iloc[indices].reset_index(drop=True)
# Incertidumbre común entre escenarios para comparación consistente.
ruido = RNG.normal(0, 4.5, N_SIMULACIONES)
# Esta respuesta está alineada con el generador sintético, pero no es un
# efecto validado en el mundo real. Los parámetros deben revisarse con pilotos.
def retraso_simulado(df, accion, ruido):
    aprob = np.maximum(1, df.aprobaciones.to_numpy()-accion['aprobaciones'])
    espera = np.maximum(0, df.espera_aprobacion_dias.to_numpy()-accion['espera'])
    repro = np.maximum(0, df.reprocesos.to_numpy()-accion['reprocesos'])
    coord = np.minimum(5, df.coordinacion.to_numpy()+accion['coordinacion'])
    return np.maximum(0,
        -5 + .75*aprob + .8*espera + 2.3*repro
        + .9*df.areas.to_numpy() + 1.8*df.carga_equipo.to_numpy()
        - 2.6*coord + 1.1*df.cambios_alcance.to_numpy() + ruido)

filas = []
for nombre, accion in ALTERNATIVAS.items():
    r = retraso_simulado(cartera, accion, ruido)
    filas.append({'alternativa':nombre, 'costo':accion['costo'],
                  'retraso_medio':r.mean(), 'retraso_p90':np.quantile(r,.9),
                  'cumplimiento_pct':100*np.mean(r<=TOLERANCIA_RETRASO_PCT),
                  'factible':accion['costo']<=PRESUPUESTO_MAX})
resultados = pd.DataFrame(filas)
base_retraso = float(resultados.loc[resultados.alternativa=='Baseline','retraso_medio'].iloc[0])
resultados['reduccion_pp'] = base_retraso-resultados.retraso_medio
print(resultados.round(2).to_string(index=False))

fig, ax = plt.subplots(figsize=(11, 5))
colores = ['#7B8A97' if nombre=='Baseline' else ('#2D9C78' if fact else '#D87E37')
           for nombre, fact in zip(resultados.alternativa, resultados.factible)]
ax.barh(resultados.alternativa, resultados.retraso_medio, color=colores)
ax.axvline(TOLERANCIA_RETRASO_PCT, color='#D54B52', ls='--', label='Tolerance')
ax.invert_yaxis()
ax.set(title='Scenario Comparison: Expected Delay', xlabel='Simulated mean delay (%)')
ax.legend(); cerrar_grafica('09_scenario_comparison.png')

fig, ax = plt.subplots(figsize=(10, 5))
for _, fila in resultados.iterrows():
    ax.scatter(fila.costo/1e6, fila.cumplimiento_pct, s=130,
               c='#2D9C78' if fila.factible else '#D87E37')
    ax.annotate(fila.alternativa, (fila.costo/1e6, fila.cumplimiento_pct),
                xytext=(5,6), textcoords='offset points', fontsize=8)
ax.axvline(PRESUPUESTO_MAX/1e6, color='#D54B52', ls='--', label='Budget limit')
ax.set(title='Cost vs. Simulated On-Time Compliance',
       xlabel='Intervention cost (millions)', ylabel='Compliance (%)')
ax.legend(); cerrar_grafica('10_cost_compliance.png')

factibles = resultados[(resultados.factible) & (resultados.alternativa!='Baseline')]
if len(factibles):
    # Criterio gerencial explícito: minimizar retraso medio con presupuesto fijo.
    mejor = factibles.sort_values(['retraso_medio','costo']).iloc[0]
    recomendada = mejor.alternativa
    explicar(f'Bajo el presupuesto de {PRESUPUESTO_MAX:,.0f}, la alternativa '
             f'con menor retraso simulado es {recomendada}, con costo '
             f'{mejor.costo:,.0f}, retraso medio {mejor.retraso_medio:.1f}% '
             f'y cumplimiento {mejor.cumplimiento_pct:.1f}%. '
             'Esta selección es óptima únicamente entre las alternativas '
             'predefinidas y bajo la función de respuesta supuesta. No demuestra '
             'que sea la mejor decisión real. Antes de aprobarla se recomienda '
             'pilotar la intervención, comprobar su costo completo, verificar '
             'efectos secundarios y medir indicadores antes y después.')
else:
    recomendada = None
    explicar('No existe una alternativa de intervención factible bajo el '
             'presupuesto definido. Se requiere reformular opciones, fases '
             'de implementación o recursos disponibles.')

for _, f in resultados.iterrows():
    if f.alternativa=='Baseline':
        continue
    condicion = 'dentro' if f.factible else 'fuera'
    explicar(f'ESCENARIO {f.alternativa}: cuesta {f.costo:,.0f} y está {condicion} '
             f'del presupuesto. Reduce el retraso promedio simulado '
             f'{f.reduccion_pp:.1f} puntos porcentuales frente a la situación '
             f'base; el percentil 90 es {f.retraso_p90:.1f}% y el cumplimiento '
             f'proyectado bajo supuestos es {f.cumplimiento_pct:.1f}%. '
             'Una reducción del promedio no garantiza que los casos extremos '
             'desaparezcan ni que se alcance la tolerancia. Conviene comparar '
             'estos tres indicadores, valorar efectos sobre otras áreas y '
             'diseñar un piloto con criterios de éxito previamente definidos.')

# ================================================================
# 9. RECOMENDACIONES EJECUTIVAS Y PLAN DE VALIDACIÓN
# ================================================================
titulo('9. CONCLUSIONES EJECUTIVAS Y DECISIONES FUTURAS')
print(f'''DIAGNÓSTICO:
- Retraso promedio histórico: {datos.retraso_pct.mean():.1f}%.
- Iniciativas fuera de tolerancia: {datos.incumple.mean():.1%}.
- Segmentos identificados: {mejor_k}.
- Segmento prioritario: {cluster_prioritario}.
- Anomalías a investigar: {len(an)}.
- Cambio reciente del segmento prioritario: {cambio_pp:+.1f} puntos porcentuales.
- Alternativa preferida bajo supuestos y presupuesto: {recomendada or 'Ninguna'}.
''')
explicar('DECISIÓN ESTRATÉGICA: evitar implantar una misma solución para todas '
         'las iniciativas. Se propone diferenciar políticas de seguimiento '
         'según perfiles operacionales, verificar qué segmentos consumen más '
         'capacidad y revisar la gobernanza de decisiones. Los clusters ayudan '
         'a segmentar; no sustituyen la definición de prioridades estratégicas.')
explicar('DECISIÓN TÁCTICA: establecer un tablero mensual que muestre la '
         'participación de cada segmento, retraso promedio, percentil 90, '
         'cumplimiento y frecuencia de anomalías. Si aumenta el peso del '
         'segmento prioritario, se recomienda activar una revisión de causas '
         'y capacidad, en vez de esperar a que los proyectos terminen.')
explicar('DECISIÓN OPERACIONAL: revisar una muestra de iniciativas anómalas '
         'y otra de iniciativas representativas de cada segmento. Elaborar '
         'SIPOC o VSM de las rutas de aprobación, contrastar reprocesos y '
         'responsabilidades, y documentar los tiempos reales de espera. '
         'Posteriormente se pueden usar Ishikawa, 5 porqués y FMEA para '
         'validar las causas y sus riesgos.')
explicar('VALIDACIÓN FUTURA: realizar un piloto controlado o escalonado '
         'durante varios periodos. Registrar variables antes de iniciar '
         'cada iniciativa para construir alertas verdaderamente tempranas; '
         'evaluar los cambios de resultados frente a un grupo comparable; '
         'actualizar los clusters y verificar que los patrones sean estables. '
         'Si no existe mejora medible, se deberá reconsiderar la hipótesis '
         'de intervención, aunque la simulación inicial hubiera sido favorable.')
explicar('LIMITACIONES: los datos y las relaciones de respuesta son '
         'sintéticos. K-Means depende de la escala y del número de grupos; '
         'Isolation Forest detecta rareza, no fallas; la proyección temporal '
         'es exploratoria y la simulación prescriptiva depende de supuestos '
         'no validados. No se deben presentar porcentajes simulados como '
         'resultados empresariales comprobados ni como efectos causales.')

if GUARDAR_ARCHIVOS:
    datos.to_csv(CARPETA/'datos_sinteticos_caso_1.csv', index=False, encoding='utf-8-sig')
    resumen.to_csv(CARPETA/'resumen_clusters.csv', encoding='utf-8-sig')
    resultados.to_csv(CARPETA/'escenarios_gerenciales.csv', index=False, encoding='utf-8-sig')
    print('\nArchivos exportados en:', CARPETA.resolve())
print('\nFIN DEL LABORATORIO - CASO 1')
```
