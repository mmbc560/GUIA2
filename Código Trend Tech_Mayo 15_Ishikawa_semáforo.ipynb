```python
# ============================================================
# ISHIKAWA CON SEMÁFORO DE CAUSAS
# Trend Tech - Detección de causas críticas
# ============================================================

import matplotlib.pyplot as plt
import textwrap

# ============================================================
# 1. FUNCIÓN PARA CLASIFICAR SEMÁFORO
# ============================================================

def clasificar_semaforo(promedio):
    if promedio < 2.5:
        return "Verde", "green"
    elif promedio < 4:
        return "Naranja", "orange"
    else:
        return "Rojo", "red"

# ============================================================
# 2. FUNCIÓN PARA CORTAR TEXTOS LARGOS
# ============================================================

def ajustar_texto(texto, ancho=24):
    return "\n".join(textwrap.wrap(texto, width=ancho))

# ============================================================
# 3. PREGUNTAS DE EVALUACIÓN
# ============================================================

preguntas = [
    "¿Qué tanto impacto tiene esta causa en el problema?",
    "¿Qué tan frecuente ocurre esta causa?",
    "¿Qué tanto afecta al cliente, usuario o proceso?",
    "¿Qué tanto podría resolverse con tecnología?",
    "¿Qué tan urgente es intervenir esta causa?"
]

# ============================================================
# 4. INGRESO DEL PROBLEMA PRINCIPAL
# ============================================================

print("===================================================")
print("CONSTRUCCIÓN DEL DIAGRAMA DE ISHIKAWA CON SEMÁFORO")
print("===================================================")

problema = input("\nEscriba el problema principal de la empresa: ")

# ============================================================
# 5. INGRESO DE CATEGORÍAS Y CAUSAS
# ============================================================

while True:
    try:
        num_categorias = int(input("\n¿Cuántas categorías tendrá el Ishikawa? "))
        if num_categorias > 0:
            break
        else:
            print("Debe ingresar un número mayor que cero.")
    except ValueError:
        print("Ingrese un número válido.")

causas = {}

for i in range(num_categorias):
    categoria = input(f"\nNombre de la categoría {i+1}: ")

    while True:
        try:
            num_causas = int(input(f"¿Cuántas causas tendrá la categoría '{categoria}'? "))
            if num_causas > 0:
                break
            else:
                print("Debe ingresar un número mayor que cero.")
        except ValueError:
            print("Ingrese un número válido.")

    causas[categoria] = []

    for j in range(num_causas):
        causa = input(f"  Escriba la causa {j+1} de '{categoria}': ")
        causas[categoria].append(causa)

# ============================================================
# 6. EVALUACIÓN DE CAUSAS
# ============================================================

resultados = {}

print("\n===================================================")
print("EVALUACIÓN DE CAUSAS")
print("Califique cada pregunta de 1 a 5")
print("1 = bajo | 5 = alto")
print("===================================================")

for categoria, lista_causas in causas.items():
    resultados[categoria] = []

    for causa in lista_causas:
        print(f"\nCategoría: {categoria}")
        print(f"Causa: {causa}")

        puntajes = []

        for pregunta in preguntas:
            while True:
                try:
                    respuesta = int(input(pregunta + " "))
                    if 1 <= respuesta <= 5:
                        puntajes.append(respuesta)
                        break
                    else:
                        print("Ingrese un número entre 1 y 5.")
                except ValueError:
                    print("Ingrese un número válido.")

        promedio = sum(puntajes) / len(puntajes)
        nivel, color = clasificar_semaforo(promedio)

        resultados[categoria].append({
            "causa": causa,
            "promedio": promedio,
            "nivel": nivel,
            "color": color
        })

# ============================================================
# 7. RESULTADOS EN TEXTO
# ============================================================

print("\n===================================================")
print("RESULTADOS DEL SEMÁFORO")
print("===================================================")

for categoria, lista in resultados.items():
    print(f"\nCategoría: {categoria}")
    for item in lista:
        print(f"- {item['causa']} | Puntaje: {item['promedio']:.2f} | Semáforo: {item['nivel']}")

# ============================================================
# 8. IDENTIFICAR CAUSAS CRÍTICAS ROJAS
# ============================================================

causas_rojas = []

for categoria, lista in resultados.items():
    for item in lista:
        if item["nivel"] == "Rojo":
            causas_rojas.append((categoria, item["causa"], item["promedio"]))

print("\n===================================================")
print("CAUSAS CRÍTICAS ROJAS")
print("===================================================")

if causas_rojas:
    for categoria, causa, promedio in causas_rojas:
        print(f"- {causa} ({categoria}) | Puntaje: {promedio:.2f}")
else:
    print("No se identificaron causas rojas. Revise las causas naranjas como oportunidades de mejora.")

# ============================================================
# 9. DIBUJAR DIAGRAMA DE ISHIKAWA
# ============================================================

fig, ax = plt.subplots(figsize=(18, 10))
ax.set_xlim(0, 12)
ax.set_ylim(0, 10)
ax.axis("off")

# Línea principal
ax.plot([1, 9.5], [5, 5], color="black", linewidth=3)

# Cabeza del pescado
ax.text(
    10.1, 5,
    ajustar_texto(problema, 22),
    fontsize=12,
    fontweight="bold",
    va="center",
    ha="center",
    bbox=dict(boxstyle="round,pad=0.6", facecolor="lightgray", edgecolor="black")
)

# Posiciones dinámicas de categorías
categorias = list(resultados.keys())
n = len(categorias)

x_positions = [2 + i * (6.5 / max(n - 1, 1)) for i in range(n)]

for idx, categoria in enumerate(categorias):
    x_base = x_positions[idx]
    y_base = 5

    # Alternar categorías arriba y abajo
    if idx % 2 == 0:
        y_text = 8
        direccion = -1
    else:
        y_text = 2
        direccion = 1

    # Espina principal
    ax.plot([x_base, x_base - 0.8], [y_base, y_text], color="black", linewidth=2)

    # Nombre categoría
    ax.text(
        x_base - 0.8,
        y_text + (0.4 if y_text > 5 else -0.4),
        ajustar_texto(categoria, 16),
        fontsize=12,
        fontweight="bold",
        ha="center",
        va="center"
    )

    # Causas dentro de cada categoría
    lista_causas = resultados[categoria]

    for j, item in enumerate(lista_causas):
        x_causa = x_base - 1.3
        y_causa = y_text + direccion * j * 0.65

        ax.scatter(
            x_causa,
            y_causa,
            s=350,
            color=item["color"],
            edgecolor="black",
            zorder=3
        )

        ax.text(
            x_causa + 0.35,
            y_causa,
            ajustar_texto(f"{item['causa']} ({item['promedio']:.1f})", 28),
            fontsize=9,
            va="center"
        )

# ============================================================
# 10. LEYENDA
# ============================================================

ax.scatter(1, 0.8, s=350, color="green", edgecolor="black")
ax.text(1.35, 0.8, "Verde: baja prioridad", fontsize=11, va="center")

ax.scatter(3.8, 0.8, s=350, color="orange", edgecolor="black")
ax.text(4.15, 0.8, "Naranja: prioridad media", fontsize=11, va="center")

ax.scatter(7.2, 0.8, s=350, color="red", edgecolor="black")
ax.text(7.55, 0.8, "Rojo: causa crítica", fontsize=11, va="center")

plt.title(
    "Diagrama de Ishikawa con Semáforo de Causas",
    fontsize=18,
    fontweight="bold"
)

plt.show()

# ============================================================
# 11. PREGUNTAS PARA PASAR DE CAUSA ROJA A TECNOLOGÍA
# ============================================================

print("\n===================================================")
print("PREGUNTAS PARA CONVERTIR CAUSAS ROJAS EN SOLUCIONES TECNOLÓGICAS")
print("===================================================")

for categoria, causa, promedio in causas_rojas:
    print(f"\nCausa crítica: {causa}")
    print(f"Categoría: {categoria}")
    print("Preguntas guía:")
    print("- ¿Qué parte de esta causa podría automatizarse?")
    print("- ¿Qué información hace falta para controlar esta causa?")
    print("- ¿Qué datos deberían visualizarse en tiempo real?")
    print("- ¿Qué tecnología podría reducir esta causa?")
    print("- ¿Qué solución podría convertir esta causa en una oportunidad de mejora?")
```
