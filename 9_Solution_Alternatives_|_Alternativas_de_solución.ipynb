```python
# =============================================================================
# INVENTARIOS Y COMPRAS GLOBALES
# CASO INTEGRADOR: GLOBALBIKE COMPONENTS COLOMBIA S.A.S.
#
# ETAPA 9: ALTERNATIVAS DE SOLUCIÓN
#
# OBJETIVO:
# Traducir el diagnóstico realizado en Data Analytics y Machine Learning
# en alternativas concretas de mejora para el controlador E-500.
#
# IMPORTANTE:
# En esta etapa NO vamos a escoger todavía "la mejor alternativa".
#
# Primero vamos a:
#
# 1. Definir tres alternativas.
# 2. Estimar sus efectos.
# 3. Comparar beneficios y trade-offs.
# 4. Construir escenarios.
# 5. Preparar la información para AHP.
#
# En la ETAPA 10 utilizaremos AHP para priorizarlas.
#
# =============================================================================


# =============================================================================
# 1. IMPORTAR LIBRERÍAS
# =============================================================================

# pandas:
# permite trabajar con tablas de datos.

import pandas as pd

# numpy:
# permite realizar cálculos y simulaciones.

import numpy as np

# matplotlib:
# permite construir gráficas.

import matplotlib.pyplot as plt

# seaborn:
# facilita visualizaciones estadísticas.

import seaborn as sns


# Configuración visual.

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 6)

plt.rcParams["font.size"] = 11


# =============================================================================
# 2. RECORDAR EL PROBLEMA
# =============================================================================
#
# GlobalBike Components importa el controlador E-500.
#
# Los análisis anteriores sugieren que el riesgo de stockout
# está relacionado con diferentes factores:
#
# - Variabilidad de la demanda.
# - Error del pronóstico.
# - Inventario disponible.
# - Lead time internacional.
# - Riesgo del proveedor.
#
# Machine Learning permitió generar alertas predictivas.
#
# Pero una alerta NO es una solución.
#
# Ahora debemos responder:
#
# ¿QUÉ PUEDE HACER GLOBALBIKE?
#
# =============================================================================


print("\n" + "=" * 78)

print("GLOBALBIKE COMPONENTS | ALTERNATIVAS DE SOLUCIÓN")

print("=" * 78)

print("""
PROBLEMA:

GlobalBike necesita reducir el riesgo de stockout del controlador E-500
sin aumentar innecesariamente el costo total de la cadena.

PREGUNTA:

¿Qué alternativas podrían mejorar simultáneamente la disponibilidad,
la resiliencia y la gestión del inventario?
""")


# =============================================================================
# 3. DEFINIR LA SITUACIÓN ACTUAL
# =============================================================================
#
# Para que el ejemplo pueda ejecutarse de manera independiente,
# vamos a establecer algunos valores base.
#
# Estos valores son SIMULADOS y tienen propósito académico.
#
# En el proyecto real de los estudiantes:
#
# ESTOS SON LOS DATOS QUE DEBERÍAN CAMBIAR.
#
# -----------------------------------------------------------------------------


demanda_promedio_mensual = 1050

# Demanda promedio del controlador E-500.


desviacion_demanda = 180

# Variabilidad aproximada de la demanda mensual.


lead_time_actual_dias = 75

# Lead time promedio actual.


desviacion_lead_time = 14

# Variabilidad del lead time.


inventario_actual = 1200

# Inventario promedio aproximado.


nivel_servicio_actual = 89

# Nivel de servicio actual (%).


stockouts_anuales_actuales = 4

# Número aproximado de periodos con stockout por año.


costo_unitario_actual = 42

# Precio aproximado por controlador en USD.


dependencia_proveedor_actual = 92

# Porcentaje del abastecimiento concentrado
# en el proveedor principal.


compras_urgentes_actuales = 10

# Porcentaje aproximado de compras que deben realizarse
# de manera urgente.


# =============================================================================
# 4. DEFINIR LAS TRES ALTERNATIVAS
# =============================================================================
#
# ALTERNATIVA 1:
#
# AJUSTAR STOCK DE SEGURIDAD
#
# La empresa mantiene su estructura actual de abastecimiento,
# pero incrementa la protección frente a la incertidumbre.
#
#
# ALTERNATIVA 2:
#
# DUAL SOURCING
#
# La empresa mantiene al proveedor principal,
# pero asigna parte de la compra a un segundo proveedor.
#
#
# ALTERNATIVA 3:
#
# REABASTECIMIENTO PREDICTIVO
#
# La empresa utiliza pronósticos, alertas de Machine Learning
# y reglas dinámicas de inventario para anticipar pedidos.
#
# =============================================================================


alternativas = [

    "A1 - Stock de seguridad",

    "A2 - Dual sourcing",

    "A3 - Reabastecimiento predictivo"

]


# =============================================================================
# 5. EXPLICAR LA ALTERNATIVA 1
# =============================================================================

print("\n" + "=" * 78)

print("ALTERNATIVA 1 | AJUSTAR STOCK DE SEGURIDAD")

print("=" * 78)

print("""
La primera posibilidad consiste en aumentar la protección del inventario.

La lógica es sencilla:

MAYOR INCERTIDUMBRE
        ↓
MAYOR PROTECCIÓN
        ↓
MENOR PROBABILIDAD DE STOCKOUT

Sin embargo:

MAYOR STOCK DE SEGURIDAD
        ↓
MAYOR INVENTARIO PROMEDIO
        ↓
MAYOR CAPITAL INMOVILIZADO

Por eso aumentar inventario puede disminuir el riesgo,
pero no necesariamente es una solución gratuita.
""")


# =============================================================================
# 6. EJEMPLO SIMPLIFICADO DE STOCK DE SEGURIDAD
# =============================================================================
#
# Existen diferentes fórmulas para calcular stock de seguridad.
#
# Para fines pedagógicos utilizaremos:
#
# SS = Z × sigma_LT
#
# donde:
#
# SS = Stock de seguridad
#
# Z = factor asociado al nivel de servicio deseado
#
# sigma_LT = desviación de la demanda durante el lead time
#
#
# Si tanto demanda como lead time varían,
# una aproximación utilizada es:
#
#
# SS = Z × sqrt(
#
#       LT × sigma_D²
#
#       +
#
#       D² × sigma_LT²
#
# )
#
#
# IMPORTANTE:
#
# Debemos trabajar con unidades compatibles.
#
# Convertiremos los valores mensuales a diarios.
#
# =============================================================================


# Demanda diaria aproximada.

demanda_diaria = (
    demanda_promedio_mensual
    /
    30
)


# Desviación diaria aproximada.
#
# Para este ejemplo académico utilizamos una aproximación sencilla.
#
# En una aplicación real sería preferible calcularla
# directamente con datos diarios.

desviacion_demanda_diaria = (
    desviacion_demanda
    /
    np.sqrt(30)
)


# Factor Z para un nivel de servicio cercano al 95 %.

Z = 1.645


# Aplicamos la fórmula.

stock_seguridad_calculado = (

    Z

    *

    np.sqrt(

        lead_time_actual_dias
        *
        desviacion_demanda_diaria**2

        +

        demanda_diaria**2
        *
        desviacion_lead_time**2

    )

)


print(
    f"\nStock de seguridad estimado: "
    f"{stock_seguridad_calculado:.0f} unidades"
)


# =============================================================================
# 7. PUNTO DE REORDEN
# =============================================================================
#
# Una política sencilla de inventarios puede utilizar:
#
#
# ROP = Demanda durante el lead time + Stock de seguridad
#
#
# ROP = Reorder Point
#
# o
#
# Punto de Reorden.
#
#
# Primero calculamos:
#
# Demanda durante lead time =
#
# Demanda diaria × Lead time
#
# =============================================================================


demanda_durante_lead_time = (

    demanda_diaria

    *

    lead_time_actual_dias

)


punto_reorden = (

    demanda_durante_lead_time

    +

    stock_seguridad_calculado

)


print(
    f"Demanda esperada durante el lead time: "
    f"{demanda_durante_lead_time:.0f} unidades"
)


print(
    f"Punto de reorden estimado: "
    f"{punto_reorden:.0f} unidades"
)


print("""
INTERPRETACIÓN:

El punto de reorden no significa que debamos mantener siempre
esa cantidad de inventario.

Representa el nivel aproximado en el cual debería activarse
una orden de reposición bajo los supuestos utilizados.

La lógica es:

INVENTARIO BAJA
       ↓
ALCANZA EL ROP
       ↓
SE EMITE ORDEN
       ↓
SE CONSUME INVENTARIO DURANTE EL LEAD TIME
       ↓
LLEGA EL NUEVO PEDIDO
""")


# =============================================================================
# 8. VISUALIZAR LA LÓGICA DEL PUNTO DE REORDEN
# =============================================================================


dias = np.arange(0, 121)


# Simulamos consumo acumulado.

inventario_simulado = (

    punto_reorden

    -

    demanda_diaria * dias

)


# No permitimos valores negativos para esta visualización.

inventario_simulado = np.maximum(
    inventario_simulado,
    0
)


plt.figure(figsize=(13, 6))


plt.plot(

    dias,

    inventario_simulado,

    linewidth=3,

    color="#1565C0",

    label="Inventario estimado"

)


plt.axhline(

    stock_seguridad_calculado,

    color="#E53935",

    linestyle="--",

    linewidth=2,

    label="Stock de seguridad"

)


plt.axvline(

    lead_time_actual_dias,

    color="#F9A825",

    linestyle="--",

    linewidth=2,

    label="Lead time"

)


plt.fill_between(

    dias,

    0,

    stock_seguridad_calculado,

    color="#EF5350",

    alpha=0.15

)


plt.title(
    "A1 | Lógica simplificada del stock de seguridad",
    fontsize=16,
    fontweight="bold"
)


plt.xlabel("Días")


plt.ylabel("Unidades")


plt.legend()


plt.tight_layout()


plt.show()


print("""
INTERPRETACIÓN GRÁFICA A1:

El stock de seguridad funciona como una protección frente
a desviaciones de demanda o retrasos del abastecimiento.

Su ventaja es que puede implementarse relativamente rápido.

Su principal trade-off es financiero:

MÁS INVENTARIO → MENOS RIESGO

pero también:

MÁS INVENTARIO → MÁS CAPITAL INMOVILIZADO.
""")


# =============================================================================
# 9. EXPLICAR LA ALTERNATIVA 2
# =============================================================================


print("\n" + "=" * 78)

print("ALTERNATIVA 2 | DUAL SOURCING")

print("=" * 78)

print("""
Actualmente aproximadamente el 92 % del abastecimiento
depende del proveedor principal.

Dual sourcing significa NO necesariamente reemplazarlo.

Significa distribuir el abastecimiento entre dos fuentes.

Por ejemplo:

PROVEEDOR A
70 %

PROVEEDOR B
30 %

Esto podría disminuir la dependencia y aumentar
la resiliencia de la cadena.

Sin embargo, el segundo proveedor podría tener
un precio unitario mayor.
""")


# =============================================================================
# 10. DATOS DEL SEGUNDO PROVEEDOR
# =============================================================================
#
# Supongamos:
#
# PROVEEDOR A
#
# Precio = USD 42
# Lead time = 75 días
#
#
# PROVEEDOR B
#
# Precio = USD 45
# Lead time = 42 días
#
#
# El proveedor B es más caro,
# pero entrega más rápidamente.
#
# Esto permite introducir un concepto fundamental:
#
# PRECIO ≠ COSTO TOTAL.
#
# =============================================================================


precio_A = 42

precio_B = 45


lead_time_A = 75

lead_time_B = 42


participacion_A = 0.70

participacion_B = 0.30


# =============================================================================
# 11. CALCULAR PRECIO PROMEDIO PONDERADO
# =============================================================================
#
# Fórmula:
#
# Precio ponderado =
#
# (% A × Precio A)
#
# +
#
# (% B × Precio B)
#
# =============================================================================


precio_dual = (

    participacion_A * precio_A

    +

    participacion_B * precio_B

)


# =============================================================================
# 12. CALCULAR LEAD TIME PONDERADO
# =============================================================================


lead_time_dual = (

    participacion_A * lead_time_A

    +

    participacion_B * lead_time_B

)


print(
    f"\nPrecio actual: "
    f"USD {precio_A:.2f}"
)


print(
    f"Precio ponderado dual sourcing: "
    f"USD {precio_dual:.2f}"
)


print(
    f"\nLead time actual: "
    f"{lead_time_A:.1f} días"
)


print(
    f"Lead time ponderado dual sourcing: "
    f"{lead_time_dual:.1f} días"
)


# =============================================================================
# 13. CAMBIO PORCENTUAL DEL PRECIO
# =============================================================================


incremento_precio_dual = (

    (
        precio_dual
        -
        precio_A
    )

    /

    precio_A

) * 100


reduccion_lead_time_dual = (

    (
        lead_time_A
        -
        lead_time_dual
    )

    /

    lead_time_A

) * 100


print(
    f"\nIncremento aproximado del precio: "
    f"{incremento_precio_dual:.2f}%"
)


print(
    f"Reducción aproximada del lead time ponderado: "
    f"{reduccion_lead_time_dual:.2f}%"
)


print("""
INTERPRETACIÓN:

Dual sourcing podría aumentar ligeramente el precio promedio
del componente.

Sin embargo, podría disminuir simultáneamente:

- dependencia del proveedor,
- exposición a interrupciones,
- lead time promedio,
- necesidad de compras urgentes.

Por eso no deberíamos evaluar esta alternativa
solamente observando el precio unitario.
""")


# =============================================================================
# 14. VISUALIZAR LA DIVERSIFICACIÓN
# =============================================================================


fig, ax = plt.subplots(figsize=(8, 6))


ax.pie(

    [participacion_A, participacion_B],

    labels=[
        "Proveedor A",
        "Proveedor B"
    ],

    autopct="%1.0f%%",

    startangle=90,

    colors=[
        "#1565C0",
        "#43A047"
    ],

    wedgeprops={
        "edgecolor": "white",
        "linewidth": 2
    }

)


ax.set_title(
    "A2 | Distribución propuesta del abastecimiento",
    fontsize=16,
    fontweight="bold"
)


plt.tight_layout()

plt.show()


# =============================================================================
# 15. EXPLICAR LA ALTERNATIVA 3
# =============================================================================


print("\n" + "=" * 78)

print("ALTERNATIVA 3 | REABASTECIMIENTO PREDICTIVO")

print("=" * 78)

print("""
La tercera alternativa aprovecha el análisis realizado
con Machine Learning.

La idea NO es que el algoritmo compre automáticamente.

La lógica sería:

DATOS
   ↓
MODELO PREDICTIVO
   ↓
PROBABILIDAD DE STOCKOUT
   ↓
ALERTA
   ↓
REVISIÓN DEL RESPONSABLE
   ↓
DECISIÓN DE REABASTECIMIENTO

Esto permitiría pasar de una política completamente reactiva
a una política más anticipativa.
""")


# =============================================================================
# 16. SIMULAR PROBABILIDADES DE STOCKOUT
# =============================================================================
#
# Imaginemos que el modelo del punto 8 genera
# probabilidades semanales de stockout.
#
# Ejemplo:
#
# 15 % = riesgo bajo
# 45 % = riesgo medio
# 80 % = riesgo alto
#
# Crearemos 20 semanas de ejemplo.
#
# =============================================================================


np.random.seed(25)


semanas = np.arange(
    1,
    21
)


probabilidades = np.random.uniform(
    10,
    90,
    20
)


# =============================================================================
# 17. DEFINIR REGLAS DE DECISIÓN
# =============================================================================
#
# Para fines pedagógicos:
#
# Riesgo < 30 %
#
# → mantener política normal.
#
#
# Riesgo entre 30 % y 60 %
#
# → revisar inventario y orden abierta.
#
#
# Riesgo > 60 %
#
# → activar revisión prioritaria.
#
#
# IMPORTANTE:
#
# Estos umbrales son ilustrativos.
#
# Una empresa debería definirlos según:
#
# - costo del stockout,
# - nivel de servicio objetivo,
# - criticidad del SKU,
# - lead time,
# - tolerancia al riesgo.
#
# =============================================================================


def decision_predictiva(probabilidad):

    if probabilidad < 30:

        return "Operación normal"

    elif probabilidad < 60:

        return "Revisar inventario"

    else:

        return "Activar alerta"


decisiones = [

    decision_predictiva(x)

    for x in probabilidades

]


df_predictivo = pd.DataFrame({

    "Semana": semanas,

    "Probabilidad_stockout": probabilidades,

    "Accion": decisiones

})


print("\nEJEMPLO DE ALERTAS PREDICTIVAS:\n")


print(
    df_predictivo.round(1).to_string(
        index=False
    )
)


# =============================================================================
# 18. VISUALIZAR LAS ALERTAS
# =============================================================================


colores_alerta = [

    "#43A047"
    if x < 30

    else "#F9A825"
    if x < 60

    else "#D32F2F"

    for x in probabilidades

]


plt.figure(figsize=(13, 6))


plt.bar(

    semanas,

    probabilidades,

    color=colores_alerta

)


plt.axhline(

    30,

    linestyle="--",

    color="#616161",

    label="Inicio riesgo medio"

)


plt.axhline(

    60,

    linestyle="--",

    color="#212121",

    label="Inicio riesgo alto"

)


plt.title(
    "A3 | Sistema predictivo de alertas de stockout",
    fontsize=16,
    fontweight="bold"
)


plt.xlabel("Semana")


plt.ylabel("Probabilidad estimada de stockout (%)")


plt.ylim(0, 100)


plt.legend()


plt.tight_layout()


plt.show()


print("""
INTERPRETACIÓN GRÁFICA A3:

El modelo predictivo no reemplaza la decisión humana.

Su función es identificar situaciones que requieren atención.

Esto permite que compras e inventarios concentren esfuerzos
en los periodos con mayor riesgo.

La ventaja potencial es anticiparse.

El reto es que requiere:

- datos confiables,
- actualización periódica,
- capacidad analítica,
- reglas claras de decisión.
""")


# =============================================================================
# 19. ESTIMAR RESULTADOS ESPERADOS
# =============================================================================
#
# Ahora construiremos escenarios.
#
# MUY IMPORTANTE:
#
# LOS SIGUIENTES RESULTADOS SON SUPUESTOS DEL EJERCICIO.
#
# NO fueron demostrados por Machine Learning.
#
# NO son resultados reales de GlobalBike.
#
# Sirven para enseñar cómo comparar alternativas.
#
# En un proyecto real:
#
# estos valores deberían provenir de:
#
# - simulación,
# - pruebas piloto,
# - datos históricos,
# - benchmarking,
# - cotizaciones,
# - información de proveedores,
# - modelos de inventario.
#
# =============================================================================


escenarios = pd.DataFrame({

    "Escenario": [

        "Situación actual",

        "A1 - Stock de seguridad",

        "A2 - Dual sourcing",

        "A3 - Reabastecimiento predictivo"

    ],

    # Nivel de servicio esperado (%)

    "Nivel_servicio": [

        89,

        96,

        97,

        95

    ],

    # Stockouts aproximados por año.

    "Stockouts_año": [

        4,

        1,

        1,

        2

    ],

    # Inventario promedio estimado.

    "Inventario_promedio": [

        1200,

        1550,

        1250,

        1100

    ],

    # Lead time esperado.

    "Lead_time_dias": [

        75,

        75,

        65,

        70

    ],

    # Dependencia del proveedor principal (%).

    "Dependencia_proveedor": [

        92,

        92,

        70,

        92

    ],

    # Inversión inicial aproximada.
    #
    # Valores simulados.

    "Inversion_inicial_USD": [

        0,

        15000,

        28000,

        22000

    ],

    # Tiempo de implementación aproximado.

    "Tiempo_implementacion_meses": [

        0,

        1,

        4,

        3

    ]

})


print("\n" + "=" * 78)

print("ESCENARIOS ESTIMADOS")

print("=" * 78)


print(
    escenarios.to_string(
        index=False
    )
)


# =============================================================================
# 20. GRÁFICA: NIVEL DE SERVICIO
# =============================================================================


plt.figure(figsize=(12, 6))


colores = [

    "#757575",
    "#1565C0",
    "#43A047",
    "#8E24AA"

]


plt.bar(

    escenarios["Escenario"],

    escenarios["Nivel_servicio"],

    color=colores

)


plt.axhline(

    95,

    linestyle="--",

    color="#D32F2F",

    linewidth=2,

    label="Meta = 95%"

)


plt.ylim(
    80,
    100
)


plt.ylabel(
    "Nivel de servicio (%)"
)


plt.title(
    "Comparación del nivel de servicio esperado",
    fontsize=16,
    fontweight="bold"
)


plt.xticks(
    rotation=15
)


plt.legend()


plt.tight_layout()


plt.show()


print("""
INTERPRETACIÓN:

En este escenario académico, las tres alternativas
mejoran el nivel de servicio respecto a la situación actual.

Sin embargo, observar únicamente este indicador
no es suficiente para tomar la decisión.

Una alternativa puede ofrecer mayor servicio,
pero también requerir mayor inversión o inventario.
""")


# =============================================================================
# 21. GRÁFICA: STOCKOUTS
# =============================================================================


plt.figure(figsize=(12, 6))


plt.bar(

    escenarios["Escenario"],

    escenarios["Stockouts_año"],

    color=[
        "#D32F2F",
        "#F9A825",
        "#43A047",
        "#1565C0"
    ]

)


plt.ylabel(
    "Periodos con stockout por año"
)


plt.title(
    "Stockouts esperados por alternativa",
    fontsize=16,
    fontweight="bold"
)


plt.xticks(
    rotation=15
)


plt.tight_layout()


plt.show()


print("""
INTERPRETACIÓN:

El objetivo aquí es MINIMIZAR.

Menos stockouts significa mayor capacidad
de atender la demanda.

Sin embargo, debemos preguntar:

¿A QUÉ COSTO estamos reduciendo los stockouts?

Una estrategia que elimina stockouts acumulando inventario
podría generar un problema financiero diferente.
""")


# =============================================================================
# 22. GRÁFICA: INVENTARIO PROMEDIO
# =============================================================================


plt.figure(figsize=(12, 6))


plt.bar(

    escenarios["Escenario"],

    escenarios["Inventario_promedio"],

    color=[
        "#757575",
        "#EF6C00",
        "#43A047",
        "#1565C0"
    ]

)


plt.ylabel(
    "Unidades"
)


plt.title(
    "Inventario promedio estimado",
    fontsize=16,
    fontweight="bold"
)


plt.xticks(
    rotation=15
)


plt.tight_layout()


plt.show()


print("""
INTERPRETACIÓN:

Aquí aparece claramente un trade-off.

La alternativa de stock de seguridad
protege la operación manteniendo más inventario.

El reabastecimiento predictivo intenta reducir
la necesidad de protección permanente mediante
una mejor anticipación.

Esto no significa que una sea automáticamente mejor.

Significa que utilizan mecanismos diferentes
para enfrentar el mismo problema.
""")


# =============================================================================
# 23. GRÁFICA: DEPENDENCIA DEL PROVEEDOR
# =============================================================================


plt.figure(figsize=(12, 6))


plt.bar(

    escenarios["Escenario"],

    escenarios["Dependencia_proveedor"],

    color=[
        "#D32F2F",
        "#D32F2F",
        "#43A047",
        "#D32F2F"
    ]

)


plt.ylabel(
    "Dependencia del proveedor principal (%)"
)


plt.title(
    "Exposición al proveedor principal",
    fontsize=16,
    fontweight="bold"
)


plt.xticks(
    rotation=15
)


plt.tight_layout()


plt.show()


print("""
INTERPRETACIÓN:

Esta gráfica muestra una diferencia estratégica importante.

Aumentar stock de seguridad NO reduce directamente
la dependencia del proveedor.

Implementar Machine Learning tampoco elimina por sí solo
la concentración del abastecimiento.

Dual sourcing sí modifica estructuralmente
esa exposición al distribuir parte del volumen
entre dos proveedores.

Esta diferencia será importante cuando definamos
el criterio RIESGO DE SUMINISTRO en AHP.
""")


# =============================================================================
# 24. GRÁFICA: INVERSIÓN INICIAL
# =============================================================================


plt.figure(figsize=(12, 6))


plt.bar(

    escenarios["Escenario"],

    escenarios["Inversion_inicial_USD"],

    color=[
        "#757575",
        "#1565C0",
        "#43A047",
        "#8E24AA"
    ]

)


plt.ylabel(
    "USD"
)


plt.title(
    "Inversión inicial estimada",
    fontsize=16,
    fontweight="bold"
)


plt.xticks(
    rotation=15
)


plt.tight_layout()


plt.show()


print("""
INTERPRETACIÓN:

Las soluciones no tienen el mismo esfuerzo económico.

Stock de seguridad requiere capital adicional
principalmente asociado al inventario.

Dual sourcing puede requerir homologación,
negociación, auditorías y desarrollo del proveedor.

El reabastecimiento predictivo puede requerir
datos, capacitación y herramientas analíticas.

Por eso COSTO será uno de nuestros criterios de decisión.
""")


# =============================================================================
# 25. CONSTRUIR UNA MATRIZ DE TRADE-OFFS
# =============================================================================
#
# Vamos a resumir cualitativamente las alternativas.
#
# Escala:
#
# 1 = bajo
# 2 = medio
# 3 = alto
#
# IMPORTANTE:
#
# Esto NO es todavía AHP.
#
# Es simplemente una herramienta visual
# para identificar diferencias.
#
# =============================================================================


tradeoffs = pd.DataFrame({

    "Alternativa": [

        "Stock de seguridad",

        "Dual sourcing",

        "Reabastecimiento predictivo"

    ],

    "Costo": [

        2,
        3,
        2

    ],

    "Impacto_servicio": [

        3,
        3,
        3

    ],

    "Reduccion_riesgo": [

        2,
        3,
        2

    ],

    "Rapidez_implementacion": [

        3,
        1,
        2

    ],

    "Flexibilidad": [

        2,
        3,
        3

    ]

})


print("\n" + "=" * 78)

print("MATRIZ PRELIMINAR DE TRADE-OFFS")

print("=" * 78)


print(
    tradeoffs.to_string(
        index=False
    )
)


# =============================================================================
# 26. MAPA DE CALOR DE TRADE-OFFS
# =============================================================================


datos_heatmap = (

    tradeoffs

    .set_index("Alternativa")

)


plt.figure(figsize=(11, 6))


sns.heatmap(

    datos_heatmap,

    annot=True,

    cmap="YlGnBu",

    linewidths=0.5,

    cbar_kws={
        "label": "Valor relativo"
    }

)


plt.title(
    "Perfil preliminar de las alternativas",
    fontsize=16,
    fontweight="bold"
)


plt.xlabel(
    "Dimensión"
)


plt.ylabel(
    ""
)


plt.tight_layout()


plt.show()


print("""
INTERPRETACIÓN:

Ninguna alternativa domina necesariamente
en todas las dimensiones.

Una puede ser rápida pero aumentar inventario.

Otra puede reducir el riesgo del proveedor,
pero necesitar mayor inversión.

Otra puede ofrecer flexibilidad,
pero requerir capacidad analítica.

Esto explica por qué necesitamos
un método MULTICRITERIO.
""")


# =============================================================================
# 27. COMPARAR CAMBIO RESPECTO A LA SITUACIÓN ACTUAL
# =============================================================================
#
# Calcularemos cuánto cambia cada alternativa
# frente a la situación actual.
#
# Fórmula general:
#
#
# CAMBIO % =
#
# (Nuevo - Actual)
# ---------------- × 100
#      Actual
#
#
# IMPORTANTE:
#
# Un porcentaje positivo NO siempre significa mejora.
#
# Ejemplo:
#
# +10 % en nivel de servicio = potencialmente favorable.
#
# +10 % en stockouts = desfavorable.
#
# Por eso siempre debemos interpretar
# la DIRECCIÓN DESEADA del indicador.
#
# =============================================================================


actual = escenarios.iloc[0]


comparacion = escenarios.iloc[1:].copy()


comparacion[
    "Cambio_servicio_%"
] = (

    (
        comparacion["Nivel_servicio"]
        -
        actual["Nivel_servicio"]
    )

    /

    actual["Nivel_servicio"]

) * 100


comparacion[
    "Reduccion_stockouts_%"
] = (

    (
        actual["Stockouts_año"]
        -
        comparacion["Stockouts_año"]
    )

    /

    actual["Stockouts_año"]

) * 100


comparacion[
    "Cambio_inventario_%"
] = (

    (
        comparacion["Inventario_promedio"]
        -
        actual["Inventario_promedio"]
    )

    /

    actual["Inventario_promedio"]

) * 100


comparacion[
    "Reduccion_dependencia_%"
] = (

    (
        actual["Dependencia_proveedor"]
        -
        comparacion["Dependencia_proveedor"]
    )

    /

    actual["Dependencia_proveedor"]

) * 100


print("\n" + "=" * 78)

print("CAMBIO RESPECTO A LA SITUACIÓN ACTUAL")

print("=" * 78)


columnas_comparacion = [

    "Escenario",

    "Cambio_servicio_%",

    "Reduccion_stockouts_%",

    "Cambio_inventario_%",

    "Reduccion_dependencia_%"

]


print(

    comparacion[
        columnas_comparacion
    ]

    .round(1)

    .to_string(
        index=False
    )

)


# =============================================================================
# 28. GRÁFICA DE BENEFICIOS RELATIVOS
# =============================================================================


beneficios = comparacion[[

    "Escenario",

    "Cambio_servicio_%",

    "Reduccion_stockouts_%",

    "Reduccion_dependencia_%"

]].copy()


beneficios = beneficios.set_index(
    "Escenario"
)


beneficios.plot(

    kind="bar",

    figsize=(13, 7),

    color=[
        "#1565C0",
        "#43A047",
        "#8E24AA"
    ]

)


plt.title(
    "Beneficios relativos frente a la situación actual",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel(
    "Cambio (%)"
)


plt.xlabel(
    ""
)


plt.xticks(
    rotation=15
)


plt.axhline(
    0,
    color="#212121",
    linewidth=1
)


plt.legend(
    title="Indicador"
)


plt.tight_layout()


plt.show()


print("""
INTERPRETACIÓN:

Esta gráfica permite observar que cada alternativa
ataca el problema de una manera diferente.

Por ejemplo:

STOCK DE SEGURIDAD
protege principalmente mediante inventario.

DUAL SOURCING
actúa sobre el riesgo y la dependencia del suministro.

REABASTECIMIENTO PREDICTIVO
busca anticiparse mediante información.

La pregunta ya no es:

"¿Cuál alternativa funciona?"

Las tres podrían funcionar.

La pregunta ahora es:

"¿CUÁL RESPONDE MEJOR A LAS PRIORIDADES
ESTRATÉGICAS DE GLOBALBIKE?"
""")


# =============================================================================
# 29. CONSTRUIR LA MATRIZ QUE UTILIZAREMOS EN AHP
# =============================================================================
#
# En el siguiente numeral trabajaremos con cinco criterios:
#
# C1 = Costo total
#
# C2 = Nivel de servicio
#
# C3 = Riesgo de suministro
#
# C4 = Tiempo de implementación
#
# C5 = Flexibilidad
#
#
# ATENCIÓN:
#
# No asignaremos todavía los pesos.
#
# Los pesos deben surgir de las comparaciones pareadas
# realizadas mediante AHP.
#
# =============================================================================


criterios_ahp = pd.DataFrame({

    "Criterio": [

        "Costo total",

        "Nivel de servicio",

        "Riesgo de suministro",

        "Tiempo de implementación",

        "Flexibilidad"

    ],

    "Pregunta_gerencial": [

        "¿Cuánto cuesta implementar y sostener la alternativa?",

        "¿Cuánto contribuye a garantizar disponibilidad?",

        "¿Cuánto reduce la exposición a interrupciones?",

        "¿Qué tan rápido puede implementarse?",

        "¿Qué tan adaptable es ante cambios?"

    ]

})


print("\n" + "=" * 78)

print("CRITERIOS PROPUESTOS PARA AHP")

print("=" * 78)


print(
    criterios_ahp.to_string(
        index=False
    )
)


# =============================================================================
# 30. RESUMEN VISUAL FINAL
# =============================================================================
#
# Crearemos una gráfica sencilla:
#
# cada alternativa se representa mediante
# cinco dimensiones.
#
# Todavía NO son pesos AHP.
#
# Son perfiles preliminares.
#
# =============================================================================


categorias = [

    "Costo",

    "Servicio",

    "Riesgo",

    "Rapidez",

    "Flexibilidad"

]


# Para esta gráfica:
#
# valores más altos significan
# un perfil relativamente más favorable.
#
# Por eso invertimos conceptualmente "Costo":
#
# 3 = menor carga relativa
# 1 = mayor carga relativa.
#
# Son valores pedagógicos.


perfil_A1 = [
    2,
    3,
    2,
    3,
    2
]


perfil_A2 = [
    1,
    3,
    3,
    1,
    3
]


perfil_A3 = [
    2,
    3,
    2,
    2,
    3
]


# Preparación de radar.


N = len(categorias)


angulos = np.linspace(
    0,
    2 * np.pi,
    N,
    endpoint=False
).tolist()


angulos += angulos[:1]


perfil_A1 += perfil_A1[:1]

perfil_A2 += perfil_A2[:1]

perfil_A3 += perfil_A3[:1]


fig = plt.figure(
    figsize=(9, 8)
)


ax = plt.subplot(
    111,
    polar=True
)


ax.plot(

    angulos,

    perfil_A1,

    linewidth=2,

    label="A1 - Stock seguridad",

    color="#1565C0"

)


ax.fill(

    angulos,

    perfil_A1,

    alpha=0.08,

    color="#1565C0"

)


ax.plot(

    angulos,

    perfil_A2,

    linewidth=2,

    label="A2 - Dual sourcing",

    color="#43A047"

)


ax.fill(

    angulos,

    perfil_A2,

    alpha=0.08,

    color="#43A047"

)


ax.plot(

    angulos,

    perfil_A3,

    linewidth=2,

    label="A3 - Reabastecimiento predictivo",

    color="#8E24AA"

)


ax.fill(

    angulos,

    perfil_A3,

    alpha=0.08,

    color="#8E24AA"

)


ax.set_xticks(
    angulos[:-1]
)


ax.set_xticklabels(
    categorias
)


ax.set_ylim(
    0,
    3
)


plt.title(

    "Perfil preliminar de las alternativas",

    fontsize=16,

    fontweight="bold",

    pad=25

)


plt.legend(

    loc="upper right",

    bbox_to_anchor=(1.4, 1.15)

)


plt.tight_layout()


plt.show()


print("""
INTERPRETACIÓN DEL RADAR:

Esta gráfica NO selecciona la alternativa ganadora.

Solamente permite observar que las soluciones
tienen perfiles diferentes.

A1:
puede implementarse rápidamente,
pero aumenta la dependencia del inventario.

A2:
reduce la concentración del suministro,
pero requiere desarrollo de un segundo proveedor.

A3:
aporta anticipación y flexibilidad,
pero necesita capacidades de datos y analítica.

Precisamente porque existen estos trade-offs,
la decisión requiere múltiples criterios.
""")


# =============================================================================
# 31. GENERAR UNA TABLA RESUMEN PARA LA GERENCIA
# =============================================================================


resumen_final = pd.DataFrame({

    "Alternativa": [

        "A1 - Stock de seguridad",

        "A2 - Dual sourcing",

        "A3 - Reabastecimiento predictivo"

    ],

    "Mecanismo": [

        "Mayor protección mediante inventario",

        "Diversificación del abastecimiento",

        "Anticipación mediante datos y alertas"

    ],

    "Principal_beneficio": [

        "Reduce exposición a variabilidad",

        "Reduce dependencia del proveedor",

        "Permite anticipar riesgo"

    ],

    "Principal_tradeoff": [

        "Mayor capital inmovilizado",

        "Mayor complejidad y costo inicial",

        "Requiere datos y capacidad analítica"

    ]

})


print("\n" + "=" * 78)

print("RESUMEN EJECUTIVO DE ALTERNATIVAS")

print("=" * 78)


print(
    resumen_final.to_string(
        index=False
    )
)


# =============================================================================
# 32. GUARDAR RESULTADOS
# =============================================================================


escenarios.to_csv(

    "GlobalBike_Escenarios_Solucion.csv",

    index=False,

    encoding="utf-8-sig"

)


tradeoffs.to_csv(

    "GlobalBike_Tradeoffs.csv",

    index=False,

    encoding="utf-8-sig"

)


criterios_ahp.to_csv(

    "GlobalBike_Criterios_AHP.csv",

    index=False,

    encoding="utf-8-sig"

)


resumen_final.to_csv(

    "GlobalBike_Alternativas.csv",

    index=False,

    encoding="utf-8-sig"

)


print("\nARCHIVOS GENERADOS:")

print("""
1. GlobalBike_Escenarios_Solucion.csv
2. GlobalBike_Tradeoffs.csv
3. GlobalBike_Criterios_AHP.csv
4. GlobalBike_Alternativas.csv
""")


# =============================================================================
# 33. CONCLUSIÓN GERENCIAL AUTOMÁTICA
# =============================================================================


print("\n" + "=" * 78)

print("CONCLUSIÓN | ETAPA 9")

print("=" * 78)


print(f"""
El diagnóstico realizado sobre GlobalBike permitió pasar
de la identificación del problema a la formulación
de tres alternativas de solución.

------------------------------------------------------------

A1 | AJUSTAR STOCK DE SEGURIDAD

Stock de seguridad estimado:
{stock_seguridad_calculado:.0f} unidades.

Mecanismo:
proteger la operación mediante mayor disponibilidad
de inventario.

Principal trade-off:
mayor capital inmovilizado.

------------------------------------------------------------

A2 | DUAL SOURCING

Distribución propuesta:

Proveedor A = {participacion_A:.0%}

Proveedor B = {participacion_B:.0%}

El precio promedio ponderado aumentaría aproximadamente:

{incremento_precio_dual:.2f}%

mientras que el lead time ponderado disminuiría
aproximadamente:

{reduccion_lead_time_dual:.2f}%.

Mecanismo:
reducir la dependencia de una única fuente
de abastecimiento.

Principal trade-off:
mayor complejidad de gestión y posible incremento de costo.

------------------------------------------------------------

A3 | REABASTECIMIENTO PREDICTIVO

Mecanismo:

DATOS
   ↓
MACHINE LEARNING
   ↓
ALERTA
   ↓
DECISIÓN DE REABASTECIMIENTO

Principal beneficio:
anticipar situaciones de riesgo.

Principal trade-off:
requiere calidad de datos y capacidad analítica.

------------------------------------------------------------

CONCLUSIÓN:

Las tres alternativas podrían contribuir a reducir
el riesgo de stockout, pero lo hacen de maneras diferentes.

Por lo tanto, todavía NO tenemos suficiente información
para afirmar cuál debería seleccionarse.

Necesitamos evaluar simultáneamente:

COSTO

NIVEL DE SERVICIO

RIESGO DE SUMINISTRO

TIEMPO DE IMPLEMENTACIÓN

FLEXIBILIDAD


Esto nos conduce directamente al:

PUNTO 10

ANALYTIC HIERARCHY PROCESS - AHP
""")


# =============================================================================
# FIN DEL PUNTO 9
# =============================================================================
#
#
#                PROBLEMA
#
#                    ↓
#
#               DIAGNÓSTICO
#
#                    ↓
#
#            MACHINE LEARNING
#
#                    ↓
#
#             ALERTA DE RIESGO
#
#                    ↓
#
#       ┌────────────┼────────────┐
#
#       ↓            ↓            ↓
#
# STOCK SEGURIDAD  DUAL       REABASTECIMIENTO
#                 SOURCING       PREDICTIVO
#
#       └────────────┼────────────┘
#
#                    ↓
#
#            MÚLTIPLES TRADE-OFFS
#
#                    ↓
#
#                   AHP
#
#                    ↓
#
#            DECISIÓN PRIORIZADA
#
# =============================================================================
```
