```python
# =============================================================================
# =============================================================================
#
# INVENTARIOS Y COMPRAS GLOBALES
# CASO INTEGRADOR: GLOBALBIKE COMPONENTS COLOMBIA S.A.S.
#
# PUNTO 12
# FROM ANALYSIS TO ACTION
# DEL ANÁLISIS A LA ACCIÓN
#
# =============================================================================
# =============================================================================
#
# OBJETIVO DE ESTA ETAPA
# -----------------------------------------------------------------------------
#
# En las etapas anteriores se realizó:
#
# 1. Diagnóstico del problema.
# 2. Análisis de datos.
# 3. Machine Learning.
# 4. Formulación de alternativas.
# 5. Priorización mediante AHP.
# 6. Integración de resultados.
#
# Ahora debemos responder una pregunta diferente:
#
# ¿CÓMO CONVERTIMOS LA DECISIÓN EN ACCIÓN?
#
#
# En esta etapa construiremos:
#
# - Línea base de indicadores.
# - Metas.
# - Brechas.
# - Plan de implementación.
# - Cronograma tipo Gantt.
# - Matriz RACI.
# - Matriz de riesgos.
# - Acciones de mitigación.
# - Simulación de un piloto.
# - Seguimiento de KPIs.
# - Semáforo de cumplimiento.
# - Evaluación económica.
# - ROI.
# - Criterios para decidir si escalar.
# - Dashboard ejecutivo.
# - Conclusión gerencial.
#
#
# IMPORTANTE:
# -----------------------------------------------------------------------------
#
# TODOS LOS DATOS SON FICTICIOS Y DIDÁCTICOS.
#
# No representan información real de una empresa.
#
# En sus proyectos, los estudiantes deben reemplazar estos valores
# por información obtenida de sus propios casos.
#
# =============================================================================


# =============================================================================
# 1. IMPORTAR LIBRERÍAS
# =============================================================================
#
# pandas:
# Manejo de tablas y bases de datos.
#
# numpy:
# Operaciones numéricas.
#
# matplotlib:
# Construcción de gráficas.
#
# seaborn:
# Gráficas estadísticas con mejor presentación.
#
# =============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Configuración visual general.

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["font.size"] = 11


# =============================================================================
# 2. RECORDAR LA ALTERNATIVA PRIORIZADA
# =============================================================================
#
# En el ejemplo anterior se analizaron:
#
# A1 = Ajustar stock de seguridad.
#
# A2 = Dual sourcing.
#
# A3 = Reabastecimiento predictivo.
#
#
# El resultado ilustrativo del AHP fue:
#
# A1 = 24 %
# A2 = 43 %
# A3 = 33 %
#
#
# Por tanto, en este ejercicio:
#
# DUAL SOURCING
#
# obtuvo la mayor prioridad.
#
#
# ATENCIÓN:
#
# Esto NO significa que dual sourcing sea siempre
# la mejor estrategia.
#
# El resultado depende de:
#
# - criterios seleccionados,
# - pesos,
# - comparaciones,
# - información disponible,
# - contexto empresarial.
#
# =============================================================================

alternativa_priorizada = "A2 - Dual sourcing"
prioridad_ahp = 43.0


print("\n" + "=" * 90)
print("GLOBALBIKE COMPONENTS | FROM ANALYSIS TO ACTION")
print("=" * 90)

print(f"""
ALTERNATIVA PRIORIZADA:

{alternativa_priorizada}

Prioridad AHP del ejemplo:

{prioridad_ahp:.1f} %

Ahora debemos transformar esta recomendación
en un PLAN DE IMPLEMENTACIÓN.
""")


# =============================================================================
# 3. DEFINIR LA LÍNEA BASE
# =============================================================================
#
# LÍNEA BASE:
#
# Es la situación existente ANTES de implementar
# una estrategia.
#
# Es fundamental porque posteriormente podremos comparar:
#
# ANTES
#
# vs.
#
# DESPUÉS
#
#
# En nuestro ejemplo utilizaremos:
#
# Nivel de servicio = 89 %
# Stockouts = 4 al año
# Lead time = 75 días
# Dependencia proveedor principal = 92 %
# MAPE = 18.4 %
# Compras urgentes = USD 58.000/año
#
# =============================================================================

nivel_servicio_actual = 89.0
stockouts_actuales = 4
lead_time_actual = 75.0
dependencia_actual = 92.0
mape_actual = 18.4
compras_urgentes_actuales = 58000


# =============================================================================
# 4. DEFINIR LAS METAS
# =============================================================================
#
# Las METAS representan la situación deseada.
#
# IMPORTANTE:
#
# Una meta NO es un resultado alcanzado.
#
# Es simplemente el valor que la empresa espera lograr.
#
# =============================================================================

nivel_servicio_meta = 95.0
stockouts_meta = 1
lead_time_meta = 60.0
dependencia_meta = 70.0
mape_meta = 12.0
compras_urgentes_meta = 20000


# =============================================================================
# 5. CREAR TABLA ACTUAL VS META
# =============================================================================

kpis = pd.DataFrame({
    "Indicador": [
        "Nivel de servicio (%)",
        "Stockouts por año",
        "Lead time promedio (días)",
        "Dependencia proveedor principal (%)",
        "MAPE (%)",
        "Compras urgentes (USD)"
    ],

    "Actual": [
        nivel_servicio_actual,
        stockouts_actuales,
        lead_time_actual,
        dependencia_actual,
        mape_actual,
        compras_urgentes_actuales
    ],

    "Meta": [
        nivel_servicio_meta,
        stockouts_meta,
        lead_time_meta,
        dependencia_meta,
        mape_meta,
        compras_urgentes_meta
    ],

    "Direccion": [
        "Aumentar",
        "Disminuir",
        "Disminuir",
        "Disminuir",
        "Disminuir",
        "Disminuir"
    ]
})


print("\n" + "=" * 90)
print("LÍNEA BASE Y METAS")
print("=" * 90)

print(kpis.to_string(index=False))


# =============================================================================
# 6. CALCULAR EL CAMBIO REQUERIDO
# =============================================================================
#
# Para indicadores que deben AUMENTAR:
#
#                  META - ACTUAL
# Cambio (%) = --------------------- x 100
#                     ACTUAL
#
#
# Para indicadores que deben DISMINUIR:
#
#                   ACTUAL - META
# Cambio (%) = ---------------------- x 100
#                      ACTUAL
#
#
# Queremos que un porcentaje POSITIVO represente
# siempre una mejora esperada.
#
# =============================================================================

def calcular_cambio_requerido(actual, meta, direccion):

    if direccion == "Aumentar":
        cambio = ((meta - actual) / actual) * 100

    else:
        cambio = ((actual - meta) / actual) * 100

    return cambio


kpis["Cambio_requerido_%"] = kpis.apply(
    lambda fila: calcular_cambio_requerido(
        fila["Actual"],
        fila["Meta"],
        fila["Direccion"]
    ),
    axis=1
)


print("\nCAMBIO REQUERIDO PARA ALCANZAR LA META:\n")

print(
    kpis[
        ["Indicador", "Actual", "Meta", "Cambio_requerido_%"]
    ]
    .round(1)
    .to_string(index=False)
)


# =============================================================================
# 7. GRÁFICA DE BRECHAS
# =============================================================================
#
# Como los indicadores utilizan unidades diferentes,
# no debemos compararlos directamente.
#
# Por eso mostraremos el porcentaje de cambio requerido.
#
# =============================================================================

plt.figure(figsize=(13, 6))

colores_brecha = [
    "#1565C0",
    "#E53935",
    "#8E24AA",
    "#F9A825",
    "#00897B",
    "#43A047"
]

plt.bar(
    kpis["Indicador"],
    kpis["Cambio_requerido_%"],
    color=colores_brecha
)

plt.title(
    "Magnitud del cambio requerido para alcanzar las metas",
    fontsize=16,
    fontweight="bold"
)

plt.ylabel("Cambio requerido (%)")
plt.xlabel("")

plt.xticks(rotation=25, ha="right")

for i, valor in enumerate(kpis["Cambio_requerido_%"]):
    plt.text(
        i,
        valor + 1,
        f"{valor:.1f}%",
        ha="center",
        fontweight="bold"
    )

plt.tight_layout()
plt.show()


print("""
INTERPRETACIÓN:

Esta gráfica NO compara directamente los valores originales.

Compara la magnitud relativa del cambio requerido.

Por ejemplo:

- El nivel de servicio necesita aumentar.
- Los stockouts necesitan disminuir.
- El lead time necesita disminuir.
- La dependencia del proveedor necesita disminuir.
- El MAPE necesita disminuir.
- Las compras urgentes necesitan disminuir.

Esto transforma una recomendación general
en objetivos cuantificables.
""")


# =============================================================================
# 8. CREAR EL PLAN DE IMPLEMENTACIÓN
# =============================================================================
#
# La recomendación:
#
# "Implementar dual sourcing"
#
# todavía NO es un plan.
#
#
# Necesitamos definir:
#
# QUÉ se hará.
# QUIÉN lo hará.
# CUÁNDO.
# CUÁNTO DURARÁ.
# QUÉ resultado debe producir.
#
#
# Diseñaremos cinco fases.
#
# =============================================================================

plan = pd.DataFrame({
    "Fase": [
        "1. Preparación",
        "2. Búsqueda",
        "3. Homologación",
        "4. Piloto",
        "5. Evaluación"
    ],

    "Actividad": [
        "Validar demanda, inventario, costos y desempeño del proveedor actual",
        "Identificar y preseleccionar proveedores alternativos",
        "Evaluar calidad, TCO, capacidad, lead time y riesgo",
        "Implementar piloto de abastecimiento con distribución 70/30",
        "Comparar resultados y decidir posible escalamiento"
    ],

    "Responsable": [
        "Planeación + Inventarios",
        "Compras Globales",
        "Compras + Calidad",
        "Supply Chain + Compras",
        "Comité Supply Chain"
    ],

    "Inicio_semana": [
        1,
        3,
        6,
        10,
        15
    ],

    "Duracion_semanas": [
        2,
        3,
        4,
        5,
        2
    ]
})


# Calculamos automáticamente cuándo termina cada fase.
#
# Fórmula:
#
# FIN = INICIO + DURACIÓN - 1

plan["Fin_semana"] = (
    plan["Inicio_semana"]
    + plan["Duracion_semanas"]
    - 1
)


print("\n" + "=" * 90)
print("PLAN DE IMPLEMENTACIÓN")
print("=" * 90)

print(plan.to_string(index=False))


# =============================================================================
# 9. CREAR ENTREGABLES
# =============================================================================
#
# Cada fase debería producir algo verificable.
#
# Eso evita planes con actividades demasiado generales.
#
# =============================================================================

entregables = pd.DataFrame({
    "Fase": plan["Fase"],

    "Entregable": [
        "Base validada y línea base de KPIs",
        "Lista corta de proveedores potenciales",
        "Matriz de evaluación y proveedor homologado",
        "Informe de resultados del piloto 70/30",
        "Informe final y decisión de escalamiento"
    ]
})


print("\n" + "=" * 90)
print("ENTREGABLES POR FASE")
print("=" * 90)

print(entregables.to_string(index=False))


# =============================================================================
# 10. CREAR DIAGRAMA DE GANTT
# =============================================================================
#
# El Gantt muestra:
#
# - momento de inicio,
# - duración,
# - secuencia de implementación.
#
# =============================================================================

plt.figure(figsize=(14, 7))

colores_fases = [
    "#1565C0",
    "#00897B",
    "#F9A825",
    "#8E24AA",
    "#43A047"
]

for i, fila in plan.iterrows():

    plt.barh(
        fila["Fase"],
        fila["Duracion_semanas"],
        left=fila["Inicio_semana"],
        color=colores_fases[i],
        edgecolor="white",
        height=0.6
    )

    plt.text(
        fila["Inicio_semana"] + fila["Duracion_semanas"] / 2,
        i,
        fila["Responsable"],
        ha="center",
        va="center",
        fontsize=9,
        color="white",
        fontweight="bold"
    )


plt.title(
    "GlobalBike | Roadmap de implementación",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel("Semana")
plt.ylabel("")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()


print("""
INTERPRETACIÓN DEL GANTT:

La empresa no cambia inmediatamente toda su estrategia
de abastecimiento.

Primero:

1. Valida información.
2. Busca proveedores.
3. Evalúa y homologa.
4. Realiza un piloto.
5. Mide resultados.
6. Decide si escala.

Esto disminuye el riesgo de implementar
una solución sin haberla validado.
""")


# =============================================================================
# 11. CONSTRUIR MATRIZ RACI
# =============================================================================
#
# RACI ayuda a definir responsabilidades.
#
# R = Responsible
#     Ejecuta la actividad.
#
# A = Accountable
#     Tiene la responsabilidad final.
#
# C = Consulted
#     Debe ser consultado.
#
# I = Informed
#     Debe mantenerse informado.
#
# =============================================================================

raci = pd.DataFrame({
    "Actividad": [
        "Validar datos",
        "Buscar proveedor",
        "Evaluar proveedor",
        "Ejecutar piloto",
        "Evaluar resultados"
    ],

    "Compras": [
        "C",
        "R/A",
        "R",
        "R",
        "C"
    ],

    "Planeación": [
        "R/A",
        "C",
        "C",
        "R",
        "R"
    ],

    "Calidad": [
        "I",
        "C",
        "R/A",
        "C",
        "C"
    ],

    "Finanzas": [
        "C",
        "C",
        "C",
        "I",
        "C"
    ],

    "Operaciones": [
        "C",
        "I",
        "C",
        "R/A",
        "R/A"
    ]
})


print("\n" + "=" * 90)
print("MATRIZ RACI")
print("=" * 90)

print(raci.to_string(index=False))


# =============================================================================
# 12. IDENTIFICAR RIESGOS DE IMPLEMENTACIÓN
# =============================================================================
#
# Implementar una alternativa también genera riesgos.
#
# Utilizaremos:
#
# PROBABILIDAD: 1 a 5
#
# IMPACTO: 1 a 5
#
#
# Fórmula:
#
# NIVEL DE RIESGO = PROBABILIDAD x IMPACTO
#
# =============================================================================

riesgos = pd.DataFrame({
    "Riesgo": [
        "Proveedor alternativo no cumple calidad",
        "Costo total mayor al esperado",
        "Retraso en homologación",
        "Proveedor actual modifica condiciones",
        "Datos insuficientes para evaluar piloto",
        "Resistencia interna al cambio"
    ],

    "Probabilidad": [
        3,
        3,
        4,
        2,
        2,
        2
    ],

    "Impacto": [
        5,
        4,
        4,
        4,
        3,
        3
    ]
})


riesgos["Nivel_riesgo"] = (
    riesgos["Probabilidad"]
    * riesgos["Impacto"]
)


# =============================================================================
# 13. CLASIFICAR LOS RIESGOS
# =============================================================================
#
# Escala utilizada:
#
# 1 - 4   = Bajo
# 5 - 9   = Medio
# 10 - 15 = Alto
# 16 - 25 = Crítico
#
# =============================================================================

def clasificar_riesgo(valor):

    if valor <= 4:
        return "Bajo"

    elif valor <= 9:
        return "Medio"

    elif valor <= 15:
        return "Alto"

    else:
        return "Crítico"


riesgos["Clasificacion"] = riesgos["Nivel_riesgo"].apply(
    clasificar_riesgo
)


print("\n" + "=" * 90)
print("MATRIZ DE RIESGOS")
print("=" * 90)

print(
    riesgos
    .sort_values("Nivel_riesgo", ascending=False)
    .to_string(index=False)
)


# =============================================================================
# 14. GRÁFICA DE RIESGOS
# =============================================================================

plt.figure(figsize=(12, 7))

colores_riesgo = {
    "Bajo": "#43A047",
    "Medio": "#FDD835",
    "Alto": "#FB8C00",
    "Crítico": "#E53935"
}


for clasificacion in ["Bajo", "Medio", "Alto", "Crítico"]:

    datos = riesgos[
        riesgos["Clasificacion"] == clasificacion
    ]

    plt.scatter(
        datos["Probabilidad"],
        datos["Impacto"],
        s=datos["Nivel_riesgo"] * 60,
        color=colores_riesgo[clasificacion],
        label=clasificacion,
        alpha=0.75,
        edgecolors="black"
    )


for i, fila in riesgos.iterrows():

    plt.text(
        fila["Probabilidad"] + 0.05,
        fila["Impacto"] + 0.05,
        f"R{i + 1}",
        fontsize=10,
        fontweight="bold"
    )


plt.xlim(0.5, 5.5)
plt.ylim(0.5, 5.5)

plt.xticks(range(1, 6))
plt.yticks(range(1, 6))

plt.xlabel("Probabilidad")
plt.ylabel("Impacto")

plt.title(
    "Mapa de riesgos de implementación",
    fontsize=17,
    fontweight="bold"
)

plt.legend(title="Clasificación")

plt.tight_layout()
plt.show()


print("""
INTERPRETACIÓN DEL MAPA DE RIESGOS:

Los riesgos ubicados hacia la zona superior derecha
requieren mayor atención.

¿Por qué?

Porque combinan:

ALTA PROBABILIDAD

+

ALTO IMPACTO.

Una buena propuesta empresarial no solamente explica
los beneficios esperados.

También identifica:

- qué podría salir mal,
- qué tan importante sería,
- cómo podría controlarse.
""")


# =============================================================================
# 15. DEFINIR ACCIONES DE MITIGACIÓN
# =============================================================================

mitigacion = pd.DataFrame({
    "Riesgo": riesgos["Riesgo"],

    "Accion_mitigacion": [
        "Realizar pruebas técnicas y lote piloto antes de aprobar",
        "Evaluar TCO y establecer límites económicos",
        "Definir cronograma y seguimiento semanal",
        "Mantener negociación y contratos claramente definidos",
        "Definir KPIs y captura de datos antes del piloto",
        "Realizar comunicación, capacitación y participación del equipo"
    ]
})


print("\n" + "=" * 90)
print("ACCIONES DE MITIGACIÓN")
print("=" * 90)

print(mitigacion.to_string(index=False))


# =============================================================================
# 16. SIMULAR EL PILOTO
# =============================================================================
#
# Imaginaremos que GlobalBike ejecuta un piloto
# durante seis meses.
#
# ATENCIÓN:
#
# LOS SIGUIENTES DATOS SON COMPLETAMENTE SIMULADOS.
#
# No representan resultados reales.
#
# Se utilizan para mostrar cómo debería monitorearse
# una estrategia después de implementarla.
#
# =============================================================================

seguimiento = pd.DataFrame({
    "Mes": [
        "Mes 1",
        "Mes 2",
        "Mes 3",
        "Mes 4",
        "Mes 5",
        "Mes 6"
    ],

    "Nivel_servicio": [
        90,
        91,
        92,
        94,
        95,
        96
    ],

    "Lead_time": [
        74,
        72,
        69,
        66,
        62,
        59
    ],

    "Dependencia_proveedor": [
        90,
        86,
        82,
        78,
        73,
        69
    ],

    "Compras_urgentes_USD": [
        4800,
        4200,
        3500,
        2700,
        1900,
        1400
    ]
})


print("\n" + "=" * 90)
print("DATOS SIMULADOS DEL PILOTO")
print("=" * 90)

print(seguimiento.to_string(index=False))


# =============================================================================
# 17. GRÁFICA: NIVEL DE SERVICIO
# =============================================================================

plt.figure(figsize=(12, 6))

plt.plot(
    seguimiento["Mes"],
    seguimiento["Nivel_servicio"],
    marker="o",
    linewidth=3,
    markersize=8,
    color="#1565C0",
    label="Nivel de servicio"
)

plt.axhline(
    nivel_servicio_meta,
    color="#43A047",
    linestyle="--",
    linewidth=2,
    label=f"Meta = {nivel_servicio_meta:.0f}%"
)

plt.title(
    "Seguimiento del nivel de servicio",
    fontsize=17,
    fontweight="bold"
)

plt.ylabel("Nivel de servicio (%)")
plt.xlabel("")

plt.ylim(85, 100)

plt.legend()

plt.tight_layout()
plt.show()


print("""
INTERPRETACIÓN:

En esta simulación el nivel de servicio
mejora progresivamente.

La línea verde representa la meta del 95 %.

Es importante observar que:

ALCANZAR una meta

no necesariamente significa

SOSTENER una meta.

Por eso los indicadores deben monitorearse
durante varios periodos.
""")


# =============================================================================
# 18. GRÁFICA: LEAD TIME
# =============================================================================

plt.figure(figsize=(12, 6))

plt.plot(
    seguimiento["Mes"],
    seguimiento["Lead_time"],
    marker="o",
    linewidth=3,
    markersize=8,
    color="#8E24AA",
    label="Lead time"
)

plt.axhline(
    lead_time_meta,
    color="#43A047",
    linestyle="--",
    linewidth=2,
    label=f"Meta = {lead_time_meta:.0f} días"
)

plt.title(
    "Seguimiento del lead time",
    fontsize=17,
    fontweight="bold"
)

plt.ylabel("Días")
plt.xlabel("")

plt.legend()

plt.tight_layout()
plt.show()


print("""
INTERPRETACIÓN:

En este indicador la dirección deseada
es DESCENDENTE.

Reducir el lead time puede disminuir:

- exposición durante la reposición,
- necesidad de inventario de protección,
- riesgo de ruptura.

Sin embargo, siempre debemos revisar también
el costo asociado a esa reducción.
""")


# =============================================================================
# 19. GRÁFICA: DEPENDENCIA DEL PROVEEDOR
# =============================================================================

plt.figure(figsize=(12, 6))

plt.plot(
    seguimiento["Mes"],
    seguimiento["Dependencia_proveedor"],
    marker="o",
    linewidth=3,
    markersize=8,
    color="#E53935",
    label="Dependencia proveedor principal"
)

plt.axhline(
    dependencia_meta,
    color="#43A047",
    linestyle="--",
    linewidth=2,
    label=f"Meta = {dependencia_meta:.0f}%"
)

plt.title(
    "Evolución de la dependencia del proveedor principal",
    fontsize=17,
    fontweight="bold"
)

plt.ylabel("Dependencia (%)")
plt.xlabel("")

plt.legend()

plt.tight_layout()
plt.show()


print("""
INTERPRETACIÓN:

Esta gráfica se relaciona directamente
con el propósito del dual sourcing.

La estrategia NO busca necesariamente eliminar
al proveedor principal.

Busca disminuir una concentración excesiva.

Por ejemplo:

ANTES:

Proveedor principal = 92 %

DESPUÉS:

Proveedor principal ≈ 70 %
Proveedor alternativo ≈ 30 %

Esto puede aumentar la resiliencia,
aunque también incrementa la complejidad
de la gestión de proveedores.
""")


# =============================================================================
# 20. GRÁFICA: COMPRAS URGENTES
# =============================================================================

plt.figure(figsize=(12, 6))

plt.bar(
    seguimiento["Mes"],
    seguimiento["Compras_urgentes_USD"],
    color=[
        "#E53935",
        "#EF6C00",
        "#F9A825",
        "#FDD835",
        "#7CB342",
        "#43A047"
    ]
)

plt.title(
    "Evolución mensual de compras urgentes",
    fontsize=17,
    fontweight="bold"
)

plt.ylabel("USD")
plt.xlabel("")

for i, valor in enumerate(seguimiento["Compras_urgentes_USD"]):

    plt.text(
        i,
        valor + 100,
        f"${valor:,.0f}",
        ha="center",
        fontweight="bold"
    )

plt.tight_layout()
plt.show()


print("""
INTERPRETACIÓN:

Las compras urgentes suelen ser una señal
de que la organización está reaccionando
a problemas de disponibilidad.

Su reducción puede representar:

- menor uso de transporte urgente,
- menos compras de emergencia,
- menor presión operacional,
- mayor estabilidad del abastecimiento.

Pero debe analizarse junto con el inventario.

Reducir compras urgentes aumentando excesivamente
el inventario tampoco sería necesariamente eficiente.
""")


# =============================================================================
# 21. TOMAR RESULTADOS DEL ÚLTIMO MES
# =============================================================================

ultimo_mes = seguimiento.iloc[-1]


# =============================================================================
# 22. CREAR TABLA DE RESULTADOS DEL PILOTO
# =============================================================================

resultados_piloto = pd.DataFrame({
    "Indicador": [
        "Nivel de servicio",
        "Lead time",
        "Dependencia proveedor"
    ],

    "Inicial": [
        nivel_servicio_actual,
        lead_time_actual,
        dependencia_actual
    ],

    "Meta": [
        nivel_servicio_meta,
        lead_time_meta,
        dependencia_meta
    ],

    "Resultado": [
        ultimo_mes["Nivel_servicio"],
        ultimo_mes["Lead_time"],
        ultimo_mes["Dependencia_proveedor"]
    ],

    "Direccion": [
        "Aumentar",
        "Disminuir",
        "Disminuir"
    ]
})


# =============================================================================
# 23. CALCULAR CUMPLIMIENTO DE LA META
# =============================================================================
#
# Para un indicador que debe AUMENTAR:
#
#                    RESULTADO - INICIAL
# Cumplimiento = ---------------------------- x 100
#                       META - INICIAL
#
#
# Ejemplo:
#
# Inicial = 89
# Meta = 95
# Resultado = 92
#
# (92 - 89) / (95 - 89) = 0.50
#
# Cumplimiento = 50 %
#
#
# Para un indicador que debe DISMINUIR:
#
#                     INICIAL - RESULTADO
# Cumplimiento = ---------------------------- x 100
#                       INICIAL - META
#
# =============================================================================

def calcular_cumplimiento(inicial, meta, resultado, direccion):

    if direccion == "Aumentar":

        denominador = meta - inicial

        if denominador == 0:
            return 100.0

        cumplimiento = (
            (resultado - inicial)
            / denominador
        ) * 100

    else:

        denominador = inicial - meta

        if denominador == 0:
            return 100.0

        cumplimiento = (
            (inicial - resultado)
            / denominador
        ) * 100

    return cumplimiento


resultados_piloto["Cumplimiento_%"] = resultados_piloto.apply(
    lambda fila: calcular_cumplimiento(
        fila["Inicial"],
        fila["Meta"],
        fila["Resultado"],
        fila["Direccion"]
    ),
    axis=1
)


# =============================================================================
# 24. CREAR SEMÁFORO
# =============================================================================
#
# Utilizaremos:
#
# VERDE:
# Cumplimiento >= 100 %
#
# AMARILLO:
# Cumplimiento entre 80 % y 99.9 %
#
# ROJO:
# Cumplimiento < 80 %
#
# =============================================================================

def asignar_semaforo(cumplimiento):

    if cumplimiento >= 100:
        return "VERDE"

    elif cumplimiento >= 80:
        return "AMARILLO"

    else:
        return "ROJO"


resultados_piloto["Semaforo"] = (
    resultados_piloto["Cumplimiento_%"]
    .apply(asignar_semaforo)
)


print("\n" + "=" * 90)
print("SEMÁFORO DE KPIs")
print("=" * 90)

print(
    resultados_piloto
    .round(1)
    .to_string(index=False)
)


# =============================================================================
# 25. GRÁFICA DEL SEMÁFORO
# =============================================================================

mapa_colores = {
    "VERDE": "#43A047",
    "AMARILLO": "#FDD835",
    "ROJO": "#E53935"
}


colores_semaforo = [
    mapa_colores[estado]
    for estado in resultados_piloto["Semaforo"]
]


plt.figure(figsize=(11, 6))

plt.bar(
    resultados_piloto["Indicador"],
    resultados_piloto["Cumplimiento_%"],
    color=colores_semaforo
)

plt.axhline(
    100,
    linestyle="--",
    color="#212121",
    linewidth=2,
    label="Meta = 100% de cumplimiento"
)

plt.axhline(
    80,
    linestyle=":",
    color="#616161",
    linewidth=2,
    label="Umbral amarillo = 80%"
)

plt.ylabel("Cumplimiento de la mejora esperada (%)")

plt.title(
    "Semáforo de seguimiento del piloto",
    fontsize=17,
    fontweight="bold"
)

plt.legend()

plt.tight_layout()
plt.show()


print("""
INTERPRETACIÓN DEL SEMÁFORO:

VERDE:

La meta fue alcanzada o superada.

AMARILLO:

Existe un avance importante,
pero todavía no se alcanza completamente la meta.

ROJO:

La mejora todavía se encuentra
por debajo del nivel esperado.

IMPORTANTE:

El semáforo facilita la comunicación ejecutiva,
pero NO sustituye el análisis detallado.
""")


# =============================================================================
# 26. EVALUACIÓN ECONÓMICA
# =============================================================================
#
# Ahora conectaremos la mejora operacional
# con una dimensión económica.
#
#
# Compras urgentes actuales:
#
# USD 58.000 / año
#
#
# Meta:
#
# USD 20.000 / año
#
#
# Reducción potencial:
#
# 58.000 - 20.000
#
# =
#
# USD 38.000
#
# =============================================================================

ahorro_bruto = (
    compras_urgentes_actuales
    - compras_urgentes_meta
)


print("\n" + "=" * 90)
print("EVALUACIÓN ECONÓMICA")
print("=" * 90)

print(f"""
Compras urgentes actuales:

USD {compras_urgentes_actuales:,.0f}

Compras urgentes objetivo:

USD {compras_urgentes_meta:,.0f}

Reducción potencial:

USD {ahorro_bruto:,.0f}
""")


# =============================================================================
# 27. COSTO DE IMPLEMENTACIÓN
# =============================================================================
#
# Supongamos que implementar el piloto requiere:
#
# - evaluación del proveedor,
# - auditoría,
# - homologación,
# - pruebas,
# - negociación,
# - ajustes logísticos.
#
#
# Costo estimado:
#
# USD 22.000
#
# =============================================================================

costo_implementacion = 22000


# =============================================================================
# 28. BENEFICIO NETO
# =============================================================================
#
# Fórmula:
#
# BENEFICIO NETO =
#
# BENEFICIO BRUTO - COSTO DE IMPLEMENTACIÓN
#
# =============================================================================

beneficio_neto_estimado = (
    ahorro_bruto
    - costo_implementacion
)


print(f"""
Costo estimado de implementación:

USD {costo_implementacion:,.0f}

Beneficio neto estimado:

USD {beneficio_neto_estimado:,.0f}
""")


# =============================================================================
# 29. CALCULAR ROI
# =============================================================================
#
# ROI = Return on Investment
#
#
# Fórmula simplificada:
#
#             BENEFICIO NETO
# ROI = --------------------------- x 100
#                INVERSIÓN
#
#
# IMPORTANTE:
#
# Este es un cálculo didáctico simplificado.
#
# Una evaluación financiera completa podría incluir:
#
# - VAN,
# - TIR,
# - flujo de caja,
# - costo de capital,
# - impuestos,
# - costos recurrentes,
# - horizonte de evaluación.
#
# =============================================================================

ROI = (
    beneficio_neto_estimado
    / costo_implementacion
) * 100


print(f"ROI estimado del ejemplo: {ROI:.1f}%")


# =============================================================================
# 30. GRÁFICA ECONÓMICA
# =============================================================================

categorias_economicas = [
    "Compras urgentes\nactuales",
    "Compras urgentes\nmeta",
    "Costo de\nimplementación",
    "Beneficio neto\nestimado"
]

valores_economicos = [
    compras_urgentes_actuales,
    compras_urgentes_meta,
    costo_implementacion,
    beneficio_neto_estimado
]


plt.figure(figsize=(12, 6))

plt.bar(
    categorias_economicas,
    valores_economicos,
    color=[
        "#E53935",
        "#F9A825",
        "#1565C0",
        "#43A047"
    ]
)

plt.ylabel("USD")

plt.title(
    "Evaluación económica simplificada",
    fontsize=17,
    fontweight="bold"
)

for i, valor in enumerate(valores_economicos):

    plt.text(
        i,
        valor + 1000,
        f"USD {valor:,.0f}",
        ha="center",
        fontweight="bold"
    )

plt.tight_layout()
plt.show()


print(f"""
INTERPRETACIÓN ECONÓMICA:

La reducción potencial de compras urgentes
es aproximadamente:

USD {ahorro_bruto:,.0f}

El costo estimado de implementación es:

USD {costo_implementacion:,.0f}

Por tanto, el beneficio neto simplificado sería:

USD {beneficio_neto_estimado:,.0f}

y el ROI estimado:

{ROI:.1f}%

ATENCIÓN:

Estos valores son supuestos del ejercicio.

No deben presentarse como resultados financieros reales.
""")


# =============================================================================
# 31. CREAR CRITERIOS DE ESCALAMIENTO
# =============================================================================
#
# Ahora debemos decidir:
#
# ¿EL PILOTO DEBERÍA ESCALAR?
#
#
# Para evitar una decisión basada solamente
# en percepción, definiremos reglas.
#
#
# Criterios:
#
# 1. Nivel de servicio >= 95 %
#
# 2. Lead time <= 60 días
#
# 3. Dependencia proveedor <= 70 %
#
# 4. Beneficio económico > 0
#
# 5. No existir riesgos críticos sin controlar
#
# =============================================================================

condicion_servicio = (
    ultimo_mes["Nivel_servicio"]
    >= nivel_servicio_meta
)

condicion_lead_time = (
    ultimo_mes["Lead_time"]
    <= lead_time_meta
)

condicion_dependencia = (
    ultimo_mes["Dependencia_proveedor"]
    <= dependencia_meta
)

condicion_economica = (
    beneficio_neto_estimado > 0
)


numero_riesgos_criticos = (
    riesgos["Clasificacion"] == "Crítico"
).sum()


condicion_riesgos = (
    numero_riesgos_criticos == 0
)


# =============================================================================
# 32. TABLA DE CRITERIOS DE ESCALAMIENTO
# =============================================================================

criterios_escalamiento = pd.DataFrame({
    "Criterio": [
        "Nivel de servicio >= meta",
        "Lead time <= meta",
        "Dependencia <= meta",
        "Beneficio económico positivo",
        "Sin riesgos críticos"
    ],

    "Cumple": [
        condicion_servicio,
        condicion_lead_time,
        condicion_dependencia,
        condicion_economica,
        condicion_riesgos
    ]
})


print("\n" + "=" * 90)
print("CRITERIOS DE ESCALAMIENTO")
print("=" * 90)

print(criterios_escalamiento.to_string(index=False))


# =============================================================================
# 33. PORCENTAJE DE CRITERIOS CUMPLIDOS
# =============================================================================
#
# En Python:
#
# True = 1
# False = 0
#
# Por eso podemos calcular el promedio
# de la columna "Cumple".
#
# =============================================================================

porcentaje_cumplimiento = (
    criterios_escalamiento["Cumple"].mean()
    * 100
)


print(
    f"\nPorcentaje de criterios cumplidos: "
    f"{porcentaje_cumplimiento:.1f}%"
)


# =============================================================================
# 34. REGLA DE DECISIÓN
# =============================================================================
#
# Para fines pedagógicos:
#
# 100 %:
#
# CANDIDATO A ESCALAMIENTO
#
#
# 80 % a menos de 100 %:
#
# CONTINUAR PILOTO Y AJUSTAR
#
#
# Menos de 80 %:
#
# REVISAR PROPUESTA
#
# =============================================================================

if porcentaje_cumplimiento == 100:

    decision_final = "CANDIDATO A ESCALAMIENTO"

elif porcentaje_cumplimiento >= 80:

    decision_final = "CONTINUAR PILOTO Y AJUSTAR"

else:

    decision_final = "REVISAR LA PROPUESTA"


print("\nDECISIÓN DEL EJEMPLO:")
print(decision_final)


# =============================================================================
# 35. GRÁFICA DE CRITERIOS DE ESCALAMIENTO
# =============================================================================

criterios_plot = criterios_escalamiento.copy()

criterios_plot["Valor"] = criterios_plot["Cumple"].astype(int) * 100

criterios_plot["Estado"] = criterios_plot["Cumple"].map({
    True: "Cumple",
    False: "No cumple"
})


colores_criterios = criterios_plot["Cumple"].map({
    True: "#43A047",
    False: "#E53935"
})


plt.figure(figsize=(12, 6))

plt.barh(
    criterios_plot["Criterio"],
    criterios_plot["Valor"],
    color=colores_criterios
)

plt.xlim(0, 110)

plt.xlabel("Cumplimiento")

plt.title(
    "Criterios para decidir el escalamiento",
    fontsize=17,
    fontweight="bold"
)

for i, fila in criterios_plot.iterrows():

    plt.text(
        102 if fila["Cumple"] else 3,
        i,
        fila["Estado"],
        va="center",
        fontweight="bold"
    )

plt.tight_layout()
plt.show()


# =============================================================================
# 36. DASHBOARD EJECUTIVO FINAL
# =============================================================================
#
# Ahora resumiremos todo el ejercicio
# en una sola pantalla.
#
# Esto es útil porque la alta dirección
# normalmente no necesita revisar todas
# las líneas del código.
#
# Necesita comprender:
#
# - qué se decidió,
# - qué pasó,
# - qué indicadores mejoraron,
# - qué riesgos existen,
# - si conviene avanzar.
#
# =============================================================================

fig, ax = plt.subplots(figsize=(16, 9))

ax.axis("off")


# -----------------------------------------------------------------------------
# TÍTULO
# -----------------------------------------------------------------------------

ax.text(
    0.50,
    0.94,
    "GLOBALBIKE | FROM ANALYSIS TO ACTION",
    transform=ax.transAxes,
    ha="center",
    fontsize=21,
    fontweight="bold"
)


# -----------------------------------------------------------------------------
# ALTERNATIVA
# -----------------------------------------------------------------------------

ax.text(
    0.05,
    0.82,
    "ALTERNATIVA",
    transform=ax.transAxes,
    fontsize=13,
    fontweight="bold",
    color="#1565C0"
)

ax.text(
    0.05,
    0.75,
    alternativa_priorizada,
    transform=ax.transAxes,
    fontsize=14,
    fontweight="bold"
)


# -----------------------------------------------------------------------------
# SERVICIO
# -----------------------------------------------------------------------------

ax.text(
    0.30,
    0.82,
    "NIVEL DE SERVICIO",
    transform=ax.transAxes,
    fontsize=13,
    fontweight="bold",
    color="#43A047"
)

ax.text(
    0.30,
    0.75,
    f"{nivel_servicio_actual:.0f}% → "
    f"{ultimo_mes['Nivel_servicio']:.0f}%",
    transform=ax.transAxes,
    fontsize=17,
    fontweight="bold"
)


# -----------------------------------------------------------------------------
# LEAD TIME
# -----------------------------------------------------------------------------

ax.text(
    0.55,
    0.82,
    "LEAD TIME",
    transform=ax.transAxes,
    fontsize=13,
    fontweight="bold",
    color="#8E24AA"
)

ax.text(
    0.55,
    0.75,
    f"{lead_time_actual:.0f} → "
    f"{ultimo_mes['Lead_time']:.0f} días",
    transform=ax.transAxes,
    fontsize=17,
    fontweight="bold"
)


# -----------------------------------------------------------------------------
# DEPENDENCIA
# -----------------------------------------------------------------------------

ax.text(
    0.78,
    0.82,
    "DEPENDENCIA",
    transform=ax.transAxes,
    fontsize=13,
    fontweight="bold",
    color="#E53935"
)

ax.text(
    0.78,
    0.75,
    f"{dependencia_actual:.0f}% → "
    f"{ultimo_mes['Dependencia_proveedor']:.0f}%",
    transform=ax.transAxes,
    fontsize=17,
    fontweight="bold"
)


# -----------------------------------------------------------------------------
# ROI
# -----------------------------------------------------------------------------

ax.text(
    0.18,
    0.53,
    "ROI ESTIMADO",
    transform=ax.transAxes,
    fontsize=13,
    fontweight="bold",
    color="#F9A825"
)

ax.text(
    0.18,
    0.45,
    f"{ROI:.1f}%",
    transform=ax.transAxes,
    fontsize=28,
    fontweight="bold"
)


# -----------------------------------------------------------------------------
# RIESGOS CRÍTICOS
# -----------------------------------------------------------------------------

ax.text(
    0.45,
    0.53,
    "RIESGOS CRÍTICOS",
    transform=ax.transAxes,
    fontsize=13,
    fontweight="bold",
    color="#E53935"
)

ax.text(
    0.45,
    0.45,
    str(numero_riesgos_criticos),
    transform=ax.transAxes,
    fontsize=28,
    fontweight="bold"
)


# -----------------------------------------------------------------------------
# CRITERIOS CUMPLIDOS
# -----------------------------------------------------------------------------

ax.text(
    0.72,
    0.53,
    "CRITERIOS CUMPLIDOS",
    transform=ax.transAxes,
    fontsize=13,
    fontweight="bold",
    color="#1565C0"
)

ax.text(
    0.72,
    0.45,
    f"{porcentaje_cumplimiento:.0f}%",
    transform=ax.transAxes,
    fontsize=28,
    fontweight="bold"
)


# -----------------------------------------------------------------------------
# DECISIÓN FINAL
# -----------------------------------------------------------------------------

ax.text(
    0.50,
    0.27,
    "DECISIÓN DE IMPLEMENTACIÓN",
    transform=ax.transAxes,
    ha="center",
    fontsize=15,
    fontweight="bold"
)


# Elegimos color según decisión.

if decision_final == "CANDIDATO A ESCALAMIENTO":

    color_decision = "#43A047"

elif decision_final == "CONTINUAR PILOTO Y AJUSTAR":

    color_decision = "#F9A825"

else:

    color_decision = "#E53935"


ax.text(
    0.50,
    0.18,
    decision_final,
    transform=ax.transAxes,
    ha="center",
    fontsize=23,
    fontweight="bold",
    color=color_decision
)


ax.text(
    0.50,
    0.08,
    "Resultados simulados con fines exclusivamente didácticos",
    transform=ax.transAxes,
    ha="center",
    fontsize=10,
    style="italic"
)


plt.show()


# =============================================================================
# 37. CONCLUSIÓN GERENCIAL AUTOMÁTICA
# =============================================================================

print("\n" + "=" * 90)
print("CONCLUSIÓN GERENCIAL")
print("=" * 90)


print(f"""
El caso de GlobalBike permitió transformar
una recomendación analítica en una propuesta
estructurada de implementación.

La alternativa priorizada mediante AHP fue:

{alternativa_priorizada}

con una prioridad aproximada de:

{prioridad_ahp:.1f}%.


La propuesta fue estructurada mediante cinco fases:

1. Preparación.

2. Búsqueda del proveedor alternativo.

3. Evaluación y homologación.

4. Implementación del piloto.

5. Evaluación y decisión de escalamiento.


En los DATOS SIMULADOS del piloto,
el nivel de servicio evolucionó desde:

{nivel_servicio_actual:.0f}%

hasta:

{ultimo_mes['Nivel_servicio']:.0f}%.


El lead time evolucionó desde:

{lead_time_actual:.0f} días

hasta:

{ultimo_mes['Lead_time']:.0f} días.


La dependencia del proveedor principal
evolucionó desde:

{dependencia_actual:.0f}%

hasta:

{ultimo_mes['Dependencia_proveedor']:.0f}%.


El ahorro bruto potencial del escenario sería:

USD {ahorro_bruto:,.0f}


El costo estimado de implementación sería:

USD {costo_implementacion:,.0f}


El beneficio neto simplificado sería:

USD {beneficio_neto_estimado:,.0f}


El ROI estimado del escenario sería:

{ROI:.1f}%.


El porcentaje de criterios de escalamiento
cumplidos es:

{porcentaje_cumplimiento:.1f}%.


Por tanto, bajo los SUPUESTOS
y DATOS SIMULADOS del ejercicio,
la clasificación sería:

{decision_final}


IMPORTANTE:

Esta conclusión NO demuestra que el dual sourcing
producirá necesariamente estos resultados.

Los valores representan un escenario académico.

En una empresa real, los resultados deberían
validarse mediante un piloto y datos observados.
""")


# =============================================================================
# 38. CREAR TABLA RESUMEN PARA PRESENTACIÓN
# =============================================================================
#
# Esta tabla puede llevarse posteriormente
# a Google Sheets, Power BI o Looker Studio.
#
# =============================================================================

resumen_ejecutivo = pd.DataFrame({
    "Elemento": [
        "Problema",
        "Alternativa priorizada",
        "Nivel de servicio",
        "Lead time",
        "Dependencia proveedor",
        "Beneficio neto estimado",
        "ROI estimado",
        "Decisión"
    ],

    "Resultado": [
        "Riesgo de stockout y dependencia de abastecimiento",
        alternativa_priorizada,
        f"{nivel_servicio_actual:.0f}% → {ultimo_mes['Nivel_servicio']:.0f}%",
        f"{lead_time_actual:.0f} → {ultimo_mes['Lead_time']:.0f} días",
        f"{dependencia_actual:.0f}% → {ultimo_mes['Dependencia_proveedor']:.0f}%",
        f"USD {beneficio_neto_estimado:,.0f}",
        f"{ROI:.1f}%",
        decision_final
    ]
})


print("\n" + "=" * 90)
print("RESUMEN EJECUTIVO")
print("=" * 90)

print(resumen_ejecutivo.to_string(index=False))


# =============================================================================
# 39. EXPORTAR RESULTADOS
# =============================================================================
#
# Los archivos generados pueden utilizarse posteriormente
# en otras herramientas.
#
# =============================================================================

plan.to_csv(
    "GlobalBike_Plan_Implementacion.csv",
    index=False,
    encoding="utf-8-sig"
)

entregables.to_csv(
    "GlobalBike_Entregables.csv",
    index=False,
    encoding="utf-8-sig"
)

raci.to_csv(
    "GlobalBike_RACI.csv",
    index=False,
    encoding="utf-8-sig"
)

kpis.to_csv(
    "GlobalBike_KPIs.csv",
    index=False,
    encoding="utf-8-sig"
)

riesgos.to_csv(
    "GlobalBike_Riesgos.csv",
    index=False,
    encoding="utf-8-sig"
)

mitigacion.to_csv(
    "GlobalBike_Mitigacion.csv",
    index=False,
    encoding="utf-8-sig"
)

seguimiento.to_csv(
    "GlobalBike_Seguimiento_Piloto.csv",
    index=False,
    encoding="utf-8-sig"
)

resultados_piloto.to_csv(
    "GlobalBike_Semaforo_KPIs.csv",
    index=False,
    encoding="utf-8-sig"
)

criterios_escalamiento.to_csv(
    "GlobalBike_Criterios_Escalamiento.csv",
    index=False,
    encoding="utf-8-sig"
)

resumen_ejecutivo.to_csv(
    "GlobalBike_Resumen_Ejecutivo.csv",
    index=False,
    encoding="utf-8-sig"
)


print("\n" + "=" * 90)
print("ARCHIVOS GENERADOS")
print("=" * 90)

print("""
1. GlobalBike_Plan_Implementacion.csv

2. GlobalBike_Entregables.csv

3. GlobalBike_RACI.csv

4. GlobalBike_KPIs.csv

5. GlobalBike_Riesgos.csv

6. GlobalBike_Mitigacion.csv

7. GlobalBike_Seguimiento_Piloto.csv

8. GlobalBike_Semaforo_KPIs.csv

9. GlobalBike_Criterios_Escalamiento.csv

10. GlobalBike_Resumen_Ejecutivo.csv
""")


# =============================================================================
# 40. MAPA COMPLETO DEL CASO INTEGRADOR
# =============================================================================

print("""
==========================================================================================

                    INVENTARIOS Y COMPRAS GLOBALES

                       CASO GLOBALBIKE

==========================================================================================


                       PROBLEMA EMPRESARIAL
                                │
                                ↓
                       EVIDENCIA CIENTÍFICA
                       SCOPUS + VOSVIEWER
                                │
                                ↓
                   DIAGNÓSTICO DE INVENTARIOS
                                │
                                ↓
                      KPIs Y SEGMENTACIÓN
                                │
                                ↓
                     PENSAMIENTO SISTÉMICO
                             VENSIM
                                │
                                ↓
                         DATA ANALYTICS
                             PYTHON
                                │
                                ↓
                       MACHINE LEARNING
                                │
                                ↓
                  PREDICCIÓN DEL RIESGO
                                │
                                ↓
                 ALTERNATIVAS DE SOLUCIÓN
                                │
                                ↓
                              AHP
                                │
                                ↓
                   PRIORIZACIÓN MULTICRITERIO
                                │
                                ↓
                    DECISIÓN EMPRESARIAL
                                │
                                ↓
                    PLAN DE IMPLEMENTACIÓN
                                │
                                ↓
                             PILOTO
                                │
                                ↓
                     MONITOREO DE KPIs
                                │
                                ↓
                   EVALUACIÓN DE RIESGOS
                                │
                                ↓
                    EVALUACIÓN ECONÓMICA
                                │
                                ↓
                  DECISIÓN DE ESCALAMIENTO
                                │
                                ↓
                    MEJORA Y APRENDIZAJE


==========================================================================================

                         FROM ANALYSIS TO ACTION

==========================================================================================
""")


# =============================================================================
# 41. MENSAJE FINAL PARA LOS ESTUDIANTES
# =============================================================================

print("""
MENSAJE FINAL DEL CASO:

El propósito de utilizar Python,
Machine Learning, Vensim, bibliometría o AHP
no consiste en utilizar la mayor cantidad posible
de herramientas.

Una herramienta agrega valor cuando ayuda
a responder una pregunta concreta.


VOSviewer ayudó a preguntar:

¿QUÉ DICE LA EVIDENCIA CIENTÍFICA?


Vensim ayudó a preguntar:

¿CÓMO SE RELACIONAN LAS VARIABLES?


Python y Data Analytics ayudaron a preguntar:

¿QUÉ ESTÁ PASANDO?


Machine Learning ayudó a preguntar:

¿PODEMOS ANTICIPAR EL RIESGO?


El análisis de alternativas ayudó a preguntar:

¿QUÉ PODEMOS HACER?


AHP ayudó a preguntar:

¿CÓMO PODEMOS PRIORIZAR LAS ALTERNATIVAS?


La integración ayudó a preguntar:

¿QUÉ SIGNIFICAN TODOS LOS RESULTADOS JUNTOS?


Y finalmente el plan de implementación permitió preguntar:

¿CÓMO CONVERTIMOS LA DECISIÓN EN ACCIÓN?


Por eso la secuencia completa puede resumirse como:


PROBLEMA

        ↓

EVIDENCIA

        ↓

ANÁLISIS

        ↓

COMPRENSIÓN

        ↓

PREDICCIÓN

        ↓

ALTERNATIVAS

        ↓

PRIORIZACIÓN

        ↓

DECISIÓN

        ↓

ACCIÓN

        ↓

MEDICIÓN

        ↓

APRENDIZAJE


==========================================================================================

                         FIN DEL CASO INTEGRADOR

==========================================================================================
""")
```
