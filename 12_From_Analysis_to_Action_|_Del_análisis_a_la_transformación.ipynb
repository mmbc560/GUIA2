```python
# =============================================================================
# INVENTARIOS Y COMPRAS GLOBALES
# CASO INTEGRADOR: GLOBALBIKE COMPONENTS COLOMBIA S.A.S.
#
# ETAPA 12: FROM ANALYSIS TO ACTION
# DEL ANÁLISIS A LA ACCIÓN
# =============================================================================
#
# OBJETIVO:
#
# Convertir la alternativa priorizada mediante AHP
# en un PLAN DE IMPLEMENTACIÓN medible.
#
#
# Hasta ahora recorrimos:
#
# DATA ANALYTICS
#       ↓
# MACHINE LEARNING
#       ↓
# ALTERNATIVAS
#       ↓
# AHP
#       ↓
# DECISIÓN
#
#
# Ahora debemos responder:
#
# ¿CÓMO IMPLEMENTAMOS LA DECISIÓN?
#
#
# Una recomendación empresarial completa debería indicar:
#
# 1. QUÉ se hará.
#
# 2. POR QUÉ se hará.
#
# 3. CÓMO se hará.
#
# 4. QUIÉN será responsable.
#
# 5. CUÁNDO se realizará.
#
# 6. QUÉ recursos se necesitan.
#
# 7. CÓMO se medirá.
#
# 8. QUÉ riesgos existen.
#
# 9. CUÁNDO se revisará.
#
# 10. QUÉ condiciones permitirán escalar la solución.
#
#
# IMPORTANTE:
#
# Los datos de este caso son DIDÁCTICOS.
#
# Las metas, costos y resultados futuros son escenarios
# ilustrativos y NO resultados reales ya obtenidos.
#
# =============================================================================


# =============================================================================
# 1. IMPORTAR LIBRERÍAS
# =============================================================================

import pandas as pd

import numpy as np

import matplotlib.pyplot as plt

import seaborn as sns

from datetime import datetime, timedelta


# Configuración visual.

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 6)

plt.rcParams["font.size"] = 11


# =============================================================================
# 2. RECORDAR LA DECISIÓN DEL PUNTO 11
# =============================================================================
#
# En nuestro ejemplo académico:
#
# A1 = Stock de seguridad
#
# A2 = Dual sourcing
#
# A3 = Reabastecimiento predictivo
#
#
# El AHP del ejemplo priorizó:
#
# A2 = DUAL SOURCING
#
# con aproximadamente 43 %.
#
#
# ATENCIÓN:
#
# En el proyecto de cada estudiante esta variable
# debe reemplazarse por SU alternativa priorizada.
#
# =============================================================================


alternativa_priorizada = "A2 - Dual sourcing"

prioridad_ahp = 43.0


print("\n" + "=" * 82)

print("GLOBALBIKE COMPONENTS | FROM ANALYSIS TO ACTION")

print("=" * 82)


print(
    f"""

ALTERNATIVA PRIORIZADA:

{alternativa_priorizada}

Prioridad AHP del ejemplo:

{prioridad_ahp:.1f}%


PREGUNTA DE ESTA ETAPA:

¿Cómo transformar esta recomendación
en un plan concreto, medible y controlable?

"""
)


# =============================================================================
# 3. DEFINIR LA SITUACIÓN ACTUAL
# =============================================================================
#
# Recuperamos algunos indicadores utilizados
# durante el caso.
#
# Estos representan la LÍNEA BASE.
#
# LÍNEA BASE =
#
# situación existente antes de implementar
# la propuesta.
#
# Sin línea base es muy difícil demostrar posteriormente
# si la intervención realmente produjo mejoras.
#
# =============================================================================


nivel_servicio_actual = 89.0

stockouts_actuales = 4

lead_time_actual = 75.0

dependencia_actual = 92.0

mape_actual = 18.4

compras_urgentes_actuales = 58000


# =============================================================================
# 4. DEFINIR METAS
# =============================================================================
#
# Las metas NO son resultados garantizados.
#
# Representan la situación deseada.
#
# Para el ejercicio plantearemos:
#
# Nivel de servicio:
# 89 % -> 95 %
#
# Stockouts:
# 4 -> máximo 1 por año
#
# Lead time:
# 75 -> 60 días
#
# Dependencia proveedor principal:
# 92 % -> 70 %
#
# MAPE:
# 18.4 % -> 12 %
#
# Compras urgentes:
# USD 58.000 -> USD 20.000
#
# =============================================================================


nivel_servicio_meta = 95.0

stockouts_meta = 1

lead_time_meta = 60.0

dependencia_meta = 70.0

mape_meta = 12.0

compras_urgentes_meta = 20000


# =============================================================================
# 5. CREAR TABLA ACTUAL VS META
# =============================================================================


kpis = pd.DataFrame({

    "Indicador": [

        "Nivel de servicio (%)",

        "Stockouts por año",

        "Lead time promedio (días)",

        "Dependencia proveedor principal (%)",

        "MAPE (%)",

        "Compras urgentes (USD)"

    ],

    "Actual": [

        nivel_servicio_actual,

        stockouts_actuales,

        lead_time_actual,

        dependencia_actual,

        mape_actual,

        compras_urgentes_actuales

    ],

    "Meta": [

        nivel_servicio_meta,

        stockouts_meta,

        lead_time_meta,

        dependencia_meta,

        mape_meta,

        compras_urgentes_meta

    ],

    # Dirección deseada:
    #
    # algunos indicadores deben aumentar,
    # mientras otros deben disminuir.

    "Direccion": [

        "Aumentar",

        "Disminuir",

        "Disminuir",

        "Disminuir",

        "Disminuir",

        "Disminuir"

    ]

})


print("\n" + "=" * 82)

print("LÍNEA BASE Y METAS")

print("=" * 82)


print(
    kpis.to_string(
        index=False
    )
)


# =============================================================================
# 6. CALCULAR EL CAMBIO REQUERIDO
# =============================================================================
#
# Si el indicador debe AUMENTAR:
#
#
#                 META - ACTUAL
# MEJORA (%) = --------------------- × 100
#                    ACTUAL
#
#
# Si el indicador debe DISMINUIR:
#
#
#                  ACTUAL - META
# REDUCCIÓN (%) = -------------------- × 100
#                     ACTUAL
#
#
# ¿POR QUÉ CAMBIAMOS LA FÓRMULA?
#
# Porque queremos que un porcentaje POSITIVO represente
# siempre la magnitud de la mejora deseada.
#
# =============================================================================


def calcular_cambio_requerido(actual, meta, direccion):

    if direccion == "Aumentar":

        cambio = (
            (meta - actual)
            /
            actual
        ) * 100

    else:

        cambio = (
            (actual - meta)
            /
            actual
        ) * 100

    return cambio


kpis["Cambio_requerido_%"] = kpis.apply(

    lambda fila:

    calcular_cambio_requerido(

        fila["Actual"],

        fila["Meta"],

        fila["Direccion"]

    ),

    axis=1

)


print("\nCAMBIO REQUERIDO:\n")


print(

    kpis[
        [
            "Indicador",
            "Cambio_requerido_%"
        ]
    ]

    .round(1)

    .to_string(
        index=False
    )

)


# =============================================================================
# 7. INTERPRETACIÓN DE LAS BRECHAS
# =============================================================================


print("""

INTERPRETACIÓN:

Esta tabla permite transformar una recomendación general
en objetivos cuantificables.

Por ejemplo:

No basta con decir:

"mejorar el nivel de servicio".

Debemos establecer:

ACTUAL = 89 %

META = 95 %


Tampoco basta con decir:

"reducir la dependencia del proveedor".

Debemos establecer:

ACTUAL = 92 %

META = 70 %


Esto convierte una intención en un objetivo medible.

""")


# =============================================================================
# 8. CREAR EL PLAN DE IMPLEMENTACIÓN
# =============================================================================
#
# Diseñaremos cinco fases.
#
#
# FASE 1
# Preparación y validación de datos.
#
#
# FASE 2
# Búsqueda y preselección del proveedor alternativo.
#
#
# FASE 3
# Evaluación y homologación.
#
#
# FASE 4
# Piloto de dual sourcing.
#
#
# FASE 5
# Evaluación y decisión de escalamiento.
#
#
# IMPORTANTE:
#
# Los responsables son roles, no personas específicas.
#
# =============================================================================


plan = pd.DataFrame({

    "Fase": [

        "1. Preparación",

        "2. Búsqueda proveedor",

        "3. Homologación",

        "4. Piloto",

        "5. Evaluación"

    ],

    "Actividad": [

        "Validar demanda, inventario, costos y desempeño del proveedor actual",

        "Identificar y preseleccionar proveedores alternativos",

        "Evaluar calidad, costo total, capacidad, lead time y riesgo",

        "Implementar abastecimiento piloto con distribución 70/30",

        "Comparar resultados y decidir escalamiento"

    ],

    "Responsable": [

        "Planeación + Inventarios",

        "Compras Globales",

        "Compras + Calidad + Operaciones",

        "Supply Chain + Compras",

        "Comité de Supply Chain"

    ],

    "Inicio_semana": [

        1,

        3,

        6,

        10,

        15

    ],

    "Duracion_semanas": [

        2,

        3,

        4,

        5,

        2

    ]

})


# Calculamos automáticamente la semana final.
#
# Fórmula:
#
# Fin = Inicio + Duración - 1


plan["Fin_semana"] = (

    plan["Inicio_semana"]

    +

    plan["Duracion_semanas"]

    -

    1

)


print("\n" + "=" * 82)

print("PLAN DE IMPLEMENTACIÓN")

print("=" * 82)


print(
    plan.to_string(
        index=False
    )
)


# =============================================================================
# 9. CREAR UN DIAGRAMA DE GANTT
# =============================================================================
#
# Un Gantt permite visualizar:
#
# - cuándo comienza cada actividad,
# - cuánto dura,
# - qué actividades se superponen,
# - cuándo debería terminar el proyecto.
#
#
# La longitud de cada barra representa:
#
# DURACIÓN.
#
# =============================================================================


plt.figure(
    figsize=(14, 7)
)


colores_fases = [

    "#1565C0",

    "#00897B",

    "#F9A825",

    "#8E24AA",

    "#43A047"

]


for i, fila in plan.iterrows():

    plt.barh(

        fila["Fase"],

        fila["Duracion_semanas"],

        left=fila["Inicio_semana"],

        color=colores_fases[i],

        edgecolor="white",

        height=0.6

    )


    # Escribimos el responsable dentro o al lado
    # de la barra.

    plt.text(

        fila["Inicio_semana"]
        +
        fila["Duracion_semanas"] / 2,

        i,

        fila["Responsable"],

        ha="center",

        va="center",

        fontsize=9,

        color="white",

        fontweight="bold"

    )


plt.title(

    "GlobalBike | Roadmap de implementación",

    fontsize=17,

    fontweight="bold"

)


plt.xlabel(
    "Semana"
)


plt.ylabel(
    ""
)


plt.gca().invert_yaxis()


plt.tight_layout()

plt.show()


print("""

INTERPRETACIÓN DEL GANTT:

El plan no propone cambiar inmediatamente
todo el abastecimiento.

La implementación comienza con información,
continúa con búsqueda y homologación,
y posteriormente desarrolla un piloto.

Esta secuencia reduce el riesgo de implementar
una solución sin haberla validado previamente.

""")


# =============================================================================
# 10. DEFINIR ENTREGABLES POR FASE
# =============================================================================
#
# Una actividad debería producir algo verificable.
#
# A eso lo llamaremos ENTREGABLE.
#
# =============================================================================


entregables = pd.DataFrame({

    "Fase": plan["Fase"],

    "Entregable": [

        "Base de datos validada y línea base de KPIs",

        "Shortlist de proveedores potenciales",

        "Matriz de evaluación y proveedor homologado",

        "Resultados del piloto 70/30",

        "Informe de resultados y decisión de escalamiento"

    ]

})


print("\n" + "=" * 82)

print("ENTREGABLES")

print("=" * 82)


print(
    entregables.to_string(
        index=False
    )
)


# =============================================================================
# 11. CONSTRUIR MATRIZ RACI SIMPLIFICADA
# =============================================================================
#
# RACI es una herramienta útil para definir responsabilidades.
#
#
# R = RESPONSIBLE
#     Ejecuta la actividad.
#
# A = ACCOUNTABLE
#     Tiene la responsabilidad final.
#
# C = CONSULTED
#     Debe ser consultado.
#
# I = INFORMED
#     Debe ser informado.
#
#
# Para mantener el ejemplo sencillo utilizaremos:
#
# Compras
# Planeación
# Calidad
# Finanzas
# Operaciones
#
# =============================================================================


raci = pd.DataFrame({

    "Actividad": [

        "Validar datos",

        "Buscar proveedor",

        "Evaluar proveedor",

        "Ejecutar piloto",

        "Evaluar resultados"

    ],

    "Compras": [

        "C",

        "R/A",

        "R",

        "R",

        "C"

    ],

    "Planeación": [

        "R/A",

        "C",

        "C",

        "R",

        "R"

    ],

    "Calidad": [

        "I",

        "C",

        "R/A",

        "C",

        "C"

    ],

    "Finanzas": [

        "C",

        "C",

        "C",

        "I",

        "C"

    ],

    "Operaciones": [

        "C",

        "I",

        "C",

        "R/A",

        "R/A"

    ]

})


print("\n" + "=" * 82)

print("MATRIZ RACI SIMPLIFICADA")

print("=" * 82)


print(
    raci.to_string(
        index=False
    )
)


# =============================================================================
# 12. IDENTIFICAR RIESGOS DE IMPLEMENTACIÓN
# =============================================================================
#
# Implementar una solución también genera riesgos.
#
# Vamos a evaluar:
#
# PROBABILIDAD
#
# e
#
# IMPACTO.
#
#
# Utilizaremos una escala:
#
# 1 = Bajo
# 2 = Medio
# 3 = Alto
# 4 = Muy alto
# 5 = Crítico
#
#
# NIVEL DE RIESGO:
#
# RIESGO = PROBABILIDAD × IMPACTO
#
#
# Ejemplo:
#
# Probabilidad = 4
#
# Impacto = 5
#
# Riesgo = 20
#
# =============================================================================


riesgos = pd.DataFrame({

    "Riesgo": [

        "Proveedor alternativo no cumple calidad",

        "Costo total mayor al esperado",

        "Retraso en homologación",

        "Proveedor actual modifica condiciones",

        "Datos insuficientes para evaluar piloto",

        "Resistencia interna al cambio"

    ],

    "Probabilidad": [

        3,

        3,

        4,

        2,

        2,

        2

    ],

    "Impacto": [

        5,

        4,

        4,

        4,

        3,

        3

    ]

})


# Fórmula del nivel de riesgo.

riesgos["Nivel_riesgo"] = (

    riesgos["Probabilidad"]

    *

    riesgos["Impacto"]

)


# =============================================================================
# 13. CLASIFICAR LOS RIESGOS
# =============================================================================
#
# Crearemos una clasificación pedagógica:
#
# 1 - 4   = Bajo
#
# 5 - 9   = Medio
#
# 10 - 15 = Alto
#
# 16 - 25 = Crítico
#
# =============================================================================


def clasificar_riesgo(valor):

    if valor <= 4:

        return "Bajo"

    elif valor <= 9:

        return "Medio"

    elif valor <= 15:

        return "Alto"

    else:

        return "Crítico"


riesgos["Clasificacion"] = (

    riesgos["Nivel_riesgo"]

    .apply(
        clasificar_riesgo
    )

)


print("\n" + "=" * 82)

print("MATRIZ DE RIESGOS")

print("=" * 82)


print(

    riesgos

    .sort_values(
        "Nivel_riesgo",
        ascending=False
    )

    .to_string(
        index=False
    )

)


# =============================================================================
# 14. MAPA DE RIESGOS
# =============================================================================
#
# Eje X:
#
# Probabilidad.
#
# Eje Y:
#
# Impacto.
#
# Tamaño del punto:
#
# Nivel de riesgo.
#
# =============================================================================


plt.figure(
    figsize=(11, 8)
)


colores_riesgo = {

    "Bajo": "#43A047",

    "Medio": "#FDD835",

    "Alto": "#FB8C00",

    "Crítico": "#E53935"

}


for clasificacion in [
    "Bajo",
    "Medio",
    "Alto",
    "Crítico"
]:

    datos = riesgos[
        riesgos["Clasificacion"]
        ==
        clasificacion
    ]


    plt.scatter(

        datos["Probabilidad"],

        datos["Impacto"],

        s=datos["Nivel_riesgo"] * 50,

        color=colores_riesgo[
            clasificacion
        ],

        label=clasificacion,

        alpha=0.75,

        edgecolors="black"

    )


# Añadimos número identificador.

for i, fila in riesgos.iterrows():

    plt.text(

        fila["Probabilidad"] + 0.05,

        fila["Impacto"] + 0.05,

        f"R{i + 1}",

        fontsize=10,

        fontweight="bold"

    )


plt.xlim(
    0.5,
    5.5
)


plt.ylim(
    0.5,
    5.5
)


plt.xticks(
    range(1, 6)
)


plt.yticks(
    range(1, 6)
)


plt.xlabel(
    "Probabilidad"
)


plt.ylabel(
    "Impacto"
)


plt.title(

    "Mapa de riesgos de implementación",

    fontsize=17,

    fontweight="bold"

)


plt.legend(
    title="Clasificación"
)


plt.tight_layout()

plt.show()


print("""

INTERPRETACIÓN DEL MAPA DE RIESGOS:

Los riesgos ubicados hacia la parte superior derecha
requieren mayor atención porque combinan:

ALTA PROBABILIDAD

y

ALTO IMPACTO.

La propuesta no debería ocultar estos riesgos.

Una recomendación gerencial sólida debe explicar
también qué podría salir mal y cómo se controlará.

""")


# =============================================================================
# 15. DEFINIR ACCIONES DE MITIGACIÓN
# =============================================================================


mitigacion = pd.DataFrame({

    "Riesgo": riesgos["Riesgo"],

    "Accion_mitigacion": [

        "Pruebas técnicas y lote piloto antes de aprobar",

        "Evaluar TCO y establecer límites económicos",

        "Definir cronograma y seguimiento semanal",

        "Mantener negociación y contratos claros",

        "Definir KPIs y captura de datos antes del piloto",

        "Comunicación, capacitación y participación del equipo"

    ]

})


print("\n" + "=" * 82)

print("ACCIONES DE MITIGACIÓN")

print("=" * 82)


print(
    mitigacion.to_string(
        index=False
    )
)


# =============================================================================
# 16. SIMULAR SEGUIMIENTO DEL PILOTO
# =============================================================================
#
# Ahora imaginaremos que el piloto dura seis meses.
#
# ATENCIÓN:
#
# ESTOS DATOS SON SIMULADOS.
#
# Sirven únicamente para enseñar cómo podría
# monitorearse la implementación.
#
#
# Queremos observar:
#
# - nivel de servicio,
# - lead time,
# - dependencia del proveedor,
# - compras urgentes.
#
# =============================================================================


meses = [

    "Mes 1",

    "Mes 2",

    "Mes 3",

    "Mes 4",

    "Mes 5",

    "Mes 6"

]


seguimiento = pd.DataFrame({

    "Mes": meses,

    "Nivel_servicio": [

        90,

        91,

        92,

        94,

        95,

        96

    ],

    "Lead_time": [

        74,

        72,

        69,

        66,

        62,

        59

    ],

    "Dependencia_proveedor": [

        90,

        86,

        82,

        78,

        73,

        69

    ],

    "Compras_urgentes_USD": [

        4800,

        4200,

        3500,

        2700,

        1900,

        1400

    ]

})


print("\n" + "=" * 82)

print("DATOS SIMULADOS DEL PILOTO")

print("=" * 82)


print(
    seguimiento.to_string(
        index=False
    )
)


# =============================================================================
# 17. GRÁFICA DEL NIVEL DE SERVICIO
# =============================================================================


plt.figure(
    figsize=(12, 6)
)


plt.plot(

    seguimiento["Mes"],

    seguimiento["Nivel_servicio"],

    marker="o",

    linewidth=3,

    markersize=8,

    color="#1565C0",

    label="Nivel de servicio"

)


plt.axhline(

    nivel_servicio_meta,

    color="#43A047",

    linestyle="--",

    linewidth=2,

    label=f"Meta = {nivel_servicio_meta:.0f}%"

)


plt.title(

    "Seguimiento del nivel de servicio",

    fontsize=17,

    fontweight="bold"

)


plt.ylabel(
    "Nivel de servicio (%)"
)


plt.ylim(
    85,
    100
)


plt.legend()


plt.tight_layout()

plt.show()


print("""

INTERPRETACIÓN:

En esta SIMULACIÓN el nivel de servicio
presenta una tendencia creciente.

La línea punteada representa la meta.

La pregunta gerencial sería:

¿En qué momento el piloto alcanza
y posteriormente sostiene la meta?

No basta con alcanzarla durante un solo periodo.

Es importante verificar su estabilidad.

""")


# =============================================================================
# 18. GRÁFICA DEL LEAD TIME
# =============================================================================


plt.figure(
    figsize=(12, 6)
)


plt.plot(

    seguimiento["Mes"],

    seguimiento["Lead_time"],

    marker="o",

    linewidth=3,

    markersize=8,

    color="#8E24AA",

    label="Lead time"

)


plt.axhline(

    lead_time_meta,

    color="#43A047",

    linestyle="--",

    linewidth=2,

    label=f"Meta = {lead_time_meta:.0f} días"

)


plt.title(

    "Seguimiento del lead time",

    fontsize=17,

    fontweight="bold"

)


plt.ylabel(
    "Días"
)


plt.legend()


plt.tight_layout()

plt.show()


print("""

INTERPRETACIÓN:

Aquí la dirección deseada es DESCENDENTE.

Una reducción del lead time podría disminuir
la exposición de la empresa durante el periodo
de reposición.

Sin embargo, nuevamente debemos comprobar
si la reducción es sostenible y si no genera
un incremento excesivo del costo total.

""")


# =============================================================================
# 19. GRÁFICA DE DEPENDENCIA DEL PROVEEDOR
# =============================================================================


plt.figure(
    figsize=(12, 6)
)


plt.plot(

    seguimiento["Mes"],

    seguimiento["Dependencia_proveedor"],

    marker="o",

    linewidth=3,

    markersize=8,

    color="#E53935",

    label="Dependencia proveedor principal"

)


plt.axhline(

    dependencia_meta,

    color="#43A047",

    linestyle="--",

    linewidth=2,

    label=f"Meta = {dependencia_meta:.0f}%"

)


plt.title(

    "Evolución de la dependencia del proveedor principal",

    fontsize=17,

    fontweight="bold"

)


plt.ylabel(
    "Dependencia (%)"
)


plt.legend()


plt.tight_layout()

plt.show())


# =============================================================================
# 20. INTERPRETAR DEPENDENCIA
# =============================================================================


print("""

INTERPRETACIÓN:

Esta gráfica está directamente relacionada
con el propósito del dual sourcing.

La estrategia no necesariamente busca eliminar
al proveedor principal.

Busca disminuir una concentración excesiva.

En este ejemplo:

92 %

podría evolucionar hacia aproximadamente:

70 % proveedor principal

30 % proveedor alternativo.

Esto aumenta la diversificación,
aunque también aumenta la complejidad de gestión.

""")


# =============================================================================
# 21. SEMÁFORO DE KPIs
# =============================================================================
#
# Construiremos un sistema sencillo de seguimiento.
#
#
# VERDE:
# meta alcanzada.
#
#
# AMARILLO:
# avance importante, pero todavía no alcanza la meta.
#
#
# ROJO:
# desempeño todavía alejado de la meta.
#
#
# Para determinar AMARILLO utilizaremos
# un cumplimiento >= 80 % de la mejora requerida.
#
# =============================================================================


# Tomamos el último mes del piloto.

ultimo_mes = seguimiento.iloc[-1]


resultados_piloto = pd.DataFrame({

    "Indicador": [

        "Nivel de servicio",

        "Lead time",

        "Dependencia proveedor"

    ],

    "Inicial": [

        nivel_servicio_actual,

        lead_time_actual,

        dependencia_actual

    ],

    "Meta": [

        nivel_servicio_meta,

        lead_time_meta,

        dependencia_meta

    ],

    "Resultado": [

        ultimo_mes["Nivel_servicio"],

        ultimo_mes["Lead_time"],

        ultimo_mes["Dependencia_proveedor"]

    ],

    "Direccion": [

        "Aumentar",

        "Disminuir",

        "Disminuir"

    ]

})


# =============================================================================
# 22. FUNCIÓN DE CUMPLIMIENTO
# =============================================================================
#
# Para un indicador que debe AUMENTAR:
#
#
#                  Resultado - Inicial
# Cumplimiento = ------------------------- × 100
#                    Meta - Inicial
#
#
# Para un indicador que debe DISMINUIR:
#
#
#                   Inicial - Resultado
# Cumplimiento = ------------------------- × 100
#                    Inicial - Meta
#
#
# Ejemplo:
#
# Servicio:
#
# Inicial = 89
#
# Meta = 95
#
# Resultado = 92
#
#
# Cumplimiento:
#
# (92 - 89)
# ----------- × 100
# (95 - 89)
#
# =
#
# 50 %
#
# =============================================================================


def calcular_cumplimiento(
    inicial,
    meta,
    resultado,
    direccion
):

    if direccion == "Aumentar":

        denominador = (
            meta - inicial
        )

        if denominador == 0:

            return 100

        cumplimiento = (

            (
                resultado
                -
                inicial
            )

            /

            denominador

        ) * 100

    else:

        denominador = (
            inicial - meta
        )

        if denominador == 0:

            return 100

        cumplimiento = (

            (
                inicial
                -
                resultado
            )

            /

            denominador

        ) * 100


    # Permitimos valores mayores a 100
    # si se supera la meta.

    return cumplimiento


resultados_piloto[
    "Cumplimiento_%"
] = resultados_piloto.apply(

    lambda fila:

    calcular_cumplimiento(

        fila["Inicial"],

        fila["Meta"],

        fila["Resultado"],

        fila["Direccion"]

    ),

    axis=1

)


# =============================================================================
# 23. CREAR SEMÁFORO
# =============================================================================


def semaforo(cumplimiento):

    if cumplimiento >= 100:

        return "VERDE"

    elif cumplimiento >= 80:

        return "AMARILLO"

    else:

        return "ROJO"


resultados_piloto[
    "Semaforo"
] = (

    resultados_piloto[
        "Cumplimiento_%"
    ]

    .apply(
        semaforo
    )

)


print("\n" + "=" * 82)

print("SEMÁFORO DE RESULTADOS")

print("=" * 82)


print(

    resultados_piloto

    .round(1)

    .to_string(
        index=False
    )

)


# =============================================================================
# 24. GRÁFICA DEL SEMÁFORO
# =============================================================================


mapa_colores = {

    "VERDE": "#43A047",

    "AMARILLO": "#FDD835",

    "ROJO": "#E53935"

}


colores = [

    mapa_colores[x]

    for x in resultados_piloto[
        "Semaforo"
    ]

]


plt.figure(
    figsize=(11, 6)
)


plt.bar(

    resultados_piloto[
        "Indicador"
    ],

    resultados_piloto[
        "Cumplimiento_%"
    ],

    color=colores

)


plt.axhline(

    100,

    linestyle="--",

    color="#212121",

    label="Meta = 100% de cumplimiento"

)


plt.axhline(

    80,

    linestyle=":",

    color="#616161",

    label="Umbral amarillo = 80%"

)


plt.ylabel(
    "Cumplimiento de la mejora esperada (%)"
)


plt.title(

    "Semáforo de seguimiento del piloto",

    fontsize=17,

    fontweight="bold"

)


plt.legend()


plt.tight_layout()

plt.show()


print("""

INTERPRETACIÓN DEL SEMÁFORO:

VERDE no significa que el proyecto haya terminado.

Significa que el indicador alcanzó
la meta definida para el piloto.

AMARILLO indica que existe avance,
pero todavía debemos observar el indicador.

ROJO indica que el resultado se encuentra
por debajo del avance esperado.

El semáforo facilita la comunicación ejecutiva,
pero nunca debería sustituir el análisis detallado.

""")


# =============================================================================
# 25. EVALUAR EL IMPACTO ECONÓMICO
# =============================================================================
#
# Vamos a introducir un cálculo sencillo.
#
# Compras urgentes actuales:
#
# USD 58.000 / año.
#
#
# Meta:
#
# USD 20.000 / año.
#
#
# AHORRO POTENCIAL:
#
# Costo actual - costo esperado.
#
#
# IMPORTANTE:
#
# Esto es ahorro POTENCIAL.
#
# Todavía debemos restar los costos
# de implementar la alternativa.
#
# =============================================================================


ahorro_bruto = (

    compras_urgentes_actuales

    -

    compras_urgentes_meta

)


print(
    f"""

Compras urgentes actuales:

USD {compras_urgentes_actuales:,.0f}


Meta:

USD {compras_urgentes_meta:,.0f}


Reducción potencial:

USD {ahorro_bruto:,.0f}

"""
)


# =============================================================================
# 26. INCLUIR COSTO DE IMPLEMENTACIÓN
# =============================================================================
#
# Supongamos que:
#
# homologación,
# pruebas,
# auditoría,
# negociación,
# ajustes logísticos
#
# cuestan:
#
# USD 22.000.
#
# =============================================================================


costo_implementacion = 22000


beneficio_neto_estimado = (

    ahorro_bruto

    -

    costo_implementacion

)


print(
    f"""

Costo estimado de implementación:

USD {costo_implementacion:,.0f}


Beneficio neto estimado del primer año:

USD {beneficio_neto_estimado:,.0f}

"""
)


# =============================================================================
# 27. CALCULAR ROI
# =============================================================================
#
# Fórmula:
#
#
#        Beneficio neto
# ROI = ---------------- × 100
#          Inversión
#
#
# donde:
#
# Beneficio neto =
#
# Beneficios - inversión.
#
#
# ATENCIÓN:
#
# Es una estimación simplificada.
#
# Un análisis financiero real podría incluir:
#
# - flujo de caja,
# - costo de capital,
# - VAN,
# - TIR,
# - impuestos,
# - horizonte temporal,
# - costos recurrentes.
#
# =============================================================================


ROI = (

    beneficio_neto_estimado

    /

    costo_implementacion

) * 100


print(
    f"ROI estimado del ejemplo: {ROI:.1f}%"
)


# =============================================================================
# 28. GRÁFICA ECONÓMICA
# =============================================================================


categorias_economicas = [

    "Compras urgentes\nactuales",

    "Compras urgentes\nmeta",

    "Costo de\nimplementación",

    "Beneficio neto\nestimado"

]


valores_economicos = [

    compras_urgentes_actuales,

    compras_urgentes_meta,

    costo_implementacion,

    beneficio_neto_estimado

]


plt.figure(
    figsize=(12, 6)
)


plt.bar(

    categorias_economicas,

    valores_economicos,

    color=[

        "#E53935",

        "#F9A825",

        "#1565C0",

        "#43A047"

    ]

)


plt.ylabel(
    "USD"
)


plt.title(

    "Evaluación económica simplificada",

    fontsize=17,

    fontweight="bold"

)


for i, valor in enumerate(
    valores_economicos
):

    plt.text(

        i,

        valor + 1000,

        f"USD {valor:,.0f}",

        ha="center",

        fontweight="bold"

    )


plt.tight_layout()

plt.show()


print("""

INTERPRETACIÓN ECONÓMICA:

Una mejora operacional no debería evaluarse
únicamente desde indicadores logísticos.

También debemos preguntar:

¿CUÁNTO CUESTA IMPLEMENTARLA?

y:

¿QUÉ BENEFICIO ECONÓMICO PODRÍA GENERAR?

El ROI presentado aquí es una estimación didáctica.

En una empresa real debería construirse
con información financiera validada.

""")


# =============================================================================
# 29. CRITERIOS PARA DECIDIR SI ESCALAR
# =============================================================================
#
# Ahora definimos reglas de decisión.
#
# Por ejemplo:
#
# Escalar si:
#
# 1. Nivel de servicio >= 95 %
#
# 2. Lead time <= 60 días
#
# 3. Dependencia proveedor <= 70 %
#
# 4. No existen riesgos críticos sin control.
#
# 5. Beneficio económico esperado > 0.
#
# =============================================================================


condicion_servicio = (

    ultimo_mes["Nivel_servicio"]

    >=

    nivel_servicio_meta

)


condicion_lead_time = (

    ultimo_mes["Lead_time"]

    <=

    lead_time_meta

)


condicion_dependencia = (

    ultimo_mes["Dependencia_proveedor"]

    <=

    dependencia_meta

)


condicion_economica = (

    beneficio_neto_estimado

    >

    0

)


numero_riesgos_criticos = (

    riesgos[
        "Clasificacion"
    ]

    ==
    "Crítico"

).sum()


condicion_riesgos = (

    numero_riesgos_criticos

    ==

    0

)


# =============================================================================
# 30. TABLA DE CRITERIOS DE ESCALAMIENTO
# =============================================================================


criterios_escalamiento = pd.DataFrame({

    "Criterio": [

        "Nivel de servicio >= meta",

        "Lead time <= meta",

        "Dependencia <= meta",

        "Beneficio económico positivo",

        "Sin riesgos críticos"

    ],

    "Cumple": [

        condicion_servicio,

        condicion_lead_time,

        condicion_dependencia,

        condicion_economica,

        condicion_riesgos

    ]

})


print("\n" + "=" * 82)

print("CRITERIOS DE ESCALAMIENTO")

print("=" * 82)


print(
    criterios_escalamiento.to_string(
        index=False
    )
)


# =============================================================================
# 31. CALCULAR PORCENTAJE DE CRITERIOS CUMPLIDOS
# =============================================================================


porcentaje_cumplimiento = (

    criterios_escalamiento[
        "Cumple"
    ]

    .mean()

    * 100

)


print(
    f"\nCriterios cumplidos: "
    f"{porcentaje_cumplimiento:.1f}%"
)


# =============================================================================
# 32. REGLA FINAL DE DECISIÓN
# =============================================================================
#
# Para fines didácticos:
#
# 100 %:
# candidato a escalamiento.
#
# 80 % a <100 %:
# continuar piloto y corregir.
#
# <80 %:
# revisar la propuesta.
#
# =============================================================================


if porcentaje_cumplimiento == 100:

    decision_final = (
        "CANDIDATO A ESCALAMIENTO"
    )

elif porcentaje_cumplimiento >= 80:

    decision_final = (
        "CONTINUAR PILOTO Y AJUSTAR"
    )

else:

    decision_final = (
        "REVISAR LA PROPUESTA"
    )


print(
    f"\nDECISIÓN DEL EJEMPLO: "
    f"{decision_final}"
)


# =============================================================================
# 33. DASHBOARD FINAL
# =============================================================================
#
# Esta pantalla resume:
#
# - decisión,
# - resultados,
# - economía,
# - riesgos,
# - escalamiento.
#
# =============================================================================


fig, ax = plt.subplots(
    figsize=(16, 9)
)


ax.axis(
    "off"
)


# TÍTULO

ax.text(

    0.50,

    0.94,

    "GLOBALBIKE | FROM ANALYSIS TO ACTION",

    transform=ax.transAxes,

    ha="center",

    fontsize=21,

    fontweight="bold"

)


# ALTERNATIVA

ax.text(

    0.05,

    0.82,

    "ALTERNATIVA",

    transform=ax.transAxes,

    fontsize=13,

    fontweight="bold",

    color="#1565C0"

)


ax.text(

    0.05,

    0.75,

    alternativa_priorizada,

    transform=ax.transAxes,

    fontsize=14

)


# SERVICIO

ax.text(

    0.30,

    0.82,

    "SERVICIO",

    transform=ax.transAxes,

    fontsize=13,

    fontweight="bold",

    color="#43A047"

)


ax.text(

    0.30,

    0.75,

    f"{nivel_servicio_actual:.0f}% → "
    f"{ultimo_mes['Nivel_servicio']:.0f}%",

    transform=ax.transAxes,

    fontsize=16,

    fontweight="bold"

)


# LEAD TIME

ax.text(

    0.52,

    0.82,

    "LEAD TIME",

    transform=ax.transAxes,

    fontsize=13,

    fontweight="bold",

    color="#8E24AA"

)


ax.text(

    0.52,

    0.75,

    f"{lead_time_actual:.0f} → "
    f"{ultimo_mes['Lead_time']:.0f} días",

    transform=ax.transAxes,

    fontsize=16,

    fontweight="bold"

)


# DEPENDENCIA

ax.text(

    0.74,

    0.82,

    "DEPENDENCIA",

    transform=ax.transAxes,

    fontsize=13,

    fontweight="bold",

    color="#E53935"

)


ax.text(

    0.74,

    0.75,

    f"{dependencia_actual:.0f}% → "
    f"{ultimo_mes['Dependencia_proveedor']:.0f}%",

    transform=ax.transAxes,

    fontsize=16,

    fontweight="bold"

)


# ROI

ax.text(

    0.18,

    0.53,

    "ROI ESTIMADO",

    transform=ax.transAxes,

    fontsize=13,

    fontweight="bold",

    color="#F9A825"

)


ax.text(

    0.18,

    0.45,

    f"{ROI:.1f}%",

    transform=ax.transAxes,

    fontsize=26,

    fontweight="bold"

)


# RIESGOS

ax.text(

    0.45,

    0.53,

    "RIESGOS CRÍTICOS",

    transform=ax.transAxes,

    fontsize=13,

    fontweight="bold",

    color="#E53935"

)


ax.text(

    0.45,

    0.45,

    str(numero_riesgos_criticos),

    transform=ax.transAxes,

    fontsize=26,

    fontweight="bold"

)


# CRITERIOS

ax.text(

    0.70,

    0.53,

    "CRITERIOS CUMPLIDOS",

    transform=ax.transAxes,

    fontsize=13,

    fontweight="bold",

    color="#1565C0"

)


ax.text(

    0.70,

    0.45,

    f"{porcentaje_cumplimiento:.0f}%",

    transform=ax.transAxes,

    fontsize=26,

    fontweight="bold"

)


# DECISIÓN FINAL

ax.text(

    0.50,

    0.25,

    "DECISIÓN DE IMPLEMENTACIÓN",

    transform=ax.transAxes,

    ha="center",

    fontsize=15,

    fontweight="bold"

)


ax.text(

    0.50,

    0.16,

    decision_final,

    transform=ax.transAxes,

    ha="center",

    fontsize=23,

    fontweight="bold",

    color="#43A047"

)


plt.show()


# =============================================================================
# 34. GENERAR CONCLUSIÓN GERENCIAL AUTOMÁTICA
# =============================================================================


print("\n" + "=" * 82)

print("CONCLUSIÓN GERENCIAL")

print("=" * 82)


print(
    f"""

El proceso de análisis permitió transformar
un problema de inventarios y abastecimiento
en una propuesta estructurada de intervención.

La alternativa priorizada mediante AHP fue:

{alternativa_priorizada}

con una prioridad aproximada de:

{prioridad_ahp:.1f}%.


La propuesta se estructuró mediante un piloto
que incluye:

1. Validación de información.

2. Búsqueda de proveedores.

3. Evaluación y homologación.

4. Implementación piloto.

5. Medición de resultados.

6. Decisión de escalamiento.


En los datos SIMULADOS del piloto,
el nivel de servicio evolucionó desde:

{nivel_servicio_actual:.0f}%

hasta:

{ultimo_mes['Nivel_servicio']:.0f}%.


El lead time evolucionó desde:

{lead_time_actual:.0f} días

hasta:

{ultimo_mes['Lead_time']:.0f} días.


La dependencia del proveedor principal
evolucionó desde:

{dependencia_actual:.0f}%

hasta:

{ultimo_mes['Dependencia_proveedor']:.0f}%.


El beneficio neto estimado del escenario
es aproximadamente:

USD {beneficio_neto_estimado:,.0f}


con un ROI simplificado estimado de:

{ROI:.1f}%.


El porcentaje de criterios de escalamiento
cumplidos es:

{porcentaje_cumplimiento:.1f}%.


Por tanto, bajo los SUPUESTOS
y DATOS SIMULADOS del ejercicio,
la clasificación final sería:

{decision_final}


IMPORTANTE:

Esta conclusión NO demuestra que el dual sourcing
producirá necesariamente estos resultados.

Los valores representan un escenario académico
que debería validarse mediante datos reales
durante la implementación.

"""
)


# =============================================================================
# 35. EXPORTAR RESULTADOS
# =============================================================================
#
# Los estudiantes pueden descargar estos archivos
# desde Google Colab y posteriormente utilizarlos
# en:
#
# - Google Sheets,
# - Excel,
# - Power BI,
# - Looker Studio,
# - Google Sites.
#
# =============================================================================


plan.to_csv(

    "GlobalBike_Plan_Implementacion.csv",

    index=False,

    encoding="utf-8-sig"

)


kpis.to_csv(

    "GlobalBike_KPIs.csv",

    index=False,

    encoding="utf-8-sig"

)


riesgos.to_csv(

    "GlobalBike_Riesgos.csv",

    index=False,

    encoding="utf-8-sig"

)


seguimiento.to_csv(

    "GlobalBike_Seguimiento_Piloto.csv",

    index=False,

    encoding="utf-8-sig"

)


resultados_piloto.to_csv(

    "GlobalBike_Semaforo_KPIs.csv",

    index=False,

    encoding="utf-8-sig"

)


criterios_escalamiento.to_csv(

    "GlobalBike_Criterios_Escalamiento.csv",

    index=False,

    encoding="utf-8-sig"

)


print("""

ARCHIVOS GENERADOS:

1. GlobalBike_Plan_Implementacion.csv

2. GlobalBike_KPIs.csv

3. GlobalBike_Riesgos.csv

4. GlobalBike_Seguimiento_Piloto.csv

5. GlobalBike_Semaforo_KPIs.csv

6. GlobalBike_Criterios_Escalamiento.csv

""")


# =============================================================================
# 36. MAPA FINAL DE TODO EL CURSO
# =============================================================================


print("""

===============================================================================

            INVENTARIOS Y COMPRAS GLOBALES
             CASO INTEGRADOR GLOBALBIKE

===============================================================================


                   PROBLEMA EMPRESARIAL
                            │
                            ↓
                   EVIDENCIA CIENTÍFICA
                    SCOPUS + VOSVIEWER
                            │
                            ↓
                DIAGNÓSTICO DE INVENTARIOS
                            │
                            ↓
                  KPIs + SEGMENTACIÓN
                            │
                            ↓
                  PENSAMIENTO SISTÉMICO
                         VENSIM
                            │
                            ↓
                      DATA ANALYTICS
                         PYTHON
                            │
                            ↓
                    MACHINE LEARNING
                            │
                            ↓
                PREDICCIÓN DEL RIESGO
                            │
                            ↓
               ALTERNATIVAS DE SOLUCIÓN
                            │
                            ↓
                           AHP
                            │
                            ↓
                PRIORIZACIÓN MULTICRITERIO
                            │
                            ↓
                 DECISIÓN EMPRESARIAL
                            │
                            ↓
                   PLAN DE IMPLEMENTACIÓN
                            │
                            ↓
                          PILOTO
                            │
                            ↓
                   MONITOREO DE KPIs
                            │
                            ↓
                  EVALUACIÓN DE RIESGOS
                            │
                            ↓
                  DECISIÓN DE ESCALAMIENTO


===============================================================================

                DEL ANÁLISIS A LA ACCIÓN

===============================================================================

""")


# =============================================================================
# 37. MENSAJE FINAL PARA LOS ESTUDIANTES
# =============================================================================


print("""

MENSAJE FINAL:

El propósito de utilizar Python, Machine Learning,
Vensim, bibliometría o AHP no es utilizar
la mayor cantidad posible de herramientas.

Una herramienta agrega valor cuando ayuda
a responder una pregunta concreta.

En este caso:

VOSviewer ayudó a comprender
qué dice la evidencia científica.

Vensim ayudó a comprender
cómo interactúan las variables.

Python ayudó a comprender
qué muestran los datos.

Machine Learning ayudó a explorar
qué podría anticiparse.

Las alternativas permitieron transformar
los hallazgos en posibles acciones.

AHP ayudó a estructurar
una decisión multicriterio.

Y el plan de implementación permitió
convertir la decisión en acción.

Por eso la lógica completa es:

PROBLEMA

→ EVIDENCIA

→ ANÁLISIS

→ COMPRENSIÓN

→ PREDICCIÓN

→ ALTERNATIVAS

→ DECISIÓN

→ ACCIÓN

→ MEDICIÓN

→ APRENDIZAJE.

===============================================================================

FIN DEL CASO INTEGRADOR

===============================================================================

""")
```
