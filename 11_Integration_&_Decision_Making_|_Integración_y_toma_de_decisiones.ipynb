```python
# =============================================================================
# INVENTARIOS Y COMPRAS GLOBALES
# CASO INTEGRADOR: GLOBALBIKE COMPONENTS COLOMBIA S.A.S.
#
# ETAPA 11: INTEGRATION & DECISION MAKING
# INTEGRACIÓN DE RESULTADOS Y TOMA DE DECISIONES
# =============================================================================
#
# OBJETIVO DE ESTA ETAPA
# -----------------------------------------------------------------------------
#
# Hasta este momento hemos utilizado diferentes herramientas:
#
# PUNTO 7  -> DATA ANALYTICS
#
# PUNTO 8  -> MACHINE LEARNING
#
# PUNTO 9  -> ALTERNATIVAS DE SOLUCIÓN
#
# PUNTO 10 -> AHP
#
#
# El error que queremos evitar ahora es presentar cada análisis
# como si fuera una actividad independiente.
#
# El objetivo de esta etapa es responder:
#
# ¿CÓMO SE CONECTAN TODOS LOS RESULTADOS PARA SUSTENTAR
# UNA DECISIÓN GERENCIAL?
#
#
# La lógica será:
#
# PROBLEMA
#     ↓
# EVIDENCIA
#     ↓
# DIAGNÓSTICO
#     ↓
# PREDICCIÓN
#     ↓
# ALTERNATIVAS
#     ↓
# PRIORIZACIÓN
#     ↓
# DECISIÓN
#
#
# IMPORTANTE:
# -----------------------------------------------------------------------------
#
# Los valores utilizados en este código son ILUSTRATIVOS.
#
# El estudiante debe reemplazarlos por:
#
# - resultados de su diagnóstico,
# - resultados de sus KPIs,
# - resultados de Machine Learning,
# - alternativas de su proyecto,
# - pesos obtenidos mediante AHP,
# - prioridades obtenidas mediante AHP.
#
# NO se deben presentar datos simulados como evidencia real
# de una empresa.
#
# =============================================================================


# =============================================================================
# 1. IMPORTAR LIBRERÍAS
# =============================================================================

# pandas:
# permite crear y manipular tablas.

import pandas as pd


# numpy:
# permite realizar cálculos numéricos.

import numpy as np


# matplotlib:
# permite construir visualizaciones.

import matplotlib.pyplot as plt


# seaborn:
# facilita la construcción de gráficas estadísticas.

import seaborn as sns


# Configuración visual general.

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 6)

plt.rcParams["font.size"] = 11


# =============================================================================
# 2. DEFINIR EL PROBLEMA EMPRESARIAL
# =============================================================================
#
# Antes de mirar cualquier gráfica debemos recordar
# cuál era el problema.
#
# GlobalBike importa el controlador E-500.
#
# La empresa presenta:
#
# - variabilidad de demanda,
# - errores de pronóstico,
# - lead time internacional variable,
# - dependencia elevada del proveedor principal,
# - stockouts,
# - inventario elevado en algunos periodos.
#
#
# PREGUNTA GERENCIAL:
#
# ¿Qué estrategia de inventarios y abastecimiento global
# debería implementar GlobalBike para reducir el riesgo
# de stockout y mantener un nivel de servicio adecuado
# sin incrementar innecesariamente el costo total?
#
# =============================================================================


problema = """

GlobalBike presenta simultáneamente riesgo de stockout,
variabilidad del lead time, errores de pronóstico
y elevada dependencia del proveedor principal
para el controlador E-500.

"""


pregunta_gerencial = """

¿Qué estrategia de inventarios y abastecimiento global
debería implementar GlobalBike para reducir el riesgo
de stockout y mejorar el nivel de servicio,
considerando simultáneamente costo, riesgo,
tiempo de implementación y flexibilidad?

"""


print("\n" + "=" * 80)

print("GLOBALBIKE COMPONENTS | INTEGRATION & DECISION MAKING")

print("=" * 80)


print("\nPROBLEMA:")

print(problema)


print("\nPREGUNTA GERENCIAL:")

print(pregunta_gerencial)


# =============================================================================
# 3. CONSOLIDAR RESULTADOS DEL DATA ANALYTICS
# =============================================================================
#
# En el punto 7 respondimos principalmente:
#
# ¿QUÉ ESTÁ PASANDO?
#
#
# Para el ejemplo utilizaremos algunos indicadores
# representativos.
#
# ATENCIÓN:
#
# Estos valores son ILUSTRATIVOS.
#
# Si el estudiante ejecutó los códigos anteriores,
# debe reemplazarlos por sus resultados reales.
#
# =============================================================================


nivel_servicio_actual = 89

# Porcentaje de demanda atendida.


meta_nivel_servicio = 95

# Meta hipotética de la empresa.


mape_actual = 18.4

# Error porcentual absoluto medio del forecast.


lead_time_actual = 75

# Lead time promedio en días.


stockouts_anuales = 4

# Número aproximado de periodos con stockout.


dependencia_proveedor = 92

# Porcentaje del abastecimiento concentrado
# en el proveedor principal.


inventario_promedio = 1200

# Inventario promedio del controlador E-500.


# =============================================================================
# 4. CALCULAR LAS BRECHAS
# =============================================================================
#
# Una BRECHA permite comparar:
#
# SITUACIÓN ACTUAL
#
# contra
#
# SITUACIÓN DESEADA.
#
#
# Fórmula sencilla:
#
# BRECHA = META - RESULTADO ACTUAL
#
#
# Ejemplo:
#
# Meta servicio = 95 %
#
# Servicio actual = 89 %
#
# Brecha = 6 puntos porcentuales.
#
# =============================================================================


brecha_servicio = (

    meta_nivel_servicio
    -
    nivel_servicio_actual

)


print("\n" + "=" * 80)

print("RESULTADOS DEL DIAGNÓSTICO")

print("=" * 80)


print(
    f"""
Nivel de servicio actual:      {nivel_servicio_actual:.1f}%

Meta de servicio:              {meta_nivel_servicio:.1f}%

Brecha de servicio:            {brecha_servicio:.1f} puntos porcentuales

MAPE:                          {mape_actual:.1f}%

Lead time promedio:            {lead_time_actual:.0f} días

Stockouts aproximados/año:     {stockouts_anuales}

Dependencia proveedor:         {dependencia_proveedor:.0f}%

Inventario promedio:           {inventario_promedio:,.0f} unidades
"""
)


# =============================================================================
# 5. CREAR UN DASHBOARD DE KPIs
# =============================================================================
#
# Vamos a construir una gráfica tipo "tarjetas".
#
# Esto permite resumir rápidamente la situación actual.
#
# =============================================================================


indicadores = [

    "Nivel de\nservicio",

    "MAPE",

    "Lead time",

    "Stockouts\naño",

    "Dependencia\nproveedor"

]


valores_texto = [

    f"{nivel_servicio_actual:.0f}%",

    f"{mape_actual:.1f}%",

    f"{lead_time_actual:.0f} días",

    f"{stockouts_anuales}",

    f"{dependencia_proveedor:.0f}%"

]


colores_kpi = [

    "#E53935",

    "#F9A825",

    "#1565C0",

    "#E53935",

    "#8E24AA"

]


fig, ax = plt.subplots(
    figsize=(15, 4)
)


ax.axis(
    "off"
)


for i in range(
    len(indicadores)
):

    # Dibujamos un rectángulo que funciona
    # como tarjeta visual.

    rectangulo = plt.Rectangle(

        (i * 0.2 + 0.01, 0.15),

        0.18,

        0.70,

        transform=ax.transAxes,

        color=colores_kpi[i],

        alpha=0.90

    )


    ax.add_patch(
        rectangulo
    )


    # Valor principal.

    ax.text(

        i * 0.2 + 0.10,

        0.58,

        valores_texto[i],

        transform=ax.transAxes,

        ha="center",

        va="center",

        fontsize=20,

        fontweight="bold",

        color="white"

    )


    # Nombre del indicador.

    ax.text(

        i * 0.2 + 0.10,

        0.32,

        indicadores[i],

        transform=ax.transAxes,

        ha="center",

        va="center",

        fontsize=10,

        color="white"

    )


plt.title(

    "GlobalBike | Situación actual del controlador E-500",

    fontsize=17,

    fontweight="bold",

    pad=15

)


plt.show()


# =============================================================================
# 6. INTERPRETACIÓN DEL DIAGNÓSTICO
# =============================================================================


print("""

INTERPRETACIÓN DEL DIAGNÓSTICO:

Los indicadores no deberían analizarse de manera independiente.

Por ejemplo:

Un nivel de servicio inferior a la meta podría estar relacionado
con diferentes condiciones:

- disponibilidad insuficiente,
- errores del forecast,
- variabilidad de la demanda,
- lead time elevado,
- retrasos del proveedor.

De la misma forma, simplemente aumentar inventario
podría mejorar el servicio, pero también incrementar
el capital inmovilizado.

Por eso el problema requiere una visión integrada.

""")


# =============================================================================
# 7. INCORPORAR LOS RESULTADOS DE MACHINE LEARNING
# =============================================================================
#
# En el punto 8 cambiamos la pregunta:
#
# ¿QUÉ ESTÁ PASANDO?
#
#             ↓
#
# ¿PODEMOS ANTICIPAR EL RIESGO?
#
#
# Supongamos que el modelo Random Forest obtuvo
# los siguientes resultados ilustrativos.
#
# IMPORTANTE:
#
# Deben reemplazarse por los resultados obtenidos
# realmente al ejecutar el modelo.
#
# =============================================================================


accuracy = 0.82

precision = 0.78

recall = 0.84

f1 = 0.81

auc = 0.86


# =============================================================================
# 8. VARIABLES IMPORTANTES DEL MODELO
# =============================================================================
#
# Para fines didácticos utilizaremos un perfil ilustrativo.
#
# Los porcentajes deben sumar 100 %.
#
# IMPORTANTE:
#
# IMPORTANCIA PREDICTIVA NO SIGNIFICA CAUSALIDAD.
#
# =============================================================================


importancia_variables = pd.DataFrame({

    "Variable": [

        "Inventario inicial",

        "Lead time",

        "Error histórico forecast",

        "Pronóstico",

        "Riesgo proveedor",

        "APE histórico"

    ],

    "Importancia": [

        0.28,

        0.24,

        0.18,

        0.13,

        0.10,

        0.07

    ]

})


# Ordenamos de mayor a menor.

importancia_variables = (

    importancia_variables

    .sort_values(
        "Importancia",
        ascending=False
    )

)


# =============================================================================
# 9. GRÁFICA DEL MACHINE LEARNING
# =============================================================================


plt.figure(
    figsize=(11, 6)
)


sns.barplot(

    data=importancia_variables,

    x="Importancia",

    y="Variable",

    hue="Variable",

    palette="viridis",

    legend=False

)


plt.title(

    "Machine Learning | Variables con mayor importancia predictiva",

    fontsize=16,

    fontweight="bold"

)


plt.xlabel(
    "Importancia relativa"
)


plt.ylabel(
    ""
)


plt.tight_layout()

plt.show()


print(
    f"""

INTERPRETACIÓN DEL MACHINE LEARNING:

El modelo obtuvo, para este ejemplo:

Accuracy:  {accuracy:.1%}

Precision: {precision:.1%}

Recall:    {recall:.1%}

F1-score:  {f1:.1%}

AUC:       {auc:.2f}


Las variables con mayor importancia predictiva
aparecen en la gráfica.

Esto permite pasar de una gestión puramente reactiva
a una lógica de ALERTA.

Sin embargo:

MACHINE LEARNING NO SELECCIONA AUTOMÁTICAMENTE
LA ESTRATEGIA DE INVENTARIOS O COMPRAS.

El modelo ayuda a identificar riesgo.

La decisión sigue requiriendo evaluación gerencial.

"""
)


# =============================================================================
# 10. CONECTAR DATA ANALYTICS CON MACHINE LEARNING
# =============================================================================
#
# Ahora vamos a mostrar explícitamente
# qué aporta cada herramienta.
#
# =============================================================================


integracion_herramientas = pd.DataFrame({

    "Herramienta": [

        "Data Analytics",

        "Machine Learning",

        "Alternativas",

        "AHP"

    ],

    "Pregunta": [

        "¿Qué está pasando?",

        "¿Podemos anticipar el riesgo?",

        "¿Qué podemos hacer?",

        "¿Qué alternativa priorizamos?"

    ],

    "Resultado": [

        "Diagnóstico y KPIs",

        "Probabilidad de stockout",

        "Opciones de intervención",

        "Prioridad multicriterio"

    ]

})


print("\n" + "=" * 80)

print("INTEGRACIÓN DE HERRAMIENTAS")

print("=" * 80)


print(

    integracion_herramientas.to_string(
        index=False
    )

)


# =============================================================================
# 11. RECUPERAR LAS TRES ALTERNATIVAS
# =============================================================================
#
# En el punto 9 construimos:
#
# A1 = Stock de seguridad.
#
# A2 = Dual sourcing.
#
# A3 = Reabastecimiento predictivo.
#
# =============================================================================


alternativas = [

    "A1 - Stock de seguridad",

    "A2 - Dual sourcing",

    "A3 - Reabastecimiento predictivo"

]


# =============================================================================
# 12. RECORDAR QUÉ PROBLEMA ATACA CADA ALTERNATIVA
# =============================================================================


matriz_soluciones = pd.DataFrame({

    "Alternativa": alternativas,

    "Mecanismo": [

        "Protección mediante inventario",

        "Diversificación del abastecimiento",

        "Anticipación mediante datos"

    ],

    "Problema_principal": [

        "Variabilidad e incertidumbre",

        "Dependencia del proveedor",

        "Detección tardía del riesgo"

    ],

    "Beneficio": [

        "Mayor disponibilidad",

        "Mayor resiliencia",

        "Mayor anticipación"

    ],

    "Trade_off": [

        "Mayor inventario",

        "Mayor complejidad",

        "Mayor exigencia analítica"

    ]

})


print("\n" + "=" * 80)

print("ALTERNATIVAS DE SOLUCIÓN")

print("=" * 80)


print(

    matriz_soluciones.to_string(
        index=False
    )

)


# =============================================================================
# 13. RESULTADOS DEL AHP
# =============================================================================
#
# Ahora incorporaremos los resultados del punto 10.
#
# IMPORTANTE:
#
# Estos valores son ilustrativos.
#
# Si se ejecuta el código AHP anterior,
# deben utilizarse los valores realmente obtenidos.
#
#
# Supongamos:
#
# A1 = 24 %
#
# A2 = 43 %
#
# A3 = 33 %
#
#
# Los tres deben sumar:
#
# 100 %.
#
# =============================================================================


resultado_ahp = pd.DataFrame({

    "Alternativa": alternativas,

    "Prioridad_AHP": [

        0.24,

        0.43,

        0.33

    ]

})


resultado_ahp[
    "Prioridad_%"
] = (

    resultado_ahp[
        "Prioridad_AHP"
    ]

    * 100

)


# =============================================================================
# 14. VERIFICAR QUE LAS PRIORIDADES SUMEN 100 %
# =============================================================================
#
# Esta pequeña validación evita errores.
#
# =============================================================================


suma_prioridades = (

    resultado_ahp[
        "Prioridad_AHP"
    ]

    .sum()

)


print(
    f"\nSuma prioridades AHP = "
    f"{suma_prioridades:.2f}"
)


if np.isclose(
    suma_prioridades,
    1.0
):

    print(
        "Validación correcta: las prioridades suman 100%."
    )

else:

    print(
        "ATENCIÓN: revisar las prioridades."
    )


# =============================================================================
# 15. GRÁFICA AHP
# =============================================================================


resultado_ahp_ordenado = (

    resultado_ahp

    .sort_values(
        "Prioridad_%",
        ascending=False
    )

)


plt.figure(
    figsize=(11, 6)
)


plt.bar(

    resultado_ahp_ordenado[
        "Alternativa"
    ],

    resultado_ahp_ordenado[
        "Prioridad_%"
    ],

    color=[

        "#43A047",

        "#1565C0",

        "#8E24AA"

    ]

)


plt.title(

    "Resultado multicriterio | AHP",

    fontsize=16,

    fontweight="bold"

)


plt.ylabel(
    "Prioridad global (%)"
)


plt.xlabel(
    ""
)


plt.xticks(
    rotation=10
)


for i, valor in enumerate(

    resultado_ahp_ordenado[
        "Prioridad_%"
    ]

):

    plt.text(

        i,

        valor + 1,

        f"{valor:.1f}%",

        ha="center",

        fontweight="bold"

    )


plt.tight_layout()

plt.show()


# =============================================================================
# 16. IDENTIFICAR LA ALTERNATIVA PRIORIZADA
# =============================================================================


alternativa_priorizada = (

    resultado_ahp_ordenado

    .iloc[0][
        "Alternativa"
    ]

)


prioridad_alternativa = (

    resultado_ahp_ordenado

    .iloc[0][
        "Prioridad_%"
    ]

)


print(
    f"""

RESULTADO AHP:

La alternativa con mayor prioridad
en este escenario académico es:

{alternativa_priorizada}

Prioridad:

{prioridad_alternativa:.1f}%


ATENCIÓN:

Esto NO significa:

"Python decidió que esta es la mejor alternativa."

Significa:

"Dadas las comparaciones realizadas,
los criterios utilizados y sus pesos,
esta alternativa obtuvo la mayor prioridad."

"""
)


# =============================================================================
# 17. INTEGRAR TODA LA EVIDENCIA
# =============================================================================
#
# Ahora construiremos una MATRIZ DE EVIDENCIAS.
#
# Esta tabla es muy importante porque obliga
# a conectar cada resultado con una implicación gerencial.
#
# =============================================================================


matriz_evidencia = pd.DataFrame({

    "Evidencia": [

        "Nivel de servicio",

        "Forecast",

        "Lead time",

        "Dependencia proveedor",

        "Machine Learning",

        "AHP"

    ],

    "Resultado": [

        f"{nivel_servicio_actual}% frente a meta de {meta_nivel_servicio}%",

        f"MAPE = {mape_actual}%",

        f"{lead_time_actual} días promedio",

        f"{dependencia_proveedor}% proveedor principal",

        "Inventario y lead time aportan información predictiva",

        f"{alternativa_priorizada} = {prioridad_alternativa:.1f}%"

    ],

    "Interpretacion": [

        "Existe una brecha de servicio",

        "Existe incertidumbre en el pronóstico",

        "Existe exposición al tiempo de reposición",

        "Existe concentración del abastecimiento",

        "El riesgo puede anticiparse parcialmente",

        "La decisión multicriterio prioriza una alternativa"

    ]

})


print("\n" + "=" * 80)

print("MATRIZ DE EVIDENCIAS")

print("=" * 80)


print(

    matriz_evidencia.to_string(
        index=False
    )

)


# =============================================================================
# 18. UNA IDEA MUY IMPORTANTE:
# EVIDENCIA ≠ DECISIÓN
# =============================================================================
#
# Un indicador por sí solo no constituye
# una recomendación.
#
#
# EJEMPLO INCORRECTO:
#
# "El lead time es 75 días,
# por tanto debemos implementar dual sourcing."
#
#
# Falta razonamiento.
#
#
# UNA LÓGICA MÁS COMPLETA SERÍA:
#
# Lead time variable
#
#       +
#
# Dependencia del proveedor
#
#       +
#
# Stockouts
#
#       +
#
# Importancia del riesgo en AHP
#
#       ↓
#
# justifican EVALUAR una estrategia
# que reduzca la exposición del abastecimiento.
#
#
# Esta diferencia es fundamental
# para un proyecto de posgrado.
#
# =============================================================================


# =============================================================================
# 19. CONSTRUIR LA CADENA DE RAZONAMIENTO
# =============================================================================


cadena_razonamiento = pd.DataFrame({

    "Etapa": [

        "1. Problema",

        "2. Diagnóstico",

        "3. Predicción",

        "4. Alternativas",

        "5. Priorización",

        "6. Decisión"

    ],

    "Pregunta": [

        "¿Qué necesita resolver la empresa?",

        "¿Qué muestran los datos?",

        "¿Podemos anticipar el riesgo?",

        "¿Qué opciones existen?",

        "¿Cómo se comparan?",

        "¿Qué debería evaluarse para implementar?"

    ],

    "Resultado": [

        "Stockouts + costo + riesgo",

        "KPIs y patrones",

        "Probabilidad de stockout",

        "A1, A2 y A3",

        "Resultado AHP",

        alternativa_priorizada

    ]

})


print("\n" + "=" * 80)

print("CADENA DE RAZONAMIENTO")

print("=" * 80)


print(

    cadena_razonamiento.to_string(
        index=False
    )

)


# =============================================================================
# 20. CREAR UNA VISUALIZACIÓN DE LA CADENA DE DECISIÓN
# =============================================================================
#
# Construiremos un diagrama sencillo utilizando matplotlib.
#
# =============================================================================


etapas = [

    "PROBLEMA",

    "DIAGNÓSTICO",

    "MACHINE\nLEARNING",

    "ALTERNATIVAS",

    "AHP",

    "DECISIÓN"

]


colores_etapas = [

    "#E53935",

    "#F9A825",

    "#1565C0",

    "#00897B",

    "#8E24AA",

    "#43A047"

]


fig, ax = plt.subplots(
    figsize=(16, 5)
)


ax.axis(
    "off"
)


# Posiciones horizontales.

posiciones = np.linspace(
    0.08,
    0.92,
    len(etapas)
)


for i, (
    x,
    etapa,
    color
) in enumerate(
    zip(
        posiciones,
        etapas,
        colores_etapas
    )
):

    # Círculo.

    circulo = plt.Circle(

        (x, 0.55),

        0.065,

        transform=ax.transAxes,

        color=color,

        alpha=0.95

    )


    ax.add_patch(
        circulo
    )


    # Número de etapa.

    ax.text(

        x,

        0.55,

        str(i + 1),

        transform=ax.transAxes,

        ha="center",

        va="center",

        fontsize=16,

        fontweight="bold",

        color="white"

    )


    # Nombre.

    ax.text(

        x,

        0.32,

        etapa,

        transform=ax.transAxes,

        ha="center",

        va="center",

        fontsize=11,

        fontweight="bold"

    )


    # Flecha hacia la siguiente etapa.

    if i < len(etapas) - 1:

        ax.annotate(

            "",

            xy=(
                posiciones[i + 1] - 0.07,
                0.55
            ),

            xytext=(
                x + 0.07,
                0.55
            ),

            xycoords=ax.transAxes,

            textcoords=ax.transAxes,

            arrowprops=dict(

                arrowstyle="->",

                linewidth=2,

                color="#616161"

            )

        )


plt.title(

    "De los datos a la decisión | GlobalBike Components",

    fontsize=18,

    fontweight="bold"

)


plt.show()


# =============================================================================
# 21. DEFINIR INDICADORES ESPERADOS
# =============================================================================
#
# Una recomendación empresarial no debería terminar diciendo:
#
# "Implementar alternativa X".
#
# También debería establecer:
#
# ¿QUÉ ESPERAMOS MEJORAR?
#
#
# Construiremos una tabla:
#
# ACTUAL
#
# vs
#
# META.
#
#
# IMPORTANTE:
#
# Las metas siguientes son HIPÓTESIS DEL CASO.
#
# No deben presentarse como resultados ya alcanzados.
#
# =============================================================================


indicadores_transformacion = pd.DataFrame({

    "Indicador": [

        "Nivel de servicio (%)",

        "Stockouts por año",

        "Lead time promedio (días)",

        "Dependencia proveedor principal (%)",

        "MAPE (%)"

    ],

    "Actual": [

        89,

        4,

        75,

        92,

        18.4

    ],

    "Meta": [

        95,

        1,

        60,

        70,

        12

    ],

    "Direccion_deseada": [

        "Aumentar",

        "Disminuir",

        "Disminuir",

        "Disminuir",

        "Disminuir"

    ]

})


print("\n" + "=" * 80)

print("INDICADORES DE TRANSFORMACIÓN")

print("=" * 80)


print(

    indicadores_transformacion.to_string(
        index=False
    )

)


# =============================================================================
# 22. CALCULAR LA MEJORA REQUERIDA
# =============================================================================
#
# Para indicadores que deben aumentar:
#
# Mejora requerida =
#
# (Meta - Actual) / Actual × 100
#
#
# Para indicadores que deben disminuir:
#
# Reducción requerida =
#
# (Actual - Meta) / Actual × 100
#
#
# Esto permite expresar la magnitud
# del cambio esperado.
#
# =============================================================================


def calcular_mejora(
    actual,
    meta,
    direccion
):

    if direccion == "Aumentar":

        return (

            (
                meta
                -
                actual
            )

            /

            actual

        ) * 100

    else:

        return (

            (
                actual
                -
                meta
            )

            /

            actual

        ) * 100


indicadores_transformacion[
    "Cambio_requerido_%"
] = indicadores_transformacion.apply(

    lambda fila:

    calcular_mejora(

        fila["Actual"],

        fila["Meta"],

        fila["Direccion_deseada"]

    ),

    axis=1

)


print("\nCAMBIO REQUERIDO:\n")


print(

    indicadores_transformacion[
        [
            "Indicador",
            "Cambio_requerido_%"
        ]
    ]

    .round(1)

    .to_string(
        index=False
    )

)


# =============================================================================
# 23. GRÁFICA ACTUAL VS META
# =============================================================================
#
# Como los indicadores utilizan unidades diferentes,
# no conviene poner directamente:
#
# 89 %
# 4 stockouts
# 75 días
#
# en una misma escala.
#
# Por eso construiremos un ÍNDICE NORMALIZADO.
#
#
# ACTUAL = 100
#
# y mostraremos qué representa la meta
# respecto a la situación actual.
#
# =============================================================================


indice_actual = np.ones(
    len(indicadores_transformacion)
) * 100


indice_meta = []


for _, fila in indicadores_transformacion.iterrows():

    # Si queremos aumentar:
    #
    # meta / actual × 100

    if fila["Direccion_deseada"] == "Aumentar":

        valor = (

            fila["Meta"]

            /

            fila["Actual"]

        ) * 100


    # Si queremos disminuir:
    #
    # también mostramos meta / actual × 100.
    #
    # Un valor inferior a 100 representa reducción.

    else:

        valor = (

            fila["Meta"]

            /

            fila["Actual"]

        ) * 100


    indice_meta.append(
        valor
    )


x = np.arange(
    len(
        indicadores_transformacion
    )
)


ancho = 0.35


plt.figure(
    figsize=(14, 7)
)


plt.bar(

    x - ancho / 2,

    indice_actual,

    width=ancho,

    label="Situación actual = 100",

    color="#757575"

)


plt.bar(

    x + ancho / 2,

    indice_meta,

    width=ancho,

    label="Meta relativa",

    color="#43A047"

)


plt.axhline(

    100,

    linestyle="--",

    color="#212121",

    linewidth=1

)


plt.xticks(

    x,

    indicadores_transformacion[
        "Indicador"
    ],

    rotation=20

)


plt.ylabel(
    "Índice relativo"
)


plt.title(

    "Situación actual vs meta de transformación",

    fontsize=16,

    fontweight="bold"

)


plt.legend()


plt.tight_layout()

plt.show()


print("""

INTERPRETACIÓN:

La gráfica no pretende comparar directamente
unidades diferentes.

La situación actual se transforma en un índice = 100.

De esta manera podemos visualizar
la magnitud relativa del cambio esperado.

IMPORTANTE:

Las metas son objetivos.

NO son resultados garantizados.

La empresa tendría que implementar,
medir y posteriormente comprobar
si realmente se alcanzaron.

""")


# =============================================================================
# 24. CONSTRUIR UNA MATRIZ:
# EVIDENCIA → IMPLICACIÓN → DECISIÓN
# =============================================================================
#
# Esta puede ser una de las tablas más útiles
# para los estudiantes.
#
# =============================================================================


evidencia_decision = pd.DataFrame({

    "Evidencia": [

        "Nivel de servicio por debajo de meta",

        "MAPE relativamente elevado",

        "Lead time internacional variable",

        "Dependencia del proveedor principal",

        "Modelo predictivo identifica señales de riesgo",

        "AHP prioriza una alternativa"

    ],

    "Implicacion": [

        "Existe riesgo de demanda no atendida",

        "La planeación contiene incertidumbre",

        "La reposición está expuesta a variabilidad",

        "Existe vulnerabilidad ante interrupciones",

        "Es posible anticipar algunos periodos críticos",

        "Los criterios estratégicos favorecen una opción"

    ],

    "Uso_en_la_decision": [

        "Proteger disponibilidad",

        "Fortalecer forecasting",

        "Revisar política de abastecimiento",

        "Evaluar diversificación",

        "Incorporar alertas al proceso",

        "Definir alternativa para piloto"

    ]

})


print("\n" + "=" * 80)

print("EVIDENCIA → IMPLICACIÓN → DECISIÓN")

print("=" * 80)


print(

    evidencia_decision.to_string(
        index=False
    )

)


# =============================================================================
# 25. DEFINIR LA DECISIÓN COMO HIPÓTESIS DE IMPLEMENTACIÓN
# =============================================================================
#
# Aquí debemos ser cuidadosos.
#
# El AHP priorizó una alternativa.
#
# Pero eso NO significa que mañana
# debamos implementarla al 100 %.
#
#
# Una decisión prudente puede ser:
#
# "DESARROLLAR UN PILOTO"
#
# para validar:
#
# - costos,
# - resultados,
# - impacto,
# - riesgos,
# - capacidad de implementación.
#
# =============================================================================


decision = f"""

Desarrollar un piloto de:

{alternativa_priorizada}

para el controlador E-500,

manteniendo seguimiento sobre:

- nivel de servicio,
- stockouts,
- costo total,
- lead time,
- dependencia del proveedor,
- desempeño del forecast.

"""


print("\n" + "=" * 80)

print("DECISIÓN PROPUESTA PARA EL EJERCICIO")

print("=" * 80)


print(decision)


# =============================================================================
# 26. ¿POR QUÉ UN PILOTO?
# =============================================================================


print("""

¿POR QUÉ HABLAMOS DE PILOTO?

Porque los análisis anteriores apoyan una decisión,
pero todavía existen supuestos.

Por ejemplo:

AHP utiliza juicios.

Machine Learning utiliza patrones históricos.

Las metas son estimaciones.

Los escenarios contienen supuestos.

Por tanto:

ANÁLISIS
    ↓
DECISIÓN
    ↓
PILOTO
    ↓
MEDICIÓN
    ↓
APRENDIZAJE
    ↓
ESCALAMIENTO

es una secuencia más prudente que:

ANÁLISIS
    ↓
IMPLEMENTAR TODO INMEDIATAMENTE.

""")


# =============================================================================
# 27. CONSTRUIR UN DASHBOARD EJECUTIVO FINAL
# =============================================================================
#
# Vamos a combinar:
#
# - situación actual,
# - Machine Learning,
# - AHP,
# - decisión.
#
#
# Esta gráfica puede utilizarse como una lámina
# de cierre del análisis.
#
# =============================================================================


fig, ax = plt.subplots(
    figsize=(16, 9)
)


ax.axis(
    "off"
)


# -------------------------------------------------------------------------
# TÍTULO
# -------------------------------------------------------------------------


ax.text(

    0.5,

    0.94,

    "GLOBALBIKE COMPONENTS | EXECUTIVE DECISION DASHBOARD",

    transform=ax.transAxes,

    ha="center",

    fontsize=20,

    fontweight="bold"

)


# -------------------------------------------------------------------------
# BLOQUE 1: PROBLEMA
# -------------------------------------------------------------------------


ax.text(

    0.03,

    0.82,

    "1. PROBLEMA",

    transform=ax.transAxes,

    fontsize=14,

    fontweight="bold",

    color="#E53935"

)


ax.text(

    0.03,

    0.72,

    f"""
Servicio: {nivel_servicio_actual:.0f}%
Stockouts/año: {stockouts_anuales}
Lead time: {lead_time_actual:.0f} días
Dependencia proveedor: {dependencia_proveedor:.0f}%
""",

    transform=ax.transAxes,

    fontsize=12

)


# -------------------------------------------------------------------------
# BLOQUE 2: MACHINE LEARNING
# -------------------------------------------------------------------------


ax.text(

    0.28,

    0.82,

    "2. MACHINE LEARNING",

    transform=ax.transAxes,

    fontsize=14,

    fontweight="bold",

    color="#1565C0"

)


ax.text(

    0.28,

    0.72,

    f"""
Accuracy: {accuracy:.0%}
Recall: {recall:.0%}
AUC: {auc:.2f}

Uso:
anticipar riesgo
""",

    transform=ax.transAxes,

    fontsize=12

)


# -------------------------------------------------------------------------
# BLOQUE 3: ALTERNATIVAS
# -------------------------------------------------------------------------


ax.text(

    0.53,

    0.82,

    "3. ALTERNATIVAS",

    transform=ax.transAxes,

    fontsize=14,

    fontweight="bold",

    color="#00897B"

)


ax.text(

    0.53,

    0.72,

    """
A1 Stock seguridad
A2 Dual sourcing
A3 Reabastecimiento predictivo
""",

    transform=ax.transAxes,

    fontsize=12

)


# -------------------------------------------------------------------------
# BLOQUE 4: AHP
# -------------------------------------------------------------------------


ax.text(

    0.78,

    0.82,

    "4. AHP",

    transform=ax.transAxes,

    fontsize=14,

    fontweight="bold",

    color="#8E24AA"

)


ax.text(

    0.78,

    0.72,

    f"""
Mayor prioridad:

{alternativa_priorizada}

{prioridad_alternativa:.1f}%
""",

    transform=ax.transAxes,

    fontsize=12

)


# -------------------------------------------------------------------------
# FLECHA
# -------------------------------------------------------------------------


ax.annotate(

    "",

    xy=(0.50, 0.48),

    xytext=(0.50, 0.62),

    xycoords=ax.transAxes,

    arrowprops=dict(

        arrowstyle="->",

        linewidth=3,

        color="#616161"

    )

)


# -------------------------------------------------------------------------
# DECISIÓN
# -------------------------------------------------------------------------


ax.text(

    0.50,

    0.40,

    "DECISIÓN INTEGRADA",

    transform=ax.transAxes,

    ha="center",

    fontsize=17,

    fontweight="bold",

    color="#43A047"

)


ax.text(

    0.50,

    0.30,

    f"""
Desarrollar un piloto de
{alternativa_priorizada}

y evaluar su impacto antes del escalamiento.
""",

    transform=ax.transAxes,

    ha="center",

    fontsize=14

)


# -------------------------------------------------------------------------
# KPIs
# -------------------------------------------------------------------------


ax.text(

    0.50,

    0.14,

    """
KPIs de seguimiento:
Nivel de servicio | Stockouts | Lead time | Costo total |
Dependencia proveedor | MAPE
""",

    transform=ax.transAxes,

    ha="center",

    fontsize=12,

    fontweight="bold"

)


plt.show()


# =============================================================================
# 28. GENERAR UNA NARRATIVA GERENCIAL AUTOMÁTICA
# =============================================================================
#
# Esta parte ayuda a los estudiantes a comprender
# cómo deberían explicar los resultados.
#
# NO deberían limitarse a mostrar gráficas.
#
# Deben construir una HISTORIA DE DECISIÓN.
#
# =============================================================================


print("\n" + "=" * 80)

print("NARRATIVA GERENCIAL DEL CASO")

print("=" * 80)


print(
    f"""

1. PROBLEMA

GlobalBike presenta una situación en la que el controlador E-500
combina riesgo de stockout con incertidumbre de demanda
y abastecimiento internacional.


2. EVIDENCIA DEL DIAGNÓSTICO

El nivel de servicio actual es aproximadamente
{nivel_servicio_actual:.1f}% frente a una meta de
{meta_nivel_servicio:.1f}%.

El forecast presenta un MAPE aproximado de
{mape_actual:.1f}%.

El lead time promedio es de aproximadamente
{lead_time_actual:.0f} días.

Adicionalmente, alrededor del
{dependencia_proveedor:.0f}% del abastecimiento
se concentra en el proveedor principal.


3. MACHINE LEARNING

El modelo predictivo permitió explorar
la posibilidad de anticipar periodos
con mayor riesgo de stockout.

El recall ilustrativo fue de
{recall:.1%},

lo cual representa la capacidad del modelo
para identificar una proporción de los eventos
de stockout observados.


4. ALTERNATIVAS

Se analizaron tres mecanismos diferentes:

A1:
protección mediante stock de seguridad.

A2:
diversificación mediante dual sourcing.

A3:
anticipación mediante reabastecimiento predictivo.


5. AHP

Las alternativas fueron evaluadas considerando
múltiples criterios.

Bajo los juicios utilizados en el ejemplo,
la alternativa con mayor prioridad fue:

{alternativa_priorizada}

con:

{prioridad_alternativa:.1f}%.


6. DECISIÓN

El resultado no debería interpretarse
como una orden automática de implementación.

La decisión propuesta consiste en desarrollar
un piloto de la alternativa priorizada,
medir su desempeño y validar los supuestos
antes de escalarla.


7. PRINCIPIO GERENCIAL

La decisión no proviene de una sola herramienta.

Proviene de la integración entre:

DATOS
+
ANALÍTICA
+
PREDICCIÓN
+
CONOCIMIENTO DEL NEGOCIO
+
EVALUACIÓN MULTICRITERIO.

"""
)


# =============================================================================
# 29. GENERAR PREGUNTAS PARA LA DISCUSIÓN EN CLASE
# =============================================================================


preguntas = [

    "¿Qué evidencia demuestra que existe realmente un problema?",

    "¿Qué resultado del diagnóstico consideran más crítico?",

    "¿Qué información adicional solicitarían antes de decidir?",

    "¿Qué aporta Machine Learning que no aportaba el análisis descriptivo?",

    "¿Qué limitaciones tiene el modelo predictivo?",

    "¿Qué alternativa ataca directamente el riesgo del proveedor?",

    "¿Qué alternativa podría implementarse más rápidamente?",

    "¿Qué supuestos afectan el resultado del AHP?",

    "¿Cambiaría la decisión si el costo fuera el criterio dominante?",

    "¿Qué medirían durante un piloto?"

]


print("\n" + "=" * 80)

print("PREGUNTAS PARA DISCUSIÓN")

print("=" * 80)


for i, pregunta in enumerate(
    preguntas,
    start=1
):

    print(
        f"{i}. {pregunta}"
    )


# =============================================================================
# 30. GUARDAR LOS RESULTADOS
# =============================================================================


matriz_evidencia.to_csv(

    "GlobalBike_Matriz_Evidencias.csv",

    index=False,

    encoding="utf-8-sig"

)


evidencia_decision.to_csv(

    "GlobalBike_Evidencia_Decision.csv",

    index=False,

    encoding="utf-8-sig"

)


indicadores_transformacion.to_csv(

    "GlobalBike_KPIs_Transformacion.csv",

    index=False,

    encoding="utf-8-sig"

)


resultado_ahp.to_csv(

    "GlobalBike_Resultado_AHP_Integrado.csv",

    index=False,

    encoding="utf-8-sig"

)


print("\nARCHIVOS GENERADOS:")


print("""

1. GlobalBike_Matriz_Evidencias.csv

2. GlobalBike_Evidencia_Decision.csv

3. GlobalBike_KPIs_Transformacion.csv

4. GlobalBike_Resultado_AHP_Integrado.csv

""")


# =============================================================================
# 31. CONCLUSIÓN FINAL
# =============================================================================


print("\n" + "=" * 80)

print("CONCLUSIÓN | INTEGRATION & DECISION MAKING")

print("=" * 80)


print(
    f"""

El análisis de GlobalBike permitió integrar
diferentes herramientas utilizadas durante el curso.

DATA ANALYTICS permitió responder:

¿QUÉ ESTÁ PASANDO?


MACHINE LEARNING permitió responder:

¿PODEMOS ANTICIPAR EL RIESGO?


EL ANÁLISIS DE ALTERNATIVAS permitió responder:

¿QUÉ PODRÍAMOS HACER?


AHP permitió responder:

¿CÓMO PRIORIZAMOS LAS ALTERNATIVAS
CONSIDERANDO MÚLTIPLES CRITERIOS?


Bajo los supuestos del ejemplo,
la alternativa con mayor prioridad fue:

{alternativa_priorizada}

con aproximadamente:

{prioridad_alternativa:.1f}%.


Sin embargo, la decisión final no debería
depender únicamente del resultado matemático.

También debe considerar:

- viabilidad operativa,
- disponibilidad de recursos,
- información de proveedores,
- restricciones financieras,
- riesgos de implementación,
- conocimiento de los responsables del proceso.


Por eso proponemos pasar de:

RECOMENDACIÓN

a:

PILOTO

y posteriormente a:

MEDICIÓN Y VALIDACIÓN.


Ahora solamente queda una última pregunta:

¿CÓMO CONVERTIMOS ESTA DECISIÓN
EN UN PLAN CONCRETO DE IMPLEMENTACIÓN?


Esto nos lleva al:

PUNTO 12

FROM ANALYSIS TO ACTION.

"""
)


# =============================================================================
# 32. MAPA FINAL
# =============================================================================


print("""

===============================================================================

                     GLOBALBIKE COMPONENTS

===============================================================================


                     PROBLEMA EMPRESARIAL
                              │
                              ↓
                      EVIDENCIA CIENTÍFICA
                              │
                              ↓
                        DATA ANALYTICS
                              │
                              ↓
                           KPIs
                              │
                              ↓
                    MACHINE LEARNING
                              │
                              ↓
                    ALERTA DE STOCKOUT
                              │
                              ↓
                  ALTERNATIVAS DE SOLUCIÓN
                              │
                              ↓
                             AHP
                              │
                              ↓
                    DECISIÓN PRIORIZADA
                              │
                              ↓
                           PILOTO
                              │
                              ↓
                    MEDICIÓN DE RESULTADOS
                              │
                              ↓
                         ESCALAMIENTO


===============================================================================

""")
```
