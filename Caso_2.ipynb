```python

# -*- coding: utf-8 -*-
"""
CASO 2 | EXCELENCIA OPERACIONAL Y ANALÍTICA NO SUPERVISADA

Objetivo:
Identificar perfiles ocultos de solicitudes administrativas,
detectar anomalías, construir alertas anticipatorias y evaluar
alternativas gerenciales mediante simulación Monte Carlo.

Todos los datos y los efectos de intervención son sintéticos.
Los resultados no constituyen evidencia causal ni representan
el desempeño de una institución real.

Ejecución:
Google Colab o Python con numpy, pandas, matplotlib,
seaborn y scikit-learn.
"""

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.ensemble import IsolationForest
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

warnings.filterwarnings("ignore", category=FutureWarning)


# ============================================================
# 0. CONFIGURACIÓN EDITABLE
# ============================================================
# En el primer ensayo se recomienda modificar únicamente
# los parámetros de esta sección.

SEMILLA = 42
N_SOLICITUDES = 900
MESES = 12

# Umbral de servicio ilustrativo, no institucional.
SLA_HORAS = 48.0

# Restricciones y supuestos financieros.
PRESUPUESTO_MAX_COP = 18_000_000
COSTO_HORA_OPERATIVA_COP = 42_000

# Número de escenarios aleatorios para Monte Carlo.
N_SIMULACIONES = 4000

# Proporción de registros que Isolation Forest señalará
# aproximadamente como atípicos.
CONTAMINACION = 0.06

# Rango de clusters que se evaluará.
K_MIN = 2
K_MAX = 6

DIRECTORIO = "Caso_2_Resultados"
os.makedirs(DIRECTORIO, exist_ok=True)

rng = np.random.default_rng(SEMILLA)

sns.set_theme(style="whitegrid", palette="colorblind")

plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "savefig.facecolor": "white",
    "font.size": 10,
    "axes.titlesize": 14,
    "axes.titleweight": "bold"
})


# ============================================================
# FUNCIONES AUXILIARES
# ============================================================

def guardar(nombre):
    """
    Guarda cada gráfica en PNG, con fondo blanco,
    y posteriormente la presenta en pantalla.
    """
    plt.tight_layout()
    plt.savefig(
        os.path.join(DIRECTORIO, nombre + ".png"),
        dpi=190,
        bbox_inches="tight",
        facecolor="white"
    )
    plt.show()
    plt.close()


def interpretar(titulo, texto):
    """
    Presenta interpretaciones extensas para facilitar
    la lectura de resultados y la discusión gerencial.
    """
    print("\n" + "=" * 90)
    print(titulo.upper())
    print("-" * 90)
    print(texto.strip())
    print()


# ============================================================
# 1. CONSTRUCCIÓN DE LA BASE DE DATOS SINTÉTICA
# ============================================================
# Cada fila representa una solicitud administrativa.
#
# Se generan distintos comportamientos operacionales para
# demostrar cómo el aprendizaje no supervisado puede
# descubrir perfiles sin conocer etiquetas previas.
#
# La variable "perfiles" se utiliza solamente para generar
# diversidad sintética. No se entrega a K-Means.
#
# El tiempo de resolución se utiliza después del clustering
# para evaluar externamente los segmentos descubiertos.

perfiles = rng.choice(
    [0, 1, 2],
    size=N_SOLICITUDES,
    p=[0.44, 0.34, 0.22]
)

mes = rng.integers(
    1,
    MESES + 1,
    size=N_SOLICITUDES
)

# Complejidad administrativa, de 1 a 5.
complejidad = np.clip(
    np.rint(
        rng.normal(
            np.select(
                [perfiles == 0, perfiles == 1],
                [1.7, 3.1],
                default=4.0
            ),
            0.7
        )
    ),
    1,
    5
).astype(int)

# Cantidad de campos o documentos faltantes.
campos_faltantes = np.clip(
    rng.poisson(
        np.select(
            [perfiles == 0, perfiles == 1],
            [0.4, 1.8],
            default=2.6
        )
    ),
    0,
    7
)

# Número de cambios de responsable.
reasignaciones = np.clip(
    rng.poisson(
        np.select(
            [perfiles == 0, perfiles == 1],
            [0.2, 1.2],
            default=2.0
        )
    ),
    0,
    6
)

# Carga de trabajo del analista.
# Se incluye una presión creciente ilustrativa.
carga_analista = np.clip(
    rng.normal(
        np.select(
            [perfiles == 0, perfiles == 1],
            [8, 13],
            default=17
        ) + 0.14 * (mes - 1),
        2.3
    ),
    2,
    28
)

# Seguimientos manuales realizados.
seguimiento_manual = np.clip(
    rng.poisson(
        np.select(
            [perfiles == 0, perfiles == 1],
            [0.5, 2.1],
            default=3.0
        )
    ),
    0,
    9
)

# Dependencias de otras áreas.
dependencias = np.clip(
    rng.poisson(
        np.select(
            [perfiles == 0, perfiles == 1],
            [0.25, 1.1],
            default=2.0
        )
    ),
    0,
    6
)

# Variabilidad no explicada por las variables anteriores.
ruido = rng.normal(
    0,
    5.0,
    N_SOLICITUDES
)

# Fórmula exclusivamente sintética.
# Los coeficientes NO son efectos causales estimados.
tiempo_h = np.maximum(
    2,
    4
    + 3.7 * complejidad
    + 4.5 * campos_faltantes
    + 5.5 * reasignaciones
    + 1.0 * carga_analista
    + 3.0 * seguimiento_manual
    + 5.0 * dependencias
    + ruido
)

base = pd.DataFrame({
    "solicitud_id": [
        f"SOL-{i:04d}"
        for i in range(1, N_SOLICITUDES + 1)
    ],
    "mes": mes,
    "complejidad_1_5": complejidad,
    "campos_faltantes": campos_faltantes,
    "reasignaciones": reasignaciones,
    "carga_analista_casos": np.round(
        carga_analista, 1
    ),
    "seguimientos_manuales": seguimiento_manual,
    "dependencias_otras_areas": dependencias,
    "tiempo_resolucion_h": np.round(tiempo_h, 2)
})

base["cumple_sla"] = (
    base["tiempo_resolucion_h"] <= SLA_HORAS
)

VARIABLES = [
    "complejidad_1_5",
    "campos_faltantes",
    "reasignaciones",
    "carga_analista_casos",
    "seguimientos_manuales",
    "dependencias_otras_areas"
]

base.to_csv(
    os.path.join(
        DIRECTORIO,
        "01_datos_sinteticos.csv"
    ),
    index=False
)

interpretar(
    "01. Contexto y calidad del dato",
    f"""
Se dispone de {len(base):,} solicitudes sintéticas
distribuidas en {MESES} meses.

El tiempo medio de resolución es
{base.tiempo_resolucion_h.mean():.1f} horas.

El cumplimiento respecto al SLA ilustrativo de
{SLA_HORAS:.0f} horas es
{100 * base.cumple_sla.mean():.1f}%.

Estos valores no describen ninguna institución real.

Para trasladar el laboratorio a una operación existente
se requieren registros de recepción, clasificación,
documentación, responsables, reasignaciones, fechas de
cierre y trazabilidad de cada estado.

También se debe distinguir entre tiempo calendario,
tiempo hábil y tiempo efectivo de procesamiento.

La primera decisión gerencial consiste en construir
una línea base verificable antes de atribuir los
retrasos a una causa determinada.
"""
)


# ============================================================
# 2. DIAGNÓSTICO DESCRIPTIVO
# ============================================================

fig, ax = plt.subplots(figsize=(10, 5))

sns.histplot(
    data=base,
    x="tiempo_resolucion_h",
    bins=35,
    color="#2379B5",
    ax=ax
)

ax.axvline(
    SLA_HORAS,
    color="#D24A43",
    linestyle="--",
    linewidth=2,
    label=f"Illustrative SLA: {SLA_HORAS:.0f} h"
)

ax.set(
    title="01 | Service Request Resolution Time",
    xlabel="Resolution time (hours)",
    ylabel="Requests"
)

ax.legend()
guardar("01_Resolution_Time")

interpretar(
    "02. Lectura del tiempo de resolución",
    f"""
El promedio de
{base.tiempo_resolucion_h.mean():.1f} horas
no describe por sí solo la experiencia del usuario.

La distribución permite examinar la dispersión,
la concentración de solicitudes y la proporción
que supera el compromiso de servicio.

Antes de incrementar la planta de personal,
conviene investigar si los casos tardíos tienen
mayor complejidad, documentación incompleta,
reasignaciones o dependencias externas.

Una decisión futura sería establecer indicadores
diferenciados de tiempo total, tiempo en espera,
primera respuesta, devoluciones y cumplimiento
por tipo de solicitud.

Esto permitiría distinguir problemas de capacidad
de problemas de diseño del proceso.
"""
)


# ============================================================
# 3. ESTANDARIZACIÓN Y SELECCIÓN DE CLUSTERS
# ============================================================
# K-Means utiliza distancias entre observaciones.
# Por eso las variables deben estar en escalas comparables.
#
# El tiempo de resolución NO participa en el agrupamiento.

escalador = StandardScaler()
X = escalador.fit_transform(base[VARIABLES])

k_values = list(
    range(
        K_MIN,
        min(K_MAX, len(base) - 1) + 1
    )
)

scores = []

for k in k_values:

    km = KMeans(
        n_clusters=k,
        random_state=SEMILLA,
        n_init=15
    )

    labels = km.fit_predict(X)

    puntuacion = silhouette_score(
        X,
        labels,
        sample_size=min(700, len(base)),
        random_state=SEMILLA
    )

    scores.append(puntuacion)

k_elegido = k_values[
    int(np.argmax(scores))
]

fig, ax = plt.subplots(figsize=(8, 4.5))

ax.plot(
    k_values,
    scores,
    marker="o",
    linewidth=2,
    color="#1A78B0"
)

ax.axvline(
    k_elegido,
    color="#E18B24",
    linestyle="--",
    label=f"Selected k = {k_elegido}"
)

ax.set(
    title="02 | Cluster Selection by Silhouette Score",
    xlabel="Number of clusters",
    ylabel="Silhouette score"
)

ax.legend()
guardar("02_Silhouette")

interpretar(
    "03. Elección del número de clusters",
    f"""
Se compararon soluciones entre {K_MIN} y
{max(k_values)} grupos mediante el índice silhouette.

La configuración seleccionada fue k={k_elegido},
con un índice de {max(scores):.3f}.

Un índice relativamente alto indica que los
grupos presentan una separación favorable
respecto a las alternativas evaluadas.

Sin embargo, no demuestra que correspondan a
categorías organizacionales reales.

La selección debe contrastarse con el tamaño,
la estabilidad y la utilidad de los segmentos.

La recomendación es comprobar si cada grupo
representa un comportamiento operacional que
justifique una estrategia de atención diferenciada.
"""
)


# ============================================================
# 4. K-MEANS Y VISUALIZACIÓN CON PCA
# ============================================================

modelo = KMeans(
    n_clusters=k_elegido,
    random_state=SEMILLA,
    n_init=20
)

base["cluster"] = modelo.fit_predict(X)

# PCA se utiliza únicamente para representar gráficamente
# las seis dimensiones en un plano de dos dimensiones.
pca = PCA(
    n_components=2,
    random_state=SEMILLA
)

xy = pca.fit_transform(X)

fig, ax = plt.subplots(figsize=(9, 6))

sns.scatterplot(
    x=xy[:, 0],
    y=xy[:, 1],
    hue=base["cluster"].astype(str),
    palette="tab10",
    s=32,
    alpha=0.7,
    ax=ax
)

ax.set(
    title="03 | K-Means Operational Segments (PCA View)",
    xlabel="Principal component 1",
    ylabel="Principal component 2"
)

ax.legend(title="Cluster")
guardar("03_Clusters_PCA")


# ============================================================
# 5. MAPA DE CALOR Y PERFILES OPERACIONALES
# ============================================================

perfil = base.groupby("cluster")[VARIABLES].mean()

media = base[VARIABLES].mean()

desv = base[VARIABLES].std().replace(0, 1)

perfil_z = (perfil - media) / desv

fig, ax = plt.subplots(
    figsize=(11, max(3.8, 1.2 * k_elegido))
)

sns.heatmap(
    perfil_z,
    annot=True,
    fmt=".2f",
    cmap="RdYlBu_r",
    center=0,
    linewidths=0.6,
    ax=ax,
    cbar_kws={
        "label": "Standardized deviation"
    }
)

ax.set(
    title="04 | Operational Profile by Cluster",
    xlabel="Operational variables",
    ylabel="Cluster"
)

guardar("04_Cluster_Heatmap")

resumen = base.groupby("cluster").agg(
    solicitudes=("solicitud_id", "size"),
    tiempo_promedio_h=("tiempo_resolucion_h", "mean"),
    cumplimiento_sla=("cumple_sla", "mean")
).round(3)

resumen["cumplimiento_sla_pct"] = (
    100 * resumen["cumplimiento_sla"]
)

print("\nRESUMEN POR CLUSTER")
print(resumen.to_string())

peor = int(
    resumen["tiempo_promedio_h"].idxmax()
)

mejor = int(
    resumen["tiempo_promedio_h"].idxmin()
)

for c in perfil.index:

    variables_dominantes = (
        perfil_z.loc[c]
        .sort_values(ascending=False)
        .head(2)
        .index.tolist()
    )

    interpretar(
        f"05. Interpretación ejecutiva del cluster {c}",
        f"""
El cluster {c} agrupa
{int(resumen.loc[c, 'solicitudes'])} solicitudes.

Su tiempo promedio es de
{resumen.loc[c, 'tiempo_promedio_h']:.1f} horas.

El cumplimiento ilustrativo del SLA alcanza
{resumen.loc[c, 'cumplimiento_sla_pct']:.1f}%.

Las características relativamente más elevadas
son: {', '.join(variables_dominantes)}.

Esto sugiere investigar si los procedimientos
de recepción, validación, asignación y
escalamiento deberían diferenciarse.

La gerencia no debería interpretar las
variables dominantes como causas comprobadas.

Se recomienda revisar expedientes individuales,
entrevistar a los analistas y medir los tiempos
de permanencia en cada estado del proceso.

Si se confirman diferencias relevantes, podría
diseñarse un flujo especializado para este
segmento y evaluar su efecto mediante un piloto.
"""
    )

interpretar(
    "06. Comparación transversal y mejora",
    f"""
El cluster {peor} presenta el mayor tiempo
promedio de resolución.

El cluster {mejor} registra el menor.

Esta diferencia sugiere que una política
uniforme de atención podría utilizar los
recursos de manera poco eficiente.

Una oportunidad futura sería separar las
solicitudes de resolución sencilla de aquellas
que requieren conocimiento especializado,
documentación adicional o intervención
de otras áreas.

No se recomienda clasificar automáticamente
a los responsables como eficientes o
ineficientes según el cluster.

Los segmentos describen características
de las solicitudes y sus condiciones
operacionales, no el desempeño individual
de los trabajadores.

La organización debería contrastar esta
segmentación con el VSM, el análisis de
cuellos de botella y las prioridades
seleccionadas previamente mediante AHP.
"""
)


# ============================================================
# 6. ISOLATION FOREST: DETECCIÓN DE ANOMALÍAS
# ============================================================

iso = IsolationForest(
    n_estimators=250,
    contamination=CONTAMINACION,
    random_state=SEMILLA
)

base["anomalia"] = (
    iso.fit_predict(X) == -1
)

# Un valor mayor representa mayor atipicidad.
base["anomaly_score"] = (
    -iso.decision_function(X)
)

fig, ax = plt.subplots(figsize=(9, 6))

normal = ~base["anomalia"]

ax.scatter(
    xy[normal, 0],
    xy[normal, 1],
    s=23,
    alpha=0.45,
    color="#247FB8",
    label="Typical"
)

ax.scatter(
    xy[~normal, 0],
    xy[~normal, 1],
    s=56,
    facecolors="none",
    edgecolors="#D94D4D",
    linewidth=1.6,
    label="Anomaly"
)

ax.set(
    title="05 | Isolation Forest: Unusual Requests",
    xlabel="Principal component 1",
    ylabel="Principal component 2"
)

ax.legend()
guardar("05_Isolation_Forest")

num_anom = int(
    base["anomalia"].sum()
)

interpretar(
    "07. Anomalías y riesgo operacional",
    f"""
Isolation Forest señaló {num_anom}
solicitudes, equivalentes al
{100 * base.anomalia.mean():.1f}% del conjunto.

El porcentaje está condicionado por
CONTAMINACION y no representa una tasa
real de errores o incumplimientos.

La recomendación es revisar inicialmente
los quince casos con mayor puntuación
de anomalía.

En esa revisión se debería establecer si
existen errores de registro, documentación
atípica, acumulación excepcional de trabajo,
reasignaciones inusuales o dependencias
administrativas poco frecuentes.

También podrían encontrarse solicitudes
correctamente gestionadas, pero diferentes
a las demás.

Si determinados patrones se relacionan
posteriormente con incidentes confirmados,
podrían diseñarse alertas preventivas.

No sería apropiado rechazar solicitudes
o tomar medidas disciplinarias utilizando
únicamente la clasificación de anomalía.
"""
)

print("\n15 SOLICITUDES MÁS ATÍPICAS")
print(
    base.sort_values(
        "anomaly_score",
        ascending=False
    )
    .head(15)
    .to_string(index=False)
)


# ============================================================
# 7. EVALUACIÓN EXTERNA DEL DESEMPEÑO
# ============================================================
# Se compara el tiempo de resolución DESPUÉS de construir
# los clusters. Así se evita utilizar el indicador de
# desempeño para definir directamente los segmentos.

fig, ax = plt.subplots(figsize=(9, 5))

orden = (
    resumen
    .sort_values("tiempo_promedio_h")
    .index
)

sns.barplot(
    data=base,
    x="cluster",
    y="tiempo_resolucion_h",
    order=orden,
    estimator="mean",
    errorbar=None,
    color="#358BC4",
    ax=ax
)

ax.axhline(
    SLA_HORAS,
    color="#D24A43",
    linestyle="--",
    linewidth=2,
    label="Illustrative SLA"
)

ax.set(
    title="06 | Resolution Performance by Operational Segment",
    xlabel="Cluster",
    ylabel="Mean resolution time (hours)"
)

ax.legend()
guardar("06_Performance_By_Cluster")

interpretar(
    "08. Validación externa del perfil operacional",
    f"""
El tiempo de resolución no participó
en la formación de los clusters.

Se utilizó posteriormente para evaluar
si los perfiles descubiertos presentan
diferencias de desempeño.

El cluster {peor} registra el mayor
tiempo promedio:
{resumen.loc[peor, 'tiempo_promedio_h']:.1f} horas.

Este resultado permite priorizar
una investigación operacional específica.

Se recomienda construir un Value Stream
Mapping para ese segmento y distinguir
actividades que agregan valor de esperas,
devoluciones y transferencias.

La comparación entre grupos no demuestra
que sus características sean las causas
de los retrasos.

También pueden influir factores omitidos,
como estacionalidad, complejidad normativa,
disponibilidad de terceros o cambios
en la demanda.

La decisión futura debe basarse en
la validación de estas hipótesis.
"""
)


# ============================================================
# 8. EVOLUCIÓN TEMPORAL DE LOS SEGMENTOS
# ============================================================

participacion = (
    pd.crosstab(
        base["mes"],
        base["cluster"],
        normalize="index"
    ) * 100
)

fig, ax = plt.subplots(figsize=(10, 5))

for c in participacion.columns:

    ax.plot(
        participacion.index,
        participacion[c],
        marker="o",
        linewidth=2,
        label=f"Cluster {c}"
    )

ax.set(
    title="07 | Monthly Evolution of Operational Segments",
    xlabel="Month",
    ylabel="Share of requests (%)",
    xticks=range(1, MESES + 1)
)

ax.legend(ncol=3)
guardar("07_Cluster_Trends")

tasa_anom = (
    base.groupby("mes")["anomalia"].mean() * 100
)

fig, ax = plt.subplots(figsize=(10, 4.5))

ax.plot(
    tasa_anom.index,
    tasa_anom.values,
    marker="o",
    color="#C65C32",
    linewidth=2
)

ax.axhline(
    100 * CONTAMINACION,
    color="#555555",
    linestyle=":",
    label="Overall model setting"
)

ax.set(
    title="08 | Monthly Anomaly Monitoring",
    xlabel="Month",
    ylabel="Flagged requests (%)",
    xticks=range(1, MESES + 1)
)

ax.legend()
guardar("08_Anomaly_Trends")

reciente = (
    participacion[peor]
    .tail(3)
    .mean()
)

if MESES >= 6:
    anterior = (
        participacion[peor]
        .iloc[-6:-3]
        .mean()
    )
else:
    anterior = (
        participacion[peor]
        .head(3)
        .mean()
    )

interpretar(
    "09. Señales de alerta y análisis temporal",
    f"""
El segmento de mayor tiempo,
cluster {peor}, representa
{reciente:.1f}% de las solicitudes
en los tres meses más recientes.

En los tres meses anteriores
representaba {anterior:.1f}%.

La variación fue de
{reciente - anterior:+.1f}
puntos porcentuales.

Este cambio describe una evolución
histórica y no constituye una
predicción futura validada.

Se recomienda investigar si coincide
con ciclos administrativos, cambios
en procedimientos, mayor volumen
de solicitudes o reducción temporal
de capacidad.

Si se confirma un comportamiento
recurrente, podría implementarse
un tablero de control mensual.

Las alertas deberían basarse en
umbrales históricos y considerar
la estacionalidad, evitando
interpretar fluctuaciones normales
como deterioros estructurales.
"""
)


# ============================================================
# 9. ALERTA ANTICIPATORIA PARA UNA SOLICITUD NUEVA
# ============================================================
# Se utiliza similitud con segmentos históricos.
#
# No se calcula una predicción individual calibrada
# del tiempo de resolución.
#
# IMPORTANTE:
# Para una aplicación real, solo deben utilizarse
# variables disponibles en el momento de la alerta.
#
# Si las reasignaciones o los seguimientos todavía
# no han ocurrido, no deben introducirse como si
# fueran datos conocidos al recibir la solicitud.

nueva = pd.DataFrame([{
    "complejidad_1_5": 4,
    "campos_faltantes": 2,
    "reasignaciones": 2,
    "carga_analista_casos": 17,
    "seguimientos_manuales": 3,
    "dependencias_otras_areas": 2
}])

# Se reutilizan los modelos ajustados anteriormente.
# No se modifica el escalador con el caso nuevo.

X_nueva = escalador.transform(
    nueva[VARIABLES]
)

cluster_nuevo = int(
    modelo.predict(X_nueva)[0]
)

anomalia_nueva = bool(
    iso.predict(X_nueva)[0] == -1
)

referencia = base.loc[
    base["cluster"] == cluster_nuevo
]

interpretar(
    "10. Alerta anticipatoria para una nueva solicitud",
    f"""
La solicitud de ejemplo fue asignada
al cluster {cluster_nuevo}.

Históricamente, ese segmento presenta
{referencia.tiempo_resolucion_h.mean():.1f}
horas promedio de resolución.

Su cumplimiento ilustrativo alcanza
{100 * referencia.cumple_sla.mean():.1f}%.

La nueva observación
{'SÍ' if anomalia_nueva else 'NO'}
fue marcada como anómala.

El resultado constituye una alerta
basada en similitud con casos históricos.

No debe interpretarse como una
predicción individual exacta ni como
una probabilidad calibrada.

La organización podría utilizarlo
para revisar anticipadamente
documentación, clasificación,
capacidad disponible y dependencias.

Antes de implementarlo, deberá
comprobarse si la alerta realmente
identifica solicitudes que requieren
atención preventiva.

La validación debe realizarse con
solicitudes posteriores y utilizando
únicamente información disponible
en el momento de emitir la alerta.
"""
)


# ============================================================
# 10. SIMULACIÓN PRESCRIPTIVA CON MONTE CARLO
# ============================================================
# Los efectos de las alternativas son supuestos.
#
# El ejercicio compara escenarios gerenciales,
# pero no demuestra que las intervenciones
# produzcan estos resultados en la realidad.
#
# Las alternativas pueden modificarse para
# explorar otras decisiones.

ESCENARIOS = {

    "Current process": {
        "costo": 0,
        "faltantes": 0,
        "reasig": 0,
        "seguim": 0,
        "carga": 0,
        "dep": 0
    },

    "Standardized intake portal": {
        "costo": 8_000_000,
        "faltantes": 0.55,
        "reasig": 0.10,
        "seguim": 0.10,
        "carga": 0,
        "dep": 0
    },

    "Workflow automation": {
        "costo": 13_000_000,
        "faltantes": 0.10,
        "reasig": 0.35,
        "seguim": 0.60,
        "carga": 0.05,
        "dep": 0.10
    },

    "Tiered service model": {
        "costo": 11_000_000,
        "faltantes": 0.10,
        "reasig": 0.45,
        "seguim": 0.20,
        "carga": 0.20,
        "dep": 0.25
    },

    "Workload balancing": {
        "costo": 9_000_000,
        "faltantes": 0,
        "reasig": 0.10,
        "seguim": 0.15,
        "carga": 0.35,
        "dep": 0.05
    },

    "Integrated redesign": {
        "costo": 27_000_000,
        "faltantes": 0.60,
        "reasig": 0.55,
        "seguim": 0.65,
        "carga": 0.30,
        "dep": 0.35
    }
}

# Se seleccionan solicitudes sintéticas con reemplazo.
# Se utiliza la misma muestra y el mismo ruido
# para comparar todas las alternativas.

idx = rng.integers(
    0,
    len(base),
    size=N_SIMULACIONES
)

muestra = base.iloc[idx]

ruido_mc = rng.normal(
    0,
    6,
    size=N_SIMULACIONES
)

filas = []

for nombre, p in ESCENARIOS.items():

    # Cada porcentaje representa una reducción
    # hipotética de la variable correspondiente.

    t = np.maximum(
        2,
        4
        + 3.7 * muestra["complejidad_1_5"].to_numpy()
        + 4.5 * muestra["campos_faltantes"].to_numpy()
          * (1 - p["faltantes"])
        + 5.5 * muestra["reasignaciones"].to_numpy()
          * (1 - p["reasig"])
        + 1.0 * muestra["carga_analista_casos"].to_numpy()
          * (1 - p["carga"])
        + 3.0 * muestra["seguimientos_manuales"].to_numpy()
          * (1 - p["seguim"])
        + 5.0 * muestra["dependencias_otras_areas"].to_numpy()
          * (1 - p["dep"])
        + ruido_mc
    )

    filas.append({
        "alternativa": nombre,
        "costo_cop": p["costo"],
        "tiempo_medio_h": t.mean(),
        "p90_h": np.percentile(t, 90),
        "cumplimiento_pct": 100 * np.mean(
            t <= SLA_HORAS
        ),
        "ahorro_h_por_solicitud": 0
    })

esc = pd.DataFrame(filas)

base_t = esc.loc[
    esc["alternativa"] == "Current process",
    "tiempo_medio_h"
].iloc[0]

esc["ahorro_h_por_solicitud"] = (
    base_t - esc["tiempo_medio_h"]
)

# Este valor representa potencial monetario ilustrativo.
# No equivale necesariamente a un ahorro real de caja.
esc["ahorro_operativo_estimado_cop"] = (
    esc["ahorro_h_por_solicitud"]
    * N_SOLICITUDES
    * COSTO_HORA_OPERATIVA_COP
)

esc["factible"] = (
    esc["costo_cop"] <= PRESUPUESTO_MAX_COP
)

esc.to_csv(
    os.path.join(
        DIRECTORIO,
        "02_escenarios_montecarlo.csv"
    ),
    index=False
)


# ============================================================
# 11. COMPARACIÓN GRÁFICA DE ALTERNATIVAS
# ============================================================

fig, ax = plt.subplots(figsize=(11, 5.5))

colores = [
    "#2383B9" if f else "#D1A044"
    for f in esc["factible"]
]

ax.barh(
    esc["alternativa"],
    esc["tiempo_medio_h"],
    color=colores
)

ax.axvline(
    SLA_HORAS,
    color="#CC4B48",
    linestyle="--",
    label="Illustrative SLA"
)

ax.invert_yaxis()

ax.set(
    title="09 | Simulated Resolution Time by Intervention",
    xlabel="Mean resolution time (hours)",
    ylabel=""
)

ax.legend()
guardar("09_Scenario_Resolution")

fig, ax = plt.subplots(figsize=(10, 6))

for _, r in esc.iterrows():

    ax.scatter(
        r["costo_cop"] / 1e6,
        r["cumplimiento_pct"],
        s=140,
        color=(
            "#2383B9"
            if r["factible"]
            else "#E08B2E"
        )
    )

    ax.annotate(
        r["alternativa"],
        (
            r["costo_cop"] / 1e6,
            r["cumplimiento_pct"]
        ),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=8
    )

ax.axvline(
    PRESUPUESTO_MAX_COP / 1e6,
    color="#D04A49",
    linestyle="--",
    label="Budget limit"
)

ax.set(
    title="10 | Cost–Service Trade-off (Hypothetical)",
    xlabel="Implementation cost (million COP)",
    ylabel="Simulated SLA compliance (%)"
)

ax.legend()
guardar("10_Cost_Service_Tradeoff")


# ============================================================
# 12. SELECCIÓN DE LA ALTERNATIVA FACTIBLE
# ============================================================
# Se prioriza el cumplimiento simulado del SLA.
# El tiempo medio funciona como segundo criterio.
#
# Esto es una regla de decisión ilustrativa.
# No constituye una optimización matemática completa.

factibles = (
    esc.loc[esc["factible"]]
    .sort_values(
        ["cumplimiento_pct", "tiempo_medio_h"],
        ascending=[False, True]
    )
)

recomendada = factibles.iloc[0]

print("\nCOMPARACIÓN DE ESCENARIOS")
print(
    esc.round(2).to_string(index=False)
)

interpretar(
    "11. Interpretación prescriptiva y presupuesto",
    f"""
Con un presupuesto ilustrativo de
COP {PRESUPUESTO_MAX_COP:,.0f},
la alternativa factible mejor posicionada
por cumplimiento simulado es:

{recomendada.alternativa}

Su tiempo medio simulado es
{recomendada.tiempo_medio_h:.1f} horas,
frente a {base_t:.1f} horas
del proceso actual.

Su cumplimiento simulado del SLA
es {recomendada.cumplimiento_pct:.1f}%.

La reducción estimada es de
{recomendada.ahorro_h_por_solicitud:.1f}
horas por solicitud.

Estos resultados permiten comparar
las alternativas bajo los supuestos
establecidos, pero no constituyen
evidencia de su efectividad real.

La gerencia debería contrastar la
alternativa con los criterios y pesos
utilizados previamente en AHP.

También convendría evaluar si una
solución de mayor inversión podría
generar beneficios adicionales en
calidad, trazabilidad o experiencia
del usuario.

La decisión de inversión debería
tomarse después de validar los
efectos mediante un piloto y
comprobar los costos reales.
"""
)


# ============================================================
# 13. ANÁLISIS DE SENSIBILIDAD DEL SLA
# ============================================================
# Se estudia cómo cambiaría el cumplimiento
# si se utilizaran distintos compromisos
# de tiempo de resolución.

p = ESCENARIOS[
    recomendada["alternativa"]
]

t_rec = np.maximum(
    2,
    4
    + 3.7 * muestra["complejidad_1_5"].to_numpy()
    + 4.5 * muestra["campos_faltantes"].to_numpy()
      * (1 - p["faltantes"])
    + 5.5 * muestra["reasignaciones"].to_numpy()
      * (1 - p["reasig"])
    + muestra["carga_analista_casos"].to_numpy()
      * (1 - p["carga"])
    + 3 * muestra["seguimientos_manuales"].to_numpy()
      * (1 - p["seguim"])
    + 5 * muestra["dependencias_otras_areas"].to_numpy()
      * (1 - p["dep"])
    + ruido_mc
)

for umbral in [24, 48, 72]:

    cumplimiento = (
        100 * np.mean(t_rec <= umbral)
    )

    print(
        f"SLA hipotético {umbral:>2} h | "
        f"Cumplimiento simulado: "
        f"{cumplimiento:.1f}%"
    )

interpretar(
    "12. Sensibilidad y riesgos de decisión",
    """
La proporción de solicitudes que cumple
el SLA depende del umbral establecido.

Un plazo más flexible puede aumentar
matemáticamente el porcentaje de cumplimiento,
pero no necesariamente mejorar la
experiencia del usuario.

Por ello, los compromisos de servicio
no deberían modificarse únicamente
para presentar indicadores favorables.

También convendría evaluar cómo cambia
la alternativa recomendada si aumenta
el volumen de solicitudes, disminuye
la disponibilidad de analistas o se
incrementa la complejidad.

Este análisis permitiría comprobar
si la solución es robusta frente a
condiciones operacionales diferentes.

Si una pequeña variación de los supuestos
cambia completamente la recomendación,
la gerencia debería solicitar más
información antes de invertir.
"""
)


# ============================================================
# 14. CONCLUSIONES EJECUTIVAS
# ============================================================

interpretar(
    "13. Conclusión ejecutiva integral",
    f"""
HALLAZGOS PRINCIPALES

Se identificaron {k_elegido}
perfiles operacionales y
{num_anom} solicitudes atípicas.

El segmento con mayor tiempo
promedio corresponde al cluster {peor}.

La evolución mensual muestra una
variación de {reciente - anterior:+.1f}
puntos porcentuales en la participación
reciente de ese segmento.

Estos resultados aportan señales
para investigar, pero no prueban
un deterioro futuro.

DECISIONES ESTRATÉGICAS

Se recomienda establecer estándares
de servicio verificables, mejorar la
trazabilidad de extremo a extremo y
evaluar una arquitectura de atención
diferenciada por complejidad.

Las inversiones deberían compararse
considerando experiencia del usuario,
costo total, calidad y capacidad.

DECISIONES TÁCTICAS

Se propone diseñar un piloto de seis
a ocho semanas con solicitudes del
segmento que presenta mayores tiempos.

El piloto debería medir documentación
completa al ingreso, reasignaciones,
carga por analista, tiempo en espera
y cumplimiento de plazos.

DECISIONES OPERACIONALES

Se recomienda aplicar controles de
completitud, reglas claras de
clasificación y visibilidad del
responsable de cada solicitud.

Las solicitudes detenidas deberían
contar con alertas de seguimiento.

También se recomienda revisar
semanalmente las anomalías confirmadas
y documentar sus características.

EVALUACIÓN DE RESULTADOS

La intervención debería compararse
con una línea base previa.

Cuando sea posible, convendría
utilizar un grupo de comparación
para separar el efecto del piloto
de cambios estacionales o variaciones
de la demanda.

El agrupamiento deberá revisarse
periódicamente para comprobar que
los segmentos continúan siendo
interpretables y útiles.

LIMITACIONES

Todos los registros utilizados son
sintéticos.

Los efectos de las intervenciones
son supuestos de demostración.

Ningún resultado debe comunicarse
como diagnóstico institucional,
ahorro comprobado o retorno de
inversión validado.
"""
)


# ============================================================
# 15. EXPORTACIÓN FINAL
# ============================================================

base.to_csv(
    os.path.join(
        DIRECTORIO,
        "03_datos_con_clusters_anomalias.csv"
    ),
    index=False
)

resumen.to_csv(
    os.path.join(
        DIRECTORIO,
        "04_resumen_clusters.csv"
    )
)

print("\nARCHIVOS GENERADOS EN:")
print(os.path.abspath(DIRECTORIO))

print(
    "\nFinalizado correctamente. "
    "Revisar las 10 gráficas PNG "
    "y los 4 archivos CSV."
)
```
