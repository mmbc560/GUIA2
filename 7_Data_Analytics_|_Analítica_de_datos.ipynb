```python
# =============================================================================
# INVENTARIOS Y COMPRAS GLOBALES
# CASO INTEGRADOR: GLOBALBIKE COMPONENTS COLOMBIA S.A.S.
#
# ETAPA 7: DATA ANALYTICS - DIAGNÓSTICO DEL PROBLEMA
# =============================================================================
#
# CONTEXTO DEL CASO
# -----------------------------------------------------------------------------
# GlobalBike Components Colombia S.A.S. ensambla bicicletas eléctricas.
#
# Uno de sus componentes críticos es el controlador electrónico E-500,
# adquirido principalmente a un proveedor internacional.
#
# La empresa ha identificado una situación aparentemente contradictoria:
#
#   - En algunos periodos tiene exceso de inventario.
#   - En otros periodos presenta faltantes o stockouts.
#   - La demanda presenta variabilidad.
#   - El pronóstico no siempre coincide con la demanda real.
#   - El lead time internacional también presenta variabilidad.
#
# PREGUNTA DE ESTA ETAPA:
#
# ¿Qué nos dicen los datos acerca del problema de inventarios y abastecimiento
# del controlador E-500?
#
# IMPORTANTE:
# En esta etapa NO vamos a buscar todavía la solución.
# Primero vamos a diagnosticar el problema utilizando datos.
#
# Posteriormente utilizaremos estos mismos datos para:
#
# ETAPA 8  -> Machine Learning
# ETAPA 9  -> Alternativas de solución
# ETAPA 10 -> AHP
#
# =============================================================================


# =============================================================================
# 1. IMPORTAR LAS LIBRERÍAS
# =============================================================================

# pandas:
# Permite trabajar con bases de datos en forma de tablas.
# Es similar a trabajar con una hoja de Excel.

import pandas as pd


# numpy:
# Permite realizar operaciones matemáticas y generar datos simulados.

import numpy as np


# matplotlib:
# Permite construir gráficas.

import matplotlib.pyplot as plt


# seaborn:
# Facilita la creación de gráficas estadísticas visualmente atractivas.

import seaborn as sns


# Configuración visual general.

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 6)

plt.rcParams["font.size"] = 11


# =============================================================================
# 2. CREAR LA BASE DE DATOS DEL CASO
# =============================================================================
#
# Para fines académicos vamos a crear una base ficticia de 36 meses.
#
# Esto representa tres años de operación del controlador E-500.
#
# Cada fila representa un mes.
#
# Las principales variables serán:
#
# Demanda_real
# Pronostico
# Lead_time
# Inventario_inicial
# Compras_recibidas
# Precio_unitario
#
# Posteriormente calcularemos otras variables.


# -----------------------------------------------------------------------------
# SEMILLA ALEATORIA
# -----------------------------------------------------------------------------
#
# np.random.seed() permite que los números aleatorios sean siempre iguales
# cada vez que ejecutamos el código.
#
# Esto es muy importante para una clase:
# todos los estudiantes obtendrán exactamente los mismos resultados.

np.random.seed(42)


# Creamos 36 meses.

fechas = pd.date_range(
    start="2024-01-01",
    periods=36,
    freq="MS"
)


# =============================================================================
# 3. SIMULAR LA DEMANDA REAL
# =============================================================================
#
# Supongamos que la demanda promedio está alrededor de 1.050 unidades.
#
# Pero la demanda no es constante.
#
# Vamos a incorporar:
#
# 1. Una demanda base.
# 2. Una pequeña tendencia de crecimiento.
# 3. Estacionalidad.
# 4. Variabilidad aleatoria.
#
# Esto hace que el comportamiento sea más parecido al de una empresa real.


demanda_base = 1050


# Tendencia:
# La demanda aumenta gradualmente durante los 36 meses.

tendencia = np.linspace(0, 180, 36)


# Estacionalidad:
# Utilizamos una función seno para simular meses de mayor y menor demanda.

estacionalidad = 150 * np.sin(
    2 * np.pi * np.arange(36) / 12
)


# Ruido:
# Representa situaciones que no podemos predecir perfectamente.

ruido_demanda = np.random.normal(
    loc=0,
    scale=90,
    size=36
)


# Fórmula general:
#
# DEMANDA =
# demanda base
# + tendencia
# + estacionalidad
# + variabilidad aleatoria

demanda_real = (
    demanda_base
    + tendencia
    + estacionalidad
    + ruido_demanda
).round().astype(int)


# Evitamos valores demasiado pequeños.

demanda_real = np.maximum(
    demanda_real,
    600
)


# =============================================================================
# 4. SIMULAR EL PRONÓSTICO
# =============================================================================
#
# El pronóstico intenta anticipar la demanda.
#
# Pero ningún forecast es perfecto.
#
# Vamos a introducir un error aleatorio.
#
# Si:
#
# Demanda real = 1.200
# Pronóstico    = 1.000
#
# entonces la empresa SUBESTIMÓ la demanda.
#
# Esto puede contribuir a generar stockouts.


error_forecast_simulado = np.random.normal(
    loc=-20,
    scale=130,
    size=36
)


pronostico = (
    demanda_real
    + error_forecast_simulado
).round().astype(int)


pronostico = np.maximum(
    pronostico,
    500
)


# =============================================================================
# 5. SIMULAR EL LEAD TIME
# =============================================================================
#
# LEAD TIME:
#
# Tiempo transcurrido desde que se realiza la orden de compra
# hasta que el producto está disponible para utilizarse.
#
# En GlobalBike el lead time internacional puede variar
# aproximadamente entre 55 y 105 días.
#
# Un lead time elevado o variable puede obligar a la empresa a:
#
# - mantener mayor inventario,
# - aumentar stock de seguridad,
# - anticipar pedidos,
# - asumir mayor riesgo de stockout.


lead_time = np.random.normal(
    loc=75,
    scale=14,
    size=36
)


lead_time = np.clip(
    lead_time,
    55,
    105
).round().astype(int)


# =============================================================================
# 6. SIMULAR EL INVENTARIO INICIAL
# =============================================================================
#
# Para simplificar el ejercicio académico,
# simularemos niveles de inventario inicial.
#
# Estos niveles representan las unidades disponibles
# al comenzar cada mes.


inventario_inicial = np.random.normal(
    loc=1350,
    scale=350,
    size=36
)


inventario_inicial = np.clip(
    inventario_inicial,
    500,
    2300
).round().astype(int)


# =============================================================================
# 7. SIMULAR LAS COMPRAS RECIBIDAS
# =============================================================================
#
# Las compras recibidas corresponden a las unidades que llegan
# durante cada periodo.
#
# Para nuestro ejemplo, estarán relacionadas con el pronóstico,
# pero tendrán cierta variabilidad.


compras_recibidas = (
    pronostico
    * np.random.uniform(
        0.75,
        1.10,
        36
    )
).round().astype(int)


# =============================================================================
# 8. PRECIO UNITARIO
# =============================================================================
#
# Vamos a asumir un precio promedio aproximado de USD 42
# por controlador.
#
# El precio presenta pequeñas variaciones.


precio_unitario = np.random.normal(
    loc=42,
    scale=2.5,
    size=36
).round(2)


# =============================================================================
# 9. CONSTRUIR EL DATAFRAME
# =============================================================================
#
# Un DataFrame es una tabla de datos.
#
# Podemos imaginarlo como una hoja de Excel dentro de Python.


df = pd.DataFrame({

    "Fecha": fechas,

    "Demanda_real": demanda_real,

    "Pronostico": pronostico,

    "Lead_time_dias": lead_time,

    "Inventario_inicial": inventario_inicial,

    "Compras_recibidas": compras_recibidas,

    "Precio_USD": precio_unitario

})


# Creamos variables adicionales de tiempo.

df["Año"] = df["Fecha"].dt.year

df["Mes"] = df["Fecha"].dt.month

df["Periodo"] = df["Fecha"].dt.strftime("%Y-%m")


# Mostramos las primeras filas.

print("\nPRIMERAS FILAS DE LA BASE DE DATOS\n")

print(df.head())


# =============================================================================
# 10. CALCULAR EL ERROR DEL PRONÓSTICO
# =============================================================================
#
# ERROR DEL PRONÓSTICO:
#
# Error = Demanda real - Pronóstico
#
# Interpretación:
#
# Error > 0
# La demanda real fue MAYOR al pronóstico.
# La empresa subestimó la demanda.
#
# Error < 0
# La demanda real fue MENOR al pronóstico.
# La empresa sobreestimó la demanda.


df["Error_pronostico"] = (
    df["Demanda_real"]
    - df["Pronostico"]
)


# =============================================================================
# 11. CALCULAR ERROR PORCENTUAL ABSOLUTO
# =============================================================================
#
# Fórmula:
#
#            |Real - Pronóstico|
# APE = ----------------------------- x 100
#                  Real
#
# APE = Absolute Percentage Error
#
# Utilizamos valor absoluto porque queremos medir
# el tamaño del error independientemente de su dirección.


df["APE"] = (
    abs(
        df["Demanda_real"]
        - df["Pronostico"]
    )
    /
    df["Demanda_real"]
) * 100


# =============================================================================
# 12. CALCULAR MAPE
# =============================================================================
#
# MAPE = Mean Absolute Percentage Error
#
# Es el promedio de los errores porcentuales absolutos.
#
# Fórmula conceptual:
#
#           1       |Real - Pronóstico|
# MAPE = -------- Σ --------------------- x 100
#           n              Real
#
#
# EJEMPLO:
#
# MAPE = 15 %
#
# significa que, en promedio,
# el pronóstico se desvía aproximadamente 15 %
# respecto a la demanda real.
#
# IMPORTANTE:
# No debemos interpretar automáticamente un MAPE
# como "bueno" o "malo".
#
# Su interpretación depende del sector,
# comportamiento de la demanda y contexto empresarial.


MAPE = df["APE"].mean()


# =============================================================================
# 13. CALCULAR DISPONIBILIDAD TOTAL
# =============================================================================
#
# Disponibilidad =
#
# Inventario inicial + Compras recibidas


df["Disponible"] = (
    df["Inventario_inicial"]
    + df["Compras_recibidas"]
)


# =============================================================================
# 14. CALCULAR VENTAS / DEMANDA ATENDIDA
# =============================================================================
#
# La empresa solamente puede atender la demanda
# hasta donde tenga unidades disponibles.
#
# Por eso:
#
# Demanda atendida =
#
# mínimo entre:
#
# Demanda real
# Disponible


df["Demanda_atendida"] = np.minimum(
    df["Demanda_real"],
    df["Disponible"]
)


# =============================================================================
# 15. CALCULAR STOCKOUT
# =============================================================================
#
# Stockout:
#
# unidades de demanda que no pudieron ser atendidas.
#
# Fórmula:
#
# Stockout =
#
# MAX(
#      Demanda real - Disponible,
#      0
# )
#
#
# Si:
#
# Demanda = 1.500
# Disponible = 1.300
#
# Stockout = 200 unidades.
#
#
# Si:
#
# Demanda = 1.000
# Disponible = 1.300
#
# Stockout = 0.


df["Stockout"] = np.maximum(
    df["Demanda_real"]
    - df["Disponible"],
    0
)


# =============================================================================
# 16. INVENTARIO FINAL
# =============================================================================
#
# Fórmula:
#
# Inventario final =
#
# Inventario inicial
# + Compras recibidas
# - Demanda atendida


df["Inventario_final"] = (

    df["Inventario_inicial"]

    + df["Compras_recibidas"]

    - df["Demanda_atendida"]

)


# =============================================================================
# 17. NIVEL DE SERVICIO
# =============================================================================
#
# Vamos a utilizar un indicador sencillo:
#
#                       Demanda atendida
# Nivel de servicio = --------------------- x 100
#                        Demanda real
#
#
# Ejemplo:
#
# Demanda real = 1.000
#
# Demanda atendida = 950
#
# Nivel de servicio = 95 %


df["Nivel_servicio"] = (

    df["Demanda_atendida"]

    /

    df["Demanda_real"]

) * 100


# =============================================================================
# 18. COBERTURA DEL INVENTARIO
# =============================================================================
#
# Queremos saber aproximadamente cuántos meses
# puede cubrir el inventario final.
#
# Fórmula simplificada:
#
#                    Inventario final
# Cobertura = --------------------------------
#             Demanda promedio mensual
#
#
# Ejemplo:
#
# Inventario = 2.000
#
# Demanda promedio = 1.000
#
# Cobertura = 2 meses


demanda_promedio = df["Demanda_real"].mean()


df["Cobertura_meses"] = (

    df["Inventario_final"]

    /

    demanda_promedio

)


# =============================================================================
# 19. COSTO DEL INVENTARIO
# =============================================================================
#
# Para simplificar:
#
# Valor inventario =
#
# Inventario final x precio unitario
#
#
# IMPORTANTE:
#
# Esto NO representa todavía el costo total de mantener inventario.
#
# Simplemente representa el valor monetario aproximado
# del inventario almacenado.


df["Valor_inventario_USD"] = (

    df["Inventario_final"]

    *

    df["Precio_USD"]

)


# =============================================================================
# 20. CREAR UNA VARIABLE DE RIESGO
# =============================================================================
#
# Esta variable será útil posteriormente para Machine Learning.
#
# Por ahora solamente la construiremos.
#
# Definimos:
#
# Riesgo_stockout = 1
#
# cuando ocurre un stockout.
#
# Riesgo_stockout = 0
#
# cuando no ocurre.


df["Riesgo_stockout"] = np.where(
    df["Stockout"] > 0,
    1,
    0
)


# =============================================================================
# 21. RESUMEN DE INDICADORES
# =============================================================================


nivel_servicio_global = (

    df["Demanda_atendida"].sum()

    /

    df["Demanda_real"].sum()

) * 100


stockout_total = df["Stockout"].sum()


periodos_stockout = (
    df["Stockout"] > 0
).sum()


inventario_promedio = df["Inventario_final"].mean()


lead_time_promedio = df["Lead_time_dias"].mean()


lead_time_std = df["Lead_time_dias"].std()


valor_inventario_promedio = (
    df["Valor_inventario_USD"].mean()
)


# =============================================================================
# 22. MOSTRAR KPIs
# =============================================================================


print("\n")
print("=" * 70)
print("DASHBOARD GERENCIAL - GLOBALBIKE COMPONENTS")
print("=" * 70)


print(
    f"\nDemanda promedio mensual: "
    f"{demanda_promedio:,.0f} unidades"
)


print(
    f"MAPE del pronóstico: "
    f"{MAPE:.2f}%"
)


print(
    f"Nivel de servicio global: "
    f"{nivel_servicio_global:.2f}%"
)


print(
    f"Stockout acumulado: "
    f"{stockout_total:,.0f} unidades"
)


print(
    f"Periodos con stockout: "
    f"{periodos_stockout} de {len(df)}"
)


print(
    f"Inventario final promedio: "
    f"{inventario_promedio:,.0f} unidades"
)


print(
    f"Lead time promedio: "
    f"{lead_time_promedio:.1f} días"
)


print(
    f"Desviación del lead time: "
    f"{lead_time_std:.1f} días"
)


print(
    f"Valor promedio del inventario: "
    f"USD {valor_inventario_promedio:,.0f}"
)


print("\n" + "=" * 70)


# =============================================================================
# 23. GRÁFICA 1
# DEMANDA REAL VS PRONÓSTICO
# =============================================================================
#
# OBJETIVO:
#
# Observar visualmente qué tan bien el pronóstico
# sigue el comportamiento de la demanda.


plt.figure(figsize=(14, 6))


plt.plot(

    df["Fecha"],

    df["Demanda_real"],

    marker="o",

    linewidth=2.5,

    color="#1565C0",

    label="Demanda real"

)


plt.plot(

    df["Fecha"],

    df["Pronostico"],

    marker="o",

    linestyle="--",

    linewidth=2,

    color="#EF6C00",

    label="Pronóstico"

)


plt.title(
    "GlobalBike Components | Demanda real vs Pronóstico",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel("Unidades")


plt.legend()


plt.xticks(rotation=45)


plt.tight_layout()


plt.show()


# INTERPRETACIÓN AUTOMÁTICA

print("\nINTERPRETACIÓN GRÁFICA 1")

print(
    f"""
El MAPE obtenido es {MAPE:.2f}%.

La gráfica permite observar los periodos donde el pronóstico
se separa de la demanda real.

Cuando la línea del pronóstico se encuentra por debajo
de la demanda real, GlobalBike está subestimando la demanda.

Si esta situación coincide con bajos niveles de inventario
o lead times elevados, el riesgo de stockout puede aumentar.

La pregunta gerencial no debe ser solamente:

"¿Qué tan preciso es nuestro forecast?"

sino:

"¿Qué consecuencias tiene el error del forecast
sobre las decisiones de inventario y compras?"
"""
)


# =============================================================================
# 24. GRÁFICA 2
# ERROR PORCENTUAL DEL PRONÓSTICO
# =============================================================================


plt.figure(figsize=(14, 6))


colores_error = np.where(

    df["APE"] > MAPE,

    "#D32F2F",

    "#43A047"

)


plt.bar(

    df["Periodo"],

    df["APE"],

    color=colores_error

)


plt.axhline(

    MAPE,

    color="#212121",

    linestyle="--",

    linewidth=2,

    label=f"MAPE promedio = {MAPE:.1f}%"

)


plt.title(
    "Error porcentual absoluto del pronóstico",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel("Error (%)")


plt.xlabel("Periodo")


plt.xticks(rotation=70)


plt.legend()


plt.tight_layout()


plt.show()


print("\nINTERPRETACIÓN GRÁFICA 2")

print(
    """
Las barras rojas representan periodos cuyo error porcentual
está por encima del error promedio.

Estos meses merecen especial atención porque podrían estar
asociados con decisiones de compra menos precisas.

Una pregunta interesante para el análisis posterior será:

¿Los periodos con mayor error del forecast coinciden
con periodos de stockout?
"""
)


# =============================================================================
# 25. GRÁFICA 3
# COMPORTAMIENTO DEL LEAD TIME
# =============================================================================


plt.figure(figsize=(14, 6))


plt.plot(

    df["Fecha"],

    df["Lead_time_dias"],

    marker="o",

    linewidth=2.5,

    color="#6A1B9A"

)


plt.axhline(

    lead_time_promedio,

    color="#FF8F00",

    linestyle="--",

    linewidth=2,

    label=f"Promedio = {lead_time_promedio:.1f} días"

)


plt.fill_between(

    df["Fecha"],

    55,

    df["Lead_time_dias"],

    color="#CE93D8",

    alpha=0.25

)


plt.title(
    "Variabilidad del Lead Time Internacional",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel("Días")


plt.xlabel("Periodo")


plt.legend()


plt.xticks(rotation=45)


plt.tight_layout()


plt.show()


print("\nINTERPRETACIÓN GRÁFICA 3")

print(
    f"""
El lead time promedio es de {lead_time_promedio:.1f} días,
con una desviación estándar aproximada de {lead_time_std:.1f} días.

Esto significa que GlobalBike no enfrenta solamente
un lead time relativamente largo.

También enfrenta INCERTIDUMBRE en el tiempo de reposición.

Desde la gestión de inventarios esto es importante porque
mayor incertidumbre puede requerir mayor protección,
por ejemplo mediante stock de seguridad.

Sin embargo, aumentar inventario también tiene un costo.

Por eso aparece un trade-off:

MAYOR PROTECCIÓN
        ↕
MAYOR INVENTARIO
"""
)


# =============================================================================
# 26. GRÁFICA 4
# INVENTARIO FINAL
# =============================================================================


plt.figure(figsize=(14, 6))


colores_inventario = [

    "#C62828" if x < 300

    else "#F9A825" if x < 700

    else "#2E7D32"

    for x in df["Inventario_final"]

]


plt.bar(

    df["Periodo"],

    df["Inventario_final"],

    color=colores_inventario

)


plt.title(
    "Inventario final del controlador E-500",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel("Unidades")


plt.xlabel("Periodo")


plt.xticks(rotation=70)


plt.tight_layout()


plt.show()


print("\nINTERPRETACIÓN GRÁFICA 4")

print(
    """
Esta gráfica permite observar la evolución del inventario.

Para fines pedagógicos:

VERDE:
inventario relativamente alto.

AMARILLO:
inventario que comienza a requerir atención.

ROJO:
nivel de inventario muy bajo.

IMPORTANTE:

Estos límites son ilustrativos.

En una empresa real deberían definirse utilizando
políticas de inventario, demanda, lead time,
nivel de servicio objetivo y stock de seguridad.
"""
)


# =============================================================================
# 27. GRÁFICA 5
# STOCKOUTS
# =============================================================================


plt.figure(figsize=(14, 6))


plt.bar(

    df["Periodo"],

    df["Stockout"],

    color="#D32F2F"

)


plt.title(
    "Stockouts del controlador E-500",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel("Unidades no atendidas")


plt.xlabel("Periodo")


plt.xticks(rotation=70)


plt.tight_layout()


plt.show()


print("\nINTERPRETACIÓN GRÁFICA 5")

print(
    f"""
Durante los 36 periodos analizados se presentaron
{periodos_stockout} periodos con stockout.

El stockout acumulado fue de aproximadamente
{stockout_total:,.0f} unidades.

Ahora aparece una pregunta mucho más interesante:

¿QUÉ VARIABLES ESTÁN ASOCIADAS CON ESTOS STOCKOUTS?

¿Lead time?

¿Error del forecast?

¿Inventario inicial?

¿Demanda?

Esta pregunta será fundamental cuando avancemos
hacia Machine Learning.
"""
)


# =============================================================================
# 28. GRÁFICA 6
# NIVEL DE SERVICIO
# =============================================================================


plt.figure(figsize=(14, 6))


plt.plot(

    df["Fecha"],

    df["Nivel_servicio"],

    marker="o",

    linewidth=2.5,

    color="#00897B"

)


# Meta empresarial hipotética:
# 95 %

meta_servicio = 95


plt.axhline(

    meta_servicio,

    color="#D32F2F",

    linestyle="--",

    linewidth=2,

    label="Meta = 95%"

)


plt.title(
    "Nivel de servicio del controlador E-500",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel("Nivel de servicio (%)")


plt.ylim(0, 105)


plt.legend()


plt.xticks(rotation=45)


plt.tight_layout()


plt.show()


print("\nINTERPRETACIÓN GRÁFICA 6")

print(
    f"""
El nivel de servicio global obtenido es
{nivel_servicio_global:.2f}%.

La meta utilizada para el ejemplo es 95%.

Cuando el nivel de servicio cae,
la empresa no está atendiendo completamente
la demanda del controlador E-500.

Desde una perspectiva gerencial,
el reto consiste en mejorar este indicador
sin incrementar innecesariamente el inventario.
"""
)


# =============================================================================
# 29. GRÁFICA 7
# COBERTURA DEL INVENTARIO
# =============================================================================


plt.figure(figsize=(14, 6))


plt.bar(

    df["Periodo"],

    df["Cobertura_meses"],

    color="#0288D1"

)


plt.axhline(

    1,

    color="#F57C00",

    linestyle="--",

    linewidth=2,

    label="1 mes de cobertura"

)


plt.title(
    "Cobertura aproximada del inventario",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel("Meses de cobertura")


plt.xlabel("Periodo")


plt.xticks(rotation=70)


plt.legend()


plt.tight_layout()


plt.show()


print("\nINTERPRETACIÓN GRÁFICA 7")

print(
    """
La cobertura indica aproximadamente cuántos meses
de demanda podrían atenderse con el inventario disponible.

Una cobertura muy baja puede aumentar el riesgo de stockout.

Pero una cobertura excesivamente alta puede significar:

- capital inmovilizado,
- mayores costos de almacenamiento,
- riesgo de obsolescencia.

Por eso el objetivo no es maximizar inventario.

El objetivo es encontrar un nivel adecuado
para las características de la cadena.
"""
)


# =============================================================================
# 30. GRÁFICA 8
# RELACIÓN ENTRE LEAD TIME Y NIVEL DE SERVICIO
# =============================================================================


plt.figure(figsize=(10, 7))


sns.scatterplot(

    data=df,

    x="Lead_time_dias",

    y="Nivel_servicio",

    size="Demanda_real",

    hue="Riesgo_stockout",

    palette={
        0: "#2E7D32",
        1: "#D32F2F"
    },

    sizes=(70, 250),

    alpha=0.8

)


plt.title(
    "Lead Time vs Nivel de Servicio",
    fontsize=16,
    fontweight="bold"
)


plt.xlabel("Lead time (días)")


plt.ylabel("Nivel de servicio (%)")


plt.tight_layout()


plt.show()


print("\nINTERPRETACIÓN GRÁFICA 8")

print(
    """
Cada punto representa un periodo.

Los puntos rojos corresponden a periodos donde
se presentó algún stockout.

Esta gráfica nos permite explorar si los lead times elevados
aparecen asociados con menores niveles de servicio.

IMPORTANTE:

Una relación visual NO demuestra causalidad.

Para afirmar que una variable causa otra
necesitaríamos un análisis mucho más profundo.

Aquí solamente estamos buscando patrones
que orienten nuestra investigación.
"""
)


# =============================================================================
# 31. MATRIZ DE CORRELACIONES
# =============================================================================
#
# La correlación mide el grado de relación LINEAL
# entre dos variables.
#
# Su valor puede estar entre -1 y +1.
#
#
# Cerca de +1:
# relación positiva fuerte.
#
# Cerca de -1:
# relación negativa fuerte.
#
# Cerca de 0:
# poca relación lineal.
#
#
# MUY IMPORTANTE:
#
# CORRELACIÓN NO SIGNIFICA CAUSALIDAD.


variables_correlacion = [

    "Demanda_real",

    "Pronostico",

    "Lead_time_dias",

    "Inventario_inicial",

    "Stockout",

    "Nivel_servicio",

    "APE"

]


correlacion = (
    df[variables_correlacion]
    .corr()
)


plt.figure(figsize=(11, 8))


sns.heatmap(

    correlacion,

    annot=True,

    cmap="RdYlGn",

    center=0,

    fmt=".2f",

    linewidths=0.5

)


plt.title(
    "Matriz de correlaciones | GlobalBike Components",
    fontsize=16,
    fontweight="bold"
)


plt.tight_layout()


plt.show()


print("\nINTERPRETACIÓN MATRIZ DE CORRELACIONES")

print(
    """
La matriz permite explorar qué variables se mueven juntas.

Por ejemplo:

Una correlación positiva entre dos variables indica que
cuando una aumenta, la otra tiende también a aumentar.

Una correlación negativa indica que cuando una aumenta,
la otra tiende a disminuir.

Pero debemos recordar:

CORRELACIÓN ≠ CAUSALIDAD.

La matriz sirve para generar preguntas,
no para demostrar automáticamente causas.
"""
)


# =============================================================================
# 32. IDENTIFICAR LOS PERIODOS MÁS CRÍTICOS
# =============================================================================
#
# Vamos a mostrar solamente los periodos
# donde ocurrió stockout.


periodos_criticos = df[
    df["Stockout"] > 0
][[

    "Periodo",

    "Demanda_real",

    "Pronostico",

    "Error_pronostico",

    "Lead_time_dias",

    "Inventario_inicial",

    "Stockout",

    "Nivel_servicio"

]]


print("\n")
print("=" * 70)
print("PERIODOS CRÍTICOS")
print("=" * 70)


if len(periodos_criticos) > 0:

    print(
        periodos_criticos.to_string(
            index=False
        )
    )

else:

    print(
        "No se presentaron stockouts."
    )


# =============================================================================
# 33. COMPARAR PERIODOS CON Y SIN STOCKOUT
# =============================================================================
#
# Esta comparación es muy importante.
#
# Queremos preguntarnos:
#
# ¿Qué cambia cuando aparece un stockout?
#
#
# Compararemos:
#
# demanda,
# error del forecast,
# lead time,
# inventario inicial.


comparacion = df.groupby(
    "Riesgo_stockout"
)[[

    "Demanda_real",

    "APE",

    "Lead_time_dias",

    "Inventario_inicial"

]].mean()


comparacion.index = [

    "Sin stockout",

    "Con stockout"

][:len(comparacion)]


print("\n")
print("=" * 70)
print("COMPARACIÓN DE PERIODOS")
print("=" * 70)


print(
    comparacion.round(2)
)


# =============================================================================
# 34. GRÁFICA DE COMPARACIÓN
# =============================================================================


comparacion_normalizada = (

    comparacion

    /

    comparacion.max()

) * 100


comparacion_normalizada.T.plot(

    kind="bar",

    figsize=(12, 7),

    color=[
        "#43A047",
        "#E53935"
    ]

)


plt.title(
    "Perfil relativo: periodos con y sin stockout",
    fontsize=16,
    fontweight="bold"
)


plt.ylabel("Índice relativo (0-100)")


plt.xlabel("Variable")


plt.xticks(rotation=30)


plt.legend(
    title="Situación"
)


plt.tight_layout()


plt.show()


print("\nINTERPRETACIÓN")

print(
    """
Esta gráfica no muestra unidades originales.

Los indicadores fueron normalizados para facilitar
la comparación visual.

La pregunta que debemos hacernos es:

¿Los periodos con stockout parecen tener
características diferentes de los periodos normales?

Si encontramos diferencias,
estas variables podrían convertirse en candidatos
para nuestro modelo predictivo.
"""
)


# =============================================================================
# 35. RESUMEN AUTOMÁTICO DEL DIAGNÓSTICO
# =============================================================================


print("\n")
print("=" * 70)
print("DIAGNÓSTICO FINAL - ETAPA DATA ANALYTICS")
print("=" * 70)


print(
    f"""

1. DEMANDA

La demanda promedio mensual del E-500 es aproximadamente
{demanda_promedio:,.0f} unidades.


2. FORECAST

El MAPE es aproximadamente
{MAPE:.2f}%.

Esto evidencia que existe incertidumbre
en la estimación de la demanda.


3. LEAD TIME

El lead time promedio es aproximadamente
{lead_time_promedio:.1f} días.

Además, presenta variabilidad,
lo que incrementa la incertidumbre
del proceso de abastecimiento.


4. INVENTARIO

El inventario final promedio es aproximadamente
{inventario_promedio:,.0f} unidades.

Sin embargo, el promedio por sí solo
no muestra completamente el problema,
porque existen diferencias importantes
entre periodos.


5. STOCKOUT

Se presentaron
{periodos_stockout} periodos con stockout.

El faltante acumulado fue aproximadamente
{stockout_total:,.0f} unidades.


6. NIVEL DE SERVICIO

El nivel de servicio global fue aproximadamente
{nivel_servicio_global:.2f}%.


7. CONCLUSIÓN

GlobalBike enfrenta un problema que no parece depender
de una sola variable.

El comportamiento de la demanda,
los errores del pronóstico,
el inventario disponible
y las condiciones de abastecimiento
deben analizarse conjuntamente.

El siguiente paso será utilizar MACHINE LEARNING
para responder una nueva pregunta:

¿PODEMOS ANTICIPAR EL RIESGO DE STOCKOUT
ANTES DE QUE OCURRA?

"""
)


# =============================================================================
# 36. GUARDAR LA BASE DE DATOS
# =============================================================================
#
# Finalmente guardamos todos los resultados.
#
# Esto permitirá utilizar exactamente la misma base
# en las siguientes etapas del ejercicio.


df.to_csv(

    "GlobalBike_Data_Analytics.csv",

    index=False,

    encoding="utf-8-sig"

)


print(
    "\nArchivo creado correctamente:"
    " GlobalBike_Data_Analytics.csv"
)


# =============================================================================
# FIN ETAPA 7
# =============================================================================
#
# LO QUE ACABAMOS DE HACER:
#
# DATOS
#   ↓
# KPIs
#   ↓
# VISUALIZACIONES
#   ↓
# IDENTIFICACIÓN DE PATRONES
#   ↓
# DIAGNÓSTICO
#
#
# SIGUIENTE ETAPA:
#
#               MACHINE LEARNING
#
#                       ↓
#
#       ¿Podemos anticipar un stockout?
#
# =============================================================================
```
