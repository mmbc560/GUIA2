```python
# ============================================================
# DIAGNÓSTICO DE CAPACIDADES DIGITALES - MODELO DCM
# Actividad para estudiantes Trend Tech
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ============================================================
# PARTE 1. LOS ESTUDIANTES SOLO DEBEN CAMBIAR ESTA SECCIÓN
# ============================================================

nombre_empresa = "Nombre de la empresa analizada"

datos = {
    "Capacidad_DCM": [
        "Connected Customer",
        "Intelligent Supply",
        "Smart Operations",
        "Synchronized Planning",
        "Dynamic Fulfillment",
        "Digital Development"
    ],

    # Cambiar únicamente estos números según la empresa analizada.
    # Escala:
    # 1 = Muy bajo: casi no hay evidencia.
    # 2 = Bajo: hay señales mínimas, pero poco desarrolladas.
    # 3 = Medio: hay algunas herramientas digitales, pero no integradas.
    # 4 = Alto: hay uso claro de tecnología, datos o automatización.
    # 5 = Muy alto: la empresa muestra capacidades digitales avanzadas.

    "Puntaje": [
        4,  # Connected Customer
        3,  # Intelligent Supply
        2,  # Smart Operations
        3,  # Synchronized Planning
        4,  # Dynamic Fulfillment
        2   # Digital Development
    ],

    # Aquí pueden escribir la evidencia encontrada en la página web, noticias, LinkedIn o informes.
    "Evidencia": [
        "Tiene tienda online y canales digitales de atención",
        "Menciona proveedores y trazabilidad, pero con poca evidencia",
        "No se observa información clara sobre automatización operativa",
        "Publica información sobre cobertura e inventarios",
        "Ofrece entregas a domicilio y seguimiento de pedidos",
        "Tiene poca evidencia visible de innovación digital"
    ]
}

# ============================================================
# PARTE 2. NO MODIFICAR
# Esta parte genera las tablas, gráficas e interpretación.
# ============================================================

df = pd.DataFrame(datos)

# ============================================================
# COLORES TIPO SEMÁFORO
# Verde = Alto
# Amarillo = Medio
# Rojo = Bajo
# ============================================================

def color_semaforo(valor):
    if valor >= 4:
        return "green"
    elif valor == 3:
        return "gold"
    else:
        return "red"

colores = [color_semaforo(x) for x in df["Puntaje"]]

print("Empresa analizada:", nombre_empresa)
print("\nTabla de diagnóstico:")
display(df)

# ============================================================
# GRÁFICO 1: DIAGNÓSTICO GENERAL
# ============================================================

plt.figure(figsize=(11,5))
plt.bar(df["Capacidad_DCM"], df["Puntaje"], color=colores)

plt.title("Diagnóstico de capacidades digitales DCM - " + nombre_empresa)
plt.xlabel("Capacidades digitales DCM")
plt.ylabel("Nivel de madurez")

plt.ylim(0,5)
plt.xticks(rotation=30, ha="right")
plt.grid(axis="y", alpha=0.3)

for i, valor in enumerate(df["Puntaje"]):
    plt.text(i, valor + 0.1, str(valor), ha="center", fontsize=11)

plt.show()

# ============================================================
# GRÁFICO 2: OPORTUNIDADES DE MEJORA
# ============================================================

df["Brecha_de_mejora"] = 5 - df["Puntaje"]

df_brecha = df.sort_values("Brecha_de_mejora", ascending=False)

# Colores tipo semáforo para brechas
# Brecha alta = rojo
# Brecha media = amarillo
# Brecha baja = verde

def color_brecha(valor):
    if valor >= 3:
        return "red"
    elif valor == 2:
        return "gold"
    else:
        return "green"

colores_brecha = [color_brecha(x) for x in df_brecha["Brecha_de_mejora"]]

plt.figure(figsize=(11,5))
plt.bar(df_brecha["Capacidad_DCM"], df_brecha["Brecha_de_mejora"], color=colores_brecha)

plt.title("Oportunidades de mejora digital - " + nombre_empresa)
plt.xlabel("Capacidades digitales DCM")
plt.ylabel("Brecha frente al nivel ideal")

plt.ylim(0,5)
plt.xticks(rotation=30, ha="right")
plt.grid(axis="y", alpha=0.3)

for i, valor in enumerate(df_brecha["Brecha_de_mejora"]):
    plt.text(i, valor + 0.1, str(valor), ha="center", fontsize=11)

plt.show()

# ============================================================
# GRÁFICO 3: PERFIL RADAR
# ============================================================

categorias = df["Capacidad_DCM"].tolist()
valores = df["Puntaje"].tolist()

valores += valores[:1]

angulos = np.linspace(0, 2 * np.pi, len(categorias), endpoint=False).tolist()
angulos += angulos[:1]

plt.figure(figsize=(8,8))

ax = plt.subplot(111, polar=True)

ax.plot(angulos, valores, color="green", linewidth=3)
ax.fill(angulos, valores, color="gold", alpha=0.35)

ax.set_xticks(angulos[:-1])
ax.set_xticklabels(categorias, fontsize=10)

ax.set_ylim(0,5)

ax.set_title("Perfil digital DCM - " + nombre_empresa, fontsize=14)

ax.grid(True)

plt.show()

# ============================================================
# INTERPRETACIÓN AUTOMÁTICA
# ============================================================

capacidad_mas_fuerte = df.loc[df["Puntaje"].idxmax()]
capacidad_mas_debil = df.loc[df["Puntaje"].idxmin()]

print("\nInterpretación automática:")

print("\nLa capacidad más fuerte es:")
print(capacidad_mas_fuerte["Capacidad_DCM"])
print("Puntaje:", capacidad_mas_fuerte["Puntaje"])
print("Evidencia:", capacidad_mas_fuerte["Evidencia"])

print("\nLa principal oportunidad de mejora está en:")
print(capacidad_mas_debil["Capacidad_DCM"])
print("Puntaje:", capacidad_mas_debil["Puntaje"])
print("Evidencia:", capacidad_mas_debil["Evidencia"])

print("\nPregunta final para el grupo:")
print("¿Dónde está perdiendo valor la empresa por no fortalecer esta capacidad digital y qué solución tecnológica podrían proponer?")
```
