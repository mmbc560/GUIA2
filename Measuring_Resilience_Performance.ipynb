```python
# ================================================================
# SUPPLY CHAIN RESILIENCE STRATEGY ANALYSIS
# ================================================================
#
# OBJETIVO
#
# Comparar diferentes estrategias de resiliencia de la cadena
# de suministro utilizando indicadores operacionales y financieros.
#
# Queremos responder:
#
# 1. ¿Qué estrategia ofrece mayor nivel de servicio?
# 2. ¿Cuál permite una recuperación más rápida?
# 3. ¿Cuál reduce más los stockouts?
# 4. ¿Cuál requiere mayor inversión en inventario?
# 5. ¿Cuál tiene mayor costo total?
# 6. ¿Qué estrategia ofrece el mejor equilibrio entre:
#
#       COST
#       SERVICE
#       INVENTORY
#       RISK
#       RECOVERY
#
#
# IMPORTANTE:
#
# La estrategia más resiliente NO necesariamente será
# la estrategia con mayor valor para el negocio.
#
# ================================================================


# ================================================================
# 1. IMPORTAR LIBRERÍAS
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ================================================================
# 2. DEFINIR LAS ESTRATEGIAS
# ================================================================
#
# Vamos a comparar cinco estrategias.
#
#
# STRATEGY 1 - Current Policy
#
# La empresa mantiene su política actual.
# Bajo costo, pero protección limitada.
#
#
# STRATEGY 2 - Higher Safety Stock
#
# La empresa aumenta el inventario de seguridad.
# Mayor protección, pero mayor capital de trabajo.
#
#
# STRATEGY 3 - Dual Sourcing
#
# La empresa divide el abastecimiento entre dos proveedores.
# Reduce dependencia y mejora capacidad de respuesta.
#
#
# STRATEGY 4 - Regional Backup Supplier
#
# La empresa conserva su proveedor principal global,
# pero desarrolla un proveedor regional de respaldo.
#
#
# STRATEGY 5 - Maximum Resilience
#
# Combinación de:
#
# - mayor safety stock
# - dual sourcing
# - proveedor regional
# - mayor redundancia
#
# Esta estrategia ofrece mucha protección,
# pero también puede ser la más costosa.
#
# ================================================================


strategies = [

    "Current Policy",

    "Higher Safety Stock",

    "Dual Sourcing",

    "Regional Backup",

    "Maximum Resilience"

]


# ================================================================
# 3. CREAR DATOS PARA CADA ESTRATEGIA
# ================================================================
#
# Los siguientes datos son ficticios y tienen fines pedagógicos.
#
# Cada estrategia tendrá diferentes resultados en:
#
# - Service Level
# - Recovery Time
# - Stockout Frequency
# - Inventory Investment
# - Supplier Reliability
# - Total Cost
#
# ================================================================


data = {

    "Strategy": strategies,


    # ------------------------------------------------------------
    # SERVICE LEVEL
    #
    # Porcentaje de demanda atendida adecuadamente.
    #
    # Mayor valor = mejor.
    #
    # ------------------------------------------------------------

    "Service_Level": [

        91,
        96,
        97,
        98,
        99

    ],


    # ------------------------------------------------------------
    # RECOVERY TIME
    #
    # Número de días que tarda la cadena de suministro
    # en regresar a condiciones aceptables después
    # de una disrupción.
    #
    # Menor valor = mejor.
    #
    # ------------------------------------------------------------

    "Recovery_Time_Days": [

        24,
        18,
        11,
        8,
        5

    ],


    # ------------------------------------------------------------
    # STOCKOUT FREQUENCY
    #
    # Porcentaje de periodos en los cuales
    # ocurre un faltante.
    #
    # Menor valor = mejor.
    #
    # ------------------------------------------------------------

    "Stockout_Frequency": [

        8.5,
        4.8,
        3.4,
        2.6,
        1.5

    ],


    # ------------------------------------------------------------
    # INVENTORY INVESTMENT
    #
    # Dinero promedio invertido en inventario.
    #
    # Supongamos valores en miles de dólares.
    #
    # Menor inversión no siempre es mejor:
    #
    # un inventario demasiado bajo puede aumentar riesgo.
    #
    # ------------------------------------------------------------

    "Inventory_Investment": [

        310,
        430,
        355,
        370,
        480

    ],


    # ------------------------------------------------------------
    # SUPPLIER RELIABILITY
    #
    # Indicador de confiabilidad del abastecimiento.
    #
    # Mayor valor = mejor.
    #
    # ------------------------------------------------------------

    "Supplier_Reliability": [

        84,
        86,
        94,
        96,
        98

    ],


    # ------------------------------------------------------------
    # TOTAL COST
    #
    # Costo anual total de la estrategia.
    #
    # Incluye de forma simplificada:
    #
    # - compra
    # - transporte
    # - inventario
    # - costos de resiliencia
    # - costos de stockout
    # - recuperación
    #
    # Valores en miles de dólares.
    #
    # ------------------------------------------------------------

    "Total_Cost": [

        1450,
        1540,
        1495,
        1515,
        1660

    ]

}


df = pd.DataFrame(data)


# ================================================================
# 4. MOSTRAR BASE DE DATOS
# ================================================================


print("\n" + "=" * 100)

print("SUPPLY CHAIN RESILIENCE STRATEGIES")

print("=" * 100)


print(

    df.to_string(
        index=False
    )

)


# ================================================================
# 5. GRÁFICA 1
#
# SERVICE LEVEL
# ================================================================
#
# Esta gráfica muestra qué estrategia protege mejor
# la capacidad de atender al cliente.
#
# ================================================================


service_colors = [

    "#E74C3C",   # rojo
    "#F39C12",   # naranja
    "#3498DB",   # azul
    "#1ABC9C",   # turquesa
    "#2ECC71"    # verde

]


plt.figure(
    figsize=(12,7)
)


bars = plt.bar(

    df["Strategy"],

    df["Service_Level"],

    color=service_colors

)


plt.title(

    "Service Level by Resilience Strategy",

    fontsize=18,

    fontweight="bold"

)


plt.ylabel(
    "Service Level (%)"
)


plt.ylim(
    85,
    101
)


# Línea de referencia.
#
# Supongamos que la meta de servicio es 96%.

plt.axhline(

    96,

    linestyle="--",

    color="#34495E",

    label="Target Service Level = 96%"

)


# Colocar porcentaje encima de cada barra.

for bar in bars:

    height = bar.get_height()

    plt.text(

        bar.get_x()
        + bar.get_width()/2,

        height + 0.3,

        f"{height:.1f}%",

        ha="center",

        fontweight="bold"

    )


plt.xticks(
    rotation=15
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# INTERPRETACIÓN
# ================================================================
#
# Las estrategias más resilientes normalmente
# ofrecen mayor nivel de servicio.
#
# Pero debemos preguntarnos:
#
# ¿Cuánto nos cuesta obtener ese nivel adicional de servicio?
#
# ================================================================


# ================================================================
# 6. GRÁFICA 2
#
# RECOVERY TIME
# ================================================================
#
# Esta gráfica representa una dimensión fundamental
# de la resiliencia:
#
# ¿Qué tan rápido puede recuperarse la cadena?
#
# ================================================================


plt.figure(
    figsize=(12,7)
)


recovery_colors = [

    "#C0392B",
    "#E67E22",
    "#F1C40F",
    "#3498DB",
    "#2ECC71"

]


bars = plt.bar(

    df["Strategy"],

    df["Recovery_Time_Days"],

    color=recovery_colors

)


plt.title(

    "Recovery Time After a Supply Disruption",

    fontsize=18,

    fontweight="bold"

)


plt.ylabel(
    "Recovery Time (Days)"
)


for bar in bars:

    height = bar.get_height()

    plt.text(

        bar.get_x()
        + bar.get_width()/2,

        height + 0.4,

        f"{height:.0f} days",

        ha="center",

        fontweight="bold"

    )


plt.xticks(
    rotation=15
)


plt.grid(
    axis="y",
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# INTERPRETACIÓN
# ================================================================
#
# Un menor tiempo de recuperación significa
# una cadena de suministro más resiliente.
#
# Ejemplo:
#
# Current Policy:
# 24 días para recuperar desempeño.
#
# Maximum Resilience:
# 5 días.
#
#
# Sin embargo:
#
# lograr una recuperación tan rápida
# probablemente requiere más recursos.
#
# ================================================================


# ================================================================
# 7. GRÁFICA 3
#
# STOCKOUT FREQUENCY
# ================================================================
#
# Esta gráfica permite observar
# cuántas veces aparecen problemas de disponibilidad.
#
# ================================================================


plt.figure(
    figsize=(12,7)
)


stockout_colors = [

    "#E74C3C",
    "#E67E22",
    "#F1C40F",
    "#3498DB",
    "#2ECC71"

]


bars = plt.bar(

    df["Strategy"],

    df["Stockout_Frequency"],

    color=stockout_colors

)


plt.title(

    "Stockout Frequency by Strategy",

    fontsize=18,

    fontweight="bold"

)


plt.ylabel(
    "Stockout Frequency (%)"
)


for bar in bars:

    height = bar.get_height()

    plt.text(

        bar.get_x()
        + bar.get_width()/2,

        height + 0.15,

        f"{height:.1f}%",

        ha="center",

        fontweight="bold"

    )


plt.xticks(
    rotation=15
)


plt.grid(
    axis="y",
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# INTERPRETACIÓN
# ================================================================
#
# Las estrategias más resilientes reducen
# significativamente la frecuencia de faltantes.
#
#
# Pero esta reducción normalmente implica:
#
# - mayor inventario
# - mayor capacidad
# - proveedores adicionales
# - mayor costo
#
# ================================================================


# ================================================================
# 8. GRÁFICA 4
#
# INVENTORY INVESTMENT
# ================================================================
#
# Aquí aparece un trade-off importante.
#
#
# Más protección puede requerir
# más capital de trabajo.
#
# ================================================================


plt.figure(
    figsize=(12,7)
)


inventory_colors = [

    "#2ECC71",
    "#E67E22",
    "#3498DB",
    "#1ABC9C",
    "#E74C3C"

]


bars = plt.bar(

    df["Strategy"],

    df["Inventory_Investment"],

    color=inventory_colors

)


plt.title(

    "Inventory Investment Required",

    fontsize=18,

    fontweight="bold"

)


plt.ylabel(
    "Inventory Investment ($000)"
)


for bar in bars:

    height = bar.get_height()

    plt.text(

        bar.get_x()
        + bar.get_width()/2,

        height + 6,

        f"${height:.0f}k",

        ha="center",

        fontweight="bold"

    )


plt.xticks(
    rotation=15
)


plt.grid(
    axis="y",
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# INTERPRETACIÓN
# ================================================================
#
# Observe que Maximum Resilience
# requiere la mayor inversión.
#
#
# Esto ayuda a explicar que:
#
# "More resilience is not free."
#
#
# La empresa debe preguntarse si el beneficio
# justifica la inversión adicional.
#
# ================================================================


# ================================================================
# 9. GRÁFICA 5
#
# SUPPLIER RELIABILITY
# ================================================================


plt.figure(
    figsize=(12,7)
)


bars = plt.bar(

    df["Strategy"],

    df["Supplier_Reliability"],

    color=service_colors

)


plt.title(

    "Supplier Reliability by Strategy",

    fontsize=18,

    fontweight="bold"

)


plt.ylabel(
    "Supplier Reliability (%)"
)


plt.ylim(
    75,
    101
)


for bar in bars:

    height = bar.get_height()

    plt.text(

        bar.get_x()
        + bar.get_width()/2,

        height + 0.3,

        f"{height:.0f}%",

        ha="center",

        fontweight="bold"

    )


plt.xticks(
    rotation=15
)


plt.grid(
    axis="y",
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# 10. GRÁFICA 6
#
# TOTAL COST
# ================================================================
#
# Ahora observamos el costo financiero
# de implementar cada estrategia.
#
# ================================================================


plt.figure(
    figsize=(12,7)
)


cost_colors = [

    "#2ECC71",
    "#F39C12",
    "#3498DB",
    "#1ABC9C",
    "#E74C3C"

]


bars = plt.bar(

    df["Strategy"],

    df["Total_Cost"],

    color=cost_colors

)


plt.title(

    "Total Annual Supply Chain Cost",

    fontsize=18,

    fontweight="bold"

)


plt.ylabel(
    "Total Cost ($000)"
)


for bar in bars:

    height = bar.get_height()

    plt.text(

        bar.get_x()
        + bar.get_width()/2,

        height + 15,

        f"${height:.0f}k",

        ha="center",

        fontweight="bold"

    )


plt.xticks(
    rotation=15
)


plt.grid(
    axis="y",
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# INTERPRETACIÓN
# ================================================================
#
# Aquí aparece claramente el trade-off.
#
#
# Current Policy:
#
# menor protección
# pero costo relativamente bajo.
#
#
# Maximum Resilience:
#
# máximo nivel de protección
# pero también mayor costo.
#
#
# La estrategia óptima probablemente estará
# entre ambos extremos.
#
# ================================================================


# ================================================================
# 11. GRÁFICA 7
#
# COST VS SERVICE LEVEL
#
# + RECOVERY TIME COMO TAMAÑO DE BURBUJA
# ================================================================
#
# Esta gráfica integra tres dimensiones.
#
#
# X = Total Cost
#
# Y = Service Level
#
# Bubble Size = Recovery Time
#
#
# IMPORTANTE:
#
# una burbuja grande significa recuperación lenta.
#
# ================================================================


plt.figure(
    figsize=(12,8)
)


bubble_sizes = (

    df["Recovery_Time_Days"]

    * 35

)


plt.scatter(

    df["Total_Cost"],

    df["Service_Level"],

    s=bubble_sizes,

    color=service_colors,

    alpha=0.75,

    edgecolor="black",

    linewidth=1.2

)


for i, row in df.iterrows():

    plt.text(

        row["Total_Cost"] + 8,

        row["Service_Level"] + 0.15,

        row["Strategy"],

        fontsize=9

    )


plt.title(

    "Cost-Service-Recovery Trade-Off",

    fontsize=19,

    fontweight="bold"

)


plt.xlabel(
    "Total Cost ($000)"
)


plt.ylabel(
    "Service Level (%)"
)


plt.grid(
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# INTERPRETACIÓN
# ================================================================
#
# Idealmente buscamos:
#
# ALTO SERVICE LEVEL
#
# BAJO COSTO
#
# PEQUEÑO RECOVERY TIME
#
#
# Es decir:
#
# buscamos ubicarnos hacia la parte
# superior izquierda del gráfico
# con una burbuja pequeña.
#
# ================================================================


# ================================================================
# 12. NORMALIZAR INDICADORES
# ================================================================
#
# Ahora construiremos un índice combinado.
#
# Esto NO es una fórmula universal.
#
# Su propósito es ayudar a visualizar
# los trade-offs.
#
# ================================================================


# ------------------------------------------------------------
# FUNCIÓN PARA VARIABLES DONDE MÁS ES MEJOR
# ------------------------------------------------------------

def normalize_positive(series):

    return (

        (series - series.min())

        /

        (series.max() - series.min())

        * 100

    )


# ------------------------------------------------------------
# FUNCIÓN PARA VARIABLES DONDE MENOS ES MEJOR
# ------------------------------------------------------------

def normalize_negative(series):

    return (

        100

        -

        (
            (series - series.min())

            /

            (series.max() - series.min())

            * 100
        )

    )


# ================================================================
# 13. CONVERTIR TODO A ESCALA 0-100
# ================================================================


df["Service_Score"] = normalize_positive(

    df["Service_Level"]

)


df["Recovery_Score"] = normalize_negative(

    df["Recovery_Time_Days"]

)


df["Stockout_Score"] = normalize_negative(

    df["Stockout_Frequency"]

)


df["Inventory_Score"] = normalize_negative(

    df["Inventory_Investment"]

)


df["Reliability_Score"] = normalize_positive(

    df["Supplier_Reliability"]

)


df["Cost_Score"] = normalize_negative(

    df["Total_Cost"]

)


# ================================================================
# 14. CREAR RESILIENCE PERFORMANCE SCORE
# ================================================================
#
# Pesos gerenciales propuestos:
#
#
# Service Level           = 25%
#
# Recovery Time           = 20%
#
# Stockout Frequency      = 15%
#
# Inventory Investment    = 10%
#
# Supplier Reliability    = 15%
#
# Total Cost              = 15%
#
#
# SUMA = 100%
#
#
# Estos pesos pueden modificarse.
#
# ================================================================


df["Resilience_Performance_Score"] = (

    df["Service_Score"] * 0.25

    +

    df["Recovery_Score"] * 0.20

    +

    df["Stockout_Score"] * 0.15

    +

    df["Inventory_Score"] * 0.10

    +

    df["Reliability_Score"] * 0.15

    +

    df["Cost_Score"] * 0.15

)


# ================================================================
# 15. RANKING FINAL
# ================================================================


ranking = (

    df[
        [
            "Strategy",
            "Service_Level",
            "Recovery_Time_Days",
            "Stockout_Frequency",
            "Inventory_Investment",
            "Supplier_Reliability",
            "Total_Cost",
            "Resilience_Performance_Score"
        ]
    ]

    .sort_values(

        "Resilience_Performance_Score",

        ascending=False

    )

)


print("\n" + "=" * 110)

print("FINAL RESILIENCE STRATEGY RANKING")

print("=" * 110)


print(

    ranking
    .round(2)
    .to_string(index=False)

)


# ================================================================
# 16. GRÁFICA 8
#
# FINAL STRATEGY RANKING
# ================================================================


ranking_plot = ranking.sort_values(

    "Resilience_Performance_Score"

)


ranking_colors = [

    "#E74C3C",
    "#E67E22",
    "#F1C40F",
    "#3498DB",
    "#2ECC71"

]


plt.figure(
    figsize=(11,7)
)


bars = plt.barh(

    ranking_plot["Strategy"],

    ranking_plot["Resilience_Performance_Score"],

    color=ranking_colors

)


plt.title(

    "Overall Resilience Strategy Ranking",

    fontsize=19,

    fontweight="bold"

)


plt.xlabel(

    "Resilience Performance Score"

)


for bar in bars:

    width = bar.get_width()

    plt.text(

        width + 1,

        bar.get_y()
        + bar.get_height()/2,

        f"{width:.1f}",

        va="center",

        fontweight="bold"

    )


plt.grid(
    axis="x",
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# 17. RADAR CHART
#
# COMPARACIÓN MULTIDIMENSIONAL
# ================================================================
#
# Esta gráfica permite comparar simultáneamente:
#
# Service
# Recovery
# Stockout Protection
# Inventory Efficiency
# Supplier Reliability
# Cost Efficiency
#
#
# IMPORTANTE:
#
# Hemos normalizado todo para que:
#
# MÁS HACIA AFUERA = MEJOR.
#
# ================================================================


radar_variables = [

    "Service_Score",

    "Recovery_Score",

    "Stockout_Score",

    "Inventory_Score",

    "Reliability_Score",

    "Cost_Score"

]


radar_labels = [

    "Service",

    "Recovery",

    "Stockout Protection",

    "Inventory Efficiency",

    "Reliability",

    "Cost Efficiency"

]


N = len(
    radar_variables
)


angles = np.linspace(

    0,

    2 * np.pi,

    N,

    endpoint=False

).tolist()


# Cerramos el círculo.

angles += angles[:1]


radar_colors = [

    "#E74C3C",
    "#F39C12",
    "#3498DB",
    "#1ABC9C",
    "#2ECC71"

]


fig = plt.figure(
    figsize=(11,11)
)


ax = plt.subplot(
    111,
    polar=True
)


for index, row in df.iterrows():

    values = [

        row[column]

        for column in radar_variables

    ]


    # Cerramos el polígono.

    values += values[:1]


    ax.plot(

        angles,

        values,

        linewidth=2.2,

        label=row["Strategy"],

        color=radar_colors[index]

    )


    ax.fill(

        angles,

        values,

        alpha=0.05,

        color=radar_colors[index]

    )


ax.set_xticks(
    angles[:-1]
)


ax.set_xticklabels(
    radar_labels,
    fontsize=10
)


ax.set_title(

    "Resilience Strategy Profile",

    fontsize=19,

    fontweight="bold",

    pad=25

)


ax.legend(

    bbox_to_anchor=(1.40, 1.10),

    loc="upper right"

)


plt.show()


# ================================================================
# 18. GRÁFICA 10
#
# RESILIENCE FRONTIER
# ================================================================
#
# Esta gráfica es especialmente interesante
# para discusión gerencial.
#
#
# X = Total Cost
#
# Y = Resilience Performance Score
#
#
# La pregunta será:
#
# ¿Cuánto desempeño adicional obtenemos
# por cada incremento de costo?
#
# ================================================================


plt.figure(
    figsize=(12,8)
)


plt.scatter(

    df["Total_Cost"],

    df["Resilience_Performance_Score"],

    s=350,

    color=radar_colors,

    edgecolor="black",

    alpha=0.80

)


for i, row in df.iterrows():

    plt.text(

        row["Total_Cost"] + 7,

        row["Resilience_Performance_Score"] + 1,

        row["Strategy"],

        fontsize=9

    )


plt.title(

    "Cost vs Resilience Performance",

    fontsize=19,

    fontweight="bold"

)


plt.xlabel(
    "Total Cost ($000)"
)


plt.ylabel(
    "Resilience Performance Score"
)


plt.grid(
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# INTERPRETACIÓN
# ================================================================
#
# Esta gráfica ayuda a identificar algo muy importante:
#
# aumentar la resiliencia tiene un costo,
#
# pero llega un punto en donde:
#
# PAGAR MÁS
#
# puede producir
#
# MUY POCA MEJORA ADICIONAL.
#
#
# Este fenómeno puede interpretarse como:
#
# diminishing returns
#
# o
#
# rendimientos decrecientes de la resiliencia.
#
# ================================================================


# ================================================================
# 19. IDENTIFICAR MEJOR ESTRATEGIA
# ================================================================


best_strategy = ranking.iloc[0]["Strategy"]


best_score = ranking.iloc[0][
    "Resilience_Performance_Score"
]


print("\n" + "=" * 100)

print("MANAGEMENT RECOMMENDATION")

print("=" * 100)


print(

    f"""
Based on the assumptions and weights used in the model,
the strategy that provides the best overall balance is:

{best_strategy}

Overall Resilience Performance Score:

{best_score:.2f}

This does NOT mean that this strategy is universally optimal.

The decision depends on:

- product criticality
- risk tolerance
- service requirements
- inventory cost
- disruption exposure
- supplier dependency
- business strategy

Changing the assumptions or the managerial priorities
may change the final recommendation.
"""

)


# ================================================================
# 20. MANAGEMENT QUESTIONS
# ================================================================


print("\n" + "=" * 100)

print("MANAGEMENT DISCUSSION QUESTIONS")

print("=" * 100)


print("""

1. Which strategy provides the highest service level?

2. Which strategy provides the fastest recovery?

3. Which strategy minimizes stockout frequency?

4. Which strategy requires the greatest inventory investment?

5. Which strategy has the lowest total cost?

6. Is the cheapest strategy also the most resilient?

7. Is the most resilient strategy the best business decision?

8. Which strategy provides the best balance between
   cost and resilience?

9. At what point does additional resilience become
   too expensive?

10. Would you use the same resilience strategy
    for every product?

11. How would the decision change for a highly
    critical product?

12. How would the decision change for a low-value,
    stable-demand product?

13. As Supply Chain Manager, which strategy would
    you recommend and why?

""")
```
