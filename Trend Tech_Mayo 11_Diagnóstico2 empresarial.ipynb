```python
# ==========================================================
# DIAGNÓSTICO EMPRESARIAL BASADO EN SCOR - DCM
# Trend Tech 9
# Versión para estudiantes
# Incluye: caracterización, diagnóstico, barras, radar, heatmap y Pareto
# ==========================================================

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ==========================================================
# 1. DATOS GENERALES DE LA EMPRESA
# ==========================================================

print("DIAGNÓSTICO EMPRESARIAL SCOR - DCM")
print("Primero analizarán una empresa parecida a la empresa del proyecto.")
print("La idea es practicar el diagnóstico antes de aplicarlo a su empresa principal.\n")

empresa = input("Nombre de la empresa analizada: ")

sector = input(
    "Sector económico de la empresa. Ejemplo: alimentos, moda, logística, salud, educación, tecnología, comercio: "
)

producto = input(
    "Producto o servicio principal. Ejemplo: venta de ropa, domicilios, servicios médicos, software, alimentos: "
)

empleados = int(input(
    "Número aproximado de empleados. Si no conocen el dato exacto, escriban una estimación razonable: "
))

# Clasificación orientativa del tamaño empresarial.
# Esta clasificación se usa solo para fines académicos.
if empleados <= 10:
    tamano = "Microempresa"
elif empleados <= 50:
    tamano = "Pequeña empresa"
elif empleados <= 200:
    tamano = "Mediana empresa"
else:
    tamano = "Gran empresa"

print("\n====================================")
print("RESUMEN DE LA EMPRESA")
print("====================================")
print("Empresa:", empresa)
print("Sector:", sector)
print("Producto o servicio:", producto)
print("Tamaño estimado:", tamano)

# ==========================================================
# 2. INSTRUCCIONES PARA RESPONDER
# ==========================================================

print("\nINSTRUCCIONES PARA RESPONDER")
print("Cada pregunta se responde de 1 a 5.")
print("1 = Muy bajo, inexistente o no se evidencia")
print("2 = Bajo, existe muy poco")
print("3 = Medio, existe parcialmente")
print("4 = Alto, se evidencia claramente")
print("5 = Muy alto, está muy desarrollado\n")

print("IMPORTANTE:")
print("Respondan pensando en la situación ACTUAL de la empresa.")
print("No respondan según lo que les gustaría que tuviera la empresa.")
print("Usen evidencias: página web, redes sociales, noticias, observación, entrevistas o información pública.\n")

# ==========================================================
# 3. DIMENSIONES BASADAS EN SCOR - DCM
# ==========================================================

# Cada dimensión representa una capacidad empresarial y digital.
# Para cada dimensión se hacen tres preguntas.
# El código calcula el promedio de cada dimensión.

dimensiones = {
    "Estrategia digital": [
        "La empresa sabe cómo la tecnología puede ayudarle a crecer, vender más o competir mejor",
        "La empresa tiene objetivos relacionados con innovación, transformación digital o mejora tecnológica",
        "La dirección parece tomar decisiones considerando datos, tecnología o tendencias digitales"
    ],

    "Procesos": [
        "Los procesos principales parecen estar definidos, claros y organizados",
        "La empresa tiene identificadas sus actividades críticas, es decir, las que más afectan el resultado del negocio",
        "Los procesos permiten responder rápidamente a cambios del mercado, clientes o operación"
    ],

    "Datos y visibilidad": [
        "La empresa registra datos importantes sobre ventas, inventarios, clientes, servicios o resultados",
        "La información parece estar organizada y disponible para tomar decisiones",
        "La empresa utiliza indicadores, reportes o tableros para hacer seguimiento a su desempeño"
    ],

    "Automatización": [
        "La empresa reduce tareas manuales mediante herramientas digitales",
        "Existen procesos repetitivos que ya parecen estar automatizados",
        "La empresa podría ahorrar tiempo si automatizara actividades operativas"
    ],

    "Integración digital": [
        "Las áreas de la empresa parecen compartir información de forma organizada",
        "Las herramientas o sistemas de la empresa parecen estar conectados entre sí",
        "La información fluye fácilmente entre ventas, operación, servicio y administración"
    ],

    "Cliente y experiencia": [
        "La empresa conoce las necesidades, preferencias o comportamiento de sus clientes",
        "La empresa ofrece canales digitales de atención o comunicación, como web, WhatsApp, app, redes o chatbot",
        "La experiencia del cliente podría mejorar con una app, chatbot, CRM, página web o servicio digital"
    ],

    "Talento digital": [
        "Los empleados parecen tener habilidades para usar herramientas digitales",
        "El equipo de trabajo parece abierto al cambio y a nuevas formas de trabajar",
        "La empresa podría adoptar una nueva solución tecnológica sin demasiada resistencia"
    ],

    "Innovación": [
        "La empresa busca diferenciarse mediante nuevas soluciones, nuevos servicios o mejoras",
        "La empresa observa tendencias tecnológicas de su sector",
        "La empresa podría beneficiarse de IA, Power BI, Python, CRM, apps o realidad aumentada"
    ],

    "Sostenibilidad": [
        "La empresa controla desperdicios, consumo de recursos o impactos ambientales",
        "La empresa busca eficiencia en el uso de materiales, energía, tiempo o recursos",
        "La tecnología podría ayudar a mejorar la sostenibilidad de la empresa"
    ],

    "Madurez operativa": [
        "La empresa controla adecuadamente sus tiempos de operación",
        "La empresa controla costos, calidad y cumplimiento",
        "La empresa mejora continuamente sus procesos"
    ]
}

resultados = []

# ==========================================================
# 4. APLICACIÓN DEL CUESTIONARIO
# ==========================================================

for dimension, preguntas in dimensiones.items():
    print("\n==================================================")
    print("DIMENSIÓN:", dimension)
    print("==================================================")
    print("Lean cada afirmación y califiquen de 1 a 5.")
    print("Piensen en evidencias reales de la empresa antes de responder.\n")

    puntajes = []

    for pregunta in preguntas:
        print(pregunta)
        print("1 = no se evidencia")
        print("2 = se evidencia poco")
        print("3 = se evidencia parcialmente")
        print("4 = se evidencia bastante")
        print("5 = se evidencia totalmente")

        # Validación sencilla para evitar valores por fuera de la escala.
        while True:
            valor = int(input("Puntaje de 1 a 5: "))
            if valor >= 1 and valor <= 5:
                break
            else:
                print("Valor inválido. Deben escribir un número entre 1 y 5.")

        puntajes.append(valor)
        print()

    promedio = round(sum(puntajes) / len(puntajes), 2)
    resultados.append([dimension, promedio])

# Crear tabla de resultados
df = pd.DataFrame(resultados, columns=["Dimensión", "Puntaje"])

# ==========================================================
# 5. CLASIFICACIÓN DE RESULTADOS
# ==========================================================

def clasificar(puntaje):
    if puntaje >= 4:
        return "Fortaleza"
    elif puntaje >= 3:
        return "Nivel medio"
    else:
        return "Brecha crítica"

def nivel_madurez(puntaje):
    if puntaje < 2:
        return "Inicial"
    elif puntaje < 3:
        return "Básico"
    elif puntaje < 4:
        return "Intermedio"
    elif puntaje < 4.5:
        return "Avanzado"
    else:
        return "Inteligente"

df["Clasificación"] = df["Puntaje"].apply(clasificar)
df["Nivel de madurez"] = df["Puntaje"].apply(nivel_madurez)

# Colores tipo semáforo
colores = []

for puntaje in df["Puntaje"]:
    if puntaje >= 4:
        colores.append("green")
    elif puntaje >= 3:
        colores.append("orange")
    else:
        colores.append("red")

promedio_general = round(df["Puntaje"].mean(), 2)

# ==========================================================
# 6. RESULTADOS GENERALES
# ==========================================================

print("\n========================================")
print("RESULTADOS GENERALES DEL DIAGNÓSTICO")
print("========================================")
print(df)

print("\nPromedio general:", promedio_general)

if promedio_general >= 4:
    print("Diagnóstico general: empresa con alta madurez digital y operativa.")
elif promedio_general >= 3:
    print("Diagnóstico general: empresa con madurez media y oportunidades de mejora.")
else:
    print("Diagnóstico general: empresa con brechas importantes de transformación.")

print("\nCómo interpretar la tabla:")
print("- Fortaleza: capacidad bien desarrollada.")
print("- Nivel medio: capacidad presente, pero con oportunidades de mejora.")
print("- Brecha crítica: dimensión débil que puede convertirse en problema del proyecto.")

# ==========================================================
# 7. GRÁFICA 1: BARRAS TIPO SEMÁFORO
# ==========================================================

plt.figure(figsize=(13, 6))
plt.bar(df["Dimensión"], df["Puntaje"], color=colores)
plt.title("Diagnóstico SCOR - DCM por capacidad")
plt.ylabel("Puntaje promedio")
plt.ylim(0, 5)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.show()

print("\nINTERPRETACIÓN DE LA GRÁFICA DE BARRAS:")
print("Las barras verdes muestran fortalezas.")
print("Las barras naranjas muestran capacidades en desarrollo.")
print("Las barras rojas muestran brechas críticas.")
print("Las dimensiones más bajas pueden orientar el problema principal del proyecto.")

# ==========================================================
# 8. GRÁFICA 2: RADAR DE MADUREZ
# ==========================================================

categorias = df["Dimensión"].tolist()
valores = df["Puntaje"].tolist()

# Para cerrar el radar se repite el primer valor al final.
valores_radar = valores + valores[:1]

angulos = np.linspace(0, 2 * np.pi, len(categorias), endpoint=False).tolist()
angulos += angulos[:1]

plt.figure(figsize=(9, 9))
ax = plt.subplot(111, polar=True)
ax.plot(angulos, valores_radar, color="blue", linewidth=2)
ax.fill(angulos, valores_radar, color="skyblue", alpha=0.4)
ax.set_xticks(angulos[:-1])
ax.set_xticklabels(categorias, fontsize=9)
ax.set_yticks([1, 2, 3, 4, 5])
ax.set_ylim(0, 5)
plt.title("Radar de madurez empresarial SCOR - DCM")
plt.show()

print("\nINTERPRETACIÓN DEL RADAR:")
print("Entre más amplio sea el radar, mayor madurez tiene la empresa.")
print("Las zonas cercanas al centro muestran debilidades.")
print("Las zonas cercanas al borde muestran capacidades fuertes.")
print("Un radar desequilibrado indica que la empresa tiene fortalezas en unas áreas y brechas en otras.")

# ==========================================================
# 9. GRÁFICA 3: MAPA DE CALOR
# ==========================================================

plt.figure(figsize=(11, 2.5))
plt.imshow([df["Puntaje"]], cmap="RdYlGn", aspect="auto", vmin=1, vmax=5)
plt.colorbar(label="Puntaje")
plt.xticks(range(len(df["Dimensión"])), df["Dimensión"], rotation=45, ha="right")
plt.yticks([0], ["Empresa"])
plt.title("Mapa de calor de capacidades SCOR - DCM")
plt.show()

print("\nINTERPRETACIÓN DEL MAPA DE CALOR:")
print("El rojo indica debilidad o brecha crítica.")
print("El amarillo indica desarrollo medio.")
print("El verde indica fortaleza.")
print("Las zonas rojas pueden convertirse en causas del diagrama de Ishikawa.")

# ==========================================================
# 10. GRÁFICA 4: PARETO DE BRECHAS
# ==========================================================

# La brecha se calcula comparando el puntaje obtenido con el nivel ideal de 5.
# Ejemplo: si una dimensión tiene 2.5, su brecha es 5 - 2.5 = 2.5.

df_brechas = df.copy()
df_brechas["Brecha"] = 5 - df_brechas["Puntaje"]

# Se ordenan las brechas de mayor a menor.
df_brechas = df_brechas.sort_values("Brecha", ascending=False)

# Se calcula el porcentaje acumulado de las brechas.
df_brechas["Brecha acumulada %"] = (
    df_brechas["Brecha"].cumsum() / df_brechas["Brecha"].sum() * 100
)

fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.bar(df_brechas["Dimensión"], df_brechas["Brecha"], color="tomato")
ax1.set_ylabel("Brecha frente al nivel ideal")
ax1.set_ylim(0, 5)
ax1.tick_params(axis="x", rotation=45)
ax1.grid(axis="y", linestyle="--", alpha=0.4)

ax2 = ax1.twinx()
ax2.plot(
    df_brechas["Dimensión"],
    df_brechas["Brecha acumulada %"],
    color="black",
    marker="o",
    linewidth=2
)
ax2.set_ylabel("Brecha acumulada %")
ax2.set_ylim(0, 110)

plt.title("Pareto de brechas empresariales SCOR - DCM")
plt.show()

print("\nINTERPRETACIÓN DEL PARETO:")
print("El Pareto muestra cuáles dimensiones tienen las brechas más grandes.")
print("Las primeras barras son las áreas más críticas frente al nivel ideal.")
print("Estas dimensiones deberían revisarse en el mapa de procesos y en el Ishikawa.")
print("El grupo puede usar la principal brecha como punto de partida para definir el problema.")

# ==========================================================
# 11. RECOMENDACIÓN FINAL HASTA PARETO
# ==========================================================

dimension_critica = df_brechas.iloc[0]["Dimensión"]
puntaje_critico = df_brechas.iloc[0]["Puntaje"]

print("\n========================================")
print("RECOMENDACIÓN FINAL PARA EL ANÁLISIS")
print("========================================")
print("Dimensión más crítica:", dimension_critica)
print("Puntaje obtenido:", puntaje_critico)

print("\nSugerencia para el grupo:")
print("1. Revisen esta dimensión crítica en los procesos de la empresa.")
print("2. Identifiquen qué actividades pueden estar causando el problema.")
print("3. Usen esas causas para construir el diagrama de Ishikawa.")
print("4. Piensen qué solución tecnológica podría ayudar: Power BI, Python, CRM, chatbot, app o automatización.")
```
