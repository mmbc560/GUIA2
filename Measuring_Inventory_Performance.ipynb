```python
# ================================================================
# INVENTORY PERFORMANCE ANALYTICS
# Cost + Availability + Risk
# ================================================================
#
# OBJETIVO DEL EJERCICIO
#
# Analizar el desempeño del inventario desde diferentes perspectivas.
#
# Vamos a calcular:
#
# 1. Inventory Turnover
# 2. Days of Inventory
# 3. Service Level
# 4. Stockout Frequency
# 5. Carrying Cost
#
# Luego veremos que:
#
# - Mucho inventario puede mejorar el servicio,
#   pero aumentar el costo.
#
# - Muy poco inventario puede reducir costos,
#   pero aumentar los stockouts.
#
# Por eso, el objetivo NO es minimizar inventario,
# sino encontrar el mejor equilibrio entre:
#
# COST + AVAILABILITY + RISK
#
# ================================================================


# ================================================================
# 1. IMPORTAR LIBRERÍAS
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ================================================================
# 2. CREAR UNA BASE DE DATOS FICTICIA
# ================================================================
#
# Vamos a simular 12 productos.
#
# Cada producto tendrá:
#
# - ventas anuales
# - inventario promedio
# - inventario actual
# - costo unitario
# - pedidos recibidos
# - pedidos atendidos
# - número de stockouts
#
# ================================================================

np.random.seed(25)


products = [
    "Product A",
    "Product B",
    "Product C",
    "Product D",
    "Product E",
    "Product F",
    "Product G",
    "Product H",
    "Product I",
    "Product J",
    "Product K",
    "Product L"
]


inventory = pd.DataFrame({

    "Product": products,

    # Unidades vendidas durante el año
    "Annual_Sales_Units": [
        12000, 8500, 15000, 6200,
        18000, 9000, 11000, 7000,
        13500, 5000, 16000, 8000
    ],

    # Inventario promedio durante el año
    "Average_Inventory": [
        2000, 2500, 1800, 2200,
        3000, 2600, 1900, 2800,
        2100, 2400, 3500, 1700
    ],

    # Inventario actual
    "Current_Inventory": [
        2100, 2300, 1600, 2500,
        3200, 2400, 1750, 3000,
        1900, 2600, 3700, 1500
    ],

    # Costo unitario
    "Unit_Cost": [
        20, 35, 15, 50,
        12, 42, 28, 18,
        30, 55, 10, 25
    ],

    # Número total de pedidos recibidos
    "Orders_Received": [
        500, 420, 620, 350,
        700, 460, 520, 390,
        580, 300, 750, 430
    ],

    # Pedidos atendidos completamente
    "Orders_Fulfilled": [
        485, 370, 605, 290,
        680, 410, 505, 325,
        545, 235, 730, 415
    ],

    # Número de eventos de stockout
    "Stockout_Events": [
        3, 12, 2, 18,
        4, 10, 3, 16,
        7, 20, 5, 2
    ]
})


# ================================================================
# 3. CALCULAR INVENTORY TURNOVER
# ================================================================
#
# Inventory Turnover indica cuántas veces se renueva
# el inventario durante un periodo.
#
# Fórmula simplificada:
#
# Inventory Turnover =
# Annual Sales Units / Average Inventory
#
# Interpretación:
#
# Mayor rotación:
# inventario se mueve rápidamente.
#
# Menor rotación:
# inventario permanece almacenado más tiempo.
#
# ================================================================


inventory["Inventory_Turnover"] = (

    inventory["Annual_Sales_Units"] /
    inventory["Average_Inventory"]
)


# ================================================================
# 4. CALCULAR DAYS OF INVENTORY
# ================================================================
#
# Days of Inventory indica aproximadamente cuántos días
# podría cubrir la empresa con el inventario disponible.
#
# Primero calculamos la demanda diaria.
# ================================================================


inventory["Daily_Demand"] = (

    inventory["Annual_Sales_Units"] / 365
)


# Days of Inventory =
# Current Inventory / Daily Demand

inventory["Days_of_Inventory"] = (

    inventory["Current_Inventory"] /
    inventory["Daily_Demand"]
)


# ================================================================
# 5. CALCULAR SERVICE LEVEL
# ================================================================
#
# Service Level representa el porcentaje de pedidos
# atendidos correctamente.
#
# Service Level =
# Orders Fulfilled / Orders Received × 100
#
# ================================================================


inventory["Service_Level"] = (

    inventory["Orders_Fulfilled"] /
    inventory["Orders_Received"]
) * 100


# ================================================================
# 6. CALCULAR STOCKOUT FREQUENCY
# ================================================================
#
# Para comparar productos,
# expresaremos los stockouts respecto al número de pedidos.
#
# Stockout Frequency =
# Stockout Events / Orders Received × 100
#
# ================================================================


inventory["Stockout_Frequency"] = (

    inventory["Stockout_Events"] /
    inventory["Orders_Received"]
) * 100


# ================================================================
# 7. CALCULAR CARRYING COST
# ================================================================
#
# Carrying Cost representa el costo de mantener inventario.
#
# Incluye elementos como:
#
# - almacenamiento
# - seguros
# - capital inmovilizado
# - deterioro
# - obsolescencia
#
# Para simplificar, asumimos una tasa anual de 22%.
#
# Carrying Cost =
#
# Average Inventory
# × Unit Cost
# × Carrying Cost Rate
#
# ================================================================


carrying_cost_rate = 0.22


inventory["Inventory_Value"] = (

    inventory["Average_Inventory"] *
    inventory["Unit_Cost"]
)


inventory["Carrying_Cost"] = (

    inventory["Inventory_Value"] *
    carrying_cost_rate
)


# ================================================================
# 8. GRÁFICA 1
# INVENTORY TURNOVER
# ================================================================
#
# Queremos identificar productos con:
#
# - alta rotación
# - baja rotación
#
# Una rotación muy baja puede indicar:
#
# - exceso de inventario
# - baja demanda
# - capital inmovilizado
#
# ================================================================


# Definimos colores según nivel de rotación.

turnover_colors = []

for value in inventory["Inventory_Turnover"]:

    if value >= 6:
        turnover_colors.append("#2ECC71")   # Verde

    elif value >= 4:
        turnover_colors.append("#F1C40F")   # Amarillo

    else:
        turnover_colors.append("#E74C3C")   # Rojo


plt.figure(figsize=(13,6))


plt.bar(
    inventory["Product"],
    inventory["Inventory_Turnover"],
    color=turnover_colors
)


plt.title(
    "Inventory Turnover by Product",
    fontsize=17,
    fontweight="bold"
)


plt.ylabel(
    "Turns per Year"
)


plt.xticks(
    rotation=45
)


plt.axhline(
    4,
    linestyle="--",
    color="#7F8C8D"
)


plt.grid(
    axis="y",
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# 9. GRÁFICA 2
# DAYS OF INVENTORY VS SERVICE LEVEL
# ================================================================
#
# Esta gráfica es especialmente importante.
#
# EJE X = Days of Inventory
# EJE Y = Service Level
#
# Queremos analizar:
#
# ¿Tener más inventario mejora siempre el servicio?
#
# La respuesta generalmente es:
# NO necesariamente.
#
# ================================================================


# Tamaño de burbuja = valor del inventario.

bubble_size = (

    inventory["Inventory_Value"] /
    inventory["Inventory_Value"].max()
) * 1800


# Colores según nivel de servicio.

service_colors = []

for service in inventory["Service_Level"]:

    if service >= 95:
        service_colors.append("#2ECC71")

    elif service >= 90:
        service_colors.append("#F1C40F")

    else:
        service_colors.append("#E74C3C")


plt.figure(figsize=(13,7))


plt.scatter(

    inventory["Days_of_Inventory"],

    inventory["Service_Level"],

    s=bubble_size,

    c=service_colors,

    alpha=0.70,

    edgecolors="black"
)


# Añadimos etiquetas.

for i, row in inventory.iterrows():

    plt.annotate(

        row["Product"],

        (
            row["Days_of_Inventory"],
            row["Service_Level"]
        ),

        xytext=(5,5),

        textcoords="offset points"
    )


# Línea de referencia de servicio.

plt.axhline(
    95,
    linestyle="--",
    color="#34495E"
)


plt.title(
    "Inventory Coverage vs Service Level",
    fontsize=17,
    fontweight="bold"
)


plt.xlabel(
    "Days of Inventory"
)


plt.ylabel(
    "Service Level (%)"
)


plt.grid(
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# 10. GRÁFICA 3
# STOCKOUT FREQUENCY
# ================================================================
#
# Aquí queremos visualizar qué productos presentan
# mayor frecuencia de faltantes.
#
# ================================================================


stockout_colors = []

for value in inventory["Stockout_Frequency"]:

    if value <= 1:
        stockout_colors.append("#2ECC71")

    elif value <= 3:
        stockout_colors.append("#F1C40F")

    else:
        stockout_colors.append("#E74C3C")


plt.figure(figsize=(13,6))


plt.bar(
    inventory["Product"],
    inventory["Stockout_Frequency"],
    color=stockout_colors
)


plt.title(
    "Stockout Frequency by Product",
    fontsize=17,
    fontweight="bold"
)


plt.ylabel(
    "Stockout Frequency (%)"
)


plt.xticks(
    rotation=45
)


plt.grid(
    axis="y",
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# 11. GRÁFICA 4
# CARRYING COST
# ================================================================
#
# Esta gráfica muestra cuánto cuesta mantener inventario.
#
# Un producto puede tener excelente disponibilidad,
# pero generar un costo importante por almacenamiento.
#
# ================================================================


cost_colors = []

median_cost = inventory["Carrying_Cost"].median()


for value in inventory["Carrying_Cost"]:

    if value < median_cost * 0.8:
        cost_colors.append("#2ECC71")

    elif value < median_cost * 1.2:
        cost_colors.append("#F1C40F")

    else:
        cost_colors.append("#E67E22")


plt.figure(figsize=(13,6))


plt.bar(
    inventory["Product"],
    inventory["Carrying_Cost"],
    color=cost_colors
)


plt.title(
    "Annual Inventory Carrying Cost",
    fontsize=17,
    fontweight="bold"
)


plt.ylabel(
    "Annual Carrying Cost ($)"
)


plt.xticks(
    rotation=45
)


plt.grid(
    axis="y",
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# 12. CREAR UN INVENTORY PERFORMANCE SCORE
# ================================================================
#
# Ahora vamos a combinar varios KPI.
#
# Queremos una medida sencilla que considere:
#
# SERVICE LEVEL           40%
# INVENTORY TURNOVER      25%
# STOCKOUT PERFORMANCE    20%
# COST EFFICIENCY         15%
#
# Esto NO pretende ser una fórmula universal.
#
# Es un índice didáctico para mostrar cómo diferentes KPI
# pueden integrarse en una decisión gerencial.
#
# ================================================================


# Normalizamos Inventory Turnover.

inventory["Turnover_Score"] = (

    inventory["Inventory_Turnover"] /
    inventory["Inventory_Turnover"].max()
) * 100


# Convertimos Stockout Frequency:
#
# menor stockout = mejor desempeño.

inventory["Stockout_Score"] = (

    100 -
    (
        inventory["Stockout_Frequency"] /
        inventory["Stockout_Frequency"].max()
    ) * 100
)


# Convertimos Carrying Cost:
#
# menor costo = mejor desempeño.

inventory["Cost_Score"] = (

    100 -
    (
        inventory["Carrying_Cost"] /
        inventory["Carrying_Cost"].max()
    ) * 100
)


# Performance Score total.

inventory["Performance_Score"] = (

    inventory["Service_Level"] * 0.40

    +

    inventory["Turnover_Score"] * 0.25

    +

    inventory["Stockout_Score"] * 0.20

    +

    inventory["Cost_Score"] * 0.15
)


# ================================================================
# 13. GRÁFICA 5
# MANAGEMENT BALANCE MAP
# ================================================================
#
# Esta es una de las gráficas más importantes.
#
# EJE X = Carrying Cost
# EJE Y = Service Level
#
# Tamaño = Inventory Value
#
# Color = Stockout Frequency
#
# Esto permite observar simultáneamente:
#
# COST + SERVICE + RISK
#
# ================================================================


risk_colors = []

for risk in inventory["Stockout_Frequency"]:

    if risk <= 1:
        risk_colors.append("#2ECC71")

    elif risk <= 3:
        risk_colors.append("#F1C40F")

    else:
        risk_colors.append("#E74C3C")


plt.figure(figsize=(14,8))


plt.scatter(

    inventory["Carrying_Cost"],

    inventory["Service_Level"],

    s=bubble_size,

    c=risk_colors,

    alpha=0.70,

    edgecolors="black"
)


for i, row in inventory.iterrows():

    plt.annotate(

        row["Product"],

        (
            row["Carrying_Cost"],
            row["Service_Level"]
        ),

        xytext=(5,5),

        textcoords="offset points"
    )


plt.axhline(
    95,
    color="#34495E",
    linestyle="--"
)


plt.axvline(
    inventory["Carrying_Cost"].median(),
    color="#34495E",
    linestyle="--"
)


plt.title(
    "Management Balance Map: Cost, Service and Risk",
    fontsize=17,
    fontweight="bold"
)


plt.xlabel(
    "Annual Carrying Cost"
)


plt.ylabel(
    "Service Level (%)"
)


plt.grid(
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# 14. GRÁFICA 6
# INVENTORY PERFORMANCE SCORE
# ================================================================
#
# Finalmente ordenamos los productos
# según su desempeño global.
#
# ================================================================


ranking = inventory.sort_values(
    "Performance_Score",
    ascending=True
)


ranking_colors = []


for score in ranking["Performance_Score"]:

    if score >= 75:
        ranking_colors.append("#2ECC71")

    elif score >= 60:
        ranking_colors.append("#F1C40F")

    else:
        ranking_colors.append("#E74C3C")


plt.figure(figsize=(12,7))


plt.barh(

    ranking["Product"],

    ranking["Performance_Score"],

    color=ranking_colors
)


plt.title(
    "Inventory Performance Ranking",
    fontsize=17,
    fontweight="bold"
)


plt.xlabel(
    "Performance Score"
)


plt.grid(
    axis="x",
    alpha=0.20
)


plt.tight_layout()

plt.show()


# ================================================================
# 15. RESUMEN GERENCIAL
# ================================================================


print("\n" + "=" * 70)

print("INVENTORY MANAGEMENT SUMMARY")

print("=" * 70)


summary = inventory[
    [
        "Product",
        "Inventory_Turnover",
        "Days_of_Inventory",
        "Service_Level",
        "Stockout_Frequency",
        "Carrying_Cost",
        "Performance_Score"
    ]
]


print(

    summary
    .round(2)
    .sort_values(
        "Performance_Score",
        ascending=False
    )
    .to_string(index=False)
)


# ================================================================
# 16. IDENTIFICAR PRODUCTOS QUE NECESITAN ATENCIÓN
# ================================================================


critical_products = inventory[
    (
        inventory["Service_Level"] < 90
    )
    |
    (
        inventory["Stockout_Frequency"] > 3
    )
]


print("\n")

print("=" * 70)

print("PRODUCTS REQUIRING MANAGEMENT ATTENTION")

print("=" * 70)


print(

    critical_products[
        [
            "Product",
            "Days_of_Inventory",
            "Service_Level",
            "Stockout_Frequency",
            "Carrying_Cost"
        ]
    ]
    .round(2)
    .to_string(index=False)
)


# ================================================================
# 17. PREGUNTAS PARA DISCUSIÓN
# ================================================================


print("\n")

print("=" * 70)

print("MANAGEMENT QUESTIONS")

print("=" * 70)


print("""

1. Which product has the highest inventory turnover?

2. Which products have too many days of inventory?

3. Does more inventory always produce a higher service level?

4. Which product combines high carrying cost and low service?

5. Which products present the highest stockout frequency?

6. If management wanted to reduce inventory,
   where could it do so without severely affecting service?

7. If management wanted to improve service,
   which product should receive attention first?

8. Which product offers the best balance between
   cost, availability and risk?

""")
```
