```python

# -*- coding: utf-8 -*-

"""
============================================================
CASO 4
ADVANCED OPERATIONAL ANALYTICS
HOTEL RESERVATION CONFIRMATION FOR AIRLINE CREWS
============================================================

PROBLEMA:
Incumplimiento en los tiempos de confirmación de
reservas hoteleras para tripulaciones aéreas.

OBJETIVO:
Identificar perfiles operacionales, detectar anomalías,
analizar la evolución de la demanda y evaluar
alternativas gerenciales para mejorar el servicio.

METODOLOGÍA:
1. Generación de datos sintéticos.
2. Diagnóstico descriptivo.
3. K-Means.
4. Caracterización de clusters.
5. Isolation Forest.
6. Evaluación de desempeño por cluster.
7. Análisis temporal.
8. Alertas anticipatorias.
9. Monte Carlo.
10. Comparación prescriptiva.
11. Sensibilidad.
12. Conclusiones gerenciales.

ADVERTENCIA:
Todos los datos y efectos de intervención son
sintéticos e ilustrativos.

El aprendizaje no supervisado descubre patrones,
pero no demuestra relaciones causales.

Las simulaciones no representan resultados reales
ni constituyen una predicción validada.
"""

# ============================================================
# 0. LIBRERÍAS Y CONFIGURACIÓN
# ============================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.ensemble import IsolationForest
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

warnings.filterwarnings("ignore", category=FutureWarning)

# ------------------------------------------------------------
# PARÁMETROS MODIFICABLES
# ------------------------------------------------------------

SEMILLA = 42

N_RESERVAS = 1000

MESES = 12

# SLA hipotético de confirmación.
SLA_HORAS = 6.0

# Presupuesto hipotético en euros.
PRESUPUESTO_MAX_EUR = 150000

# Costo ilustrativo por hora de demora que supera el SLA.
COSTO_HORA_DEMORA_EUR = 75

N_SIMULACIONES = 5000

CONTAMINACION = 0.06

K_MIN = 2
K_MAX = 6

DIRECTORIO = "Caso_4_Resultados"

os.makedirs(DIRECTORIO, exist_ok=True)

rng = np.random.default_rng(SEMILLA)

sns.set_theme(style="whitegrid")

plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "savefig.facecolor": "white",
    "font.size": 10,
    "axes.titlesize": 14,
    "axes.titleweight": "bold"
})


# ============================================================
# FUNCIONES AUXILIARES
# ============================================================

def guardar(nombre):
    """
    Guarda una gráfica PNG independiente.

    Todas las gráficas se generan con fondo blanco,
    títulos en inglés y resolución de 190 dpi.
    """
    plt.tight_layout()

    plt.savefig(
        os.path.join(DIRECTORIO, nombre + ".png"),
        dpi=190,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.show()
    plt.close()


def interpretar(titulo, texto):
    """
    Presenta una interpretación gerencial extensa.
    """
    print("\n" + "=" * 90)
    print(titulo.upper())
    print("=" * 90)
    print(texto.strip())
    print()


# ============================================================
# 1. GENERACIÓN DE DATOS SINTÉTICOS
# ============================================================

"""
Cada fila representa una solicitud de alojamiento
para una tripulación aérea.

Se incluyen variables relacionadas con:

- Estacionalidad.
- Destino.
- Número de habitaciones requeridas.
- Ocupación hotelera.
- Disponibilidad de habitaciones.
- Anticipación de la solicitud.
- Confiabilidad histórica del proveedor.
- Disrupciones operacionales.
- Tiempo de confirmación.

Los perfiles internos se utilizan únicamente
para generar diversidad sintética.

No se entregan como etiquetas a K-Means.

El tiempo de confirmación se utilizará después
del agrupamiento para evaluar el desempeño.
"""

destinos = np.array([
    "Madrid",
    "Paris",
    "Rome",
    "Lisbon",
    "Amsterdam",
    "Barcelona"
])

mes = rng.integers(
    1,
    MESES + 1,
    size=N_RESERVAS
)

destino = rng.choice(
    destinos,
    size=N_RESERVAS
)

# Meses ilustrativos de mayor presión turística.
temporada_alta = np.isin(
    mes,
    [6, 7, 8, 12]
).astype(int)

# Perfiles generadores sintéticos.
# NO se utilizarán como variable del clustering.
perfil_generador = rng.choice(
    [0, 1, 2],
    size=N_RESERVAS,
    p=[0.40, 0.35, 0.25]
)

# Habitaciones solicitadas por tripulación.
habitaciones = np.clip(
    rng.poisson(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [5, 9],
            default=14
        )
    ) + 1,
    1,
    28
)

# Ocupación hotelera estimada al solicitar la reserva.
ocupacion = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [60, 78],
            default=91
        ) + 5 * temporada_alta,
        6
    ),
    35,
    99
)

# Habitaciones disponibles en la red consultada.
disponibilidad = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [25, 15],
            default=7
        ) - 3 * temporada_alta,
        4
    ),
    0,
    45
)

# Anticipación con la que se realiza la solicitud.
anticipacion_horas = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [60, 30],
            default=12
        ),
        9
    ),
    1,
    100
)

# Confiabilidad histórica del hotel.
confiabilidad_hotel = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [95, 84],
            default=68
        ),
        7
    ),
    35,
    100
)

# Disrupciones: cambios de vuelo, cancelaciones
# o modificaciones inesperadas.
disrupciones = np.clip(
    rng.poisson(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [0.2, 0.8],
            default=1.7
        )
    ),
    0,
    6
)

# Presión de capacidad.
# Un valor positivo indica que la demanda supera
# la disponibilidad estimada.
brecha_capacidad = np.maximum(
    0,
    habitaciones - disponibilidad
)

# Tiempo de confirmación sintético.
# Estos coeficientes son SUPUESTOS, no efectos
# causales estimados con datos reales.

ruido = rng.normal(
    0,
    0.8,
    N_RESERVAS
)

tiempo_confirmacion = np.maximum(
    0.5,

    0.8
    + 0.045 * ocupacion
    + 0.30 * brecha_capacidad
    + 0.045 * (100 - confiabilidad_hotel)
    + 0.65 * disrupciones
    + 0.035 * habitaciones
    + 0.65 * temporada_alta
    + 0.020 * np.maximum(
        0,
        24 - anticipacion_horas
    )
    + ruido
)

# Tarifa nocturna sintética.
tarifa_noche = np.clip(
    rng.normal(
        115 + 35 * temporada_alta,
        20
    ),
    65,
    250
)

base = pd.DataFrame({
    "reserva_id": [
        f"RES-{i:04d}"
        for i in range(1, N_RESERVAS + 1)
    ],
    "mes": mes,
    "destino": destino,
    "temporada_alta": temporada_alta,
    "habitaciones_requeridas": habitaciones,
    "ocupacion_hotelera_pct": ocupacion,
    "habitaciones_disponibles": disponibilidad,
    "anticipacion_horas": anticipacion_horas,
    "confiabilidad_hotel_pct": confiabilidad_hotel,
    "disrupciones": disrupciones,
    "brecha_capacidad": brecha_capacidad,
    "tarifa_noche_eur": tarifa_noche,
    "tiempo_confirmacion_h": tiempo_confirmacion
})

base["cumple_sla"] = (
    base["tiempo_confirmacion_h"] <= SLA_HORAS
)

base["exceso_sla_h"] = np.maximum(
    0,
    base["tiempo_confirmacion_h"] - SLA_HORAS
)

# Variables disponibles al recibir la solicitud.
# Se excluye el tiempo final de confirmación.
#
# Esto permite utilizar posteriormente los clusters
# para una alerta anticipatoria sin utilizar
# información del resultado futuro.

VARIABLES = [
    "temporada_alta",
    "habitaciones_requeridas",
    "ocupacion_hotelera_pct",
    "habitaciones_disponibles",
    "anticipacion_horas",
    "confiabilidad_hotel_pct",
    "disrupciones"
]

base.to_csv(
    os.path.join(
        DIRECTORIO,
        "01_Datos_Sinteticos.csv"
    ),
    index=False
)

interpretar(
    "01. Diagnóstico inicial",
    f"""
Se generaron {N_RESERVAS:,} solicitudes
de alojamiento distribuidas en
{MESES} meses.

El tiempo promedio de confirmación es:
{base.tiempo_confirmacion_h.mean():.2f} horas.

El cumplimiento del SLA ilustrativo
de {SLA_HORAS:.1f} horas es:
{100 * base.cumple_sla.mean():.1f}%.

La proporción de solicitudes
fuera del plazo es:
{100 * (1-base.cumple_sla.mean()):.1f}%.

INTERPRETACIÓN GERENCIAL

El problema no debería analizarse
únicamente mediante el promedio
general de confirmación.

Es necesario comprender si los
tiempos cambian según la temporada,
la disponibilidad hotelera, el
destino y las características
de las solicitudes.

Una demora puede originarse en
restricciones de capacidad, pero
también puede estar relacionada
con la anticipación de la reserva
o la coordinación con proveedores.

Por tanto, antes de invertir en
más hoteles, tecnología o personal,
conviene identificar perfiles
operacionales diferenciados.

La organización debería medir
la hora exacta de solicitud,
la primera respuesta del hotel,
la confirmación definitiva y
las escalaciones realizadas.

Esto permitiría separar el tiempo
de espera del proveedor del tiempo
de procesamiento administrativo.
"""
)


# ============================================================
# 2. ANÁLISIS DESCRIPTIVO
# ============================================================

fig, ax = plt.subplots(figsize=(10, 5))

sns.histplot(
    base["tiempo_confirmacion_h"],
    bins=35,
    color="#287FB8",
    ax=ax
)

ax.axvline(
    SLA_HORAS,
    color="#D14C48",
    linestyle="--",
    linewidth=2,
    label=f"Illustrative SLA: {SLA_HORAS:.0f} h"
)

ax.set(
    title="01 | Hotel Confirmation Time Distribution",
    xlabel="Confirmation time (hours)",
    ylabel="Reservation requests"
)

ax.legend()

guardar("01_Confirmation_Time")

interpretar(
    "02. Interpretación descriptiva",
    """
La distribución permite observar
la dispersión de los tiempos de
confirmación.

El interés gerencial no consiste
solamente en reducir el promedio.

También es importante disminuir
la variabilidad y evitar que
algunas tripulaciones enfrenten
esperas extraordinarias.

En este tipo de operación,
la confirmación tardía puede
generar reubicaciones, costos
adicionales y dificultades
para garantizar el descanso.

Una organización podría tener
un promedio aceptable y,
sin embargo, presentar
incumplimientos importantes
en destinos específicos.

Por eso se recomienda medir:

- Tiempo medio de confirmación.
- Percentil 90.
- Cumplimiento del SLA.
- Disponibilidad hotelera.
- Solicitudes escaladas.
- Reubicaciones.
- Costo total por reserva.

Estos indicadores deben
analizarse por destino,
temporada y proveedor.
"""
)


# ============================================================
# 3. ESTANDARIZACIÓN
# ============================================================

# K-Means utiliza distancias.
# Por eso se estandarizan las variables.

escalador = StandardScaler()

X = escalador.fit_transform(
    base[VARIABLES]
)


# ============================================================
# 4. SELECCIÓN DEL NÚMERO DE CLUSTERS
# ============================================================

valores_k = list(
    range(K_MIN, K_MAX + 1)
)

silhouettes = []

for k in valores_k:

    modelo_prueba = KMeans(
        n_clusters=k,
        random_state=SEMILLA,
        n_init=15
    )

    etiquetas = modelo_prueba.fit_predict(X)

    puntuacion = silhouette_score(
        X,
        etiquetas,
        sample_size=min(700, len(base)),
        random_state=SEMILLA
    )

    silhouettes.append(puntuacion)

k_optimo = valores_k[
    int(np.argmax(silhouettes))
]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(
    valores_k,
    silhouettes,
    marker="o",
    linewidth=2,
    color="#287FB8"
)

ax.axvline(
    k_optimo,
    color="#E49A32",
    linestyle="--",
    label=f"Selected k = {k_optimo}"
)

ax.set(
    title="02 | Optimal Cluster Selection",
    xlabel="Number of clusters",
    ylabel="Silhouette score"
)

ax.legend()

guardar("02_Silhouette")

interpretar(
    "03. Selección de clusters",
    f"""
Se evaluaron soluciones entre
{K_MIN} y {K_MAX} grupos.

El número seleccionado fue:
{k_optimo} clusters.

El índice silhouette alcanzó:
{max(silhouettes):.3f}.

INTERPRETACIÓN

El algoritmo identifica grupos
de solicitudes con características
operacionales semejantes.

Sin embargo, los clusters no
representan automáticamente
reservas eficientes o ineficientes.

Su interpretación depende de
las variables que distinguen
a cada segmento.

Una aplicación gerencial sería
identificar si existen reservas
rutinarias, solicitudes con
restricciones de capacidad y
situaciones de mayor complejidad.

La organización podría diseñar
procedimientos diferenciados
para estos perfiles.

Antes de implementar cambios,
debería verificarse la estabilidad
de los grupos y su utilidad
en diferentes temporadas.
"""
)


# ============================================================
# 5. K-MEANS
# ============================================================

modelo_kmeans = KMeans(
    n_clusters=k_optimo,
    random_state=SEMILLA,
    n_init=20
)

base["cluster"] = modelo_kmeans.fit_predict(X)

# PCA se utiliza únicamente para visualización.
# Los clusters se calculan con todas las variables.

pca = PCA(
    n_components=2,
    random_state=SEMILLA
)

X_pca = pca.fit_transform(X)

fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=base["cluster"].astype(str),
    palette="tab10",
    alpha=0.7,
    s=35,
    ax=ax
)

ax.set(
    title="03 | Hotel Reservation Operational Clusters",
    xlabel="Principal component 1",
    ylabel="Principal component 2"
)

ax.legend(title="Cluster")

guardar("03_KMeans_Clusters")


# ============================================================
# 6. CARACTERIZACIÓN DE CLUSTERS
# ============================================================

perfiles = base.groupby("cluster")[VARIABLES].mean()

media = base[VARIABLES].mean()

desviacion = base[VARIABLES].std().replace(0, 1)

perfiles_z = (
    perfiles - media
) / desviacion

fig, ax = plt.subplots(
    figsize=(12, max(4, k_optimo * 1.2))
)

sns.heatmap(
    perfiles_z,
    annot=True,
    fmt=".2f",
    cmap="RdYlBu_r",
    center=0,
    linewidths=0.5,
    ax=ax
)

ax.set(
    title="04 | Hotel Reservation Cluster Profiles",
    xlabel="Operational variables",
    ylabel="Cluster"
)

guardar("04_Cluster_Heatmap")

resumen = base.groupby("cluster").agg(
    reservas=("reserva_id", "size"),
    tiempo_promedio=("tiempo_confirmacion_h", "mean"),
    cumplimiento=("cumple_sla", "mean"),
    ocupacion=("ocupacion_hotelera_pct", "mean"),
    disponibilidad=("habitaciones_disponibles", "mean"),
    exceso_promedio=("exceso_sla_h", "mean")
)

resumen["cumplimiento_pct"] = (
    100 * resumen["cumplimiento"]
)

print("\nRESUMEN DE CLUSTERS")

print(resumen.round(2).to_string())

peor_cluster = int(
    resumen["tiempo_promedio"].idxmax()
)

mejor_cluster = int(
    resumen["tiempo_promedio"].idxmin()
)

for cluster in perfiles.index:

    dominantes = (
        perfiles_z.loc[cluster]
        .sort_values(ascending=False)
        .head(3)
        .index.tolist()
    )

    interpretar(
        f"04. Interpretación del cluster {cluster}",
        f"""
El cluster {cluster} contiene
{int(resumen.loc[cluster, 'reservas'])}
solicitudes.

El tiempo promedio de confirmación
es de:
{resumen.loc[cluster, 'tiempo_promedio']:.2f} horas.

El cumplimiento del SLA alcanza:
{resumen.loc[cluster, 'cumplimiento_pct']:.1f}%.

Las variables relativamente
más elevadas son:

{', '.join(dominantes)}.

INTERPRETACIÓN GERENCIAL

Estas características permiten
formular hipótesis sobre las
condiciones que distinguen
al segmento.

Si predominan niveles altos
de ocupación, podría ser
necesario revisar la capacidad
contratada en temporadas críticas.

Si las solicitudes tienen
poca anticipación, convendría
investigar la coordinación
entre programación aérea
y gestión hotelera.

Si aparecen condiciones
desfavorables de confiabilidad,
se debería evaluar el
desempeño de los proveedores.

No debe concluirse que
estas variables sean causas
comprobadas del retraso.

RECOMENDACIÓN ESTRATÉGICA

Evaluar políticas de reserva
diferenciadas por temporada,
destino y riesgo de capacidad.

RECOMENDACIÓN TÁCTICA

Revisar acuerdos con hoteles,
disponibilidad anticipada
y mecanismos de escalamiento.

RECOMENDACIÓN OPERACIONAL

Establecer alertas de solicitudes
que requieren verificación
prioritaria de habitaciones.

Los hallazgos deben contrastarse
con expedientes reales y con
el mapa del proceso.
"""
    )


# ============================================================
# 7. ISOLATION FOREST
# ============================================================

modelo_iso = IsolationForest(
    n_estimators=250,
    contamination=CONTAMINACION,
    random_state=SEMILLA
)

base["anomalia"] = (
    modelo_iso.fit_predict(X) == -1
)

base["anomaly_score"] = (
    -modelo_iso.decision_function(X)
)

fig, ax = plt.subplots(figsize=(10, 6))

normales = ~base["anomalia"]

ax.scatter(
    X_pca[normales, 0],
    X_pca[normales, 1],
    s=25,
    alpha=0.5,
    color="#287FB8",
    label="Typical"
)

ax.scatter(
    X_pca[~normales, 0],
    X_pca[~normales, 1],
    s=65,
    facecolors="none",
    edgecolors="#D64B48",
    linewidth=1.5,
    label="Anomaly"
)

ax.set(
    title="05 | Isolation Forest: Unusual Reservations",
    xlabel="Principal component 1",
    ylabel="Principal component 2"
)

ax.legend()

guardar("05_Isolation_Forest")

n_anomalias = int(
    base["anomalia"].sum()
)

interpretar(
    "05. Detección de anomalías",
    f"""
Isolation Forest señaló
{n_anomalias} solicitudes atípicas.

Esto equivale al
{100 * base.anomalia.mean():.1f}%
de los registros.

El porcentaje depende del
parámetro de contaminación
establecido.

INTERPRETACIÓN GERENCIAL

Una solicitud atípica puede
combinar muchas habitaciones,
ocupación elevada, escasa
anticipación o varias disrupciones.

Estas condiciones merecen
investigación, especialmente
si afectan el descanso de
las tripulaciones.

Sin embargo, una anomalía
no representa necesariamente
un incumplimiento.

Puede tratarse de una
situación extraordinaria
correctamente gestionada.

RECOMENDACIÓN

Revisar las solicitudes
más atípicas e identificar
si comparten destinos,
temporadas o proveedores.

En situaciones críticas,
podría activarse una
verificación anticipada
de disponibilidad.

También convendría registrar
las causas confirmadas
para mejorar las reglas
de seguimiento.

No se recomienda bloquear
reservas automáticamente
por ser consideradas atípicas.
"""
)

print("\n15 RESERVAS MÁS ATÍPICAS")

print(
    base.sort_values(
        "anomaly_score",
        ascending=False
    )
    .head(15)
    .to_string(index=False)
)


# ============================================================
# 8. DESEMPEÑO POR CLUSTER
# ============================================================

fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(
    data=base,
    x="cluster",
    y="tiempo_confirmacion_h",
    estimator="mean",
    errorbar=None,
    color="#368AC1",
    ax=ax
)

ax.axhline(
    SLA_HORAS,
    color="#D64B48",
    linestyle="--",
    linewidth=2,
    label="Illustrative SLA"
)

ax.set(
    title="06 | Confirmation Time by Operational Cluster",
    xlabel="Cluster",
    ylabel="Average confirmation time (hours)"
)

ax.legend()

guardar("06_Time_By_Cluster")

interpretar(
    "06. Priorización del desempeño",
    f"""
El cluster con mayor tiempo
promedio es el {peor_cluster}.

Su tiempo medio es:
{resumen.loc[peor_cluster, 'tiempo_promedio']:.2f} horas.

El cluster con menor tiempo
es el {mejor_cluster}.

Su tiempo medio es:
{resumen.loc[mejor_cluster, 'tiempo_promedio']:.2f} horas.

INTERPRETACIÓN GERENCIAL

Los segmentos presentan
diferentes condiciones
operacionales y resultados
de servicio.

Esto sugiere que una
estrategia uniforme podría
no responder adecuadamente
a todos los destinos.

Una solicitud anticipada
con disponibilidad suficiente
no necesariamente requiere
el mismo procedimiento que
una solicitud urgente durante
temporada alta.

RECOMENDACIÓN

Diseñar niveles de atención
basados en riesgo de capacidad
y urgencia operacional.

El equipo debería evaluar
si la prioridad depende del
número de habitaciones,
la proximidad de la llegada,
la disponibilidad hotelera
o la confiabilidad del proveedor.

También convendría revisar
el impacto sobre el descanso
de la tripulación y la
continuidad de los vuelos.

No basta con reducir tiempos
si la alternativa incrementa
el riesgo de alojamiento
inadecuado o incumplimiento
de condiciones de descanso.
"""
)


# ============================================================
# 9. EVOLUCIÓN TEMPORAL
# ============================================================

participacion = (
    pd.crosstab(
        base["mes"],
        base["cluster"],
        normalize="index"
    ) * 100
)

fig, ax = plt.subplots(figsize=(10, 5))

for c in participacion.columns:

    ax.plot(
        participacion.index,
        participacion[c],
        marker="o",
        linewidth=2,
        label=f"Cluster {c}"
    )

ax.set(
    title="07 | Monthly Reservation Cluster Evolution",
    xlabel="Month",
    ylabel="Share of requests (%)",
    xticks=range(1, MESES + 1)
)

ax.legend(ncol=3)

guardar("07_Cluster_Evolution")

anomalias_mes = (
    base.groupby("mes")["anomalia"].mean() * 100
)

fig, ax = plt.subplots(figsize=(10, 4.5))

ax.plot(
    anomalias_mes.index,
    anomalias_mes.values,
    marker="o",
    color="#C66A35",
    linewidth=2
)

ax.set(
    title="08 | Monthly Reservation Anomaly Rate",
    xlabel="Month",
    ylabel="Flagged requests (%)",
    xticks=range(1, MESES + 1)
)

guardar("08_Anomaly_Evolution")

reciente = (
    participacion[peor_cluster]
    .tail(3)
    .mean()
)

anterior = (
    participacion[peor_cluster]
    .iloc[-6:-3]
    .mean()
)

interpretar(
    "07. Evolución temporal",
    f"""
El segmento con mayor tiempo
promedio representa
{reciente:.1f}% de las solicitudes
en los últimos tres meses.

En los tres meses anteriores
representaba {anterior:.1f}%.

La variación fue de
{reciente - anterior:+.1f}
puntos porcentuales.

INTERPRETACIÓN

La evolución temporal permite
examinar si las condiciones
asociadas a confirmaciones
prolongadas están aumentando.

En una operación hotelera
relacionada con aerolíneas,
la estacionalidad puede ser
especialmente importante.

La presión de capacidad
puede variar según destinos,
temporadas turísticas y
programación de vuelos.

Sin embargo, una fluctuación
mensual no constituye una
predicción de deterioro futuro.

RECOMENDACIÓN

Construir un tablero de
monitoreo por destino y mes.

Comparar la participación
de los segmentos con la
ocupación hotelera y el
calendario de operaciones.

Cuando se disponga de
suficiente información
histórica, establecer
umbrales de alerta.

Estos umbrales deben
distinguir variaciones
normales de situaciones
que requieren intervención.
"""
)


# ============================================================
# 10. ALERTA ANTICIPATORIA
# ============================================================

"""
Se analiza una solicitud nueva utilizando
únicamente las variables del agrupamiento.

La alerta se basa en similitud con
solicitudes históricas.

NO es una predicción individual
calibrada del tiempo de confirmación.

Las variables deben corresponder a
información disponible al recibir
la solicitud.
"""

nueva_reserva = pd.DataFrame([{
    "temporada_alta": 1,
    "habitaciones_requeridas": 16,
    "ocupacion_hotelera_pct": 94,
    "habitaciones_disponibles": 5,
    "anticipacion_horas": 10,
    "confiabilidad_hotel_pct": 68,
    "disrupciones": 2
}])

X_nueva = escalador.transform(
    nueva_reserva[VARIABLES]
)

cluster_nuevo = int(
    modelo_kmeans.predict(X_nueva)[0]
)

anomalia_nueva = bool(
    modelo_iso.predict(X_nueva)[0] == -1
)

referencia = base.loc[
    base["cluster"] == cluster_nuevo
]

interpretar(
    "08. Alerta anticipatoria",
    f"""
La nueva solicitud pertenece
al cluster {cluster_nuevo}.

El tiempo medio histórico
sintético de ese grupo es:
{referencia.tiempo_confirmacion_h.mean():.2f} horas.

Su cumplimiento del SLA es:
{100 * referencia.cumple_sla.mean():.1f}%.

La solicitud
{'SÍ' if anomalia_nueva else 'NO'}
fue marcada como atípica.

INTERPRETACIÓN GERENCIAL

La solicitud presenta
características semejantes
a las del segmento indicado.

Esto permite reconocer
condiciones que podrían
requerir seguimiento.

No significa que el
algoritmo conozca la
hora exacta de confirmación.

RECOMENDACIÓN

Verificar anticipadamente
la disponibilidad hotelera.

Si la demanda supera
la capacidad disponible,
considerar proveedores
alternativos.

Revisar la proximidad
de la llegada de la
tripulación y las
condiciones de descanso.

Cuando corresponda,
activar un procedimiento
de escalamiento.

La utilidad de la alerta
deberá validarse con
solicitudes futuras.
"""
)


# ============================================================
# 11. MONTE CARLO: ESCENARIOS GERENCIALES
# ============================================================

"""
Se comparan intervenciones propuestas
para mejorar la confirmación hotelera.

Los efectos son SUPUESTOS DIDÁCTICOS.

No representan impactos reales
estimados mediante experimentos.

Las reducciones se aplican a componentes
del tiempo de confirmación.

La misma muestra y el mismo ruido
se utilizan en todos los escenarios
para hacer comparable el ejercicio.
"""

ESCENARIOS = {

    "Current process": {
        "costo": 0,
        "capacidad": 0,
        "confiabilidad": 0,
        "disrupciones": 0,
        "urgencia": 0,
        "coordinacion": 0
    },

    "Advance room blocks": {
        "costo": 125000,
        "capacidad": 0.55,
        "confiabilidad": 0.10,
        "disrupciones": 0.05,
        "urgencia": 0.05,
        "coordinacion": 0.05
    },

    "Backup hotel network": {
        "costo": 90000,
        "capacidad": 0.40,
        "confiabilidad": 0.20,
        "disrupciones": 0.20,
        "urgencia": 0.05,
        "coordinacion": 0.05
    },

    "Automated escalation": {
        "costo": 70000,
        "capacidad": 0.05,
        "confiabilidad": 0.05,
        "disrupciones": 0.35,
        "urgencia": 0.30,
        "coordinacion": 0.30
    },

    "Supplier diversification": {
        "costo": 110000,
        "capacidad": 0.30,
        "confiabilidad": 0.30,
        "disrupciones": 0.10,
        "urgencia": 0.05,
        "coordinacion": 0.05
    },

    "Process redesign": {
        "costo": 80000,
        "capacidad": 0.05,
        "confiabilidad": 0.10,
        "disrupciones": 0.20,
        "urgencia": 0.15,
        "coordinacion": 0.55
    },

    "Flexible room capacity": {
        "costo": 140000,
        "capacidad": 0.60,
        "confiabilidad": 0.15,
        "disrupciones": 0.10,
        "urgencia": 0.05,
        "coordinacion": 0.10
    },

    "Integrated strategy": {
        "costo": 220000,
        "capacidad": 0.65,
        "confiabilidad": 0.40,
        "disrupciones": 0.45,
        "urgencia": 0.35,
        "coordinacion": 0.50
    }
}

# Muestra con reemplazo para Monte Carlo.
indices = rng.integers(
    0,
    len(base),
    size=N_SIMULACIONES
)

muestra = base.iloc[indices]

# Ruido común entre alternativas.
ruido_mc = rng.normal(
    0,
    0.8,
    N_SIMULACIONES
)


def simular_escenario(parametros):
    """
    Calcula el tiempo de confirmación hipotético.

    La intervención modifica componentes
    específicos del proceso.

    Los efectos son supuestos.
    """

    p = parametros

    # Déficit de capacidad.
    brecha = np.maximum(
        0,
        muestra["habitaciones_requeridas"].to_numpy()
        - muestra["habitaciones_disponibles"].to_numpy()
    )

    # Tiempo de confirmación hipotético.
    tiempo = (
        0.8

        + 0.045
        * muestra["ocupacion_hotelera_pct"].to_numpy()

        + 0.30
        * brecha
        * (1 - p["capacidad"])

        + 0.045
        * (
            100
            - muestra["confiabilidad_hotel_pct"].to_numpy()
        )
        * (1 - p["confiabilidad"])

        + 0.65
        * muestra["disrupciones"].to_numpy()
        * (1 - p["disrupciones"])

        + 0.035
        * muestra["habitaciones_requeridas"].to_numpy()

        + 0.65
        * muestra["temporada_alta"].to_numpy()

        + 0.020
        * np.maximum(
            0,
            24 - muestra["anticipacion_horas"].to_numpy()
        )
        * (1 - p["urgencia"])

        - 0.9 * p["coordinacion"]

        + ruido_mc
    )

    return np.maximum(0.5, tiempo)


resultados = []

for nombre, parametros in ESCENARIOS.items():

    tiempos = simular_escenario(parametros)

    resultados.append({
        "alternativa": nombre,
        "costo_eur": parametros["costo"],
        "tiempo_promedio_h": tiempos.mean(),
        "percentil_90_h": np.percentile(tiempos, 90),
        "cumplimiento_pct": 100 * np.mean(
            tiempos <= SLA_HORAS
        ),
        "exceso_medio_sla_h": np.maximum(
            0,
            tiempos - SLA_HORAS
        ).mean()
    })

escenarios = pd.DataFrame(resultados)

base_simulada = escenarios.loc[
    escenarios["alternativa"] == "Current process"
].iloc[0]

escenarios["reduccion_horas"] = (
    base_simulada["tiempo_promedio_h"]
    - escenarios["tiempo_promedio_h"]
)

# Valor económico hipotético.
# Se aplica solo a las horas que exceden el SLA.
#
# No equivale a ahorro real de caja ni
# incluye todos los costos hoteleros.

escenarios["costo_exceso_ilustrativo_eur"] = (
    escenarios["exceso_medio_sla_h"]
    * N_RESERVAS
    * COSTO_HORA_DEMORA_EUR
)

costo_base_exceso = (
    base_simulada["exceso_medio_sla_h"]
    * N_RESERVAS
    * COSTO_HORA_DEMORA_EUR
)

escenarios["beneficio_potencial_eur"] = (
    costo_base_exceso
    - escenarios["costo_exceso_ilustrativo_eur"]
)

escenarios["beneficio_neto_hipotetico_eur"] = (
    escenarios["beneficio_potencial_eur"]
    - escenarios["costo_eur"]
)

escenarios["factible"] = (
    escenarios["costo_eur"]
    <= PRESUPUESTO_MAX_EUR
)

escenarios.to_csv(
    os.path.join(
        DIRECTORIO,
        "02_Escenarios_Monte_Carlo.csv"
    ),
    index=False
)


# ============================================================
# 12. COMPARACIÓN DE ALTERNATIVAS
# ============================================================

fig, ax = plt.subplots(figsize=(11, 6))

colores = [
    "#287FB8" if x else "#D5A04A"
    for x in escenarios["factible"]
]

ax.barh(
    escenarios["alternativa"],
    escenarios["tiempo_promedio_h"],
    color=colores
)

ax.axvline(
    SLA_HORAS,
    color="#D64B48",
    linestyle="--",
    label="Illustrative SLA"
)

ax.invert_yaxis()

ax.set(
    title="09 | Confirmation Time by Intervention",
    xlabel="Simulated average confirmation time (hours)",
    ylabel=""
)

ax.legend()

guardar("09_Scenario_Comparison")


# ============================================================
# 13. TRADE-OFF COSTO VS CUMPLIMIENTO
# ============================================================

fig, ax = plt.subplots(figsize=(11, 6))

for _, fila in escenarios.iterrows():

    ax.scatter(
        fila["costo_eur"] / 1000,
        fila["cumplimiento_pct"],
        s=140,
        color=(
            "#287FB8"
            if fila["factible"]
            else "#D99A39"
        )
    )

    ax.annotate(
        fila["alternativa"],
        (
            fila["costo_eur"] / 1000,
            fila["cumplimiento_pct"]
        ),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=8
    )

ax.axvline(
    PRESUPUESTO_MAX_EUR / 1000,
    linestyle="--",
    color="#D64B48",
    label="Budget limit"
)

ax.set(
    title="10 | Cost–Service Trade-off",
    xlabel="Implementation cost (thousand EUR)",
    ylabel="Simulated SLA compliance (%)"
)

ax.legend()

guardar("10_Cost_Service_Tradeoff")


# ============================================================
# 14. DECISIÓN GERENCIAL
# ============================================================

"""
Se selecciona la alternativa factible
con mayor cumplimiento simulado del SLA.

El menor tiempo medio se utiliza
como criterio secundario.

Esto es una regla de decisión ilustrativa,
no una optimización integral.

No reemplaza la evaluación AHP.
"""

factibles = (
    escenarios.loc[escenarios["factible"]]
    .sort_values(
        [
            "cumplimiento_pct",
            "tiempo_promedio_h"
        ],
        ascending=[False, True]
    )
)

recomendada = factibles.iloc[0]

print("\nRESULTADOS DE MONTE CARLO")

print(
    escenarios.round(2).to_string(index=False)
)

interpretar(
    "09. Interpretación prescriptiva",
    f"""
Con un presupuesto máximo
hipotético de EUR
{PRESUPUESTO_MAX_EUR:,.0f},
la alternativa factible
mejor posicionada es:

{recomendada.alternativa}

El tiempo promedio simulado
sería de:
{recomendada.tiempo_promedio_h:.2f} horas.

La reducción frente al
proceso actual sería:
{recomendada.reduccion_horas:.2f} horas.

El cumplimiento del SLA
sería:
{recomendada.cumplimiento_pct:.1f}%.

INTERPRETACIÓN GERENCIAL

La simulación permite comparar
cómo podrían comportarse
distintas alternativas bajo
los supuestos definidos.

Sin embargo, la solución
seleccionada depende de
los efectos hipotéticos
y del presupuesto disponible.

No demuestra que la
intervención produzca
esos resultados reales.

RECOMENDACIÓN ESTRATÉGICA

Contrastar esta alternativa
con la matriz AHP del equipo.

Evaluar costo total,
disponibilidad garantizada,
flexibilidad ante disrupciones
y confiabilidad hotelera.

RECOMENDACIÓN TÁCTICA

Desarrollar un piloto en
destinos seleccionados
durante periodos de alta
y baja demanda.

RECOMENDACIÓN OPERACIONAL

Medir tiempos de respuesta,
reservas confirmadas,
escalaciones y reubicaciones.

La decisión final debe
considerar especialmente
la calidad del alojamiento
y el descanso de la tripulación.

Una mejora de tiempos
no justifica comprometer
las condiciones de seguridad
y descanso.
"""
)


# ============================================================
# 15. SENSIBILIDAD DEL SLA
# ============================================================

parametros_rec = ESCENARIOS[
    recomendada["alternativa"]
]

tiempos_rec = simular_escenario(
    parametros_rec
)

print("\nSENSIBILIDAD DEL SLA")

for sla in [3, 6, 9, 12]:

    cumplimiento = 100 * np.mean(
        tiempos_rec <= sla
    )

    print(
        f"SLA {sla:>2} horas: "
        f"{cumplimiento:.1f}% de cumplimiento"
    )

interpretar(
    "10. Sensibilidad y robustez",
    """
El cumplimiento depende
del plazo de confirmación
establecido.

Un SLA más amplio puede
mejorar matemáticamente
el porcentaje de cumplimiento,
pero no necesariamente
la experiencia del cliente.

Por tanto, los plazos
no deberían modificarse
únicamente para mejorar
los indicadores.

La organización debería
evaluar escenarios adicionales:

1. Mayor ocupación hotelera.
2. Cancelaciones inesperadas.
3. Retrasos de vuelos.
4. Menor disponibilidad de hoteles.
5. Incremento de tarifas.
6. Aumento de tripulaciones.
7. Restricciones de capacidad.

También debería modificarse
la efectividad supuesta
de cada intervención.

Si una pequeña variación
cambia completamente
la alternativa recomendada,
la decisión requiere
más evidencia.

Una solución robusta
debería mantener resultados
aceptables bajo diferentes
condiciones de demanda
y disponibilidad.
"""
)


# ============================================================
# 16. CONCLUSIONES EJECUTIVAS
# ============================================================

interpretar(
    "11. Conclusiones estratégicas",
    f"""
RESULTADOS DEL DIAGNÓSTICO

Se identificaron {k_optimo}
perfiles operacionales.

Se señalaron {n_anomalias}
solicitudes atípicas.

El segmento con mayor
tiempo de confirmación
fue el cluster {peor_cluster}.

Los resultados permiten
formular hipótesis sobre
diferencias de disponibilidad,
anticipación y confiabilidad.

DECISIONES ESTRATÉGICAS

Evaluar una red de hoteles
diferenciada por destino,
temporada y criticidad.

Considerar acuerdos
anticipados de capacidad
en mercados con riesgo
de saturación.

Estudiar la diversificación
de proveedores para reducir
la dependencia de hoteles
específicos.

DECISIONES TÁCTICAS

Revisar las reglas de
confirmación y escalamiento.

Definir mecanismos de
seguimiento para solicitudes
con menor anticipación
o restricciones de capacidad.

Evaluar contratos de
capacidad flexible.

DECISIONES OPERACIONALES

Monitorear solicitudes
pendientes de confirmación.

Verificar disponibilidad
antes de escalar casos.

Registrar cambios,
cancelaciones y reubicaciones.

Revisar las anomalías
y documentar las causas
confirmadas.

PERSPECTIVA PREDICTIVA

Utilizar los clusters como
señales de similitud con
patrones históricos.

La alerta puede orientar
la priorización, pero no
predice exactamente el
tiempo de confirmación.

PERSPECTIVA PRESCRIPTIVA

Comparar alternativas
mediante escenarios
Monte Carlo y restricciones
presupuestales.

Validar los supuestos
antes de invertir.

CONEXIÓN CON EXCELENCIA OPERACIONAL

Los resultados pueden
complementarse con:

- Value Stream Mapping.
- Análisis de cuellos de botella.
- Pareto.
- Ishikawa.
- FMEA.
- DMAIC.
- AHP.

LIMITACIONES

Los datos son sintéticos.

Los efectos de las
intervenciones son supuestos.

La simulación no demuestra
causalidad ni garantiza
ahorros reales.

CONCLUSIÓN GENERAL

El aprendizaje no supervisado
permite descubrir perfiles
operacionales que pueden
pasar inadvertidos en un
diagnóstico tradicional.

La gerencia puede utilizar
esos hallazgos para formular
hipótesis, priorizar riesgos
y diseñar intervenciones
diferenciadas.

El objetivo final no es
solamente confirmar reservas
más rápido.

También es garantizar
disponibilidad, continuidad
operacional y condiciones
adecuadas para las tripulaciones.
"""
)


# ============================================================
# 17. EXPORTACIÓN FINAL
# ============================================================

base.to_csv(
    os.path.join(
        DIRECTORIO,
        "03_Reservas_Clusters_Anomalias.csv"
    ),
    index=False
)

resumen.to_csv(
    os.path.join(
        DIRECTORIO,
        "04_Resumen_Clusters.csv"
    )
)

print("\n" + "=" * 70)
print("LABORATORIO CASO 4 FINALIZADO")
print("=" * 70)

print("Resultados guardados en:")
print(os.path.abspath(DIRECTORIO))

print("\nArchivos generados:")
print("- 10 gráficas PNG individuales")
print("- Base sintética CSV")
print("- Escenarios Monte Carlo CSV")
print("- Base con clusters y anomalías")
print("- Resumen de desempeño por cluster")
```
