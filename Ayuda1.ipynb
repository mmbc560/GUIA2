```python
# ============================================================
# TEAM 1 - EVENT DEMAND PLANNING
# BUSINESS UNDERSTANDING
# ============================================================

# ------------------------------------------------------------
# 1. PROJECT INFORMATION
# ------------------------------------------------------------

PROJECT_NAME = "Event Demand Planning"

ORGANIZATION = "Event Service Organization"

SERVICE_PROCESS = "Event Planning and Service Delivery"


# ------------------------------------------------------------
# 2. BUSINESS PROBLEM
# ------------------------------------------------------------
# Describe the operational problem.
#
# The organization experiences uncertainty in event demand,
# making it difficult to anticipate the resources and capacity
# required to provide the service efficiently.
# ------------------------------------------------------------

BUSINESS_PROBLEM = """
Difficulty in anticipating event demand and translating it
into appropriate capacity and resource requirements.
"""


# ------------------------------------------------------------
# 3. MANAGERIAL DECISION
# ------------------------------------------------------------
# What decision should management improve?
# ------------------------------------------------------------

MANAGERIAL_DECISION = """
Determine the capacity and resources required for future events
according to expected demand.
"""


# ------------------------------------------------------------
# 4. TARGET VARIABLE (Y)
# ------------------------------------------------------------
# Y = The main result that we want to understand,
# explain, predict or improve.
#
# In this case:
# How much demand should the organization expect?
# ------------------------------------------------------------

TARGET_VARIABLE = "Event_Demand"


# ------------------------------------------------------------
# 5. POSSIBLE DRIVERS (X)
# ------------------------------------------------------------
# X variables = factors that may influence Y.
#
# IMPORTANT:
# At this point we are proposing possible relationships.
# We have NOT demonstrated causality.
# ------------------------------------------------------------

DRIVERS = [
    "Event_Type",
    "Day_of_Week",
    "Season",
    "Booking_Lead_Time",
    "Expected_Attendance",
    "Historical_Demand",
    "Available_Capacity"
]


# ------------------------------------------------------------
# 6. MAIN KPIs
# ------------------------------------------------------------

KPIS = [
    "Event_Demand",
    "Capacity_Utilization",
    "Service_Level",
    "Waiting_Time",
    "Cost_Per_Event"
]


# ------------------------------------------------------------
# 7. DATA REQUIRED
# ------------------------------------------------------------

DATA_REQUIRED = [
    "Historical events",
    "Number of attendees",
    "Event dates",
    "Event type",
    "Booking date",
    "Resources used",
    "Available capacity",
    "Service performance"
]


# ------------------------------------------------------------
# 8. ANALYTICS QUESTION
# ------------------------------------------------------------

ANALYTICS_QUESTION = """
What factors are associated with event demand and how can
this information support capacity and resource planning?
"""


# ------------------------------------------------------------
# 9. DISPLAY THE ANALYTICS MAP
# ------------------------------------------------------------

print("=" * 70)
print("TEAM 1 - ANALYTICS MAP")
print("=" * 70)

print("\nPROJECT:")
print(PROJECT_NAME)

print("\nBUSINESS PROBLEM:")
print(BUSINESS_PROBLEM)

print("\nMANAGERIAL DECISION:")
print(MANAGERIAL_DECISION)

print("\nTARGET VARIABLE (Y):")
print(TARGET_VARIABLE)

print("\nPOSSIBLE DRIVERS (X):")
for variable in DRIVERS:
    print("-", variable)

print("\nKPIs:")
for kpi in KPIS:
    print("-", kpi)

print("\nDATA REQUIRED:")
for data in DATA_REQUIRED:
    print("-", data)

print("\nANALYTICS QUESTION:")
print(ANALYTICS_QUESTION)
```

```python
# ============================================================
# TEAM 3 - OUTSOURCED CLEANING SERVICES
# BUSINESS UNDERSTANDING
# ============================================================


# ------------------------------------------------------------
# 1. PROJECT INFORMATION
# ------------------------------------------------------------

PROJECT_NAME = "Strategic Management of Outsourced Cleaning Services"

ORGANIZATION = "Service Organization"

SERVICE_PROCESS = "Outsourced Cleaning Service Management"


# ------------------------------------------------------------
# 2. BUSINESS PROBLEM
# ------------------------------------------------------------
#
# The organization needs to understand how outsourced
# cleaning service management relates to operational
# performance and sustainability.
#
# ------------------------------------------------------------

BUSINESS_PROBLEM = """
The organization needs to improve the management of outsourced
cleaning services in order to strengthen operational performance,
service reliability and sustainability.
"""


# ------------------------------------------------------------
# 3. MANAGERIAL DECISION
# ------------------------------------------------------------

MANAGERIAL_DECISION = """
Determine which supplier management practices and operational
capabilities should be prioritized to improve service performance
and sustainability.
"""


# ------------------------------------------------------------
# 4. TARGET VARIABLE (Y)
# ------------------------------------------------------------
#
# Here we can initially define Y as Service_Performance.
#
# Later, sustainability can be analyzed as an additional
# outcome rather than forcing everything into one indicator.
#
# ------------------------------------------------------------

TARGET_VARIABLE = "Service_Performance"


# ------------------------------------------------------------
# 5. POSSIBLE DRIVERS (X)
# ------------------------------------------------------------

DRIVERS = [
    "Supplier_Reliability",
    "SLA_Compliance",
    "Staff_Availability",
    "Training_Level",
    "Supervision_Level",
    "Response_Time",
    "Digital_Traceability",
    "Resource_Efficiency"
]


# ------------------------------------------------------------
# 6. OPERATIONAL KPIs
# ------------------------------------------------------------

OPERATIONAL_KPIS = [
    "SLA_Compliance",
    "Service_Quality",
    "Response_Time",
    "Number_of_Incidents",
    "Rework_Rate",
    "Cost_Per_Service"
]


# ------------------------------------------------------------
# 7. SUSTAINABILITY KPIs
# ------------------------------------------------------------
#
# These variables allow us to avoid reducing sustainability
# to CO2 only.
# ------------------------------------------------------------

SUSTAINABILITY_KPIS = [
    "Water_Consumption",
    "Chemical_Consumption",
    "Waste_Generation",
    "Resource_Efficiency"
]


# ------------------------------------------------------------
# 8. DATA REQUIRED
# ------------------------------------------------------------

DATA_REQUIRED = [
    "Supplier performance",
    "SLA records",
    "Service incidents",
    "Response times",
    "Staff availability",
    "Training records",
    "Cleaning resource consumption",
    "Waste generation",
    "Service costs",
    "Quality evaluations"
]


# ------------------------------------------------------------
# 9. ANALYTICS QUESTION
# ------------------------------------------------------------

ANALYTICS_QUESTION = """
What operational and supplier-management factors are associated
with better outsourced cleaning service performance and
sustainability?
"""


# ------------------------------------------------------------
# 10. DISPLAY THE ANALYTICS MAP
# ------------------------------------------------------------

print("=" * 70)
print("TEAM 3 - ANALYTICS MAP")
print("=" * 70)

print("\nPROJECT:")
print(PROJECT_NAME)

print("\nBUSINESS PROBLEM:")
print(BUSINESS_PROBLEM)

print("\nMANAGERIAL DECISION:")
print(MANAGERIAL_DECISION)

print("\nTARGET VARIABLE (Y):")
print(TARGET_VARIABLE)

print("\nPOSSIBLE DRIVERS (X):")
for variable in DRIVERS:
    print("-", variable)

print("\nOPERATIONAL KPIs:")
for kpi in OPERATIONAL_KPIS:
    print("-", kpi)

print("\nSUSTAINABILITY KPIs:")
for kpi in SUSTAINABILITY_KPIS:
    print("-", kpi)

print("\nDATA REQUIRED:")
for data in DATA_REQUIRED:
    print("-", data)

print("\nANALYTICS QUESTION:")
print(ANALYTICS_QUESTION)
```
