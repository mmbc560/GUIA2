```python
# =====================================================================
# MBA - OPERATIONAL EXCELLENCE AND SUSTAINABLE SUPPLY CHAINS
# PRACTICE: FROM A BUSINESS PROBLEM TO LINEAR REGRESSION
# =====================================================================
#
# CASE:
# Technical Support Service Center
#
# PURPOSE OF THE PRACTICE:
# This example shows how a SERVICE business problem can be progressively
# translated into an analytics problem.
#
# IMPORTANT:
# The objective is NOT to copy this case.
#
# Each team should later adapt:
#   1. The business problem
#   2. The managerial decision
#   3. The target variable (Y)
#   4. The possible drivers (X)
#   5. The KPIs
#   6. The data required
#   7. The analytics question
#   8. The simulated data
#   9. The interpretation of the results
#
# ANALYTICS JOURNEY:
#
# BUSINESS PROBLEM
#       ↓
# MANAGERIAL DECISION
#       ↓
# TARGET VARIABLE (Y)
#       ↓
# POSSIBLE DRIVERS (X)
#       ↓
# KPIs
#       ↓
# DATA REQUIRED
#       ↓
# ANALYTICS QUESTION
#       ↓
# SIMULATED DATA
#       ↓
# DESCRIPTIVE STATISTICS
#       ↓
# VARIABILITY
#       ↓
# RELATIONSHIPS
#       ↓
# LINEAR REGRESSION
#       ↓
# MANAGERIAL INSIGHT
#
# =====================================================================


# =====================================================================
# 0. IMPORT LIBRARIES
# =====================================================================

# NumPy:
# Used to generate simulated numerical data.
import numpy as np

# Pandas:
# Used to create and analyze our database.
import pandas as pd

# Matplotlib:
# Used to visualize the simulated information.
import matplotlib.pyplot as plt

# Scikit-learn:
# Used here to build the linear regression model.
from sklearn.linear_model import LinearRegression

# Metrics used to evaluate the regression model.
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


# =====================================================================
# 1. BUSINESS PROBLEM OR QUESTION FRAMING
# =====================================================================
#
# FIRST QUESTION:
#
# What is happening in the organization?
#
# We are NOT talking about algorithms yet.
# We are describing the BUSINESS situation.
#
# EXAMPLE:
#
# A technical support service receives customer requests every day.
# Some requests are solved quickly, while others take much longer.
#
# Long and highly variable resolution times can:
#
# - Reduce customer satisfaction
# - Increase operating costs
# - Generate backlog
# - Increase employee workload
# - Reduce service reliability
#
# This is therefore an OPERATIONAL EXCELLENCE problem.
#
# Notice:
# We have NOT said "we need regression".
# First, we define the problem.
# =====================================================================

PROJECT_NAME = "Technical Support Service Performance"

BUSINESS_PROBLEM = """
The technical support service experiences high variability in the
time required to resolve customer requests. Management wants to
understand the operational conditions associated with longer
resolution times and improve service consistency.
"""

print("=" * 80)
print("1. BUSINESS PROBLEM")
print("=" * 80)
print(BUSINESS_PROBLEM)


# =====================================================================
# 2. MANAGERIAL DECISION
# =====================================================================
#
# Now we ask:
#
# WHAT DECISION DOES MANAGEMENT NEED TO IMPROVE?
#
# This is very important.
#
# Analytics should support a decision.
#
# In this example, management wants to determine where operational
# improvements should be prioritized.
#
# Possible decisions could involve:
#
# - Staffing
# - Training
# - Workload distribution
# - Service capacity
# - Process standardization
# - Digital support
#
# =====================================================================

MANAGERIAL_DECISION = """
Determine which operational factors should be prioritized to reduce
resolution time and improve service consistency.
"""

print("\n" + "=" * 80)
print("2. MANAGERIAL DECISION")
print("=" * 80)
print(MANAGERIAL_DECISION)


# =====================================================================
# 3. DEFINE THE TARGET VARIABLE (Y)
# =====================================================================
#
# Remember:
#
# Y = the outcome we want to understand, explain or predict.
#
# Ask:
#
# "What result is management interested in?"
#
# In this example:
#
# Y = Resolution Time
#
# Unit:
# minutes per customer request
#
# IMPORTANT:
#
# Think about the direction of improvement.
#
# For this KPI:
#
# LOWER resolution time is generally desirable,
# provided that service quality is maintained.
#
# =====================================================================

TARGET_VARIABLE_Y = "Resolution_Time_Minutes"

print("\n" + "=" * 80)
print("3. TARGET VARIABLE (Y)")
print("=" * 80)
print(TARGET_VARIABLE_Y)


# =====================================================================
# 4. DEFINE POSSIBLE DRIVERS (X)
# =====================================================================
#
# X variables are factors that MAY be associated with changes in Y.
#
# IMPORTANT:
#
# At this stage these are hypotheses.
#
# We are NOT claiming causality.
#
# Possible X variables for this service example:
#
# 1. Request_Complexity
#       How difficult is the customer request?
#       Scale: 1 to 5
#
# 2. Agent_Experience_Months
#       How much experience does the service agent have?
#
# 3. Current_Workload
#       How many active cases is the agent handling?
#
# 4. Digital_Support_Level
#       Availability/use of digital tools, knowledge bases,
#       automation, etc.
#       Scale: 1 to 5
#
# 5. Customer_Information_Quality
#       How complete is the information initially provided?
#       Scale: 1 to 5
#
# These variables were selected because they have a plausible
# operational relationship with service resolution time.
#
# YOUR TEAM:
#
# Replace these variables with the X variables associated with
# YOUR business problem.
#
# =====================================================================

POSSIBLE_DRIVERS_X = [
    "Request_Complexity",
    "Agent_Experience_Months",
    "Current_Workload",
    "Digital_Support_Level",
    "Customer_Information_Quality"
]

print("\n" + "=" * 80)
print("4. POSSIBLE DRIVERS (X)")
print("=" * 80)

for variable in POSSIBLE_DRIVERS_X:
    print("-", variable)


# =====================================================================
# 5. DEFINE KPIs
# =====================================================================
#
# KPIs translate the business problem into measurable performance.
#
# We can consider different dimensions of Operational Excellence:
#
# TIME
# QUALITY
# PRODUCTIVITY
# SERVICE
# COST
#
# Sustainability could also be incorporated if relevant.
#
# For this example:
# =====================================================================

KPIS = [
    "Average Resolution Time",
    "Resolution Time Variability",
    "First Contact Resolution Rate",
    "Customer Satisfaction",
    "Cases Resolved per Agent"
]

print("\n" + "=" * 80)
print("5. KPIs")
print("=" * 80)

for kpi in KPIS:
    print("-", kpi)


# =====================================================================
# 6. DATA REQUIRED
# =====================================================================
#
# Now ask:
#
# "What information would we need to answer our analytics question?"
#
# Notice the difference:
#
# DATA NEEDED ≠ DATA CURRENTLY AVAILABLE
#
# In a real project, the organization would have to determine:
#
# - Which data already exists?
# - Where is it stored?
# - Is it reliable?
# - Which data is missing?
# - How frequently is it collected?
#
# =====================================================================

DATA_REQUIRED = [
    "Request ID",
    "Request date",
    "Resolution time",
    "Request complexity",
    "Agent experience",
    "Current agent workload",
    "Digital support level",
    "Customer information quality",
    "Customer satisfaction",
    "First contact resolution"
]

print("\n" + "=" * 80)
print("6. DATA REQUIRED")
print("=" * 80)

for data in DATA_REQUIRED:
    print("-", data)


# =====================================================================
# 7. ANALYTICS PROBLEM FRAMING
# =====================================================================
#
# Now we translate:
#
# BUSINESS LANGUAGE
#
# into
#
# ANALYTICS LANGUAGE.
#
# BUSINESS:
#
# "Our resolution times are too variable."
#
# ANALYTICS:
#
# "Which operational variables are associated with changes
# in resolution time?"
#
# =====================================================================

ANALYTICS_QUESTION = """
Which operational factors are associated with changes in customer
request resolution time, and which variables appear to have the
strongest relationship with service performance?
"""

print("\n" + "=" * 80)
print("7. ANALYTICS QUESTION")
print("=" * 80)
print(ANALYTICS_QUESTION)


# =====================================================================
# 8. CREATE SIMULATED DATA
# =====================================================================
#
# IMPORTANT FOR THIS CLASS:
#
# We do not have the real organization's database.
#
# Therefore, we will SIMULATE data.
#
# Simulated data allows us to:
#
# - Practice analytics
# - Understand relationships
# - Test methods
# - Explore scenarios
# - Learn Python
#
# HOWEVER:
#
# Simulated results are NOT evidence about a real organization.
#
# They are a learning laboratory.
#
# =====================================================================


# ---------------------------------------------------------------------
# Set the random seed
# ---------------------------------------------------------------------
#
# This makes the example reproducible.
#
# If everybody uses seed 42, everybody should obtain the same
# simulated numbers.
#
# Try changing 42 to another number later.
# ---------------------------------------------------------------------

np.random.seed(42)


# ---------------------------------------------------------------------
# Number of service requests
# ---------------------------------------------------------------------
#
# Imagine that we observe 500 customer service requests.
#
# STUDENTS:
#
# You can change this number.
#
# Examples:
#
# 100 observations
# 500 observations
# 1,000 observations
#
# ---------------------------------------------------------------------

N = 500


# =====================================================================
# 9. SIMULATE THE X VARIABLES
# =====================================================================


# ---------------------------------------------------------------------
# X1: REQUEST COMPLEXITY
# ---------------------------------------------------------------------
#
# Integer scale:
#
# 1 = Very simple
# 2 = Simple
# 3 = Moderate
# 4 = Complex
# 5 = Very complex
#
# ---------------------------------------------------------------------

request_complexity = np.random.randint(
    low=1,
    high=6,
    size=N
)


# ---------------------------------------------------------------------
# X2: AGENT EXPERIENCE
# ---------------------------------------------------------------------
#
# Experience measured in months.
#
# We simulate values between 1 and 60 months.
#
# ---------------------------------------------------------------------

agent_experience = np.random.randint(
    low=1,
    high=61,
    size=N
)


# ---------------------------------------------------------------------
# X3: CURRENT WORKLOAD
# ---------------------------------------------------------------------
#
# Number of active cases currently assigned to the agent.
#
# We simulate values between 2 and 20.
#
# ---------------------------------------------------------------------

current_workload = np.random.randint(
    low=2,
    high=21,
    size=N
)


# ---------------------------------------------------------------------
# X4: DIGITAL SUPPORT LEVEL
# ---------------------------------------------------------------------
#
# Scale:
#
# 1 = Very low digital support
# 5 = Very high digital support
#
# This could represent:
#
# - Knowledge bases
# - Automated recommendations
# - Digital workflows
# - Search tools
# - AI support
#
# ---------------------------------------------------------------------

digital_support = np.random.randint(
    low=1,
    high=6,
    size=N
)


# ---------------------------------------------------------------------
# X5: CUSTOMER INFORMATION QUALITY
# ---------------------------------------------------------------------
#
# Scale:
#
# 1 = Very incomplete information
# 5 = Very complete information
#
# In many service processes, incomplete information creates:
#
# - Rework
# - Additional contacts
# - Waiting
# - Errors
#
# ---------------------------------------------------------------------

information_quality = np.random.randint(
    low=1,
    high=6,
    size=N
)


# =====================================================================
# 10. CREATE THE TARGET VARIABLE Y
# =====================================================================
#
# Now we simulate RESOLUTION TIME.
#
# This is one of the most important sections to understand.
#
# Because the data are simulated, WE must define the relationships
# that exist in our artificial world.
#
# We will assume:
#
# HIGHER COMPLEXITY
#       → increases resolution time
#
# MORE EXPERIENCE
#       → reduces resolution time
#
# HIGHER WORKLOAD
#       → increases resolution time
#
# BETTER DIGITAL SUPPORT
#       → reduces resolution time
#
# BETTER CUSTOMER INFORMATION
#       → reduces resolution time
#
# We also add RANDOM NOISE.
#
# Why?
#
# Because real service systems are not perfectly predictable.
#
# There are always factors we have not measured.
#
# =====================================================================


# Random operational variability
noise = np.random.normal(
    loc=0,
    scale=8,
    size=N
)


# ---------------------------------------------------------------------
# Artificial relationship used for teaching
# ---------------------------------------------------------------------
#
# IMPORTANT:
#
# These coefficients are NOT real findings.
#
# We choose them only to construct an educational dataset.
#
# Interpretation of the artificial coefficients:
#
# +8 for complexity:
# Every additional complexity level adds approximately 8 minutes.
#
# -0.25 for experience:
# More experienced agents tend to resolve requests faster.
#
# +1.8 for workload:
# More simultaneous cases increase resolution time.
#
# -4 for digital support:
# Better digital support reduces resolution time.
#
# -3 for information quality:
# Better information reduces rework and therefore time.
#
# ---------------------------------------------------------------------

resolution_time = (
    35
    + 8.0 * request_complexity
    - 0.25 * agent_experience
    + 1.8 * current_workload
    - 4.0 * digital_support
    - 3.0 * information_quality
    + noise
)


# ---------------------------------------------------------------------
# Avoid unrealistic negative or extremely small resolution times
# ---------------------------------------------------------------------

resolution_time = np.maximum(resolution_time, 5)


# =====================================================================
# 11. CREATE THE DATAFRAME
# =====================================================================
#
# A DataFrame is similar to a table in Excel.
#
# Rows    = observations / service requests
# Columns = variables
#
# =====================================================================

df = pd.DataFrame({

    "Request_Complexity": request_complexity,

    "Agent_Experience_Months": agent_experience,

    "Current_Workload": current_workload,

    "Digital_Support_Level": digital_support,

    "Customer_Information_Quality": information_quality,

    "Resolution_Time_Minutes": resolution_time
})


# =====================================================================
# 12. LOOK AT THE DATABASE
# =====================================================================
#
# NEVER begin analysis without looking at the data.
#
# Ask:
#
# - What does one row represent?
# - What does each column represent?
# - Do the values make sense?
#
# =====================================================================

print("\n" + "=" * 80)
print("FIRST 10 OBSERVATIONS")
print("=" * 80)

print(df.head(10))


# =====================================================================
# 13. CHECK THE SIZE OF THE DATABASE
# =====================================================================

print("\nDATABASE SIZE:")
print(df.shape)

print("\nInterpretation:")
print("Rows =", df.shape[0], "service requests")
print("Columns =", df.shape[1], "variables")


# =====================================================================
# 14. CHECK DATA TYPES
# =====================================================================

print("\n" + "=" * 80)
print("DATA TYPES")
print("=" * 80)

print(df.dtypes)


# =====================================================================
# 15. CHECK MISSING VALUES
# =====================================================================
#
# This connects with DATA QUALITY.
#
# Before using analytics, we should ask:
#
# Do we have missing information?
#
# =====================================================================

print("\n" + "=" * 80)
print("MISSING VALUES")
print("=" * 80)

print(df.isnull().sum())


# =====================================================================
# 16. DESCRIPTIVE ANALYTICS
# =====================================================================
#
# QUESTION:
#
# WHAT HAPPENED?
#
# Before regression, we need to understand our data.
#
# describe() gives:
#
# count
# mean
# standard deviation
# minimum
# quartiles
# maximum
#
# =====================================================================

print("\n" + "=" * 80)
print("DESCRIPTIVE STATISTICS")
print("=" * 80)

print(df.describe().round(2))


# =====================================================================
# 17. AVERAGE
# =====================================================================
#
# Average answers:
#
# "What is the typical performance level?"
#
# =====================================================================

average_resolution = df["Resolution_Time_Minutes"].mean()

print("\nAVERAGE RESOLUTION TIME:")
print(round(average_resolution, 2), "minutes")


# =====================================================================
# 18. STANDARD DEVIATION
# =====================================================================
#
# Standard deviation answers:
#
# "How much variability exists?"
#
# This is extremely important in Operational Excellence.
#
# A process can have a reasonable average but very high variability.
#
# =====================================================================

std_resolution = df["Resolution_Time_Minutes"].std()

print("\nSTANDARD DEVIATION:")
print(round(std_resolution, 2), "minutes")


# =====================================================================
# 19. COEFFICIENT OF VARIATION - OPTIONAL
# =====================================================================
#
# The coefficient of variation expresses variability relative
# to the average.
#
# Formula:
#
# CV = Standard Deviation / Mean × 100
#
# =====================================================================

cv_resolution = (
    std_resolution /
    average_resolution
) * 100

print("\nCOEFFICIENT OF VARIATION:")
print(round(cv_resolution, 2), "%")


# =====================================================================
# 20. VISUALIZE THE TARGET VARIABLE Y
# =====================================================================
#
# A histogram helps us understand the distribution.
#
# Ask:
#
# - Are most cases concentrated around the average?
# - Is there high dispersion?
# - Are there unusually long service times?
#
# =====================================================================

plt.figure(figsize=(9, 5))

plt.hist(
    df["Resolution_Time_Minutes"],
    bins=25,
    edgecolor="black"
)

plt.title("Distribution of Service Resolution Time")
plt.xlabel("Resolution Time (Minutes)")
plt.ylabel("Number of Requests")

plt.show()


# =====================================================================
# 21. RELATIONSHIP BETWEEN ONE X AND Y
# =====================================================================
#
# Before regression, LOOK at the relationship.
#
# We will start with:
#
# X = Current Workload
# Y = Resolution Time
#
# =====================================================================

plt.figure(figsize=(9, 5))

plt.scatter(
    df["Current_Workload"],
    df["Resolution_Time_Minutes"],
    alpha=0.6
)

plt.title("Current Workload vs. Resolution Time")
plt.xlabel("Current Workload")
plt.ylabel("Resolution Time (Minutes)")

plt.show()


# =====================================================================
# 22. TRY ANOTHER RELATIONSHIP
# =====================================================================
#
# Now:
#
# X = Agent Experience
# Y = Resolution Time
#
# What direction do you expect?
#
# Positive?
# Negative?
# No relationship?
#
# =====================================================================

plt.figure(figsize=(9, 5))

plt.scatter(
    df["Agent_Experience_Months"],
    df["Resolution_Time_Minutes"],
    alpha=0.6
)

plt.title("Agent Experience vs. Resolution Time")
plt.xlabel("Agent Experience (Months)")
plt.ylabel("Resolution Time (Minutes)")

plt.show()


# =====================================================================
# 23. CORRELATION ANALYSIS
# =====================================================================
#
# Correlation gives us an initial measure of linear association.
#
# Values range approximately from:
#
# -1 = strong negative linear relationship
#
#  0 = little/no linear relationship
#
# +1 = strong positive linear relationship
#
# IMPORTANT:
#
# CORRELATION DOES NOT PROVE CAUSATION.
#
# =====================================================================

correlation_matrix = df.corr(numeric_only=True)

print("\n" + "=" * 80)
print("CORRELATION MATRIX")
print("=" * 80)

print(correlation_matrix.round(2))


# =====================================================================
# 24. CORRELATIONS WITH OUR TARGET VARIABLE Y
# =====================================================================
#
# Instead of looking at the complete matrix,
# we can focus on Y.
#
# =====================================================================

correlation_with_y = (
    correlation_matrix["Resolution_Time_Minutes"]
    .sort_values(ascending=False)
)

print("\n" + "=" * 80)
print("CORRELATIONS WITH Y = RESOLUTION TIME")
print("=" * 80)

print(correlation_with_y.round(2))


# =====================================================================
# 25. SIMPLE LINEAR REGRESSION
# =====================================================================
#
# We will begin with ONLY ONE X.
#
# Y = Resolution Time
#
# X = Current Workload
#
# The conceptual model is:
#
# Y = b0 + b1(X)
#
# Where:
#
# b0 = intercept
# b1 = coefficient/slope
#
# =====================================================================


# X must be two-dimensional for scikit-learn
X_simple = df[["Current_Workload"]]

# Y is our target variable
y = df["Resolution_Time_Minutes"]


# Create the model
simple_model = LinearRegression()


# Train the model
simple_model.fit(X_simple, y)


# Predictions
y_pred_simple = simple_model.predict(X_simple)


# =====================================================================
# 26. SIMPLE REGRESSION RESULTS
# =====================================================================

print("\n" + "=" * 80)
print("SIMPLE LINEAR REGRESSION")
print("=" * 80)

print("Dependent variable (Y): Resolution Time")
print("Independent variable (X): Current Workload")

print("\nIntercept:")
print(round(simple_model.intercept_, 3))

print("\nCoefficient for Current Workload:")
print(round(simple_model.coef_[0], 3))

print("\nR²:")
print(round(r2_score(y, y_pred_simple), 3))


# =====================================================================
# 27. INTERPRET THE SIMPLE COEFFICIENT
# =====================================================================
#
# If the coefficient is positive:
#
# Higher workload is associated with longer resolution time.
#
# If negative:
#
# Higher workload would be associated with lower resolution time.
#
# Remember:
#
# ASSOCIATION ≠ CAUSATION
#
# =====================================================================

print("\nMANAGERIAL INTERPRETATION:")

print(
    "According to this simulated dataset, each additional active case "
    "is associated with approximately",
    round(simple_model.coef_[0], 2),
    "additional minutes of resolution time when workload is analyzed alone."
)


# =====================================================================
# 28. VISUALIZE THE SIMPLE REGRESSION
# =====================================================================

plt.figure(figsize=(9, 5))

plt.scatter(
    df["Current_Workload"],
    df["Resolution_Time_Minutes"],
    alpha=0.5,
    label="Observed service requests"
)

plt.plot(
    df["Current_Workload"],
    y_pred_simple,
    linewidth=2,
    label="Regression line"
)

plt.title("Simple Linear Regression: Workload and Resolution Time")
plt.xlabel("Current Workload")
plt.ylabel("Resolution Time (Minutes)")
plt.legend()

plt.show()


# =====================================================================
# 29. MULTIPLE LINEAR REGRESSION
# =====================================================================
#
# Real operational problems normally have MULTIPLE drivers.
#
# Therefore, instead of:
#
# Y = b0 + b1X1
#
# we can estimate:
#
# Y = b0 + b1X1 + b2X2 + b3X3 + ... + error
#
# Here:
#
# Y = Resolution Time
#
# X1 = Request Complexity
# X2 = Agent Experience
# X3 = Current Workload
# X4 = Digital Support
# X5 = Information Quality
#
# =====================================================================

X_multiple = df[
    [
        "Request_Complexity",
        "Agent_Experience_Months",
        "Current_Workload",
        "Digital_Support_Level",
        "Customer_Information_Quality"
    ]
]


# Create model
multiple_model = LinearRegression()


# Train model
multiple_model.fit(X_multiple, y)


# Generate predictions
y_pred_multiple = multiple_model.predict(X_multiple)


# =====================================================================
# 30. DISPLAY THE MULTIPLE REGRESSION COEFFICIENTS
# =====================================================================

coefficients = pd.DataFrame({

    "Variable": X_multiple.columns,

    "Coefficient": multiple_model.coef_

})


# Absolute coefficient is useful only as a first visual comparison.
#
# CAUTION:
# Variables measured on different scales cannot always be directly
# compared simply by coefficient magnitude.
#
coefficients["Absolute_Coefficient"] = (
    coefficients["Coefficient"].abs()
)

coefficients = coefficients.sort_values(
    by="Absolute_Coefficient",
    ascending=False
)

print("\n" + "=" * 80)
print("MULTIPLE LINEAR REGRESSION COEFFICIENTS")
print("=" * 80)

print(coefficients.round(3))


# =====================================================================
# 31. MODEL PERFORMANCE
# =====================================================================
#
# R²:
# Proportion of variability in Y explained by the model.
#
# MAE:
# Average absolute prediction error.
#
# RMSE:
# Penalizes larger errors more strongly.
#
# IMPORTANT:
#
# A model should not be judged only by one metric.
#
# And a statistically good model is not automatically
# managerially useful.
#
# =====================================================================

r2 = r2_score(y, y_pred_multiple)

mae = mean_absolute_error(y, y_pred_multiple)

rmse = np.sqrt(
    mean_squared_error(y, y_pred_multiple)
)

print("\nMODEL PERFORMANCE")

print("R²:", round(r2, 3))

print("MAE:", round(mae, 2), "minutes")

print("RMSE:", round(rmse, 2), "minutes")


# =====================================================================
# 32. ACTUAL VS. PREDICTED
# =====================================================================
#
# If predictions are reasonably close to actual values,
# the points should tend to fall around the diagonal.
#
# =====================================================================

plt.figure(figsize=(7, 6))

plt.scatter(
    y,
    y_pred_multiple,
    alpha=0.6
)

minimum = min(y.min(), y_pred_multiple.min())
maximum = max(y.max(), y_pred_multiple.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linewidth=2
)

plt.title("Actual vs. Predicted Resolution Time")
plt.xlabel("Actual Resolution Time")
plt.ylabel("Predicted Resolution Time")

plt.show()


# =====================================================================
# 33. MANAGERIAL INTERPRETATION OF COEFFICIENTS
# =====================================================================
#
# Now we translate MODEL OUTPUT into BUSINESS LANGUAGE.
#
# This is one of the most important parts of the exercise.
#
# A manager does not need:
#
# "Coefficient = -3.97"
#
# A manager needs:
#
# "Higher digital support is associated with shorter resolution times,
# after accounting for the other variables included in the model."
#
# =====================================================================

print("\n" + "=" * 80)
print("MANAGERIAL INTERPRETATION")
print("=" * 80)

for variable, coefficient in zip(
    X_multiple.columns,
    multiple_model.coef_
):

    if coefficient > 0:

        direction = "HIGHER"

    else:

        direction = "LOWER"

    print(
        f"\n{variable}: coefficient = {coefficient:.2f}"
    )

    print(
        f"An increase in this variable is associated with {direction} "
        "resolution time, holding the other variables in the model constant."
    )


# =====================================================================
# 34. IMPORTANT WARNING: ASSOCIATION IS NOT CAUSATION
# =====================================================================
#
# This is essential for MBA students.
#
# Regression can identify statistical associations.
#
# It does NOT automatically demonstrate that X caused Y.
#
# Example:
#
# Suppose highly trained employees have better performance.
#
# Training may contribute to performance.
#
# But other factors could also explain the result:
#
# - Experience
# - Selection
# - Workload
# - Leadership
# - Technology
# - Type of customer
#
# Therefore:
#
# REGRESSION RESULT
#       ↓
# BUSINESS INTERPRETATION
#       ↓
# PROCESS KNOWLEDGE
#       ↓
# MANAGERIAL DECISION
#
# =====================================================================

print("\nIMPORTANT:")
print(
    "The regression identifies associations in the simulated data. "
    "It does not demonstrate causal relationships."
)


# =====================================================================
# 35. FROM ANALYTICS TO OPERATIONAL EXCELLENCE
# =====================================================================
#
# Now ask:
#
# What could management do with these results?
#
# Possible decisions:
#
# - Review workload allocation
# - Strengthen training
# - Improve digital tools
# - Improve information collection
# - Standardize complex service requests
#
# BUT:
#
# The model does not make the managerial decision.
#
# It provides evidence to SUPPORT the decision.
#
# =====================================================================

print("\n" + "=" * 80)
print("FROM ANALYTICS TO OPERATIONAL EXCELLENCE")
print("=" * 80)

print("""
The analytical model can help management identify operational
conditions associated with longer service resolution times.

The next managerial step would be to evaluate which of these
factors are controllable, which improvements are feasible,
and what trade-offs exist among service, cost, capacity,
quality and employee workload.
""")


# =====================================================================
# 36. STUDENT CHALLENGE - ADAPT THE CODE TO YOUR PROJECT
# =====================================================================
#
# DO NOT COPY THE SERVICE CENTER VARIABLES.
#
# Your task is to transform this code into YOUR analytics model.
#
# STEP 1
# Replace the BUSINESS PROBLEM.
#
# STEP 2
# Replace the MANAGERIAL DECISION.
#
# STEP 3
# Define your own Y.
#
# STEP 4
# Define 4-6 possible X variables.
#
# STEP 5
# Define your KPIs.
#
# STEP 6
# Define the data required.
#
# STEP 7
# Rewrite the analytics question.
#
# STEP 8
# Change the simulated distributions.
#
# STEP 9
# Decide how each X should theoretically be associated with Y.
#
# STEP 10
# Run:
#
# - Descriptive statistics
# - Mean
# - Standard deviation
# - Visualizations
# - Correlations
# - Simple regression
# - Multiple regression
#
# STEP 11
# Interpret the results in BUSINESS LANGUAGE.
#
# =====================================================================


# =====================================================================
# 37. GUIDE FOR THE THREE MBA TEAMS
# =====================================================================
#
# IMPORTANT:
#
# The following examples are NOT complete answers.
# They are guides showing what could be changed.
#
#
# ---------------------------------------------------------------------
# TEAM A - EVENT DEMAND PLANNING
# ---------------------------------------------------------------------
#
# Possible Y:
#
# Event_Demand
#
# Possible X:
#
# Event_Type
# Season
# Booking_Lead_Time
# Day_of_Week
# Marketing_Effort
#
# Students should ask:
#
# Which factors are associated with changes in event demand?
#
#
# ---------------------------------------------------------------------
# TEAM B - CONSULTING & RESEARCH NETWORKING
# ---------------------------------------------------------------------
#
# Possible Y:
#
# Collaboration_Potential
#
# Possible X:
#
# Research_Affinity
# Previous_Interactions
# Industry_Experience
# Network_Connectivity
# Knowledge_Complementarity
#
# Students should ask:
#
# Which characteristics are associated with greater
# collaboration potential?
#
#
# ---------------------------------------------------------------------
# TEAM C - OUTSOURCED CLEANING SERVICES
# ---------------------------------------------------------------------
#
# Possible Y:
#
# Service_Performance
#
# Possible X:
#
# Supplier_Reliability
# Training_Level
# Supervision_Level
# Staff_Availability
# Response_Time
# Resource_Efficiency
#
# Students should ask:
#
# Which operational and supplier-management factors are
# associated with better service performance?
#
# =====================================================================


# =====================================================================
# 38. FINAL REFLECTION
# =====================================================================
#
# Complete this sentence:
#
# "Based on our simulated data, we found that __________
# is associated with __________. This information could help
# management decide __________."
#
# Then answer:
#
# 1. What is your Y?
#
# 2. What are your most relevant X variables?
#
# 3. Which variables appear associated with Y?
#
# 4. Is the relationship positive or negative?
#
# 5. Which variables can management actually control?
#
# 6. Which variables cannot management control?
#
# 7. What managerial decision could the analysis support?
#
# 8. What additional real data would you need before making
#    an actual business decision?
#
# 9. Which Operational Excellence KPI could improve?
#
# 10. What sustainability implication could your decision have?
#
# =====================================================================
```
