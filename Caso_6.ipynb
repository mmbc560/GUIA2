```python

# -*- coding: utf-8 -*-

"""
============================================================
CASO 6
ADVANCED OPERATIONAL ANALYTICS
INDUSTRIAL SCRAP AND OPERATIONAL EXCELLENCE
============================================================

PROBLEMA:
Generación de scrap y desperdicio industrial
durante los procesos productivos.

OBJETIVO:
Identificar perfiles operacionales, detectar
lotes atípicos, predecir tasas de scrap y
evaluar alternativas de mejora.

METODOLOGÍA:
1. Datos sintéticos.
2. Diagnóstico descriptivo.
3. K-Means.
4. Caracterización de clusters.
5. Isolation Forest.
6. Análisis temporal.
7. Random Forest predictivo.
8. Alertas anticipatorias.
9. Monte Carlo.
10. Decisión prescriptiva.
11. Sensibilidad.
12. Recomendaciones gerenciales.

IMPORTANTE:
Los datos son completamente sintéticos.
No representan indicadores reales de la empresa.

Los efectos de las intervenciones son supuestos
didácticos, no impactos causales comprobados.

La simulación debe validarse con datos reales
antes de respaldar decisiones de inversión.
"""

# ============================================================
# 0. LIBRERÍAS Y CONFIGURACIÓN
# ============================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.ensemble import IsolationForest
from sklearn.decomposition import PCA
from sklearn.metrics import (
    silhouette_score,
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
from sklearn.ensemble import RandomForestRegressor

warnings.filterwarnings("ignore", category=FutureWarning)

# ------------------------------------------------------------
# PARÁMETROS MODIFICABLES POR LOS ESTUDIANTES
# ------------------------------------------------------------

SEMILLA = 42

N_LOTES = 1200

MESES = 12

# Meta hipotética de scrap.
META_SCRAP_PCT = 4.0

# Producción nominal por lote.
PRODUCCION_NOMINAL_KG = 1000

# Costo ilustrativo de scrap por kilogramo.
COSTO_SCRAP_KG_COP = 18000

# Presupuesto máximo de intervención.
PRESUPUESTO_MAX_COP = 180_000_000

# Simulaciones Monte Carlo.
N_SIMULACIONES = 5000

# Porcentaje aproximado de anomalías.
CONTAMINACION = 0.05

K_MIN = 2
K_MAX = 6

DIRECTORIO = "Caso_6_Resultados"

os.makedirs(
    DIRECTORIO,
    exist_ok=True
)

rng = np.random.default_rng(SEMILLA)

sns.set_theme(style="whitegrid")

plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "savefig.facecolor": "white",
    "font.size": 10,
    "axes.titlesize": 14,
    "axes.titleweight": "bold"
})


# ============================================================
# FUNCIONES AUXILIARES
# ============================================================

def guardar(nombre):
    """
    Guarda y presenta cada gráfica por separado.
    """
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            DIRECTORIO,
            nombre + ".png"
        ),
        dpi=190,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.show()
    plt.close()


def interpretar(titulo, texto):
    """
    Presenta una interpretación gerencial.
    """
    print("\n" + "=" * 90)
    print(titulo.upper())
    print("=" * 90)
    print(texto.strip())
    print()


# ============================================================
# 1. GENERACIÓN DE DATOS SINTÉTICOS
# ============================================================

"""
Cada fila representa un lote de producción.

VARIABLES DE PROCESO:
- Variabilidad de temperatura.
- Variabilidad de velocidad.
- Cambios de referencia.
- Paradas no programadas.
- Condición de equipos.
- Calidad de materia prima.
- Experiencia del operador.
- Carga de producción.

RESULTADOS:
- Kilogramos de scrap.
- Porcentaje de scrap.
- Kilogramos conformes.
- Costo ilustrativo del desperdicio.

Los perfiles generadores se utilizan solamente
para crear diversidad sintética.

NO se entregan como etiquetas a K-Means.

Para el modelo predictivo se utilizarán
variables medidas antes o durante la
producción, antes de conocer el scrap final.
"""

perfil_generador = rng.choice(
    [0, 1, 2],
    size=N_LOTES,
    p=[0.45, 0.35, 0.20]
)

mes = rng.integers(
    1,
    MESES + 1,
    size=N_LOTES
)

# Variabilidad de temperatura.
# Desviación estándar en grados Celsius.
variabilidad_temperatura = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [1.5, 3.2],
            default=5.5
        ),
        0.7
    ),
    0.2,
    9
)

# Variabilidad de velocidad de proceso (%).
variabilidad_velocidad = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [2, 5],
            default=9
        ),
        1.5
    ),
    0.2,
    15
)

# Cambios de referencia por lote.
cambios_referencia = np.clip(
    rng.poisson(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [0.3, 1.0],
            default=2.0
        )
    ),
    0,
    6
)

# Minutos de parada no programada.
paradas_min = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [5, 18],
            default=38
        ),
        7
    ),
    0,
    70
)

# Condición del equipo: 0-100.
# Valores altos representan mejor condición.
condicion_equipo = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [94, 80],
            default=63
        ),
        7
    ),
    35,
    100
)

# Calidad de materia prima: 0-100.
calidad_materia_prima = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [96, 87],
            default=74
        ),
        6
    ),
    45,
    100
)

# Experiencia del operador: 1-5.
experiencia_operador = np.clip(
    np.rint(
        rng.normal(
            np.select(
                [
                    perfil_generador == 0,
                    perfil_generador == 1
                ],
                [4.4, 3.3],
                default=2.4
            ),
            0.7
        )
    ),
    1,
    5
).astype(int)

# Utilización de capacidad productiva.
utilizacion_capacidad = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [76, 88],
            default=97
        ),
        5
    ),
    50,
    105
)

# ------------------------------------------------------------
# CONSTRUCCIÓN DEL PORCENTAJE DE SCRAP
# ------------------------------------------------------------

"""
La ecuación es exclusivamente sintética.

Permite representar posibles relaciones
entre variabilidad, condición del equipo,
materia prima y desperdicio.

Los coeficientes NO se han estimado
con datos reales.

Una relación dentro de esta simulación
no demuestra causalidad en la industria.
"""

ruido_scrap = rng.normal(
    0,
    0.65,
    N_LOTES
)

scrap_pct = (
    0.7
    + 0.30 * variabilidad_temperatura
    + 0.22 * variabilidad_velocidad
    + 0.30 * cambios_referencia
    + 0.035 * paradas_min
    + 0.055 * (100 - condicion_equipo)
    + 0.065 * (100 - calidad_materia_prima)
    + 0.18 * (5 - experiencia_operador)
    + 0.045 * np.maximum(
        0,
        utilizacion_capacidad - 85
    )
    + ruido_scrap
)

scrap_pct = np.clip(
    scrap_pct,
    0.2,
    35
)

# Cantidad producida por lote.
produccion_kg = np.clip(
    rng.normal(
        PRODUCCION_NOMINAL_KG,
        100,
        N_LOTES
    ),
    600,
    1400
)

scrap_kg = (
    produccion_kg * scrap_pct / 100
)

produccion_conforme_kg = (
    produccion_kg - scrap_kg
)

costo_scrap = (
    scrap_kg * COSTO_SCRAP_KG_COP
)

base = pd.DataFrame({
    "lote_id": [
        f"LOT-{i:05d}"
        for i in range(1, N_LOTES + 1)
    ],
    "mes": mes,
    "variabilidad_temperatura": variabilidad_temperatura,
    "variabilidad_velocidad_pct": variabilidad_velocidad,
    "cambios_referencia": cambios_referencia,
    "paradas_no_programadas_min": paradas_min,
    "condicion_equipo_pct": condicion_equipo,
    "calidad_materia_prima_pct": calidad_materia_prima,
    "experiencia_operador_1_5": experiencia_operador,
    "utilizacion_capacidad_pct": utilizacion_capacidad,
    "produccion_kg": produccion_kg,
    "scrap_pct": scrap_pct,
    "scrap_kg": scrap_kg,
    "produccion_conforme_kg": produccion_conforme_kg,
    "costo_scrap_cop": costo_scrap
})

base["cumple_meta"] = (
    base["scrap_pct"] <= META_SCRAP_PCT
)

base["exceso_scrap_pct"] = np.maximum(
    0,
    base["scrap_pct"] - META_SCRAP_PCT
)

# Variables explicativas disponibles
# antes de conocer el scrap final.

VARIABLES = [
    "variabilidad_temperatura",
    "variabilidad_velocidad_pct",
    "cambios_referencia",
    "paradas_no_programadas_min",
    "condicion_equipo_pct",
    "calidad_materia_prima_pct",
    "experiencia_operador_1_5",
    "utilizacion_capacidad_pct"
]

base.to_csv(
    os.path.join(
        DIRECTORIO,
        "01_Datos_Sinteticos.csv"
    ),
    index=False
)

interpretar(
    "01. Diagnóstico inicial",
    f"""
Se generaron {N_LOTES:,} lotes
de producción sintéticos.

El scrap promedio es:
{base.scrap_pct.mean():.2f}%.

El scrap acumulado es:
{base.scrap_kg.sum():,.1f} kg.

El cumplimiento de la meta
ilustrativa del {META_SCRAP_PCT:.1f}%
alcanza:
{100 * base.cumple_meta.mean():.1f}%.

El costo bruto ilustrativo
del desperdicio es:
COP {base.costo_scrap_cop.sum():,.0f}.

INTERPRETACIÓN GERENCIAL

La generación de scrap puede
representar una pérdida de
material, energía, capacidad
y tiempo productivo.

Sin embargo, el promedio
general no permite identificar
dónde se concentra el problema.

Es necesario analizar las
condiciones de proceso
asociadas con mayores
niveles de desperdicio.

También conviene diferenciar
scrap inevitable, scrap
recuperable y scrap evitable.

RECOMENDACIÓN

Construir una línea base
real por producto, máquina,
turno y lote.

Distinguir las unidades
de medida y las causas
confirmadas del desperdicio.

No utilizar el costo
ilustrativo como pérdida
financiera real sin validar
el costo neto de recuperación,
reproceso o disposición.
"""
)


# ============================================================
# 2. DIAGNÓSTICO DESCRIPTIVO
# ============================================================

fig, ax = plt.subplots(figsize=(10, 5))

sns.histplot(
    base["scrap_pct"],
    bins=35,
    color="#287FB8",
    ax=ax
)

ax.axvline(
    META_SCRAP_PCT,
    color="#D64B48",
    linestyle="--",
    linewidth=2,
    label="Illustrative scrap target"
)

ax.set(
    title="01 | Industrial Scrap Rate Distribution",
    xlabel="Scrap rate (%)",
    ylabel="Production lots"
)

ax.legend()
guardar("01_Scrap_Distribution")

interpretar(
    "02. Interpretación del scrap",
    """
La distribución permite
evaluar la variabilidad
del desperdicio entre lotes.

No basta con conocer
el porcentaje promedio.

También interesa determinar
si existen lotes con tasas
extraordinariamente elevadas.

Un proceso con resultados
muy variables puede dificultar
la planeación de materiales
y la estimación del rendimiento.

La gerencia debería monitorear:

- Scrap por lote.
- Scrap por máquina.
- Scrap por referencia.
- Scrap por turno.
- Kilogramos desperdiciados.
- Costo neto del scrap.
- Rendimiento de primera pasada.

La mejora debe enfocarse
en reducir desperdicios
sin comprometer calidad,
seguridad o productividad.

Una tasa baja de scrap
no garantiza excelencia
operacional si se logra
a costa de producir más
lentamente o generar
otros costos ocultos.
"""
)


# ============================================================
# 3. ESTANDARIZACIÓN
# ============================================================

escalador = StandardScaler()

X = escalador.fit_transform(
    base[VARIABLES]
)


# ============================================================
# 4. SELECCIÓN DE CLUSTERS
# ============================================================

valores_k = list(
    range(K_MIN, K_MAX + 1)
)

silhouettes = []

for k in valores_k:

    modelo_prueba = KMeans(
        n_clusters=k,
        random_state=SEMILLA,
        n_init=15
    )

    etiquetas = modelo_prueba.fit_predict(X)

    puntuacion = silhouette_score(
        X,
        etiquetas,
        sample_size=min(700, len(base)),
        random_state=SEMILLA
    )

    silhouettes.append(puntuacion)

k_optimo = valores_k[
    int(np.argmax(silhouettes))
]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(
    valores_k,
    silhouettes,
    marker="o",
    linewidth=2,
    color="#287FB8"
)

ax.axvline(
    k_optimo,
    color="#E49A32",
    linestyle="--",
    label=f"Selected k = {k_optimo}"
)

ax.set(
    title="02 | Optimal Cluster Selection",
    xlabel="Number of clusters",
    ylabel="Silhouette score"
)

ax.legend()
guardar("02_Silhouette")

interpretar(
    "03. Selección de clusters",
    f"""
Se evaluaron soluciones
entre {K_MIN} y {K_MAX} grupos.

Se seleccionaron:
{k_optimo} clusters.

El índice silhouette fue:
{max(silhouettes):.3f}.

INTERPRETACIÓN

El algoritmo identifica
grupos de lotes con
condiciones operacionales
semejantes.

Los clusters podrían
representar condiciones
de proceso estables,
intermedias o exigentes.

Sin embargo, esas etiquetas
no deben asignarse sin
examinar los perfiles.

La segmentación permite
explorar si diferentes
condiciones productivas
requieren estrategias
de control diferenciadas.

RECOMENDACIÓN

Validar los grupos con
ingenieros de proceso
y responsables de calidad.

Comparar los segmentos
con productos, máquinas
y turnos reales.

La gerencia debería
conservar agrupamientos
que permitan formular
acciones operacionales
claras y verificables.
"""
)


# ============================================================
# 5. K-MEANS
# ============================================================

modelo_kmeans = KMeans(
    n_clusters=k_optimo,
    random_state=SEMILLA,
    n_init=20
)

base["cluster"] = modelo_kmeans.fit_predict(X)

pca = PCA(
    n_components=2,
    random_state=SEMILLA
)

X_pca = pca.fit_transform(X)

fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=base["cluster"].astype(str),
    palette="tab10",
    alpha=0.7,
    s=35,
    ax=ax
)

ax.set(
    title="03 | Production Operational Clusters",
    xlabel="Principal component 1",
    ylabel="Principal component 2"
)

ax.legend(title="Cluster")
guardar("03_KMeans_Clusters")


# ============================================================
# 6. CARACTERIZACIÓN DE CLUSTERS
# ============================================================

perfiles = base.groupby("cluster")[VARIABLES].mean()

media = base[VARIABLES].mean()

desviacion = base[VARIABLES].std().replace(0, 1)

perfiles_z = (
    perfiles - media
) / desviacion

fig, ax = plt.subplots(
    figsize=(12, max(4, k_optimo * 1.2))
)

sns.heatmap(
    perfiles_z,
    annot=True,
    fmt=".2f",
    cmap="RdYlBu_r",
    center=0,
    linewidths=0.5,
    ax=ax
)

ax.set(
    title="04 | Industrial Process Cluster Profiles",
    xlabel="Operational variables",
    ylabel="Cluster"
)

guardar("04_Cluster_Heatmap")

resumen = base.groupby("cluster").agg(
    lotes=("lote_id", "size"),
    scrap_promedio=("scrap_pct", "mean"),
    scrap_total_kg=("scrap_kg", "sum"),
    cumplimiento=("cumple_meta", "mean"),
    paradas=("paradas_no_programadas_min", "mean"),
    condicion=("condicion_equipo_pct", "mean"),
    costo_total=("costo_scrap_cop", "sum")
)

resumen["cumplimiento_pct"] = (
    100 * resumen["cumplimiento"]
)

print("\nRESUMEN DE CLUSTERS")
print(resumen.round(2).to_string())

peor_cluster = int(
    resumen["scrap_promedio"].idxmax()
)

mejor_cluster = int(
    resumen["scrap_promedio"].idxmin()
)

for cluster in perfiles.index:

    dominantes = (
        perfiles_z.loc[cluster]
        .sort_values(ascending=False)
        .head(3)
        .index.tolist()
    )

    interpretar(
        f"04. Interpretación del cluster {cluster}",
        f"""
El cluster {cluster} contiene
{int(resumen.loc[cluster, 'lotes'])}
lotes.

Su scrap promedio es:
{resumen.loc[cluster, 'scrap_promedio']:.2f}%.

El desperdicio acumulado es:
{resumen.loc[cluster, 'scrap_total_kg']:,.1f} kg.

El cumplimiento de la meta
es:
{resumen.loc[cluster, 'cumplimiento_pct']:.1f}%.

Las variables relativamente
más elevadas son:

{', '.join(dominantes)}.

INTERPRETACIÓN GERENCIAL

Las características del
segmento permiten formular
hipótesis sobre las condiciones
asociadas con el desperdicio.

Si predominan variaciones
de temperatura o velocidad,
podría investigarse la
estabilidad del proceso.

Si existen muchas paradas,
convendría analizar el
estado de los equipos.

Si la calidad de materia
prima es desfavorable,
debería revisarse el
control de proveedores.

Las asociaciones observadas
no demuestran causalidad.

RECOMENDACIÓN

Construir un Ishikawa
específico del segmento.

Posteriormente, utilizar
Pareto para identificar
las principales causas
confirmadas.

Finalmente, diseñar
acciones de mejora
dentro de DMAIC.

El equipo debe diferenciar
los problemas de máquina,
método, material, medición,
mano de obra y entorno.
"""
    )


# ============================================================
# 7. ISOLATION FOREST
# ============================================================

modelo_iso = IsolationForest(
    n_estimators=250,
    contamination=CONTAMINACION,
    random_state=SEMILLA
)

base["anomalia"] = (
    modelo_iso.fit_predict(X) == -1
)

base["anomaly_score"] = (
    -modelo_iso.decision_function(X)
)

fig, ax = plt.subplots(figsize=(10, 6))

normales = ~base["anomalia"]

ax.scatter(
    X_pca[normales, 0],
    X_pca[normales, 1],
    s=25,
    alpha=0.5,
    color="#287FB8",
    label="Typical"
)

ax.scatter(
    X_pca[~normales, 0],
    X_pca[~normales, 1],
    s=65,
    facecolors="none",
    edgecolors="#D64B48",
    linewidth=1.5,
    label="Anomaly"
)

ax.set(
    title="05 | Isolation Forest: Unusual Production Lots",
    xlabel="Principal component 1",
    ylabel="Principal component 2"
)

ax.legend()
guardar("05_Isolation_Forest")

n_anomalias = int(
    base["anomalia"].sum()
)

interpretar(
    "05. Detección de anomalías",
    f"""
Isolation Forest señaló
{n_anomalias} lotes atípicos.

Esto representa aproximadamente:
{100 * base.anomalia.mean():.1f}%
de los registros.

INTERPRETACIÓN GERENCIAL

Una anomalía representa
una combinación de variables
poco frecuente.

Podría relacionarse con
variaciones inusuales
de temperatura, velocidad,
paradas o materia prima.

Sin embargo, una anomalía
no significa necesariamente
que el lote sea defectuoso.

También podría tratarse
de una referencia especial
o una condición operacional
válida poco frecuente.

RECOMENDACIÓN

Investigar los lotes
más atípicos.

Comparar sus registros
con especificaciones
técnicas y controles
de calidad.

Si se identifican
desviaciones relevantes,
evaluar acciones correctivas.

La detección automática
debe apoyar la inspección,
no sustituirla.

No se recomienda detener
la producción únicamente
porque un algoritmo
señale una anomalía.
"""
)

print("\n15 LOTES MÁS ATÍPICOS")

print(
    base.sort_values(
        "anomaly_score",
        ascending=False
    )
    .head(15)
    .to_string(index=False)
)


# ============================================================
# 8. SCRAP POR CLUSTER
# ============================================================

fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(
    data=base,
    x="cluster",
    y="scrap_pct",
    estimator="mean",
    errorbar=None,
    color="#368AC1",
    ax=ax
)

ax.axhline(
    META_SCRAP_PCT,
    color="#D64B48",
    linestyle="--",
    linewidth=2,
    label="Illustrative target"
)

ax.set(
    title="06 | Average Scrap Rate by Cluster",
    xlabel="Cluster",
    ylabel="Average scrap rate (%)"
)

ax.legend()
guardar("06_Scrap_By_Cluster")

interpretar(
    "06. Priorización de segmentos",
    f"""
El cluster con mayor scrap
promedio es el {peor_cluster}.

Su tasa es:
{resumen.loc[peor_cluster, 'scrap_promedio']:.2f}%.

El cluster con menor scrap
es el {mejor_cluster}.

Su tasa es:
{resumen.loc[mejor_cluster, 'scrap_promedio']:.2f}%.

INTERPRETACIÓN GERENCIAL

La diferencia entre segmentos
sugiere que el desperdicio
no se distribuye de manera
uniforme entre condiciones
operacionales.

Esto permite priorizar
la investigación.

Sin embargo, el segmento
con mayor porcentaje de scrap
no necesariamente genera
la mayor pérdida económica.

También debe considerarse
el volumen producido,
el costo del material
y la recuperabilidad
del desperdicio.

RECOMENDACIÓN

Construir una matriz
de priorización con:

- Porcentaje de scrap.
- Kilogramos desperdiciados.
- Costo neto del scrap.
- Frecuencia de ocurrencia.
- Criticidad del producto.
- Factibilidad de intervención.

La gerencia debería
priorizar el impacto
económico y operacional
total, no únicamente
el porcentaje más alto.
"""
)


# ============================================================
# 9. EVOLUCIÓN TEMPORAL
# ============================================================

participacion = (
    pd.crosstab(
        base["mes"],
        base["cluster"],
        normalize="index"
    ) * 100
)

fig, ax = plt.subplots(figsize=(10, 5))

for c in participacion.columns:

    ax.plot(
        participacion.index,
        participacion[c],
        marker="o",
        linewidth=2,
        label=f"Cluster {c}"
    )

ax.set(
    title="07 | Monthly Production Cluster Evolution",
    xlabel="Month",
    ylabel="Share of lots (%)",
    xticks=range(1, MESES + 1)
)

ax.legend(ncol=3)
guardar("07_Cluster_Evolution")

# Scrap mensual ponderado por producción.
mensual = base.groupby("mes").agg(
    scrap_kg=("scrap_kg", "sum"),
    produccion_kg=("produccion_kg", "sum")
)

mensual["scrap_pct"] = (
    100
    * mensual["scrap_kg"]
    / mensual["produccion_kg"]
)

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(
    mensual.index,
    mensual["scrap_pct"],
    marker="o",
    linewidth=2,
    color="#C66A35"
)

ax.axhline(
    META_SCRAP_PCT,
    linestyle="--",
    color="#D64B48",
    label="Illustrative target"
)

ax.set(
    title="08 | Monthly Industrial Scrap Rate",
    xlabel="Month",
    ylabel="Scrap rate (%)",
    xticks=range(1, MESES + 1)
)

ax.legend()
guardar("08_Monthly_Scrap")

interpretar(
    "07. Análisis temporal",
    f"""
El scrap mensual mínimo fue:
{mensual.scrap_pct.min():.2f}%.

El máximo fue:
{mensual.scrap_pct.max():.2f}%.

INTERPRETACIÓN GERENCIAL

El monitoreo mensual permite
examinar cambios en el
desempeño productivo.

Sin embargo, las diferencias
entre meses pueden deberse
a la mezcla de productos,
volúmenes y condiciones
de producción.

No debe atribuirse
automáticamente una variación
a la gestión del personal
o al estado de las máquinas.

RECOMENDACIÓN

Construir cartas de control
por procesos comparables.

No mezclar referencias con
especificaciones muy distintas
sin un ajuste adecuado.

Analizar por separado
variabilidad común
y señales de causas especiales.

Cuando existan suficientes
observaciones secuenciales,
podrían implementarse
alertas estadísticas.

La reducción sostenible
del scrap requiere controlar
la estabilidad del proceso,
no solamente reaccionar
a un promedio mensual alto.
"""
)


# ============================================================
# 10. MODELO PREDICTIVO RANDOM FOREST
# ============================================================

"""
A diferencia de K-Means,
Random Forest es supervisado.

Utiliza las variables del proceso
para estimar el porcentaje de scrap.

Para evitar una evaluación optimista,
se separan registros de entrenamiento
y prueba.

IMPORTANTE:
Los datos sintéticos no tienen
una secuencia cronológica real.

En una implementación industrial
se debe validar con lotes posteriores
en el tiempo, sin mezclar información
futura en el entrenamiento.
"""

# Crear orden temporal sintético.
# Dentro de cada mes se asigna un orden
# aleatorio reproducible.
base["orden_interno"] = rng.random(N_LOTES)

base = (
    base.sort_values(
        ["mes", "orden_interno"]
    )
    .reset_index(drop=True)
)

# Últimos 3 meses como prueba.
entrenamiento = base["mes"] <= 9
prueba = base["mes"] >= 10

X_train = base.loc[
    entrenamiento,
    VARIABLES
]

y_train = base.loc[
    entrenamiento,
    "scrap_pct"
]

X_test = base.loc[
    prueba,
    VARIABLES
]

y_test = base.loc[
    prueba,
    "scrap_pct"
]

modelo_rf = RandomForestRegressor(
    n_estimators=300,
    max_depth=10,
    min_samples_leaf=4,
    random_state=SEMILLA,
    n_jobs=-1
)

modelo_rf.fit(
    X_train,
    y_train
)

predicciones = modelo_rf.predict(X_test)

mae = mean_absolute_error(
    y_test,
    predicciones
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        predicciones
    )
)

r2 = r2_score(
    y_test,
    predicciones
)

# ------------------------------------------------------------
# GRÁFICA 09: REAL VS PREDICHO
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(8, 6))

ax.scatter(
    y_test,
    predicciones,
    alpha=0.55,
    color="#287FB8"
)

limite = max(
    float(y_test.max()),
    float(np.max(predicciones))
)

ax.plot(
    [0, limite],
    [0, limite],
    linestyle="--",
    color="#D64B48",
    label="Perfect prediction"
)

ax.set(
    title="09 | Actual vs Predicted Scrap Rate",
    xlabel="Actual scrap rate (%)",
    ylabel="Predicted scrap rate (%)"
)

ax.legend()
guardar("09_Actual_vs_Predicted")

interpretar(
    "08. Interpretación predictiva",
    f"""
Se entrenó un modelo
Random Forest para estimar
el porcentaje de scrap.

El conjunto de prueba
corresponde a los últimos
tres meses sintéticos.

MAE:
{mae:.3f} puntos porcentuales.

RMSE:
{rmse:.3f} puntos porcentuales.

R²:
{r2:.3f}.

INTERPRETACIÓN GERENCIAL

El MAE indica la magnitud
media del error de predicción.

Por ejemplo, un MAE de
un punto porcentual
significaría que las
predicciones se desvían,
en promedio, aproximadamente
un punto de scrap.

El RMSE penaliza más
los errores grandes.

El R² resume cuánto
de la variación del
resultado explica el modelo
en los datos de prueba.

Un buen desempeño sintético
no garantiza resultados
equivalentes en una planta real.

RECOMENDACIÓN

Evaluar el modelo con
lotes reales posteriores.

Verificar la calidad
de las mediciones.

Comparar Random Forest
con un modelo sencillo
de referencia.

Si el error es aceptable,
utilizar la predicción
como alerta de riesgo,
no como sustituto
de los controles de calidad.

El modelo debe actualizarse
si cambian materiales,
equipos o condiciones
del proceso.
"""
)


# ============================================================
# 11. IMPORTANCIA DE VARIABLES
# ============================================================

importancias = pd.Series(
    modelo_rf.feature_importances_,
    index=VARIABLES
).sort_values()

fig, ax = plt.subplots(figsize=(10, 6))

importancias.plot(
    kind="barh",
    color="#368AC1",
    ax=ax
)

ax.set(
    title="10 | Random Forest Feature Importance",
    xlabel="Relative importance",
    ylabel="Process variables"
)

guardar("10_Feature_Importance")

variable_principal = (
    importancias.idxmax()
)

interpretar(
    "09. Interpretación de variables",
    f"""
La variable con mayor
importancia predictiva es:

{variable_principal}.

INTERPRETACIÓN GERENCIAL

La importancia indica
cuánto utiliza el modelo
una variable para generar
sus predicciones.

No significa que esa
variable sea necesariamente
la causa principal del scrap.

Variables correlacionadas
pueden compartir importancia.

Además, la importancia
calculada dentro del
bosque puede presentar
sesgos y debe contrastarse
con otras técnicas.

RECOMENDACIÓN

Seleccionar las variables
más relevantes para
investigación operacional.

Contrastar los hallazgos
con conocimiento técnico,
experimentos controlados
y análisis de causa raíz.

Si se identifica una
condición de proceso
potencialmente crítica,
realizar pruebas
dentro de límites seguros
y aprobados por ingeniería.

No modificar parámetros
industriales automáticamente
a partir de la importancia
estadística del modelo.
"""
)


# ============================================================
# 12. ALERTA PREDICTIVA PARA UN LOTE NUEVO
# ============================================================

"""
Se ingresan las condiciones observadas
de un nuevo lote.

El modelo estima el scrap esperado.

Para una predicción antes de iniciar
la producción, las variables deben
corresponder a valores planificados
o mediciones disponibles previamente.

En este ejemplo, la alerta se genera
durante el seguimiento del lote,
antes de conocer su scrap final.
"""

nuevo_lote = pd.DataFrame([{
    "variabilidad_temperatura": 5.5,
    "variabilidad_velocidad_pct": 9.0,
    "cambios_referencia": 2,
    "paradas_no_programadas_min": 32,
    "condicion_equipo_pct": 65,
    "calidad_materia_prima_pct": 76,
    "experiencia_operador_1_5": 3,
    "utilizacion_capacidad_pct": 96
}])

prediccion_nueva = float(
    modelo_rf.predict(nuevo_lote)[0]
)

# Cluster de similitud.
X_nuevo = escalador.transform(
    nuevo_lote[VARIABLES]
)

cluster_nuevo = int(
    modelo_kmeans.predict(X_nuevo)[0]
)

anomalia_nueva = bool(
    modelo_iso.predict(X_nuevo)[0] == -1
)

interpretar(
    "10. Alerta anticipatoria",
    f"""
El porcentaje estimado
de scrap para el nuevo
lote es:
{prediccion_nueva:.2f}%.

La meta ilustrativa es:
{META_SCRAP_PCT:.2f}%.

El lote pertenece
al cluster:
{cluster_nuevo}.

La observación
{'SÍ' if anomalia_nueva else 'NO'}
fue señalada como atípica.

INTERPRETACIÓN GERENCIAL

La predicción permite
identificar anticipadamente
un posible riesgo de
desperdicio elevado.

Sin embargo, representa
una estimación con error.

No debe interpretarse
como un resultado seguro.

RECOMENDACIÓN

Si la predicción supera
la meta, verificar:

- Estabilidad de temperatura.
- Variación de velocidad.
- Condición de equipos.
- Calidad de materia prima.
- Cambios de referencia.
- Paradas no programadas.

Cualquier ajuste del proceso
debe realizarse dentro
de especificaciones técnicas
y procedimientos autorizados.

La predicción podría
activar una inspección
adicional o una revisión
de ingeniería.

No debería utilizarse
para cambiar parámetros
de forma autónoma.
"""
)


# ============================================================
# 13. MONTE CARLO: ALTERNATIVAS DE MEJORA
# ============================================================

"""
Se comparan ocho alternativas.

Los porcentajes son supuestos
de reducción de componentes
del scrap.

No representan resultados
de experimentos reales.

Las alternativas son:

1. Control estadístico.
2. Mantenimiento predictivo.
3. Estandarización de cambios.
4. Desarrollo de proveedores.
5. Capacitación.
6. Optimización de parámetros.
7. Rediseño Lean Six Sigma.
8. Solución integrada.

Los escenarios comparten la
misma muestra y el mismo ruido.
"""

ESCENARIOS = {

    "Current process": {
        "costo": 0,
        "temperatura": 0,
        "velocidad": 0,
        "cambios": 0,
        "paradas": 0,
        "equipo": 0,
        "materia": 0,
        "experiencia": 0,
        "capacidad": 0
    },

    "Statistical process control": {
        "costo": 65000000,
        "temperatura": 0.40,
        "velocidad": 0.35,
        "cambios": 0.05,
        "paradas": 0.10,
        "equipo": 0.05,
        "materia": 0,
        "experiencia": 0.05,
        "capacidad": 0.05
    },

    "Predictive maintenance": {
        "costo": 115000000,
        "temperatura": 0.15,
        "velocidad": 0.20,
        "cambios": 0,
        "paradas": 0.55,
        "equipo": 0.50,
        "materia": 0,
        "experiencia": 0,
        "capacidad": 0.05
    },

    "Changeover standardization": {
        "costo": 50000000,
        "temperatura": 0.10,
        "velocidad": 0.10,
        "cambios": 0.60,
        "paradas": 0.15,
        "equipo": 0,
        "materia": 0,
        "experiencia": 0.10,
        "capacidad": 0
    },

    "Supplier quality development": {
        "costo": 90000000,
        "temperatura": 0,
        "velocidad": 0,
        "cambios": 0,
        "paradas": 0.05,
        "equipo": 0,
        "materia": 0.60,
        "experiencia": 0,
        "capacidad": 0
    },

    "Operator training": {
        "costo": 40000000,
        "temperatura": 0.10,
        "velocidad": 0.10,
        "cambios": 0.15,
        "paradas": 0.10,
        "equipo": 0,
        "materia": 0,
        "experiencia": 0.60,
        "capacidad": 0
    },

    "Process parameter optimization": {
        "costo": 125000000,
        "temperatura": 0.55,
        "velocidad": 0.55,
        "cambios": 0.10,
        "paradas": 0.10,
        "equipo": 0.10,
        "materia": 0.05,
        "experiencia": 0.10,
        "capacidad": 0.15
    },

    "Lean Six Sigma redesign": {
        "costo": 150000000,
        "temperatura": 0.40,
        "velocidad": 0.40,
        "cambios": 0.45,
        "paradas": 0.35,
        "equipo": 0.20,
        "materia": 0.15,
        "experiencia": 0.30,
        "capacidad": 0.15
    },

    "Integrated transformation": {
        "costo": 260000000,
        "temperatura": 0.65,
        "velocidad": 0.65,
        "cambios": 0.60,
        "paradas": 0.65,
        "equipo": 0.60,
        "materia": 0.50,
        "experiencia": 0.55,
        "capacidad": 0.35
    }
}

indices_mc = rng.integers(
    0,
    len(base),
    size=N_SIMULACIONES
)

muestra = base.iloc[indices_mc]

ruido_mc = rng.normal(
    0,
    0.65,
    N_SIMULACIONES
)


def simular_escenario(parametros):
    """
    Calcula una tasa hipotética de scrap.

    Los efectos se aplican a los componentes
    de la ecuación sintética.

    Esto NO equivale a intervenir físicamente
    el proceso ni a demostrar causalidad.
    """

    p = parametros

    scrap = (
        0.7

        + 0.30
        * muestra["variabilidad_temperatura"].to_numpy()
        * (1 - p["temperatura"])

        + 0.22
        * muestra["variabilidad_velocidad_pct"].to_numpy()
        * (1 - p["velocidad"])

        + 0.30
        * muestra["cambios_referencia"].to_numpy()
        * (1 - p["cambios"])

        + 0.035
        * muestra["paradas_no_programadas_min"].to_numpy()
        * (1 - p["paradas"])

        + 0.055
        * (
            100
            - muestra["condicion_equipo_pct"].to_numpy()
        )
        * (1 - p["equipo"])

        + 0.065
        * (
            100
            - muestra["calidad_materia_prima_pct"].to_numpy()
        )
        * (1 - p["materia"])

        + 0.18
        * (
            5
            - muestra["experiencia_operador_1_5"].to_numpy()
        )
        * (1 - p["experiencia"])

        + 0.045
        * np.maximum(
            0,
            muestra["utilizacion_capacidad_pct"].to_numpy()
            - 85
        )
        * (1 - p["capacidad"])

        + ruido_mc
    )

    return np.clip(
        scrap,
        0.2,
        35
    )


resultados = []

for nombre, parametros in ESCENARIOS.items():

    tasas = simular_escenario(parametros)

    produccion_mc = (
        muestra["produccion_kg"].to_numpy()
    )

    scrap_kg_mc = (
        produccion_mc * tasas / 100
    )

    resultados.append({
        "alternativa": nombre,
        "costo_cop": parametros["costo"],
        "scrap_promedio_pct": tasas.mean(),
        "scrap_total_muestra_kg": scrap_kg_mc.sum(),
        "cumplimiento_pct": 100 * np.mean(
            tasas <= META_SCRAP_PCT
        ),
        "percentil_90_scrap_pct": np.percentile(
            tasas,
            90
        )
    })

escenarios = pd.DataFrame(resultados)

base_simulada = escenarios.loc[
    escenarios["alternativa"] == "Current process"
].iloc[0]

escenarios["reduccion_scrap_pp"] = (
    base_simulada["scrap_promedio_pct"]
    - escenarios["scrap_promedio_pct"]
)

# Convertir el promedio por lote simulado
# a un horizonte ilustrativo de N_LOTES.
#
# No multiplicar directamente los 5000 lotes
# simulados por el costo anual.

escenarios["scrap_esperado_horizonte_kg"] = (
    escenarios["scrap_total_muestra_kg"]
    / N_SIMULACIONES
    * N_LOTES
)

scrap_base_horizonte = (
    base_simulada["scrap_total_muestra_kg"]
    / N_SIMULACIONES
    * N_LOTES
)

escenarios["kg_evitable_hipotetico"] = (
    scrap_base_horizonte
    - escenarios["scrap_esperado_horizonte_kg"]
)

escenarios["beneficio_potencial_cop"] = (
    escenarios["kg_evitable_hipotetico"]
    * COSTO_SCRAP_KG_COP
)

escenarios["beneficio_neto_hipotetico_cop"] = (
    escenarios["beneficio_potencial_cop"]
    - escenarios["costo_cop"]
)

escenarios["factible"] = (
    escenarios["costo_cop"]
    <= PRESUPUESTO_MAX_COP
)

escenarios.to_csv(
    os.path.join(
        DIRECTORIO,
        "02_Escenarios_Monte_Carlo.csv"
    ),
    index=False
)


# ============================================================
# 14. COMPARACIÓN DE SCRAP POR ALTERNATIVA
# ============================================================

fig, ax = plt.subplots(figsize=(11, 6))

colores = [
    "#287FB8" if x else "#D5A04A"
    for x in escenarios["factible"]
]

ax.barh(
    escenarios["alternativa"],
    escenarios["scrap_promedio_pct"],
    color=colores
)

ax.axvline(
    META_SCRAP_PCT,
    color="#D64B48",
    linestyle="--",
    label="Illustrative target"
)

ax.invert_yaxis()

ax.set(
    title="11 | Simulated Scrap Rate by Intervention",
    xlabel="Average simulated scrap rate (%)",
    ylabel=""
)

ax.legend()
guardar("11_Scenario_Comparison")


# ============================================================
# 15. TRADE-OFF COSTO VS DESEMPEÑO
# ============================================================

fig, ax = plt.subplots(figsize=(11, 6))

for _, fila in escenarios.iterrows():

    ax.scatter(
        fila["costo_cop"] / 1e6,
        fila["scrap_promedio_pct"],
        s=140,
        color=(
            "#287FB8"
            if fila["factible"]
            else "#D99A39"
        )
    )

    ax.annotate(
        fila["alternativa"],
        (
            fila["costo_cop"] / 1e6,
            fila["scrap_promedio_pct"]
        ),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=8
    )

ax.axvline(
    PRESUPUESTO_MAX_COP / 1e6,
    linestyle="--",
    color="#D64B48",
    label="Budget limit"
)

ax.set(
    title="12 | Cost–Scrap Reduction Trade-off",
    xlabel="Implementation cost (million COP)",
    ylabel="Simulated scrap rate (%)"
)

ax.legend()
guardar("12_Cost_Scrap_Tradeoff")


# ============================================================
# 16. DECISIÓN PRESCRIPTIVA
# ============================================================

"""
Se selecciona la alternativa factible
con menor scrap promedio simulado.

El menor costo se utiliza como
criterio secundario.

La regla no reemplaza un AHP ni
una evaluación financiera completa.
"""

factibles = (
    escenarios.loc[escenarios["factible"]]
    .sort_values(
        [
            "scrap_promedio_pct",
            "costo_cop"
        ],
        ascending=[True, True]
    )
)

recomendada = factibles.iloc[0]

print("\nRESULTADOS MONTE CARLO")

print(
    escenarios.round(2).to_string(index=False)
)

interpretar(
    "11. Interpretación prescriptiva",
    f"""
Con un presupuesto máximo
hipotético de COP
{PRESUPUESTO_MAX_COP:,.0f},
la alternativa factible
mejor posicionada es:

{recomendada.alternativa}

Su scrap promedio simulado
sería:
{recomendada.scrap_promedio_pct:.2f}%.

La reducción frente al
proceso actual sería:
{recomendada.reduccion_scrap_pp:.2f}
puntos porcentuales.

El desperdicio evitable
hipotético en el horizonte
de {N_LOTES} lotes sería:
{recomendada.kg_evitable_hipotetico:,.1f} kg.

El beneficio potencial
ilustrativo sería:
COP {recomendada.beneficio_potencial_cop:,.0f}.

INTERPRETACIÓN GERENCIAL

La simulación permite
comparar intervenciones
bajo supuestos comunes.

Sin embargo, los resultados
dependen de los porcentajes
hipotéticos de efectividad.

No se demuestra que la
alternativa produzca
ese resultado real.

RECOMENDACIÓN ESTRATÉGICA

Contrastar la alternativa
con la matriz AHP.

Considerar criterios como:

- Reducción de scrap.
- Costo de implementación.
- Calidad del producto.
- Estabilidad del proceso.
- Factibilidad técnica.
- Impacto ambiental.
- Tiempo de implementación.

RECOMENDACIÓN TÁCTICA

Diseñar un piloto en
una referencia o línea
de producción seleccionada.

Comparar resultados
antes y después.

RECOMENDACIÓN OPERACIONAL

Monitorear variables
críticas de proceso,
kilogramos desperdiciados
y cumplimiento de calidad.

El proyecto debe demostrar
que la mejora no traslada
el desperdicio a otras
etapas del proceso.
"""
)


# ============================================================
# 17. ANÁLISIS DE SENSIBILIDAD
# ============================================================

parametros_rec = ESCENARIOS[
    recomendada["alternativa"]
]

tasas_rec = simular_escenario(
    parametros_rec
)

print("\nSENSIBILIDAD DE LA META")

for meta in [3, 4, 5, 6, 8]:

    cumplimiento = 100 * np.mean(
        tasas_rec <= meta
    )

    print(
        f"Meta {meta:>2}%: "
        f"{cumplimiento:.1f}% de lotes conformes"
    )

interpretar(
    "12. Sensibilidad y robustez",
    """
El cumplimiento depende
de la meta de scrap.

Una meta menos exigente
puede aumentar matemáticamente
el porcentaje de cumplimiento,
pero no reducir desperdicios.

La organización debería
analizar escenarios como:

1. Materia prima de menor calidad.
2. Mayor variabilidad térmica.
3. Incremento de cambios de referencia.
4. Fallas de equipos.
5. Aumento de utilización.
6. Restricciones de mantenimiento.
7. Mayor complejidad de productos.

También debe modificarse
la efectividad supuesta
de las intervenciones.

Si la alternativa recomendada
cambia ante pequeños ajustes,
la decisión es sensible
y requiere más evidencia.

Una solución robusta
debería mantener un
desempeño aceptable
ante variaciones normales
de producción.

El equipo debería evaluar
también calidad, capacidad,
consumo energético y costos.
"""
)


# ============================================================
# 18. CONCLUSIONES EJECUTIVAS
# ============================================================

interpretar(
    "13. Conclusiones estratégicas",
    f"""
RESULTADOS DEL DIAGNÓSTICO

Se identificaron {k_optimo}
perfiles operacionales.

Se señalaron {n_anomalias}
lotes atípicos.

El segmento con mayor
scrap fue el cluster
{peor_cluster}.

Se entrenó un modelo
predictivo Random Forest.

El error MAE fue:
{mae:.3f} puntos porcentuales.

DECISIONES ESTRATÉGICAS

Evaluar una estrategia
integral de reducción
de scrap basada en
estabilidad del proceso.

Priorizar inversiones
según impacto económico,
calidad y factibilidad.

Considerar el vínculo
entre desperdicio,
rentabilidad y circularidad.

DECISIONES TÁCTICAS

Revisar controles
estadísticos de proceso.

Evaluar mantenimiento
preventivo y predictivo.

Analizar calidad de
materia prima y desempeño
de proveedores.

Estandarizar cambios
de referencia.

DECISIONES OPERACIONALES

Monitorear temperatura,
velocidad y paradas.

Investigar lotes atípicos.

Registrar causas
confirmadas del scrap.

Implementar controles
dentro de especificaciones
técnicas aprobadas.

PERSPECTIVA PREDICTIVA

Utilizar Random Forest
para estimar el scrap
bajo condiciones conocidas.

Validar el desempeño
con lotes futuros.

No interpretar una
predicción como certeza.

PERSPECTIVA PRESCRIPTIVA

Comparar intervenciones
mediante Monte Carlo.

Analizar restricciones
presupuestales.

Contrastar resultados
con la matriz AHP.

CONEXIÓN CON EXCELENCIA OPERACIONAL

El laboratorio puede
complementarse con:

- Lean.
- Six Sigma.
- DMAIC.
- Pareto.
- Ishikawa.
- FMEA.
- Control Charts.
- MSA.
- AHP.
- Value Stream Mapping.

CONEXIÓN CON ECONOMÍA CIRCULAR

Reducir scrap puede
disminuir el consumo
de materias primas.

También puede reducir
residuos y necesidades
de disposición.

Sin embargo, deben
evaluarse las posibilidades
de recuperación y reciclaje
sin comprometer calidad.

LIMITACIONES

Los datos son sintéticos.

Los coeficientes del
modelo generador son
supuestos.

Las intervenciones
no tienen efectos
causales comprobados.

Los beneficios monetarios
son ilustrativos.

CONCLUSIÓN GENERAL

La analítica avanzada
permite pasar de medir
el scrap a investigar
las condiciones asociadas
con su generación.

El aprendizaje no
supervisado identifica
perfiles y anomalías.

El aprendizaje supervisado
permite estimar resultados
futuros bajo condiciones
conocidas.

Monte Carlo facilita
comparar escenarios
gerenciales hipotéticos.

La generación de valor
dependerá de validar
los resultados y convertirlos
en mejoras sostenibles.
"""
)


# ============================================================
# 19. EXPORTACIÓN FINAL
# ============================================================

base.to_csv(
    os.path.join(
        DIRECTORIO,
        "03_Lotes_Clusters_Anomalias.csv"
    ),
    index=False
)

resumen.to_csv(
    os.path.join(
        DIRECTORIO,
        "04_Resumen_Clusters.csv"
    )
)

predicciones_df = pd.DataFrame({
    "scrap_real_pct": y_test.to_numpy(),
    "scrap_predicho_pct": predicciones
})

predicciones_df.to_csv(
    os.path.join(
        DIRECTORIO,
        "05_Predicciones_Random_Forest.csv"
    ),
    index=False
)

print("\n" + "=" * 70)
print("LABORATORIO CASO 6 FINALIZADO")
print("=" * 70)

print("Resultados guardados en:")
print(os.path.abspath(DIRECTORIO))

print("\nArchivos generados:")
print("- 12 gráficas PNG independientes")
print("- Base sintética CSV")
print("- Escenarios Monte Carlo CSV")
print("- Lotes con clusters y anomalías")
print("- Resumen por cluster")
print("- Predicciones Random Forest")
```
