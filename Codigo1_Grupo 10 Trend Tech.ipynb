```python
# ==========================================================
# ANALIZADOR DE PERFILES - TREND TECH - GRUPO 10
# VERSIÓN CORREGIDA Y COMPLETA
# Compatible con Google Colab / Jupyter
# ==========================================================

import matplotlib.pyplot as plt
from collections import Counter

print("🤖 ANALIZADOR DE PERFILES - TREND TECH")
print("Grupo 10 cargado correctamente.")

# ==========================================================
# 1. BASE DE DATOS DEL GRUPO 10
# ==========================================================

estudiantes = [
    {
        "nombre": "Johan Daniel Fajardo Avila",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["empresa", "internacionalizacion", "negocios globales"],
        "habilidades": ["Liderazgo", "Resolucion de problemas"],
        "aprender": ["Inteligencia artificial", "Marketing digital"],
        "aportes": ["Liderar el equipo", "Resolver problemas complejos"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Conectar el conocimiento con accion"
    },
    {
        "nombre": "Samuel Andres Moreno Moreno",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["comercio global", "innovacion empresarial"],
        "habilidades": ["Comunicacion", "Trabajo en equipo"],
        "aprender": ["Inteligencia artificial", "Marketing digital"],
        "aportes": ["Generacion de ideas", "Analizar informacion"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Crear valor a traves de la innovacion"
    },
    {
        "nombre": "Mario Alberto Morales Torres",
        "carrera": "Mercadeo",
        "interes_personal": ["tecnologia", "redes sociales"],
        "habilidades": ["Liderazgo", "Pensamiento critico"],
        "aprender": ["Marketing digital", "Innovacion"],
        "aportes": ["Liderazgo", "Pensamiento critico"],
        "proyecto": "Empresa de marketing digital con alto impacto",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Isabella Torres Vargas",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["fashion marketing", "publicidad"],
        "habilidades": ["Organizacion y planificacion", "Diseño"],
        "aprender": ["Transformacion digital", "Marketing digital"],
        "aportes": ["Organizacion del trabajo", "Diseñar presentaciones"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Haaron Galeano Yepes",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["inteligencia artificial aplicada a negocios"],
        "habilidades": ["Trabajo en equipo", "Comunicacion"],
        "aprender": ["Inteligencia artificial", "Automatizacion de procesos"],
        "aportes": ["Generacion de ideas", "Organizacion del trabajo"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Siempre en busqueda de nuevas ideas"
    },
    {
        "nombre": "Sara Valentina Zapata Avila",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["marketing digital", "sostenibilidad", "tecnologia"],
        "habilidades": ["Diseño", "Investigacion", "Redaccion"],
        "aprender": ["Marketing digital", "Sostenibilidad y tecnologia"],
        "aportes": ["Generacion de ideas", "Analisis de informacion", "Diseñar", "Liderar el equipo"],
        "proyecto": "Crear una solucion tecnologica para una empresa",
        "frase": "Siempre en busqueda de nuevas ideas"
    },
    {
        "nombre": "Isabella Vega Cruz",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["innovacion", "emprendimiento"],
        "habilidades": ["Creatividad", "Liderazgo"],
        "aprender": ["Innovacion", "Emprendimiento"],
        "aportes": ["Diseñar presentaciones", "Liderazgo"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Mi objetivo es hacer que las cosas pasen"
    },
    {
        "nombre": "Maria Jose Tello Paez",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["tecnologia", "innovacion"],
        "habilidades": ["Analisis del mercado", "Liderazgo"],
        "aprender": [],
        "aportes": ["Entender necesidades del mercado", "Liderazgo de equipos"],
        "proyecto": "Generar valor real en las empresas",
        "frase": ""
    },
    {
        "nombre": "Juan Esteban Tabares Ruiz",
        "carrera": "Ingenieria Industrial",
        "interes_personal": ["matematica"],
        "habilidades": ["Analisis de datos", "Pensamiento logico"],
        "aprender": ["Transformacion digital"],
        "aportes": ["Analizar informacion"],
        "proyecto": "Mejorar procesos con tecnologia",
        "frase": "Los datos cuentan historias"
    },
    {
        "nombre": "Valentina Romero Gomez",
        "carrera": "Contaduria Publica",
        "interes_personal": ["economia"],
        "habilidades": ["Organizacion y planificacion", "Analisis de datos"],
        "aprender": ["Marketing digital", "Desarrollo de apps o software"],
        "aportes": ["Organizacion del trabajo", "Investigar informacion"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Los datos cuentan historias"
    },
    {
        "nombre": "Nicolas Eduardo Rojas Cordoba",
        "carrera": "Ingenieria en Ciencia de Datos",
        "interes_personal": ["musica", "causas sociales"],
        "habilidades": ["Analisis de datos", "Creacion de graficas", "Tablas", "Visualizacion"],
        "aprender": ["Power BI", "Big Data", "Data Science"],
        "aportes": ["Orden", "Empatia", "Eficiencia"],
        "proyecto": "Integrar personas al mundo tecnologico",
        "frase": "Un gran poder conlleva una gran responsabilidad"
    },
    {
        "nombre": "Sara Sofia Rodriguez Obando",
        "carrera": "Administracion de Empresas",
        "interes_personal": ["gestion", "administracion"],
        "habilidades": ["Organizacion", "Planificacion", "Diseño"],
        "aprender": ["Transformacion digital", "Innovacion", "Emprendimiento"],
        "aportes": ["Coordinar tareas", "Organizacion del trabajo"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Siempre en busqueda de nuevas ideas"
    },
    {
        "nombre": "Paula Silvana Rincon Torres",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["culturas"],
        "habilidades": ["Trabajo en equipo", "Liderazgo", "Comunicacion", "Diseño"],
        "aprender": ["Automatizacion de procesos", "Marketing digital"],
        "aportes": ["Diseñar presentaciones"],
        "proyecto": "Desarrollar una idea de negocio innovadora",
        "frase": "Siempre en busqueda de nuevas ideas"
    },
    {
        "nombre": "Catalina Riaño Toro",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["negocio"],
        "habilidades": ["Creatividad", "Resolucion de problemas"],
        "aprender": ["Innovacion", "Emprendimiento", "Automatizacion"],
        "aportes": ["Generacion de ideas", "Diseñar presentaciones"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Pensar diferente es mi ventaja"
    },
    {
        "nombre": "Santiago Ramos Correa",
        "carrera": "Economia",
        "interes_personal": ["tecnologia"],
        "habilidades": ["Analisis de datos", "Organizacion y planificacion", "Trabajo en equipo"],
        "aprender": ["Inteligencia artificial", "Emprendimiento", "Automatizacion"],
        "aportes": ["Organizacion", "Resolver problemas", "Motivar al equipo"],
        "proyecto": "Mejorar procesos con tecnologia",
        "frase": "La disciplina es la clave del exito"
    },
    {
        "nombre": "Sofia Isabella Pineda Lopez",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["manejo de empresas con tecnologia"],
        "habilidades": ["Comunicacion", "Trabajo en equipo", "Resolucion de problemas"],
        "aprender": ["Innovacion", "Emprendimiento", "Automatizacion", "Sostenibilidad y tecnologia"],
        "aportes": ["Analizar informacion", "Organizacion del trabajo"],
        "proyecto": "Crear un proyecto sostenible con tecnologia",
        "frase": "Conectar conocimiento con accion"
    },
    {
        "nombre": "Gabriela Patiño Santos",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["tecnologia", "moda"],
        "habilidades": ["Organizacion y planificacion", "Creatividad", "Innovacion"],
        "aprender": ["Marketing digital", "Sostenibilidad y tecnologia"],
        "aportes": ["Apoyar en todo lo necesario", "Resolver problemas"],
        "proyecto": "Diseñar una app o herramienta digital",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Laura Catalina Parra Camacho",
        "carrera": "Mercadeo",
        "interes_personal": ["innovacion", "tecnologia"],
        "habilidades": ["Creatividad", "Trabajo en equipo"],
        "aprender": ["Inteligencia artificial", "Marketing digital"],
        "aportes": ["Generacion de ideas", "Apoyar al equipo"],
        "proyecto": "Desarrollar una idea de negocio innovadora que integre tecnologia",
        "frase": ""
    },
    {
        "nombre": "Paula Ximena Ortiz Garzon",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["tecnologia"],
        "habilidades": ["Creatividad", "Trabajo en equipo"],
        "aprender": ["Marketing digital", "Innovacion", "Emprendimiento"],
        "aportes": ["Generacion de ideas", "Organizacion"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Valentina Obando Mendez",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["creatividad", "comunicacion"],
        "habilidades": ["Comunicacion", "Diseño"],
        "aprender": ["Transformacion digital", "Inteligencia artificial"],
        "aportes": ["Generacion de ideas", "Diseñar presentaciones"],
        "proyecto": "Desarrollar una idea innovadora para una empresa",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Juan Camilo Muñoz Mendivelso",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["marketing digital", "branding"],
        "habilidades": ["Creatividad", "Trabajo en equipo"],
        "aprender": ["Marketing digital", "Innovacion", "Emprendimiento"],
        "aportes": ["Organizacion", "Generacion de ideas"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Pensar diferente es mi ventaja"
    },
    {
        "nombre": "Samuel Esteban Mosquera Muñoz",
        "carrera": "Ingenieria Mecatronica",
        "interes_personal": ["tecnologia"],
        "habilidades": ["Resolucion de problemas", "Trabajo en equipo"],
        "aprender": ["Robotica", "Tecnologia"],
        "aportes": ["Organizacion", "Generacion de ideas", "Curiosidad"],
        "proyecto": "Crear una solucion tecnologica para empresas",
        "frase": "Siempre en busqueda de nuevas ideas"
    },
    {
        "nombre": "Jacobo Moreno Rodriguez",
        "carrera": "Administracion de Empresas",
        "interes_personal": ["gestion estrategica", "negocios digitales"],
        "habilidades": ["Trabajo en equipo", "Creatividad"],
        "aprender": ["Inteligencia artificial", "Automatizacion de procesos"],
        "aportes": ["Generacion de ideas", "Resolucion de problemas", "Motivar al equipo"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Juntos creamos valor, juntos generamos resultados"
    },
    {
        "nombre": "Juan Pablo Morales Uribe",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["negocios"],
        "habilidades": ["Creatividad", "Pensamiento logico"],
        "aprender": ["Inteligencia artificial", "Marketing digital"],
        "aportes": ["Generacion de ideas", "Resolver problemas"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Transformo ideas en soluciones"
    },
    {
        "nombre": "Natalia Morales Machado",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["estrategias innovadoras", "diseño creativo"],
        "habilidades": ["Marketing digital", "Data analytics", "Coordinacion"],
        "aprender": ["Estrategias innovadoras"],
        "aportes": ["Soluciones practicas", "Coordinacion del equipo"],
        "proyecto": "Proyecto que integre sostenibilidad y tecnologia",
        "frase": "Estrategia con proposito, innovacion con impacto"
    },
    {
        "nombre": "Juan Nikolas Mora Amezquita",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["automatizacion de marketing"],
        "habilidades": ["Comunicacion", "Creatividad"],
        "aprender": ["Automatizacion de procesos", "Inteligencia artificial"],
        "aportes": ["Generacion de ideas", "Liderar el equipo"],
        "proyecto": "Crear una solucion tecnologica para una empresa",
        "frase": "Siempre en busqueda de nuevas ideas"
    },
    {
        "nombre": "Juliana Montoya Aguirre",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["innovacion", "comunicacion", "tecnologia", "negociacion"],
        "habilidades": ["Organizacion y planificacion", "Resolucion de problemas"],
        "aprender": ["Desarrollo de apps o software", "Sostenibilidad y tecnologia"],
        "aportes": ["Generacion de ideas", "Liderar el equipo"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Sofia Martinez Molina",
        "carrera": "Mercadeo",
        "interes_personal": ["tecnologia", "creacion de contenido"],
        "habilidades": ["Creatividad", "Diseño"],
        "aprender": ["Innovacion", "Emprendimiento", "Marketing digital"],
        "aportes": ["Apoyar en todo lo necesario", "Generacion de ideas"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Daniel Manrique Roldan",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["importaciones", "exportaciones"],
        "habilidades": ["Trabajo en equipo", "Resolucion de problemas"],
        "aprender": ["Automatizacion de procesos", "Blockchain"],
        "aportes": ["Investigar informacion", "Generacion de ideas"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "La tecnologia es mi herramienta para cambiar el mundo"
    },
    {
        "nombre": "Isabella Manrique Bautista",
        "carrera": "Mercadeo",
        "interes_personal": ["tecnologia", "creacion de contenido"],
        "habilidades": ["Diseño", "Creatividad"],
        "aprender": ["Marketing digital", "Innovacion", "Emprendimiento"],
        "aportes": ["Apoyar en todo lo necesario", "Generacion de ideas"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Sarah Alejandra Mandon Galvis",
        "carrera": "Economia",
        "interes_personal": ["negocios"],
        "habilidades": ["Organizacion y planificacion", "Pensamiento logico"],
        "aprender": ["Analisis de datos", "Marketing digital"],
        "aportes": ["Analizar informacion", "Apoyar en todo lo necesario"],
        "proyecto": "Resolver un problema real con tecnologia",
        "frase": "Conectar conocimiento con accion"
    },
    {
        "nombre": "Natalia Maldonado Sanchez",
        "carrera": "Administracion de Empresas",
        "interes_personal": ["negocio"],
        "habilidades": ["Organizacion y planificacion", "Liderazgo"],
        "aprender": ["Marketing digital", "Innovacion", "Emprendimiento"],
        "aportes": ["Liderar el equipo", "Coordinar tareas"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Maria Fernanda Gonzalez Bernal",
        "carrera": "Mercadeo",
        "interes_personal": ["redes sociales", "empresas", "diseño"],
        "habilidades": ["Creatividad", "Diseño"],
        "aprender": ["Innovacion", "Emprendimiento", "Analisis de datos"],
        "aportes": ["Generacion de ideas", "Diseñar presentaciones"],
        "proyecto": "Desarrollar una idea innovadora de negocio en un proyecto",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Jacobo Gil Pinzon",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["estrategias", "analisis de datos", "arte"],
        "habilidades": ["Creatividad", "Pensamiento logico"],
        "aprender": ["Marketing digital", "Data analytics"],
        "aportes": ["Generacion de ideas", "Organizacion del trabajo"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Crear valor a traves de la innovacion"
    },
    {
        "nombre": "Sara Camila Garcia Gonzalez",
        "carrera": "Lenguas Modernas",
        "interes_personal": ["idiomas"],
        "habilidades": ["Creatividad", "Trabajo en equipo"],
        "aprender": ["Analisis de datos", "Sostenibilidad"],
        "aportes": ["Organizacion", "Generacion de ideas"],
        "proyecto": "Generar un proyecto sostenible con tecnologia",
        "frase": "Crear valor a traves de la innovacion"
    },
    {
        "nombre": "Samuel Esteban Forero Alvarez",
        "carrera": "Lenguas Modernas",
        "interes_personal": ["inteligencia artificial"],
        "habilidades": ["Comunicacion", "Diseño"],
        "aprender": ["Inteligencia artificial", "Innovacion", "Emprendimiento"],
        "aportes": ["Investigar informacion", "Apoyar en todo lo necesario"],
        "proyecto": "Mejorar procesos con tecnologia",
        "frase": "Siempre en busqueda de nuevas ideas"
    },
    {
        "nombre": "Diego Alejandro Figueredo Vela",
        "carrera": "Ingenieria Ambiental",
        "interes_personal": ["apps", "software", "innovacion", "emprendimiento"],
        "habilidades": ["Trabajo en grupo", "Uso basico de herramientas digitales"],
        "aprender": ["Sostenibilidad y tecnologia"],
        "aportes": ["Analizar informacion", "Resolver problemas"],
        "proyecto": "Desarrollar un proyecto sostenible con tecnologia",
        "frase": "Crear valor a traves de la innovacion"
    },
    {
        "nombre": "Julian Arturo Estupiñan Gonzalez",
        "carrera": "Mercadeo",
        "interes_personal": ["inteligencia artificial"],
        "habilidades": ["Diseño", "Creatividad"],
        "aprender": ["Inteligencia artificial", "Automatizacion de procesos"],
        "aportes": ["Diseñar presentaciones", "Investigar informacion"],
        "proyecto": "Diseñar una app o herramienta digital",
        "frase": "La tecnologia es mi herramienta para cambiar el mundo"
    },
    {
        "nombre": "Alicia Diaz Quintero",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["diseño"],
        "habilidades": ["Organizacion y planificacion", "Creatividad"],
        "aprender": ["Marketing digital", "Innovacion", "Emprendimiento"],
        "aportes": ["Diseñar presentaciones", "Apoyar en todo lo necesario"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Valery Dayan Diaz Dorado",
        "carrera": "Lenguas Modernas",
        "interes_personal": ["tecnologia", "interculturalidad"],
        "habilidades": ["Liderazgo", "Creatividad", "Trabajo en equipo"],
        "aprender": ["Automatizacion de procesos", "IA", "Marketing digital"],
        "aportes": ["Generacion de ideas", "Coordinacion", "Liderar el equipo"],
        "proyecto": "Mejorar procesos con tecnologia",
        "frase": "Conectar conocimiento con accion"
    },
    {
        "nombre": "Diego Alejandro Cubides Humo",
        "carrera": "Negocios Internacionales",
        "interes_personal": [],
        "habilidades": ["Analisis de datos", "Resolucion de problemas"],
        "aprender": ["Inteligencia artificial", "Analisis de datos"],
        "aportes": ["Apoyar en todo lo necesario", "Motivar al equipo"],
        "proyecto": "",
        "frase": "Crear valor a traves de la innovacion"
    },
    {
        "nombre": "Angie Tatiana Cruz Avila",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["tecnologia"],
        "habilidades": ["Comunicacion", "Diseño"],
        "aprender": ["Inteligencia artificial", "Marketing digital"],
        "aportes": ["Diseñar presentaciones", "Generacion de ideas"],
        "proyecto": "Desarrollar una idea innovadora de negocios",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Diego Raul Cortes Sepulveda",
        "carrera": "Administracion de Empresas",
        "interes_personal": ["emprendimiento"],
        "habilidades": ["Ventas", "Trabajo en equipo"],
        "aprender": ["Inteligencia artificial", "Tecnologia"],
        "aportes": ["Liderar el equipo", "Resolucion de problemas"],
        "proyecto": "Crear una solucion tecnologica para empresas",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Laura Sofia Cordoba Cortes",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["idiomas", "tecnologia"],
        "habilidades": ["Creatividad", "Diseño"],
        "aprender": ["Automatizacion de procesos", "Inteligencia artificial"],
        "aportes": ["Generacion de ideas", "Apoyar en lo necesario", "Diseñar presentaciones"],
        "proyecto": "Desarrollar una idea innovadora de negocios",
        "frase": "Crear valor a traves de la innovacion"
    },
    {
        "nombre": "Catalina Saavedra Celis",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["diseño"],
        "habilidades": ["Organizacion y planificacion", "Creatividad"],
        "aprender": ["Marketing digital", "Innovacion", "Emprendimiento"],
        "aportes": ["Generacion de ideas", "Resolver problemas"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Aprender, innovar y crear"
    },
    {
        "nombre": "Lizeth Maryana Ayala Moreno",
        "carrera": "Mercadeo y Estrategia Comercial",
        "interes_personal": ["direccion de proyectos"],
        "habilidades": ["Organizacion y planificacion", "Liderazgo"],
        "aprender": ["Marketing digital", "Sostenibilidad y tecnologia"],
        "aportes": ["Organizacion del trabajo", "Resolver problemas"],
        "proyecto": "Crear un proyecto sostenible con tecnologia",
        "frase": "Conectar conocimiento con accion"
    },
    {
        "nombre": "Juan Sebastian Alvarez Mendez",
        "carrera": "Contaduria Publica",
        "interes_personal": ["finanzas"],
        "habilidades": ["Comunicacion", "Pensamiento logico"],
        "aprender": ["Analisis de datos", "Marketing digital", "Inteligencia artificial"],
        "aportes": ["Organizacion del trabajo", "Generacion de ideas"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Transformo ideas en soluciones"
    },

    # ==========================================================
    # ESTUDIANTES QUE APARECEN EN LOS ADJUNTOS CON INFORMACION INCOMPLETA
    # ==========================================================

    {
        "nombre": "Laura Sofia Valencia Lopez",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    }
]

# ==========================================================
# 2. FUNCIONES AUXILIARES
# ==========================================================

def texto_seguro(valor):
    if isinstance(valor, list):
        if len(valor) == 0:
            return "Sin informacion"
        return ", ".join(valor)
    if valor == "":
        return "Sin informacion"
    return valor

def imprimir_separador():
    print("\n" + "=" * 80)

def mostrar_estudiante(est):
    print(f"\n👤 Nombre: {est['nombre']}")
    print(f"🎓 Carrera: {texto_seguro(est['carrera'])}")
    print(f"🌟 Interes personal: {texto_seguro(est['interes_personal'])}")
    print(f"🧠 Habilidades: {texto_seguro(est['habilidades'])}")
    print(f"🚀 Quiere aprender: {texto_seguro(est['aprender'])}")
    print(f"🤝 Puede aportar: {texto_seguro(est['aportes'])}")
    print(f"📌 Proyecto de interes: {texto_seguro(est['proyecto'])}")
    print(f"💬 Frase: {texto_seguro(est['frase'])}")

def listar_nombres():
    print("\n📋 LISTA DE ESTUDIANTES")
    for i, est in enumerate(estudiantes, start=1):
        print(f"{i}. {est['nombre']}")

def buscar_por_nombre():
    texto = input("\nEscribe una parte del nombre: ").strip().lower()
    encontrados = [e for e in estudiantes if texto in e["nombre"].lower()]

    if encontrados:
        print(f"\n✅ Se encontraron {len(encontrados)} estudiante(s).")
        for est in encontrados:
            imprimir_separador()
            mostrar_estudiante(est)
    else:
        print("\n⚠️ No se encontraron coincidencias.")

def ver_todos_los_perfiles():
    print("\n📚 TODOS LOS PERFILES DEL GRUPO")
    for est in estudiantes:
        imprimir_separador()
        mostrar_estudiante(est)

def filtrar_por_carrera():
    texto = input("\nEscribe la carrera o una parte: ").strip().lower()
    encontrados = [e for e in estudiantes if texto in e["carrera"].lower()]

    if encontrados:
        print(f"\n✅ Se encontraron {len(encontrados)} estudiante(s).")
        for est in encontrados:
            print(f"- {est['nombre']} ({texto_seguro(est['carrera'])})")
    else:
        print("\n⚠️ No se encontraron estudiantes con esa carrera.")

def filtrar_por_habilidad():
    texto = input("\nEscribe la habilidad o una parte: ").strip().lower()
    encontrados = []

    for est in estudiantes:
        for h in est["habilidades"]:
            if texto in h.lower():
                encontrados.append(est)
                break

    if encontrados:
        print(f"\n✅ Se encontraron {len(encontrados)} estudiante(s).")
        for est in encontrados:
            print(f"- {est['nombre']} | {texto_seguro(est['habilidades'])}")
    else:
        print("\n⚠️ No se encontraron estudiantes con esa habilidad.")

def filtrar_por_tecnologia():
    texto = input("\nEscribe la tecnologia o tema: ").strip().lower()
    encontrados = []

    for est in estudiantes:
        for t in est["aprender"]:
            if texto in t.lower():
                encontrados.append(est)
                break

    if encontrados:
        print(f"\n✅ Se encontraron {len(encontrados)} estudiante(s).")
        for est in encontrados:
            print(f"- {est['nombre']} | {texto_seguro(est['aprender'])}")
    else:
        print("\n⚠️ No se encontraron estudiantes con ese interes.")

def filtrar_por_aporte():
    texto = input("\nEscribe el aporte o una parte: ").strip().lower()
    encontrados = []

    for est in estudiantes:
        for a in est["aportes"]:
            if texto in a.lower():
                encontrados.append(est)
                break

    if encontrados:
        print(f"\n✅ Se encontraron {len(encontrados)} estudiante(s).")
        for est in encontrados:
            print(f"- {est['nombre']} | {texto_seguro(est['aportes'])}")
    else:
        print("\n⚠️ No se encontraron estudiantes con ese aporte.")

def ver_proyectos_interes():
    print("\n📌 PROYECTOS DE INTERES")
    for est in estudiantes:
        print(f"- {est['nombre']}: {texto_seguro(est['proyecto'])}")

def resumen_general():
    print("\n📊 RESUMEN GENERAL DEL GRUPO")
    print(f"Total de estudiantes registrados: {len(estudiantes)}")

    carreras = Counter([e["carrera"] for e in estudiantes if e["carrera"] != ""])
    habilidades = Counter([h for e in estudiantes for h in e["habilidades"]])
    aprender = Counter([a for e in estudiantes for a in e["aprender"]])
    aportes = Counter([a for e in estudiantes for a in e["aportes"]])

    print("\n🎓 Carreras representadas:")
    for carrera, cantidad in carreras.most_common():
        print(f"- {carrera}: {cantidad}")

    print("\n🧠 Habilidades mas frecuentes:")
    for habilidad, cantidad in habilidades.most_common():
        print(f"- {habilidad}: {cantidad}")

    print("\n🚀 Temas que quieren aprender:")
    for tema, cantidad in aprender.most_common():
        print(f"- {tema}: {cantidad}")

    print("\n🤝 Aportes mas frecuentes:")
    for aporte, cantidad in aportes.most_common():
        print(f"- {aporte}: {cantidad}")

def encontrar_afinidades():
    print("\n🔎 AFINIDADES ENTRE ESTUDIANTES")
    encontrados = 0

    for i in range(len(estudiantes)):
        for j in range(i + 1, len(estudiantes)):
            e1 = estudiantes[i]
            e2 = estudiantes[j]

            comunes_h = set(e1["habilidades"]) & set(e2["habilidades"])
            comunes_a = set(e1["aprender"]) & set(e2["aprender"])
            comunes_p = set(e1["aportes"]) & set(e2["aportes"])

            puntaje = len(comunes_h) + len(comunes_a) + len(comunes_p)

            if puntaje >= 2:
                encontrados += 1
                print("-" * 80)
                print(f"{e1['nombre']}  <-->  {e2['nombre']}")
                if comunes_h:
                    print("Habilidades en comun:", ", ".join(sorted(comunes_h)))
                if comunes_a:
                    print("Aprendizajes en comun:", ", ".join(sorted(comunes_a)))
                if comunes_p:
                    print("Aportes en comun:", ", ".join(sorted(comunes_p)))
                print("Puntaje de afinidad:", puntaje)

    if encontrados == 0:
        print("No se encontraron afinidades fuertes.")

def sugerir_perfiles_complementarios():
    print("\n🧩 PERFILES COMPLEMENTARIOS")
    encontrados = 0

    for i in range(len(estudiantes)):
        for j in range(i + 1, len(estudiantes)):
            e1 = estudiantes[i]
            e2 = estudiantes[j]

            h1 = set(e1["habilidades"])
            h2 = set(e2["habilidades"])

            comunes = h1 & h2
            diferentes = (h1 | h2) - comunes

            if len(comunes) >= 1 and len(diferentes) >= 2:
                encontrados += 1
                print("-" * 80)
                print(f"{e1['nombre']} + {e2['nombre']}")
                print("Coinciden en:", ", ".join(sorted(comunes)))
                print("Se complementan con:", ", ".join(sorted(diferentes)))

    if encontrados == 0:
        print("No se encontraron perfiles complementarios.")

def analizar_estudiante_individual():
    listar_nombres()
    opcion = input("\nEscribe el numero del estudiante: ").strip()

    if not opcion.isdigit():
        print("\n⚠️ Debes escribir un numero valido.")
        return

    indice = int(opcion) - 1

    if indice < 0 or indice >= len(estudiantes):
        print("\n⚠️ Numero fuera de rango.")
        return

    est = estudiantes[indice]
    imprimir_separador()
    mostrar_estudiante(est)

# ==========================================================
# 3. GRAFICAS
# ==========================================================

def grafica_carreras():
    datos = Counter([e["carrera"] for e in estudiantes if e["carrera"] != ""])
    if len(datos) == 0:
        print("No hay datos para graficar.")
        return

    etiquetas = list(datos.keys())
    valores = list(datos.values())

    plt.figure(figsize=(12, 6))
    plt.bar(etiquetas, valores, color=[
        "royalblue", "orange", "green", "red", "purple", "cyan",
        "gold", "teal", "pink", "brown"
    ])
    plt.title("Cantidad de estudiantes por carrera")
    plt.xlabel("Carreras")
    plt.ylabel("Cantidad")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

def grafica_habilidades():
    datos = Counter([h for e in estudiantes for h in e["habilidades"]])
    if len(datos) == 0:
        print("No hay datos para graficar.")
        return

    etiquetas = list(datos.keys())
    valores = list(datos.values())

    plt.figure(figsize=(12, 6))
    plt.bar(etiquetas, valores, color=[
        "magenta", "dodgerblue", "limegreen", "darkorange", "crimson",
        "mediumpurple", "gray", "olive", "coral"
    ])
    plt.title("Habilidades mas frecuentes")
    plt.xlabel("Habilidades")
    plt.ylabel("Frecuencia")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

def grafica_aprendizajes():
    datos = Counter([a for e in estudiantes for a in e["aprender"]])
    if len(datos) == 0:
        print("No hay datos para graficar.")
        return

    etiquetas = list(datos.keys())
    valores = list(datos.values())

    plt.figure(figsize=(12, 6))
    plt.bar(etiquetas, valores, color=[
        "slateblue", "tomato", "gold", "mediumseagreen", "orchid",
        "deepskyblue", "sandybrown", "lightcoral"
    ])
    plt.title("Temas o tecnologias que quieren aprender")
    plt.xlabel("Temas")
    plt.ylabel("Frecuencia")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

def grafica_aportes():
    datos = Counter([a for e in estudiantes for a in e["aportes"]])
    if len(datos) == 0:
        print("No hay datos para graficar.")
        return

    etiquetas = list(datos.keys())
    valores = list(datos.values())

    plt.figure(figsize=(12, 6))
    plt.bar(etiquetas, valores, color=[
        "navy", "skyblue", "salmon", "khaki", "turquoise",
        "plum", "lightgreen", "peru"
    ])
    plt.title("Aportes mas frecuentes al equipo")
    plt.xlabel("Aportes")
    plt.ylabel("Frecuencia")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

def menu_graficas():
    while True:
        print("\n" + "=" * 80)
        print("📈 MENU DE GRAFICAS")
        print("=" * 80)
        print("1. Grafica de carreras")
        print("2. Grafica de habilidades")
        print("3. Grafica de aprendizajes")
        print("4. Grafica de aportes")
        print("5. Volver al menu principal")
        print("=" * 80)

        opcion = input("Selecciona una opcion (1-5): ").strip()

        if opcion == "1":
            grafica_carreras()
        elif opcion == "2":
            grafica_habilidades()
        elif opcion == "3":
            grafica_aprendizajes()
        elif opcion == "4":
            grafica_aportes()
        elif opcion == "5":
            print("Volviendo al menu principal...")
            break
        else:
            print("⚠️ Opcion no valida.")

# ==========================================================
# 4. MENU PRINCIPAL
# ==========================================================

def mostrar_menu():
    print("\n" + "=" * 80)
    print("MENU DE ANALISIS DEL GRUPO 10 - TREND TECH")
    print("=" * 80)
    print("1. Ver todos los perfiles")
    print("2. Buscar estudiante por nombre")
    print("3. Analizar un estudiante individual")
    print("4. Filtrar por carrera")
    print("5. Filtrar por habilidad")
    print("6. Filtrar por tecnologia o tema que quieren aprender")
    print("7. Filtrar por aporte al equipo")
    print("8. Ver proyectos de interes")
    print("9. Ver resumen general del grupo")
    print("10. Encontrar afinidades entre estudiantes")
    print("11. Sugerir perfiles complementarios")
    print("12. Menu de graficas")
    print("13. Listar solo estudiantes sin datos")
    print("14. Salir")
    print("=" * 80)

def listar_sin_datos():
    print("\n📝 ESTUDIANTES SIN DATOS")
    vacios = []

    for est in estudiantes:
        if (
            est["carrera"] == "" and
            len(est["interes_personal"]) == 0 and
            len(est["habilidades"]) == 0 and
            len(est["aprender"]) == 0 and
            len(est["aportes"]) == 0 and
            est["proyecto"] == "" and
            est["frase"] == ""
        ):
            vacios.append(est["nombre"])

    if len(vacios) == 0:
        print("Todos tienen informacion cargada.")
    else:
        for nombre in vacios:
            print("-", nombre)

def ejecutar_menu():
    while True:
        mostrar_menu()
        opcion = input("👉 Selecciona una opcion (1-14): ").strip()

        if opcion == "1":
            ver_todos_los_perfiles()
        elif opcion == "2":
            buscar_por_nombre()
        elif opcion == "3":
            analizar_estudiante_individual()
        elif opcion == "4":
            filtrar_por_carrera()
        elif opcion == "5":
            filtrar_por_habilidad()
        elif opcion == "6":
            filtrar_por_tecnologia()
        elif opcion == "7":
            filtrar_por_aporte()
        elif opcion == "8":
            ver_proyectos_interes()
        elif opcion == "9":
            resumen_general()
        elif opcion == "10":
            encontrar_afinidades()
        elif opcion == "11":
            sugerir_perfiles_complementarios()
        elif opcion == "12":
            menu_graficas()
        elif opcion == "13":
            listar_sin_datos()
        elif opcion == "14":
            print("\n👋 Fin del analisis. Exitos con Trend Tech.")
            break
        else:
            print("\n⚠️ Opcion no valida. Intenta de nuevo.")

# ==========================================================
# 5. INICIO
# ==========================================================

ejecutar_menu()
```
