```python

# ================================================================
# UNIVERSIDAD - MBA
# EXCELENCIA OPERACIONAL Y CADENA DE SUMINISTRO SOSTENIBLE
#
# LABORATORIO 1
# DIAGRAMA DE ISHIKAWA CON SEMÁFORO DE IMPACTO
#
# OBJETIVO:
# Identificar un problema profesional, organizar sus posibles
# causas y evaluar cuáles requieren mayor atención.
#
# IMPORTANTE:
# Las valoraciones son estimaciones del estudiante.
# Una causa con impacto alto NO es necesariamente una causa raíz
# comprobada. Debe verificarse con datos y evidencia.
#
# LIBRERÍAS:
# matplotlib: construcción de gráficos.
# pandas: organización y análisis de datos.
# textwrap: división automática de textos largos.
# math: cálculos para distribuir elementos en el diagrama.
# ================================================================


# ================================================================
# PASO 0. IMPORTACIÓN DE LIBRERÍAS
# ================================================================

import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import pandas as pd
import textwrap
import math


# ================================================================
# PASO 1. CONFIGURACIÓN VISUAL
# ================================================================

# Colores del semáforo.
COLOR_ROJO = "#E53935"
COLOR_NARANJA = "#FB8C00"
COLOR_VERDE = "#43A047"

# Colores generales del diagrama.
COLOR_ESPINA = "#34495E"
COLOR_TEXTO = "#263238"
COLOR_CABEZA = "#E3F2FD"
COLOR_FONDO = "#FFFFFF"

# Fuente general para las gráficas.
plt.rcParams["font.family"] = "DejaVu Sans"

# Resolución de las figuras.
plt.rcParams["figure.dpi"] = 120


# ================================================================
# PASO 2. FUNCIONES DE APOYO
# ================================================================

def solicitar_texto(pregunta):
    """
    Solicita una respuesta escrita.

    Si el estudiante no escribe nada, el programa vuelve
    a preguntar para evitar registros vacíos.
    """

    while True:
        respuesta = input(pregunta).strip()

        if respuesta:
            return respuesta

        print("Debe escribir una respuesta para continuar.")


def solicitar_entero(pregunta, minimo, maximo):
    """
    Solicita un número entero dentro de un rango.

    Ejemplo:
    solicitar_entero("Impacto: ", 1, 5)

    Solamente permite respuestas entre 1 y 5.
    """

    while True:

        respuesta = input(pregunta).strip()

        try:
            numero = int(respuesta)

            if minimo <= numero <= maximo:
                return numero

            print(
                f"Introduzca un número entre {minimo} y {maximo}."
            )

        except ValueError:
            print("Debe introducir un número entero válido.")


def dividir_texto(texto, ancho=24):
    """
    Divide textos largos en varias líneas.

    Esto evita que las causas o categorías
    se salgan de la figura.
    """

    return "\n".join(
        textwrap.wrap(
            str(texto),
            width=ancho,
            break_long_words=False
        )
    )


def clasificar_impacto(valor):
    """
    Convierte la calificación del impacto
    en un nivel de prioridad y un color.

    1-2: Verde.
    3: Naranja.
    4-5: Rojo.
    """

    if valor >= 4:
        return "ALTA", COLOR_ROJO

    elif valor == 3:
        return "MEDIA", COLOR_NARANJA

    else:
        return "BAJA", COLOR_VERDE


# ================================================================
# PASO 3. EJEMPLO PARA LOS ESTUDIANTES
# ================================================================

# Este ejemplo se muestra antes de solicitar las respuestas.
# No es obligatorio utilizarlo.
#
# Sirve para comprender cómo deben redactarse:
# - El problema.
# - Las categorías.
# - Las causas.
# - Las calificaciones de impacto.

ejemplo_problema = (
    "Retrasos recurrentes en la entrega de informes profesionales"
)

ejemplo_categorias = {

    "Personas": [
        ("Sobrecarga de trabajo", 5),
        ("Interrupciones frecuentes", 3)
    ],

    "Procesos": [
        ("Aprobaciones innecesarias", 5),
        ("Falta de procedimientos", 4)
    ],

    "Tecnología": [
        ("Herramientas desconectadas", 3),
        ("Equipos lentos", 2)
    ],

    "Información": [
        ("Datos incompletos", 5),
        ("Archivos desorganizados", 3)
    ],

    "Gestión": [
        ("Cambios de prioridades", 4),
        ("Seguimiento insuficiente", 2)
    ],

    "Entorno": [
        ("Solicitudes inesperadas", 4),
        ("Dependencia de terceros", 2)
    ]
}


print("\n" + "=" * 72)
print("LABORATORIO: DIAGRAMA DE ISHIKAWA")
print("GESTIÓN DE OPERACIONES Y EXCELENCIA OPERACIONAL")
print("=" * 72)

print("\nPRIMERO OBSERVE EL SIGUIENTE EJEMPLO:\n")

print("PROBLEMA:")
print(ejemplo_problema)

print("\nCATEGORÍAS Y CAUSAS:")

for categoria, lista_causas in ejemplo_categorias.items():

    print(f"\n{categoria.upper()}")

    for causa, impacto in lista_causas:

        prioridad, color = clasificar_impacto(impacto)

        print(
            f"  - {causa} | Impacto: {impacto}/5 "
            f"| Prioridad: {prioridad}"
        )


print("\n" + "-" * 72)

print("¿QUÉ SIGNIFICA CALIFICAR EL IMPACTO?")

print("""
1 = Impacto muy bajo
    La causa tiene una influencia mínima sobre el problema.

2 = Impacto bajo
    La causa podría contribuir ligeramente al problema.

3 = Impacto moderado
    La causa podría contribuir de manera importante.

4 = Impacto alto
    La causa podría generar consecuencias considerables.

5 = Impacto muy alto
    La causa podría ser determinante para la gravedad
    del problema y requiere verificación prioritaria.
""")

print("""
RECOMENDACIÓN:
No califique una causa como crítica solamente porque
le parece importante.

Pregúntese:
- ¿Cómo afecta esta causa al resultado?
- ¿Qué consecuencias produce?
- ¿Qué evidencia tengo de su impacto?
- ¿Qué ocurriría si esta causa no existiera?
""")

print("=" * 72)


# ================================================================
# PASO 4. SELECCIÓN DEL MODO DE TRABAJO
# ================================================================

print("\n¿CÓMO DESEA REALIZAR EL LABORATORIO?")

print("1. Utilizar el ejemplo completo.")
print("2. Crear mi propio problema y sus causas.")

modo = solicitar_entero(
    "\nSeleccione 1 o 2: ",
    1,
    2
)


# ================================================================
# PASO 5. CAPTURA DEL PROBLEMA
# ================================================================

# Se crea una lista donde se almacenarán todas las causas.
#
# Cada causa se registrará con:
# - Categoría.
# - Nombre.
# - Impacto.
# - Nivel de prioridad.
# - Color.

registros = []


if modo == 1:

    # Se cargan automáticamente los datos del ejemplo.

    problema = ejemplo_problema

    for categoria, lista_causas in ejemplo_categorias.items():

        for causa, impacto in lista_causas:

            prioridad, color = clasificar_impacto(impacto)

            registros.append({

                "Categoría": categoria,
                "Causa": causa,
                "Impacto": impacto,
                "Prioridad": prioridad,
                "Color": color

            })


else:

    # ------------------------------------------------------------
    # PASO 5.1. FORMULACIÓN DEL PROBLEMA
    # ------------------------------------------------------------

    print("\n" + "=" * 72)
    print("ETAPA 1. IDENTIFICACIÓN DEL PROBLEMA")
    print("=" * 72)

    print("""
Describa una situación recurrente que afecte
su productividad o desempeño profesional.

Ejemplos:

- Retrasos en la entrega de informes.
- Acumulación de tareas pendientes.
- Demoras en la atención de solicitudes.
- Reprocesos frecuentes en documentos.
- Incumplimiento de fechas de entrega.

Procure describir un problema específico y observable.
""")

    problema = solicitar_texto(
        "\n¿Cuál es el problema que desea analizar?: "
    )


    # ------------------------------------------------------------
    # PASO 5.2. SELECCIÓN DE CATEGORÍAS
    # ------------------------------------------------------------

    print("\n" + "=" * 72)
    print("ETAPA 2. CATEGORÍAS DEL DIAGRAMA")
    print("=" * 72)

    print("""
Las categorías agrupan posibles causas del problema.

Para organizaciones de servicios o actividades
profesionales se sugieren:

1. Personas
2. Procesos
3. Tecnología
4. Información
5. Gestión
6. Entorno

También puede utilizar otras categorías.

Ejemplos:
- Clientes.
- Proveedores.
- Recursos.
- Planificación.
- Comunicación.
- Métodos.
- Medición.
""")

    print("\nSeleccione una alternativa:")

    print("1. Utilizar las seis categorías sugeridas.")
    print("2. Crear mis propias categorías.")

    opcion_categorias = solicitar_entero(
        "\nSeleccione 1 o 2: ",
        1,
        2
    )

    if opcion_categorias == 1:

        categorias = [
            "Personas",
            "Procesos",
            "Tecnología",
            "Información",
            "Gestión",
            "Entorno"
        ]

    else:

        numero_categorias = solicitar_entero(
            "\n¿Cuántas categorías desea utilizar? (2-10): ",
            2,
            10
        )

        categorias = []

        for i in range(numero_categorias):

            while True:

                categoria = solicitar_texto(
                    f"Nombre de la categoría {i + 1}: "
                )

                # Se evitan nombres de categoría duplicados.
                existentes = [
                    nombre.lower()
                    for nombre in categorias
                ]

                if categoria.lower() not in existentes:

                    categorias.append(categoria)
                    break

                print("Esta categoría ya existe. Use otra.")


    # ------------------------------------------------------------
    # PASO 5.3. IDENTIFICACIÓN DE CAUSAS
    # ------------------------------------------------------------

    print("\n" + "=" * 72)
    print("ETAPA 3. IDENTIFICACIÓN DE CAUSAS")
    print("=" * 72)

    print("""
Para cada categoría deberá escribir sus posibles causas.

Puede utilizar las causas obtenidas mediante IA,
pero revise que sean específicas y coherentes
con el problema seleccionado.

Evite escribir soluciones.

Ejemplo de causa:
'Aprobaciones lentas'.

Ejemplo de solución (NO corresponde):
'Implementar un nuevo sistema de aprobaciones'.
""")

    for categoria in categorias:

        print("\n" + "-" * 65)

        print(f"CATEGORÍA: {categoria.upper()}")

        numero_causas = solicitar_entero(
            "¿Cuántas causas desea registrar? (1-8): ",
            1,
            8
        )

        for j in range(numero_causas):

            print(f"\nCAUSA {j + 1}")

            causa = solicitar_texto(
                "Describa la posible causa: "
            )

            print("\nVALORACIÓN DEL IMPACTO")

            print("""
1 = Muy bajo
2 = Bajo
3 = Moderado
4 = Alto
5 = Muy alto
""")

            impacto = solicitar_entero(
                "¿Qué impacto estima que tiene esta causa? (1-5): ",
                1,
                5
            )

            prioridad, color = clasificar_impacto(impacto)

            registros.append({

                "Categoría": categoria,
                "Causa": causa,
                "Impacto": impacto,
                "Prioridad": prioridad,
                "Color": color

            })

            print(
                f"Registrada: {causa} "
                f"| Prioridad: {prioridad}"
            )


# ================================================================
# PASO 6. CONSTRUCCIÓN DE LA BASE DE DATOS
# ================================================================

# pandas permite convertir los registros en una tabla.

df = pd.DataFrame(registros)

# Se ordenan las causas por su impacto.
df_ordenado = df.sort_values(
    by="Impacto",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 72)
print("RESUMEN DE CAUSAS REGISTRADAS")
print("=" * 72)

print(
    df_ordenado[
        ["Categoría", "Causa", "Impacto", "Prioridad"]
    ].to_string(index=False)
)


# ================================================================
# PASO 7. FUNCIÓN PARA CONSTRUIR EL ISHIKAWA
# ================================================================

def dibujar_ishikawa(problema, datos):
    """
    Construye un diagrama de Ishikawa dinámico.

    CARACTERÍSTICAS:
    - El problema aparece en la cabeza derecha.
    - La espina principal se dibuja horizontalmente.
    - Las categorías se distribuyen arriba y abajo.
    - Cada categoría tiene su propia espina.
    - Las causas aparecen conectadas a la espina.
    - Cada causa tiene un nodo pequeño de semáforo.
    - El diseño se adapta al número de categorías y causas.

    NOTA:
    El semáforo representa el impacto estimado,
    no una relación causal demostrada.
    """

    # ------------------------------------------------------------
    # 7.1. PREPARACIÓN DE LAS CATEGORÍAS
    # ------------------------------------------------------------

    categorias = list(
        dict.fromkeys(datos["Categoría"].tolist())
    )

    numero_categorias = len(categorias)

    # Se alternan las categorías entre arriba y abajo.
    #
    # Ejemplo con seis categorías:
    # Categorías 1, 3 y 5: arriba.
    # Categorías 2, 4 y 6: abajo.

    categorias_superiores = categorias[::2]
    categorias_inferiores = categorias[1::2]

    numero_columnas = max(
        len(categorias_superiores),
        len(categorias_inferiores),
        1
    )

    # Se determina el número máximo de causas por categoría.
    max_causas = int(
        datos.groupby("Categoría").size().max()
    )

    # El ancho se adapta a la cantidad de categorías.
    ancho_figura = max(
        16,
        numero_columnas * 4.4 + 5
    )

    # La altura se adapta a la cantidad de causas.
    altura_figura = max(
        10,
        max_causas * 1.5 + 6
    )

    fig, ax = plt.subplots(
        figsize=(ancho_figura, altura_figura)
    )

    fig.patch.set_facecolor(COLOR_FONDO)
    ax.set_facecolor(COLOR_FONDO)

    # ------------------------------------------------------------
    # 7.2. SISTEMA DE COORDENADAS
    # ------------------------------------------------------------

    # Las posiciones se calculan proporcionalmente
    # para evitar que las ramas se amontonen.

    x_inicio = 1.5
    x_final = numero_columnas * 4.5 + 2.5

    y_centro = 0

    altura_rama = max(
        4.0,
        max_causas * 0.75 + 2.0
    )

    # ------------------------------------------------------------
    # 7.3. ESPINA PRINCIPAL
    # ------------------------------------------------------------

    ax.plot(
        [x_inicio, x_final],
        [y_centro, y_centro],
        color=COLOR_ESPINA,
        linewidth=3.5,
        solid_capstyle="round",
        zorder=2
    )

    # Flecha que apunta al problema.
    ax.annotate(
        "",
        xy=(x_final + 0.35, y_centro),
        xytext=(x_final - 0.3, y_centro),
        arrowprops=dict(
            arrowstyle="->",
            linewidth=3,
            color=COLOR_ESPINA
        )
    )

    # ------------------------------------------------------------
    # 7.4. CABEZA DEL DIAGRAMA
    # ------------------------------------------------------------

    # La cabeza contiene el problema principal.

    ax.text(
        x_final + 2.0,
        y_centro,
        dividir_texto(problema, 25),
        fontsize=11,
        fontweight="bold",
        color=COLOR_TEXTO,
        ha="center",
        va="center",
        bbox=dict(
            boxstyle="round,pad=0.8",
            facecolor=COLOR_CABEZA,
            edgecolor="#1565C0",
            linewidth=2
        ),
        zorder=5
    )

    # ------------------------------------------------------------
    # 7.5. FUNCIÓN INTERNA PARA DIBUJAR UNA CATEGORÍA
    # ------------------------------------------------------------

    def dibujar_categoria(categoria, indice, lado):
        """
        lado = 1  -> categoría superior.
        lado = -1 -> categoría inferior.
        """

        # Se ubica cada rama a lo largo de la espina.
        x_base = 3.5 + indice * 4.5

        # La rama sale hacia arriba o hacia abajo.
        x_extremo = x_base - 1.8

        y_extremo = lado * altura_rama

        # Dibujo de la espina secundaria.
        ax.plot(
            [x_base, x_extremo],
            [0, y_extremo],
            color=COLOR_ESPINA,
            linewidth=2.2,
            zorder=2
        )

        # Nombre de la categoría.
        ax.text(
            x_extremo,
            y_extremo + lado * 0.45,
            dividir_texto(categoria.upper(), 18),
            ha="center",
            va="center",
            fontsize=10,
            fontweight="bold",
            color="#1565C0"
        )

        # Se obtienen las causas de esta categoría.
        causas_categoria = datos[
            datos["Categoría"] == categoria
        ].reset_index(drop=True)

        cantidad = len(causas_categoria)

        # --------------------------------------------------------
        # 7.6. DISTRIBUCIÓN DE LAS CAUSAS
        # --------------------------------------------------------

        # Cada causa ocupa una posición diferente
        # sobre la espina de su categoría.

        for j, fila in causas_categoria.iterrows():

            causa = fila["Causa"]
            impacto = int(fila["Impacto"])
            color = fila["Color"]

            # Fracción de la rama donde irá la causa.
            #
            # No se colocan causas en los extremos
            # para dejar espacio a los títulos.

            fraccion = (j + 1) / (cantidad + 1)

            # Coordenadas del punto sobre la espina.
            x_union = (
                x_base +
                fraccion * (x_extremo - x_base)
            )

            y_union = (
                fraccion * y_extremo
            )

            # ----------------------------------------------------
            # 7.7. RAMIFICACIÓN PEQUEÑA DE CADA CAUSA
            # ----------------------------------------------------

            # Todas las causas se conectan a su espina.
            # El texto se coloca a la derecha de la
            # ramificación para facilitar la lectura.

            largo_ramificacion = 0.85

            x_causa = x_union - largo_ramificacion

            # Se utiliza una pequeña línea horizontal.
            ax.plot(
                [x_causa, x_union],
                [y_union, y_union],
                color="#78909C",
                linewidth=1.3,
                zorder=2
            )

            # ----------------------------------------------------
            # 7.8. NODO PEQUEÑO DE SEMÁFORO
            # ----------------------------------------------------

            # Se dibuja un círculo muy pequeño junto
            # a cada causa.
            #
            # Rojo: impacto 4-5.
            # Naranja: impacto 3.
            # Verde: impacto 1-2.

            ax.scatter(
                [x_causa],
                [y_union],
                s=45,
                color=color,
                edgecolors="white",
                linewidths=0.8,
                zorder=6
            )

            # ----------------------------------------------------
            # 7.9. TEXTO DE LA CAUSA
            # ----------------------------------------------------

            # El texto se coloca inmediatamente
            # a la izquierda del nodo.

            ax.annotate(
                dividir_texto(causa, 20),
                xy=(x_causa, y_union),
                xytext=(-9, 0),
                textcoords="offset points",
                ha="right",
                va="center",
                fontsize=8.2,
                color=COLOR_TEXTO,
                bbox=dict(
                    facecolor="white",
                    edgecolor="none",
                    alpha=0.92,
                    pad=0.5
                ),
                zorder=5
            )

    # ------------------------------------------------------------
    # 7.10. DIBUJO DE LAS CATEGORÍAS SUPERIORES
    # ------------------------------------------------------------

    for i, categoria in enumerate(categorias_superiores):

        dibujar_categoria(
            categoria=categoria,
            indice=i,
            lado=1
        )

    # ------------------------------------------------------------
    # 7.11. DIBUJO DE LAS CATEGORÍAS INFERIORES
    # ------------------------------------------------------------

    for i, categoria in enumerate(categorias_inferiores):

        dibujar_categoria(
            categoria=categoria,
            indice=i,
            lado=-1
        )

    # ------------------------------------------------------------
    # 7.12. LEYENDA DEL SEMÁFORO
    # ------------------------------------------------------------

    leyenda = [

        Line2D(
            [0], [0],
            marker="o",
            color="w",
            label="Rojo: impacto alto (4-5)",
            markerfacecolor=COLOR_ROJO,
            markersize=7
        ),

        Line2D(
            [0], [0],
            marker="o",
            color="w",
            label="Naranja: impacto medio (3)",
            markerfacecolor=COLOR_NARANJA,
            markersize=7
        ),

        Line2D(
            [0], [0],
            marker="o",
            color="w",
            label="Verde: impacto bajo (1-2)",
            markerfacecolor=COLOR_VERDE,
            markersize=7
        )

    ]

    ax.legend(
        handles=leyenda,
        loc="upper center",
        bbox_to_anchor=(0.5, -0.03),
        ncol=3,
        frameon=False,
        fontsize=9
    )

    # ------------------------------------------------------------
    # 7.13. TÍTULO Y LÍMITES
    # ------------------------------------------------------------

    ax.set_title(
        "DIAGRAMA DE ISHIKAWA\n"
        "Análisis de causas e impacto estimado",
        fontsize=17,
        fontweight="bold",
        pad=25
    )

    ax.set_xlim(
        -2.8,
        x_final + 4.5
    )

    ax.set_ylim(
        -altura_rama - 2.5,
        altura_rama + 2.5
    )

    # Se eliminan los ejes para mantener
    # una presentación ejecutiva.
    ax.axis("off")

    plt.tight_layout()

    # Se guarda una copia de alta resolución.
    nombre_archivo = "ishikawa_semaforo.png"

    plt.savefig(
        nombre_archivo,
        dpi=300,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.show()

    print(
        f"\nDiagrama guardado como: {nombre_archivo}"
    )


# ================================================================
# PASO 8. GENERACIÓN DEL DIAGRAMA
# ================================================================

print("\nGenerando el diagrama de Ishikawa...")

dibujar_ishikawa(
    problema=problema,
    datos=df
)


# ================================================================
# PASO 9. ANÁLISIS DE CAUSAS CRÍTICAS
# ================================================================

# Se identifican las causas con impacto 4 o 5.

causas_altas = df_ordenado[
    df_ordenado["Impacto"] >= 4
]

causas_medias = df_ordenado[
    df_ordenado["Impacto"] == 3
]

causas_bajas = df_ordenado[
    df_ordenado["Impacto"] <= 2
]


print("\n" + "=" * 72)
print("RESULTADOS DEL DIAGNÓSTICO")
print("=" * 72)

print(f"\nProblema: {problema}")

print(f"\nTotal de causas: {len(df)}")

print(f"Causas de impacto alto: {len(causas_altas)}")
print(f"Causas de impacto medio: {len(causas_medias)}")
print(f"Causas de impacto bajo: {len(causas_bajas)}")


print("\nCAUSAS QUE DEBEN INVESTIGARSE PRIMERO:")

if len(causas_altas) > 0:

    for _, fila in causas_altas.iterrows():

        print(
            f"- {fila['Causa']} "
            f"({fila['Categoría']}) "
            f"| Impacto: {fila['Impacto']}/5"
        )

else:

    print(
        "No se identificaron causas con impacto alto."
    )


# ================================================================
# PASO 10. GRÁFICO DE PRIORIZACIÓN
# ================================================================

# Se genera un gráfico de barras para complementar
# la interpretación del Ishikawa.

# Se muestran las primeras 15 causas para
# conservar la legibilidad de la gráfica.

top_causas = df_ordenado.head(15).copy()

# Se invierte el orden para que la causa
# de mayor impacto aparezca arriba.

top_causas = top_causas.iloc[::-1]

altura_barras = max(
    5,
    len(top_causas) * 0.48 + 1.5
)

fig, ax = plt.subplots(
    figsize=(12, altura_barras)
)

ax.barh(
    top_causas["Causa"],
    top_causas["Impacto"],
    color=top_causas["Color"],
    edgecolor="white"
)

# Se agrega el valor numérico a cada barra.

for i, valor in enumerate(top_causas["Impacto"]):

    ax.text(
        valor + 0.07,
        i,
        f"{valor}/5",
        va="center",
        fontsize=9
    )

ax.set_xlim(0, 5.8)

ax.set_xlabel(
    "Impacto estimado sobre el problema (1-5)"
)

ax.set_title(
    "PRIORIZACIÓN DE CAUSAS\n"
    "Impacto estimado por el estudiante",
    fontsize=14,
    fontweight="bold"
)

ax.grid(
    axis="x",
    alpha=0.18
)

ax.set_axisbelow(True)

plt.tight_layout()

plt.savefig(
    "ranking_causas.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# PASO 11. RESUMEN POR CATEGORÍA
# ================================================================

# Se calcula cuántas causas de impacto alto
# existen en cada categoría.
#
# Este indicador permite identificar dónde
# se concentran las causas prioritarias.

resumen_categorias = (
    df.assign(
        Causa_Alta=(df["Impacto"] >= 4).astype(int)
    )
    .groupby("Categoría")
    .agg(
        Total_Causas=("Causa", "count"),
        Causas_Altas=("Causa_Alta", "sum"),
        Impacto_Promedio=("Impacto", "mean")
    )
    .sort_values(
        "Causas_Altas",
        ascending=False
    )
)

print("\n" + "=" * 72)
print("RESUMEN POR CATEGORÍA")
print("=" * 72)

print(
    resumen_categorias.round(2).to_string()
)


# ================================================================
# PASO 12. INTERPRETACIÓN GERENCIAL
# ================================================================

print("\n" + "=" * 72)
print("INTERPRETACIÓN GERENCIAL")
print("=" * 72)

print("""
El diagrama permite identificar las posibles causas
del problema y organizarlas por categorías.

Los nodos rojos indican causas con mayor impacto
estimado y, por lo tanto, requieren investigación
prioritaria.

Los nodos naranjas representan causas de impacto
moderado.

Los nodos verdes representan causas de menor
impacto estimado.

IMPORTANTE:
Estas clasificaciones no demuestran causalidad.
Las causas deben verificarse mediante indicadores,
observaciones, entrevistas o datos del proceso.
""")

print("\nPREGUNTAS PARA LA REFLEXIÓN:")

print("""
1. ¿Cuáles son las tres causas más relevantes?

2. ¿Qué evidencias permitirían confirmar su impacto?

3. ¿Qué causas podrían estar relacionadas entre sí?

4. ¿Qué consecuencias tendría no intervenirlas?

5. ¿Qué indicadores permitirían comprobar
   si una intervención mejora el desempeño?
""")


# ================================================================
# PASO 13. EXPORTACIÓN DE RESULTADOS
# ================================================================

# Se guarda la tabla para que los estudiantes
# puedan utilizarla posteriormente en:
#
# - Excel.
# - Power BI.
# - Otros ejercicios de analítica.
# - El proyecto integrador del curso.

df_ordenado.drop(
    columns=["Color"]
).to_csv(
    "causas_ishikawa.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nARCHIVOS GENERADOS:")
print("- ishikawa_semaforo.png")
print("- ranking_causas.png")
print("- causas_ishikawa.csv")

print("\nLABORATORIO FINALIZADO.")
```
