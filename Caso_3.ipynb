```python

# -*- coding: utf-8 -*-
"""
CASO 3
ADVANCED OPERATIONAL ANALYTICS
PROJECT PROCUREMENT AND SUPPLY CHAIN EXCELLENCE

PROBLEMA:
Atrasos en los procesos de abastecimiento de proyectos.

OBJETIVO:
Descubrir perfiles de compras, identificar anomalías,
analizar su evolución, generar alertas anticipatorias
y comparar decisiones gerenciales mediante Monte Carlo.

ENFOQUE:
1. Diagnóstico descriptivo.
2. K-Means: aprendizaje no supervisado.
3. Isolation Forest: detección de anomalías.
4. Análisis temporal.
5. Alertas anticipatorias.
6. Simulación prescriptiva Monte Carlo.
7. Interpretaciones y recomendaciones gerenciales.

IMPORTANTE:
Todos los datos son sintéticos.
Los efectos de las alternativas son supuestos didácticos.
No representan resultados reales de Ecopetrol.

El programa puede ejecutarse en Google Colab.
"""

# ============================================================
# 0. LIBRERÍAS Y CONFIGURACIÓN
# ============================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.ensemble import IsolationForest
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

warnings.filterwarnings("ignore", category=FutureWarning)

# ------------------------------------------------------------
# PARÁMETROS QUE PUEDEN MODIFICAR LOS ESTUDIANTES
# ------------------------------------------------------------

SEMILLA = 42
N_COMPRAS = 900
MESES = 12

# SLA ilustrativo del ciclo de abastecimiento.
# Debe reemplazarse por el compromiso real del proceso.
SLA_DIAS = 45

# Presupuesto hipotético para una intervención.
PRESUPUESTO_MAX_COP = 150_000_000

# Costo financiero ilustrativo de un día de atraso.
# No representa una pérdida real comprobada.
COSTO_DIA_RETRASO_COP = 850_000

N_SIMULACIONES = 5000

# Fracción aproximada de observaciones señaladas
# como anómalas por Isolation Forest.
CONTAMINACION = 0.06

K_MIN = 2
K_MAX = 6

DIRECTORIO = "Caso_3_Resultados"
os.makedirs(DIRECTORIO, exist_ok=True)

rng = np.random.default_rng(SEMILLA)

sns.set_theme(style="whitegrid")

plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "savefig.facecolor": "white",
    "font.size": 10,
    "axes.titlesize": 14,
    "axes.titleweight": "bold"
})


# ============================================================
# FUNCIONES AUXILIARES
# ============================================================

def guardar(nombre):
    """Guarda y presenta una gráfica individual."""
    plt.tight_layout()
    plt.savefig(
        os.path.join(DIRECTORIO, nombre + ".png"),
        dpi=190,
        bbox_inches="tight",
        facecolor="white"
    )
    plt.show()
    plt.close()


def interpretar(titulo, texto):
    """Imprime interpretaciones gerenciales extensas."""
    print("\n" + "=" * 85)
    print(titulo.upper())
    print("=" * 85)
    print(texto.strip())
    print()


# ============================================================
# 1. GENERACIÓN DE DATOS SINTÉTICOS
# ============================================================
#
# Cada fila representa un proceso de abastecimiento.
#
# Variables:
# - Complejidad de especificaciones técnicas.
# - Tiempo de definición técnica.
# - Número de ofertas recibidas.
# - Tiempo de evaluación y negociación.
# - Tiempo de aprobaciones.
# - Confiabilidad del proveedor.
# - Número de modificaciones del requerimiento.
# - Criticidad del material.
# - Tiempo total del ciclo de abastecimiento.
#
# Los perfiles internos se utilizan exclusivamente
# para generar diversidad sintética.
#
# NO se entregan como etiquetas a K-Means.
# El algoritmo deberá descubrir sus propios grupos.
#
# El tiempo total se reserva para evaluar externamente
# los clusters y NO se utiliza para formarlos.

perfil_generador = rng.choice(
    [0, 1, 2],
    size=N_COMPRAS,
    p=[0.40, 0.35, 0.25]
)

mes = rng.integers(
    1, MESES + 1,
    size=N_COMPRAS
)

complejidad = np.clip(
    np.rint(
        rng.normal(
            np.select(
                [
                    perfil_generador == 0,
                    perfil_generador == 1
                ],
                [1.8, 3.2],
                default=4.3
            ),
            0.65
        )
    ),
    1,
    5
).astype(int)

# Días necesarios para definir especificaciones.
dias_especificaciones = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [4, 9],
            default=14
        ),
        2.3
    ),
    1,
    28
)

# Número de ofertas disponibles.
numero_ofertas = np.clip(
    rng.poisson(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [4.5, 3.0],
            default=1.7
        )
    ) + 1,
    1,
    10
)

# Días de evaluación y negociación.
dias_negociacion = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [5, 10],
            default=16
        ),
        3
    ),
    1,
    32
)

# Días de aprobación.
dias_aprobacion = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [3, 8],
            default=13
        ),
        2.6
    ),
    1,
    27
)

# Confiabilidad del proveedor entre 0 y 100.
confiabilidad = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [94, 80],
            default=65
        ),
        7
    ),
    30,
    100
)

# Modificaciones de requerimientos.
modificaciones = np.clip(
    rng.poisson(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [0.3, 1.2],
            default=2.2
        )
    ),
    0,
    7
)

# Criticidad del material: 1 = baja; 5 = alta.
criticidad = np.clip(
    np.rint(
        rng.normal(
            np.select(
                [
                    perfil_generador == 0,
                    perfil_generador == 1
                ],
                [2, 3],
                default=4.4
            ),
            0.8
        )
    ),
    1,
    5
).astype(int)

# Tiempo de cumplimiento del proveedor.
# Mayor confiabilidad suele asociarse en este
# ejemplo con menores demoras.
#
# Esta relación es sintética, no causal estimada.

dias_proveedor = np.maximum(
    2,
    9
    + 0.30 * (100 - confiabilidad)
    + 1.6 * criticidad
    + rng.normal(0, 3, N_COMPRAS)
)

# Se simula el ciclo total.
# Los coeficientes son supuestos didácticos.

tiempo_ciclo = np.maximum(
    3,
    dias_especificaciones
    + dias_negociacion
    + dias_aprobacion
    + dias_proveedor
    + 2.0 * modificaciones
    + rng.normal(0, 3, N_COMPRAS)
)

base = pd.DataFrame({
    "compra_id": [
        f"PRC-{i:04d}"
        for i in range(1, N_COMPRAS + 1)
    ],
    "mes": mes,
    "complejidad_1_5": complejidad,
    "dias_especificaciones": dias_especificaciones,
    "numero_ofertas": numero_ofertas,
    "dias_negociacion": dias_negociacion,
    "dias_aprobacion": dias_aprobacion,
    "confiabilidad_proveedor_pct": confiabilidad,
    "modificaciones_requerimiento": modificaciones,
    "criticidad_material_1_5": criticidad,
    "dias_cumplimiento_proveedor": dias_proveedor,
    "tiempo_ciclo_dias": tiempo_ciclo
})

base["cumple_sla"] = (
    base["tiempo_ciclo_dias"] <= SLA_DIAS
)

base["dias_exceso_sla"] = np.maximum(
    0,
    base["tiempo_ciclo_dias"] - SLA_DIAS
)

# Variables que utilizará K-Means.
# El tiempo total NO se utiliza en el agrupamiento.

VARIABLES = [
    "complejidad_1_5",
    "dias_especificaciones",
    "numero_ofertas",
    "dias_negociacion",
    "dias_aprobacion",
    "confiabilidad_proveedor_pct",
    "modificaciones_requerimiento",
    "criticidad_material_1_5",
    "dias_cumplimiento_proveedor"
]

base.to_csv(
    os.path.join(
        DIRECTORIO,
        "01_Datos_Sinteticos.csv"
    ),
    index=False
)

interpretar(
    "01. Diagnóstico inicial",
    f"""
Se generaron {N_COMPRAS:,} procesos de abastecimiento.

El tiempo promedio del ciclo es:
{base.tiempo_ciclo_dias.mean():.1f} días.

El cumplimiento del SLA ilustrativo de
{SLA_DIAS} días es:
{100 * base.cumple_sla.mean():.1f}%.

La proporción de procesos que supera
el plazo es:
{100 * (1-base.cumple_sla.mean()):.1f}%.

INTERPRETACIÓN GERENCIAL

Un promedio general puede ocultar
comportamientos muy diferentes entre compras.

Algunas solicitudes podrían presentar
dificultades desde la especificación técnica,
otras durante la negociación y otras
en la etapa de aprobación o entrega.

Por tanto, no sería conveniente asumir
que todos los retrasos deben resolverse
mediante la misma intervención.

La primera recomendación consiste en
identificar los segmentos operacionales
y determinar en cuáles se concentran
los mayores tiempos y riesgos.

Antes de utilizar registros reales,
deben definirse claramente las fechas
de inicio y cierre, los tiempos en espera,
las responsabilidades y los criterios
para identificar materiales críticos.
"""
)


# ============================================================
# 2. ANÁLISIS DESCRIPTIVO
# ============================================================

fig, ax = plt.subplots(figsize=(10, 5))

sns.histplot(
    base["tiempo_ciclo_dias"],
    bins=35,
    color="#237DB7",
    ax=ax
)

ax.axvline(
    SLA_DIAS,
    color="#D14C48",
    linestyle="--",
    linewidth=2,
    label=f"Illustrative SLA: {SLA_DIAS} days"
)

ax.set(
    title="01 | Procurement Cycle Time Distribution",
    xlabel="Procurement cycle time (days)",
    ylabel="Purchase requests"
)

ax.legend()
guardar("01_Procurement_Cycle")

interpretar(
    "02. Interpretación del tiempo de abastecimiento",
    """
La distribución muestra la variabilidad
de los tiempos de abastecimiento.

El problema gerencial no se limita
a reducir el promedio.

También interesa disminuir la
incertidumbre del proceso.

Una cadena de suministro con alta
variabilidad dificulta la programación
de proyectos, la disponibilidad de
materiales y la coordinación con
contratistas y proveedores.

Por ello, deberían medirse
simultáneamente:

- Tiempo promedio.
- Percentil 90 del ciclo.
- Cumplimiento de plazos.
- Variabilidad de las entregas.
- Incidencia de cambios técnicos.
- Tiempo en espera de aprobaciones.

Si los retrasos se concentran en
determinadas etapas, las inversiones
deberían dirigirse a esos puntos
y no necesariamente a toda la
cadena de abastecimiento.
"""
)


# ============================================================
# 3. ESTANDARIZACIÓN
# ============================================================
#
# K-Means utiliza distancias.
# Por tanto, variables como porcentajes,
# días y número de ofertas deben
# estandarizarse antes del análisis.

escalador = StandardScaler()

X = escalador.fit_transform(
    base[VARIABLES]
)


# ============================================================
# 4. SELECCIÓN DEL NÚMERO DE CLUSTERS
# ============================================================

valores_k = list(
    range(K_MIN, K_MAX + 1)
)

silhouettes = []

for k in valores_k:

    modelo_prueba = KMeans(
        n_clusters=k,
        random_state=SEMILLA,
        n_init=15
    )

    etiquetas = modelo_prueba.fit_predict(X)

    puntuacion = silhouette_score(
        X,
        etiquetas,
        sample_size=min(700, len(base)),
        random_state=SEMILLA
    )

    silhouettes.append(puntuacion)

k_optimo = valores_k[
    int(np.argmax(silhouettes))
]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(
    valores_k,
    silhouettes,
    marker="o",
    linewidth=2,
    color="#237DB7"
)

ax.axvline(
    k_optimo,
    linestyle="--",
    color="#E49A32",
    label=f"Selected k = {k_optimo}"
)

ax.set(
    title="02 | Optimal Cluster Selection",
    xlabel="Number of clusters",
    ylabel="Silhouette score"
)

ax.legend()
guardar("02_Silhouette")

interpretar(
    "03. Selección de clusters",
    f"""
El procedimiento evaluó entre
{K_MIN} y {K_MAX} grupos.

La solución seleccionada fue:
{k_optimo} clusters.

El índice silhouette fue:
{max(silhouettes):.3f}.

Este indicador permite evaluar
qué tan separados se encuentran
los grupos en el espacio de variables.

Sin embargo, una separación
estadística favorable no garantiza
que los segmentos tengan significado
gerencial.

Por ello, cada cluster debe
interpretarse según sus
características operacionales.

La recomendación es conservar
segmentos que permitan formular
estrategias de abastecimiento
claramente diferenciadas.

También debe comprobarse la
estabilidad de la segmentación
con nuevas muestras y periodos.
"""
)


# ============================================================
# 5. K-MEANS: SEGMENTACIÓN OPERACIONAL
# ============================================================

modelo_kmeans = KMeans(
    n_clusters=k_optimo,
    random_state=SEMILLA,
    n_init=20
)

base["cluster"] = modelo_kmeans.fit_predict(X)

# PCA permite visualizar en dos dimensiones
# los datos que originalmente tienen nueve.

pca = PCA(
    n_components=2,
    random_state=SEMILLA
)

X_pca = pca.fit_transform(X)

fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=base["cluster"].astype(str),
    palette="tab10",
    alpha=0.7,
    s=35,
    ax=ax
)

ax.set(
    title="03 | Procurement Operational Clusters",
    xlabel="Principal component 1",
    ylabel="Principal component 2"
)

ax.legend(title="Cluster")
guardar("03_KMeans_Clusters")


# ============================================================
# 6. CARACTERIZACIÓN DE LOS CLUSTERS
# ============================================================

perfiles = base.groupby("cluster")[VARIABLES].mean()

media = base[VARIABLES].mean()
desviacion = base[VARIABLES].std().replace(0, 1)

perfiles_z = (
    perfiles - media
) / desviacion

fig, ax = plt.subplots(
    figsize=(12, max(4, k_optimo * 1.2))
)

sns.heatmap(
    perfiles_z,
    annot=True,
    fmt=".2f",
    cmap="RdYlBu_r",
    center=0,
    linewidths=0.5,
    ax=ax
)

ax.set(
    title="04 | Procurement Cluster Profiles",
    xlabel="Operational variables",
    ylabel="Cluster"
)

guardar("04_Cluster_Heatmap")

resumen = base.groupby("cluster").agg(
    compras=("compra_id", "size"),
    ciclo_promedio=("tiempo_ciclo_dias", "mean"),
    cumplimiento=("cumple_sla", "mean"),
    criticidad=("criticidad_material_1_5", "mean"),
    exceso_promedio=("dias_exceso_sla", "mean")
)

resumen["cumplimiento_pct"] = (
    100 * resumen["cumplimiento"]
)

print("\nRESUMEN DE CLUSTERS")
print(resumen.round(2).to_string())

peor_cluster = int(
    resumen["ciclo_promedio"].idxmax()
)

mejor_cluster = int(
    resumen["ciclo_promedio"].idxmin()
)

for cluster in perfiles.index:

    dominantes = (
        perfiles_z.loc[cluster]
        .sort_values(ascending=False)
        .head(3)
        .index.tolist()
    )

    interpretar(
        f"04. Interpretación del cluster {cluster}",
        f"""
El cluster {cluster} contiene
{int(resumen.loc[cluster, 'compras'])}
procesos de abastecimiento.

Su tiempo promedio es:
{resumen.loc[cluster, 'ciclo_promedio']:.1f} días.

Su cumplimiento del SLA es:
{resumen.loc[cluster, 'cumplimiento_pct']:.1f}%.

Las características relativamente
más elevadas son:

{', '.join(dominantes)}.

INTERPRETACIÓN ESTRATÉGICA

Estas características permiten
formular hipótesis sobre los
cuellos de botella que afectan
al segmento.

No deben interpretarse como
causas comprobadas.

Si predominan los tiempos de
especificación, convendría revisar
la madurez de los requerimientos
técnicos antes de iniciar compras.

Si predominan las aprobaciones,
sería necesario evaluar niveles
de autorización, transferencias
y tiempos de espera.

Si se observan condiciones
desfavorables de proveedores,
convendría investigar cumplimiento,
capacidad y alternativas de suministro.

RECOMENDACIÓN

Construir un mapa VSM del segmento,
identificar actividades que no
agregan valor y validar las causas
mediante expedientes reales,
entrevistas y análisis de procesos.

La gerencia debería priorizar
intervenciones diferenciadas
según el perfil descubierto.
"""
    )


# ============================================================
# 7. ISOLATION FOREST
# ============================================================
#
# Detecta compras con combinaciones
# operacionales poco frecuentes.
#
# No equivale a demostrar fraude,
# incumplimiento o mala gestión.

modelo_iso = IsolationForest(
    n_estimators=250,
    contamination=CONTAMINACION,
    random_state=SEMILLA
)

base["anomalia"] = (
    modelo_iso.fit_predict(X) == -1
)

base["anomaly_score"] = (
    -modelo_iso.decision_function(X)
)

fig, ax = plt.subplots(figsize=(10, 6))

normales = ~base["anomalia"]

ax.scatter(
    X_pca[normales, 0],
    X_pca[normales, 1],
    s=25,
    alpha=0.5,
    color="#277FB8",
    label="Typical"
)

ax.scatter(
    X_pca[~normales, 0],
    X_pca[~normales, 1],
    s=65,
    facecolors="none",
    edgecolors="#D64B48",
    linewidth=1.5,
    label="Anomaly"
)

ax.set(
    title="05 | Isolation Forest: Unusual Procurement Cases",
    xlabel="Principal component 1",
    ylabel="Principal component 2"
)

ax.legend()
guardar("05_Isolation_Forest")

n_anomalias = int(
    base["anomalia"].sum()
)

interpretar(
    "05. Análisis de anomalías",
    f"""
Isolation Forest señaló
{n_anomalias} procesos atípicos.

Esto equivale aproximadamente al
{100 * base.anomalia.mean():.1f}%
de las observaciones.

Este porcentaje depende del
parámetro de contaminación
establecido en el modelo.

No representa automáticamente
la proporción de compras
defectuosas o incumplidas.

INTERPRETACIÓN GERENCIAL

Una observación atípica podría
corresponder a una compra con
especificaciones especialmente
complejas, escasas ofertas,
aprobaciones inusuales o
condiciones extraordinarias
de suministro.

También podría ser un registro
incorrecto o una operación
válida poco frecuente.

RECOMENDACIÓN

Revisar los expedientes más
atípicos y determinar si
requieren controles adicionales.

Para materiales críticos,
la investigación debería
considerar el impacto de una
eventual indisponibilidad
sobre el cronograma del proyecto.

La detección de anomalías
puede apoyar una política
de abastecimiento basada
en riesgos, pero no debe
sustituir el juicio técnico.
"""
)

print("\n15 COMPRAS MÁS ATÍPICAS")

print(
    base.sort_values(
        "anomaly_score",
        ascending=False
    )
    .head(15)
    .to_string(index=False)
)


# ============================================================
# 8. DESEMPEÑO POR CLUSTER
# ============================================================

fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(
    data=base,
    x="cluster",
    y="tiempo_ciclo_dias",
    estimator="mean",
    errorbar=None,
    color="#368AC1",
    ax=ax
)

ax.axhline(
    SLA_DIAS,
    color="#D64B48",
    linestyle="--",
    linewidth=2,
    label="Illustrative SLA"
)

ax.set(
    title="06 | Procurement Cycle Time by Cluster",
    xlabel="Cluster",
    ylabel="Average cycle time (days)"
)

ax.legend()
guardar("06_Cycle_By_Cluster")

interpretar(
    "06. Desempeño y priorización",
    f"""
El cluster con mayor tiempo
promedio es el {peor_cluster}.

Su ciclo promedio es:
{resumen.loc[peor_cluster, 'ciclo_promedio']:.1f} días.

El segmento con menor tiempo
es el {mejor_cluster}.

Su ciclo promedio es:
{resumen.loc[mejor_cluster, 'ciclo_promedio']:.1f} días.

INTERPRETACIÓN GERENCIAL

La diferencia entre segmentos
sugiere que los procesos de
abastecimiento no presentan
condiciones operacionales
homogéneas.

Una política única de
contratación, seguimiento
o aprobación podría no
ser igualmente efectiva
para todas las compras.

Por ejemplo, una compra
recurrente y estandarizada
puede requerir controles
diferentes a una adquisición
técnicamente compleja y crítica.

RECOMENDACIÓN

Diseñar estrategias diferenciadas
según criticidad, complejidad,
disponibilidad de proveedores
y riesgo de interrupción.

El análisis debería vincularse
con la programación de proyectos
y con las decisiones de
abastecimiento estratégico.

También convendría comprobar
si las compras más lentas
afectan realmente la ruta
crítica de los proyectos.

Una demora administrativa no
siempre produce el mismo
impacto económico u operacional.
"""
)


# ============================================================
# 9. EVOLUCIÓN TEMPORAL
# ============================================================

participacion = (
    pd.crosstab(
        base["mes"],
        base["cluster"],
        normalize="index"
    ) * 100
)

fig, ax = plt.subplots(figsize=(10, 5))

for c in participacion.columns:

    ax.plot(
        participacion.index,
        participacion[c],
        marker="o",
        linewidth=2,
        label=f"Cluster {c}"
    )

ax.set(
    title="07 | Monthly Procurement Cluster Evolution",
    xlabel="Month",
    ylabel="Share of purchase requests (%)",
    xticks=range(1, MESES + 1)
)

ax.legend(ncol=3)
guardar("07_Cluster_Evolution")

# Frecuencia mensual de anomalías.
anomalias_mes = (
    base.groupby("mes")["anomalia"].mean() * 100
)

fig, ax = plt.subplots(figsize=(10, 4.5))

ax.plot(
    anomalias_mes.index,
    anomalias_mes.values,
    marker="o",
    color="#C66A35",
    linewidth=2
)

ax.set(
    title="08 | Monthly Procurement Anomaly Rate",
    xlabel="Month",
    ylabel="Flagged requests (%)",
    xticks=range(1, MESES + 1)
)

guardar("08_Anomaly_Evolution")

reciente = (
    participacion[peor_cluster]
    .tail(3)
    .mean()
)

anterior = (
    participacion[peor_cluster]
    .iloc[-6:-3]
    .mean()
)

interpretar(
    "07. Interpretación temporal",
    f"""
El segmento con mayor tiempo
de abastecimiento representa
{reciente:.1f}% de las compras
en los últimos tres meses.

En los tres meses anteriores
representaba {anterior:.1f}%.

La variación fue de
{reciente - anterior:+.1f}
puntos porcentuales.

INTERPRETACIÓN

Este resultado permite evaluar
si las condiciones operacionales
asociadas a ciclos largos
están ganando participación.

Sin embargo, una variación
de corto plazo no demuestra
una tendencia permanente.

La gerencia debería investigar
si los cambios coinciden con
etapas de ejecución de proyectos,
picos de contratación, restricciones
de proveedores o variaciones
en la capacidad de compras.

APLICACIÓN FUTURA

Construir un tablero mensual
que monitoree la participación
de los segmentos y las anomalías.

Las alertas deberían utilizar
umbrales calibrados con
información histórica suficiente.

También sería conveniente
distinguir entre compras
recurrentes, especiales
y materiales críticos.
"""
)


# ============================================================
# 10. ALERTA ANTICIPATORIA PARA UNA COMPRA NUEVA
# ============================================================
#
# La alerta utiliza similitud con
# perfiles históricos.
#
# NO predice una fecha exacta
# de entrega.
#
# IMPORTANTE:
# En una implementación real,
# solo deben incluirse variables
# disponibles al momento de emitir
# la alerta.
#
# Aquí se presenta un ejemplo de
# seguimiento durante el proceso,
# cuando ya existen datos de
# negociación y aprobación.

nueva_compra = pd.DataFrame([{
    "complejidad_1_5": 5,
    "dias_especificaciones": 16,
    "numero_ofertas": 2,
    "dias_negociacion": 17,
    "dias_aprobacion": 13,
    "confiabilidad_proveedor_pct": 64,
    "modificaciones_requerimiento": 3,
    "criticidad_material_1_5": 5,
    "dias_cumplimiento_proveedor": 27
}])

X_nueva = escalador.transform(
    nueva_compra[VARIABLES]
)

cluster_nuevo = int(
    modelo_kmeans.predict(X_nueva)[0]
)

anomalia_nueva = bool(
    modelo_iso.predict(X_nueva)[0] == -1
)

referencia = base.loc[
    base["cluster"] == cluster_nuevo
]

interpretar(
    "08. Alerta anticipatoria",
    f"""
La compra de ejemplo pertenece
al cluster {cluster_nuevo}.

El tiempo promedio histórico
sintético de ese grupo es:
{referencia.tiempo_ciclo_dias.mean():.1f} días.

El cumplimiento del SLA
en ese segmento es:
{100 * referencia.cumple_sla.mean():.1f}%.

La observación
{'SÍ' if anomalia_nueva else 'NO'}
fue marcada como atípica.

INTERPRETACIÓN GERENCIAL

El sistema permite reconocer
si una compra se parece a
procesos que históricamente
presentaron ciclos prolongados.

Esto no constituye una
predicción exacta de su
fecha de entrega.

Puede utilizarse como señal
para revisar anticipadamente
los riesgos del proceso.

RECOMENDACIÓN

Verificar la calidad de las
especificaciones técnicas,
la disponibilidad de proveedores,
los tiempos de aprobación
y la criticidad del material.

Si el suministro afecta la
ruta crítica de un proyecto,
debería considerarse un
seguimiento prioritario.

Antes de automatizar alertas,
debe comprobarse su utilidad
con compras posteriores.
"""
)


# ============================================================
# 11. SIMULACIÓN MONTE CARLO
# ============================================================
#
# Se comparan alternativas gerenciales.
#
# Los porcentajes son SUPUESTOS
# de reducción de tiempos.
#
# No representan efectos reales
# de las intervenciones.
#
# Los escenarios utilizan las
# mismas observaciones y el mismo
# ruido para facilitar su comparación.

ESCENARIOS = {

    "Current process": {
        "costo": 0,
        "especificaciones": 0,
        "negociacion": 0,
        "aprobacion": 0,
        "proveedor": 0,
        "modificaciones": 0
    },

    "Integrated procurement planning": {
        "costo": 85_000_000,
        "especificaciones": 0.30,
        "negociacion": 0.10,
        "aprobacion": 0.10,
        "proveedor": 0.10,
        "modificaciones": 0.25
    },

    "Framework agreements": {
        "costo": 110_000_000,
        "especificaciones": 0.05,
        "negociacion": 0.45,
        "aprobacion": 0.10,
        "proveedor": 0.15,
        "modificaciones": 0.05
    },

    "Approval automation": {
        "costo": 70_000_000,
        "especificaciones": 0.05,
        "negociacion": 0.05,
        "aprobacion": 0.55,
        "proveedor": 0,
        "modificaciones": 0.05
    },

    "Supplier development": {
        "costo": 95_000_000,
        "especificaciones": 0,
        "negociacion": 0.05,
        "aprobacion": 0,
        "proveedor": 0.35,
        "modificaciones": 0.10
    },

    "Lean procurement redesign": {
        "costo": 120_000_000,
        "especificaciones": 0.35,
        "negociacion": 0.20,
        "aprobacion": 0.30,
        "proveedor": 0.05,
        "modificaciones": 0.30
    },

    "Risk-based procurement": {
        "costo": 90_000_000,
        "especificaciones": 0.15,
        "negociacion": 0.15,
        "aprobacion": 0.15,
        "proveedor": 0.25,
        "modificaciones": 0.10
    },

    "Strategic inventory": {
        "costo": 180_000_000,
        "especificaciones": 0,
        "negociacion": 0,
        "aprobacion": 0,
        "proveedor": 0.50,
        "modificaciones": 0
    }
}

# Se seleccionan compras con reemplazo.
indices = rng.integers(
    0,
    len(base),
    size=N_SIMULACIONES
)

muestra = base.iloc[indices]

# Mismo ruido para todos los escenarios.
ruido_mc = rng.normal(
    0,
    3,
    size=N_SIMULACIONES
)

def simular_escenario(parametros):
    """
    Calcula el ciclo hipotético de abastecimiento.

    Cada intervención modifica determinados
    componentes del proceso.

    Los efectos no han sido estimados
    a partir de un experimento real.
    """

    p = parametros

    dias = (
        muestra["dias_especificaciones"].to_numpy()
        * (1 - p["especificaciones"])

        + muestra["dias_negociacion"].to_numpy()
        * (1 - p["negociacion"])

        + muestra["dias_aprobacion"].to_numpy()
        * (1 - p["aprobacion"])

        + muestra["dias_cumplimiento_proveedor"].to_numpy()
        * (1 - p["proveedor"])

        + 2.0
        * muestra["modificaciones_requerimiento"].to_numpy()
        * (1 - p["modificaciones"])

        + ruido_mc
    )

    return np.maximum(3, dias)


resultados = []

for nombre, parametros in ESCENARIOS.items():

    dias = simular_escenario(parametros)

    resultados.append({
        "alternativa": nombre,
        "costo_cop": parametros["costo"],
        "ciclo_promedio_dias": dias.mean(),
        "percentil_90_dias": np.percentile(dias, 90),
        "cumplimiento_pct": 100 * np.mean(
            dias <= SLA_DIAS
        ),
        "exceso_medio_sla_dias": np.maximum(
            0,
            dias - SLA_DIAS
        ).mean()
    })

escenarios = pd.DataFrame(resultados)

base_simulada = escenarios.loc[
    escenarios["alternativa"] == "Current process"
].iloc[0]

escenarios["reduccion_dias"] = (
    base_simulada["ciclo_promedio_dias"]
    - escenarios["ciclo_promedio_dias"]
)

# El siguiente indicador es un valor potencial ilustrativo.
# NO representa ahorro financiero real ni ROI.
#
# Se aplica el costo por día solamente al exceso
# sobre el SLA, no a todos los días del ciclo.

escenarios["costo_exceso_ilustrativo_cop"] = (
    escenarios["exceso_medio_sla_dias"]
    * N_COMPRAS
    * COSTO_DIA_RETRASO_COP
)

costo_base_exceso = (
    base_simulada["exceso_medio_sla_dias"]
    * N_COMPRAS
    * COSTO_DIA_RETRASO_COP
)

escenarios["beneficio_potencial_cop"] = (
    costo_base_exceso
    - escenarios["costo_exceso_ilustrativo_cop"]
)

escenarios["beneficio_neto_hipotetico_cop"] = (
    escenarios["beneficio_potencial_cop"]
    - escenarios["costo_cop"]
)

escenarios["factible"] = (
    escenarios["costo_cop"]
    <= PRESUPUESTO_MAX_COP
)

escenarios.to_csv(
    os.path.join(
        DIRECTORIO,
        "02_Escenarios_Monte_Carlo.csv"
    ),
    index=False
)


# ============================================================
# 12. GRÁFICA DE TIEMPOS POR ALTERNATIVA
# ============================================================

fig, ax = plt.subplots(figsize=(11, 6))

colores = [
    "#287FB8" if x else "#D5A04A"
    for x in escenarios["factible"]
]

ax.barh(
    escenarios["alternativa"],
    escenarios["ciclo_promedio_dias"],
    color=colores
)

ax.axvline(
    SLA_DIAS,
    color="#D64B48",
    linestyle="--",
    label="Illustrative SLA"
)

ax.invert_yaxis()

ax.set(
    title="09 | Procurement Cycle by Intervention",
    xlabel="Simulated average cycle time (days)",
    ylabel=""
)

ax.legend()
guardar("09_Scenario_Comparison")


# ============================================================
# 13. TRADE-OFF COSTO VS CUMPLIMIENTO
# ============================================================

fig, ax = plt.subplots(figsize=(11, 6))

for _, fila in escenarios.iterrows():

    ax.scatter(
        fila["costo_cop"] / 1e6,
        fila["cumplimiento_pct"],
        s=140,
        color=(
            "#287FB8"
            if fila["factible"]
            else "#D99A39"
        )
    )

    ax.annotate(
        fila["alternativa"],
        (
            fila["costo_cop"] / 1e6,
            fila["cumplimiento_pct"]
        ),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=8
    )

ax.axvline(
    PRESUPUESTO_MAX_COP / 1e6,
    linestyle="--",
    color="#D64B48",
    label="Budget limit"
)

ax.set(
    title="10 | Procurement Cost–Service Trade-off",
    xlabel="Implementation cost (million COP)",
    ylabel="Simulated SLA compliance (%)"
)

ax.legend()
guardar("10_Cost_Service_Tradeoff")


# ============================================================
# 14. DECISIÓN GERENCIAL
# ============================================================
#
# Se selecciona la alternativa factible
# con mayor cumplimiento simulado del SLA.
#
# Como segundo criterio se utiliza
# el menor tiempo promedio.
#
# No constituye una optimización integral
# ni reemplaza el AHP del equipo.

factibles = (
    escenarios.loc[escenarios["factible"]]
    .sort_values(
        [
            "cumplimiento_pct",
            "ciclo_promedio_dias"
        ],
        ascending=[False, True]
    )
)

recomendada = factibles.iloc[0]

print("\nRESULTADOS DE MONTE CARLO")

print(
    escenarios.round(2).to_string(index=False)
)

interpretar(
    "09. Interpretación prescriptiva",
    f"""
Con un presupuesto máximo
ilustrativo de COP
{PRESUPUESTO_MAX_COP:,.0f},
la alternativa factible
mejor posicionada es:

{recomendada.alternativa}

El ciclo promedio simulado
sería de:
{recomendada.ciclo_promedio_dias:.1f} días.

La reducción frente al
proceso actual sería de:
{recomendada.reduccion_dias:.1f} días.

El cumplimiento del SLA
sería de:
{recomendada.cumplimiento_pct:.1f}%.

INTERPRETACIÓN GERENCIAL

La simulación permite comparar
el desempeño hipotético de
diferentes intervenciones.

Sin embargo, la alternativa
seleccionada depende de los
efectos supuestos, el presupuesto
y el criterio de decisión.

No demuestra que esa inversión
produzca el resultado calculado.

RECOMENDACIÓN

Contrastar la alternativa con
la matriz AHP del proyecto.

Revisar su viabilidad técnica,
contractual, financiera y
organizacional.

Diseñar un piloto con procesos
comparables y medir el tiempo
de abastecimiento antes
y después de la intervención.

La decisión final debería
considerar también continuidad
operacional, criticidad de
materiales y riesgo de
incumplimiento contractual.
"""
)


# ============================================================
# 15. ANÁLISIS DE SENSIBILIDAD
# ============================================================

parametros_rec = ESCENARIOS[
    recomendada["alternativa"]
]

dias_rec = simular_escenario(
    parametros_rec
)

print("\nSENSIBILIDAD DEL SLA")

for sla in [30, 45, 60, 75]:

    cumplimiento = 100 * np.mean(
        dias_rec <= sla
    )

    print(
        f"SLA {sla:>2} días: "
        f"{cumplimiento:.1f}% de cumplimiento"
    )

interpretar(
    "10. Sensibilidad y robustez",
    """
El cumplimiento del servicio
depende del plazo establecido.

Un SLA más amplio puede
mejorar matemáticamente
el porcentaje de cumplimiento,
pero no necesariamente
mejorar la operación.

Por ello, no se recomienda
modificar los compromisos
simplemente para presentar
indicadores más favorables.

La organización debería
evaluar escenarios adicionales:

1. Mayor demanda de materiales.
2. Menor disponibilidad de proveedores.
3. Cambios en especificaciones técnicas.
4. Incremento de aprobaciones.
5. Retrasos de transporte.
6. Restricciones presupuestales.

También sería conveniente
realizar una sensibilidad
sobre los efectos supuestos
de cada intervención.

Si la alternativa ganadora
cambia al modificar ligeramente
los supuestos, la decisión
requiere más evidencia.

Para materiales críticos,
el riesgo de interrupción
debe tener prioridad sobre
una reducción marginal de costos.
"""
)


# ============================================================
# 16. CONCLUSIONES EJECUTIVAS
# ============================================================

interpretar(
    "11. Conclusiones estratégicas",
    f"""
RESULTADOS DEL DIAGNÓSTICO

Se identificaron {k_optimo}
perfiles de abastecimiento.

Se señalaron {n_anomalias}
procesos operacionales atípicos.

El segmento con mayor
tiempo promedio fue el
cluster {peor_cluster}.

Estos resultados permiten
formular hipótesis sobre
las diferencias operacionales
entre procesos de compra.

DECISIONES ESTRATÉGICAS

Se recomienda evaluar una
estrategia de abastecimiento
diferenciada por criticidad,
complejidad y disponibilidad
de proveedores.

También convendría integrar
los planes de abastecimiento
con los cronogramas de proyectos.

El propósito sería anticipar
necesidades y evitar que
los materiales críticos
comprometan la continuidad
de las operaciones.

DECISIONES TÁCTICAS

Se propone revisar los
procedimientos de especificación,
negociación y aprobación.

Podrían desarrollarse pilotos
de estandarización técnica,
automatización administrativa
y seguimiento a proveedores.

Las intervenciones deben
priorizarse según el impacto
real sobre los proyectos.

DECISIONES OPERACIONALES

Implementar controles de
completitud de requerimientos.

Monitorear solicitudes con
tiempos de espera elevados.

Revisar los procesos atípicos
y documentar las causas
confirmadas.

Establecer alertas de
seguimiento para compras
críticas o con condiciones
operacionales desfavorables.

VALIDACIÓN

Los resultados del laboratorio
son sintéticos.

Las intervenciones simuladas
no tienen efectos causales
comprobados.

Para aplicar el enfoque
a datos reales deben
validarse las variables,
la calidad de los registros
y los supuestos de costos.

La alternativa recomendada
debería contrastarse con
el AHP, el análisis de
riesgos y la evaluación
económica de la organización.

CONCLUSIÓN GENERAL

La analítica no supervisada
permite pasar de un diagnóstico
general de atrasos a una
comprensión segmentada
de las condiciones de
abastecimiento.

Su principal contribución
es orientar la investigación,
la priorización de riesgos
y el diseño de decisiones
gerenciales más informadas.
"""
)


# ============================================================
# 17. EXPORTACIÓN DE RESULTADOS
# ============================================================

base.to_csv(
    os.path.join(
        DIRECTORIO,
        "03_Compras_Clusters_Anomalias.csv"
    ),
    index=False
)

resumen.to_csv(
    os.path.join(
        DIRECTORIO,
        "04_Resumen_Clusters.csv"
    )
)

print("\n" + "=" * 70)
print("LABORATORIO CASO 3 FINALIZADO")
print("=" * 70)

print("Resultados guardados en:")
print(os.path.abspath(DIRECTORIO))

print("\nArchivos generados:")
print("- 10 gráficas PNG individuales")
print("- Base sintética CSV")
print("- Resultados de Monte Carlo CSV")
print("- Base enriquecida con clusters y anomalías")
print("- Resumen de desempeño por cluster")
```
