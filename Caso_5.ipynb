```python

# -*- coding: utf-8 -*-

"""
===========================================================
CASO 5
ADVANCED OPERATIONAL ANALYTICS
IT SUPPORT TICKET QUALITY AND OPERATIONAL EXCELLENCE
===========================================================

PROBLEMA:
Recepción recurrente de tickets de soporte tecnológico
con información incompleta o ambigua.

OBJETIVO:
Descubrir perfiles de tickets, identificar anomalías,
construir alertas anticipatorias y evaluar alternativas
para reducir reprocesos y tiempos de resolución.

METODOLOGÍA:
1. Datos sintéticos.
2. Diagnóstico descriptivo.
3. K-Means.
4. Caracterización de clusters.
5. Isolation Forest.
6. Evaluación externa del MTTR.
7. Análisis temporal.
8. Alertas anticipatorias.
9. Monte Carlo.
10. Análisis costo-servicio.
11. Sensibilidad y recomendaciones.

IMPORTANTE:
Los datos y los efectos de las intervenciones
son hipotéticos y no representan resultados
reales de ninguna organización.

K-Means e Isolation Forest no son modelos
supervisados de predicción del MTTR.
"""

# ============================================================
# 0. LIBRERÍAS Y PARÁMETROS MODIFICABLES
# ============================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.ensemble import IsolationForest
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

warnings.filterwarnings("ignore", category=FutureWarning)

SEMILLA = 42
N_TICKETS = 1000
MESES = 12

# SLA ilustrativo.
SLA_HORAS = 24.0

# Presupuesto y costos hipotéticos.
PRESUPUESTO_MAX_COP = 80_000_000

# Valor ilustrativo de una hora de esfuerzo del soporte.
COSTO_HORA_SOPORTE_COP = 45000

N_SIMULACIONES = 5000
CONTAMINACION = 0.06

K_MIN = 2
K_MAX = 6

DIRECTORIO = "Caso_5_Resultados"
os.makedirs(DIRECTORIO, exist_ok=True)

rng = np.random.default_rng(SEMILLA)

sns.set_theme(style="whitegrid")

plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "savefig.facecolor": "white",
    "font.size": 10,
    "axes.titlesize": 14,
    "axes.titleweight": "bold"
})


# ============================================================
# FUNCIONES AUXILIARES
# ============================================================

def guardar(nombre):
    """
    Guarda y presenta una gráfica con fondo blanco.
    """
    plt.tight_layout()
    plt.savefig(
        os.path.join(DIRECTORIO, nombre + ".png"),
        dpi=190,
        bbox_inches="tight",
        facecolor="white"
    )
    plt.show()
    plt.close()


def interpretar(titulo, texto):
    """
    Imprime interpretaciones gerenciales.
    """
    print("\n" + "=" * 90)
    print(titulo.upper())
    print("=" * 90)
    print(texto.strip())
    print()


# ============================================================
# 1. GENERACIÓN DE DATOS SINTÉTICOS
# ============================================================

"""
Cada fila representa un ticket de soporte.

VARIABLES DISPONIBLES AL CREAR EL TICKET:
- Canal de recepción.
- Complejidad estimada.
- Porcentaje de campos completos.
- Ambigüedad de la descripción.
- Evidencia adjunta.
- Calidad de la categorización inicial.
- Experiencia del usuario con el portal.

VARIABLES OBSERVADAS DESPUÉS:
- Contactos adicionales.
- Reasignaciones.
- Tiempo de resolución.
- Cumplimiento del SLA.

Esta separación es fundamental:
los resultados posteriores NO se utilizarán
para construir los clusters ni las alertas
al momento de recibir un ticket.

Los perfiles generadores son artificiales.
No se entregan como etiquetas a K-Means.
"""

perfil_generador = rng.choice(
    [0, 1, 2],
    size=N_TICKETS,
    p=[0.43, 0.35, 0.22]
)

mes = rng.integers(
    1,
    MESES + 1,
    size=N_TICKETS
)

canales = np.array([
    "Portal",
    "Email",
    "Chat",
    "Phone"
])

canal = np.empty(N_TICKETS, dtype=object)

probabilidades_canal = {
    0: [0.65, 0.15, 0.15, 0.05],
    1: [0.30, 0.35, 0.20, 0.15],
    2: [0.15, 0.45, 0.15, 0.25]
}

for p in [0, 1, 2]:
    idx = perfil_generador == p
    canal[idx] = rng.choice(
        canales,
        size=idx.sum(),
        p=probabilidades_canal[p]
    )

# Complejidad del incidente, de 1 a 5.
complejidad = np.clip(
    np.rint(
        rng.normal(
            np.select(
                [
                    perfil_generador == 0,
                    perfil_generador == 1
                ],
                [1.8, 3.0],
                default=4.2
            ),
            0.7
        )
    ),
    1,
    5
).astype(int)

# Completitud documental al ingresar.
completitud = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [94, 72],
            default=45
        ),
        10
    ),
    10,
    100
)

# Ambigüedad de la descripción:
# 0 = muy clara; 100 = muy ambigua.
ambiguedad = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [15, 42],
            default=72
        ),
        12
    ),
    0,
    100
)

# Evidencia adjunta al ticket:
# 1 = sí; 0 = no.
evidencia = rng.binomial(
    1,
    np.select(
        [
            perfil_generador == 0,
            perfil_generador == 1
        ],
        [0.88, 0.55],
        default=0.25
    )
)

# Calidad de la categorización inicial:
# 0 a 100; valores altos indican mejor clasificación.
calidad_categoria = np.clip(
    rng.normal(
        np.select(
            [
                perfil_generador == 0,
                perfil_generador == 1
            ],
            [92, 70],
            default=43
        ),
        12
    ),
    0,
    100
)

# Familiaridad del usuario con el portal.
# Escala de 1 a 5.
experiencia_portal = np.clip(
    np.rint(
        rng.normal(
            np.select(
                [
                    perfil_generador == 0,
                    perfil_generador == 1
                ],
                [4.2, 3.0],
                default=1.8
            ),
            0.8
        )
    ),
    1,
    5
).astype(int)

# ------------------------------------------------------------
# VARIABLES OBSERVADAS DURANTE LA RESOLUCIÓN
# ------------------------------------------------------------

# Contactos adicionales solicitando aclaraciones.
lambda_contactos = (
    0.15
    + 0.018 * (100 - completitud)
    + 0.009 * ambiguedad
    + 0.18 * (1 - evidencia)
)

contactos_adicionales = np.clip(
    rng.poisson(lambda_contactos),
    0,
    9
)

# Reasignaciones entre equipos.
lambda_reasignaciones = (
    0.10
    + 0.012 * (100 - calidad_categoria)
    + 0.12 * (complejidad - 1)
)

reasignaciones = np.clip(
    rng.poisson(lambda_reasignaciones),
    0,
    7
)

# Tiempo de resolución sintético.
# Los coeficientes son supuestos didácticos,
# no efectos causales estimados.

ruido = rng.normal(
    0,
    2.5,
    N_TICKETS
)

tiempo_resolucion = np.maximum(
    1,

    2.0
    + 2.5 * complejidad
    + 0.08 * (100 - completitud)
    + 0.055 * ambiguedad
    + 2.2 * contactos_adicionales
    + 2.6 * reasignaciones
    + 1.5 * (1 - evidencia)
    + ruido
)

# ------------------------------------------------------------
# DATAFRAME
# ------------------------------------------------------------

base = pd.DataFrame({
    "ticket_id": [
        f"TCK-{i:05d}"
        for i in range(1, N_TICKETS + 1)
    ],
    "mes": mes,
    "canal": canal,
    "complejidad_1_5": complejidad,
    "completitud_pct": completitud,
    "ambiguedad_pct": ambiguedad,
    "evidencia_adjunta": evidencia,
    "calidad_categoria_pct": calidad_categoria,
    "experiencia_portal_1_5": experiencia_portal,
    "contactos_adicionales": contactos_adicionales,
    "reasignaciones": reasignaciones,
    "tiempo_resolucion_h": tiempo_resolucion
})

base["cumple_sla"] = (
    base["tiempo_resolucion_h"] <= SLA_HORAS
)

base["exceso_sla_h"] = np.maximum(
    0,
    base["tiempo_resolucion_h"] - SLA_HORAS
)

# Variables utilizadas en K-Means.
# Se excluyen contactos adicionales, reasignaciones,
# MTTR y cumplimiento del SLA porque ocurren después.

VARIABLES = [
    "complejidad_1_5",
    "completitud_pct",
    "ambiguedad_pct",
    "evidencia_adjunta",
    "calidad_categoria_pct",
    "experiencia_portal_1_5"
]

# Para incorporar el canal de recepción se utilizan
# variables indicadoras (one-hot encoding).

canales_dummy = pd.get_dummies(
    base["canal"],
    prefix="canal",
    dtype=int
)

X_original = pd.concat(
    [
        base[VARIABLES],
        canales_dummy
    ],
    axis=1
)

COLUMNAS_MODELO = X_original.columns.tolist()

base.to_csv(
    os.path.join(
        DIRECTORIO,
        "01_Datos_Sinteticos.csv"
    ),
    index=False
)

interpretar(
    "01. Diagnóstico inicial",
    f"""
Se generaron {N_TICKETS:,} tickets
de soporte tecnológico.

El tiempo medio de resolución es:
{base.tiempo_resolucion_h.mean():.2f} horas.

El cumplimiento del SLA ilustrativo
de {SLA_HORAS:.0f} horas alcanza:
{100 * base.cumple_sla.mean():.1f}%.

La completitud promedio es:
{base.completitud_pct.mean():.1f}%.

El número promedio de contactos
adicionales es:
{base.contactos_adicionales.mean():.2f}.

INTERPRETACIÓN GERENCIAL

El problema comienza en la calidad
de la información recibida.

Un ticket incompleto puede obligar
al analista a solicitar aclaraciones,
revisar antecedentes o contactar
repetidamente al usuario.

Estas actividades consumen capacidad
sin resolver directamente el incidente.

Sin embargo, no debe asumirse que
todos los tiempos prolongados se
originan en la información incompleta.

La complejidad técnica, las
reasignaciones y la disponibilidad
de especialistas también pueden
influir en la resolución.

RECOMENDACIÓN

Medir por separado:

- Completitud al ingreso.
- Calidad de la descripción.
- Contactos adicionales.
- Reasignaciones.
- MTTR.
- Cumplimiento del SLA.
- Satisfacción del usuario.

La gerencia debería construir
una línea base real antes de
aprobar inversiones tecnológicas.
"""
)


# ============================================================
# 2. ANÁLISIS DESCRIPTIVO
# ============================================================

fig, ax = plt.subplots(figsize=(10, 5))

sns.histplot(
    base["tiempo_resolucion_h"],
    bins=35,
    color="#287FB8",
    ax=ax
)

ax.axvline(
    SLA_HORAS,
    color="#D14C48",
    linestyle="--",
    linewidth=2,
    label=f"Illustrative SLA: {SLA_HORAS:.0f} h"
)

ax.set(
    title="01 | IT Ticket Resolution Time Distribution",
    xlabel="Resolution time (hours)",
    ylabel="Tickets"
)

ax.legend()
guardar("01_Resolution_Time")

interpretar(
    "02. Interpretación del MTTR",
    """
La distribución permite evaluar
la variabilidad de los tiempos
de resolución.

El MTTR es importante, pero
no debería ser el único
indicador de desempeño.

Una reducción del MTTR podría
ocultar problemas de calidad
si los tickets se cierran
sin resolver adecuadamente
la necesidad del usuario.

Por ello, el equipo debería
analizar simultáneamente
reaperturas, satisfacción,
calidad de la solución
y contactos adicionales.

Una oportunidad de mejora
consiste en prevenir los
reprocesos antes de que
el ticket llegue al analista.

Por ejemplo, un formulario
inteligente podría solicitar
información diferente según
el tipo de incidente.

Sin embargo, debe evitarse
que los controles de entrada
sean tan complejos que
desincentiven el uso del
canal formal de soporte.
"""
)


# ============================================================
# 3. ESTANDARIZACIÓN
# ============================================================

escalador = StandardScaler()

X = escalador.fit_transform(
    X_original
)


# ============================================================
# 4. SELECCIÓN DE CLUSTERS
# ============================================================

valores_k = list(
    range(K_MIN, K_MAX + 1)
)

silhouettes = []

for k in valores_k:

    modelo_prueba = KMeans(
        n_clusters=k,
        random_state=SEMILLA,
        n_init=15
    )

    etiquetas = modelo_prueba.fit_predict(X)

    puntuacion = silhouette_score(
        X,
        etiquetas,
        sample_size=min(700, len(base)),
        random_state=SEMILLA
    )

    silhouettes.append(puntuacion)

k_optimo = valores_k[
    int(np.argmax(silhouettes))
]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(
    valores_k,
    silhouettes,
    marker="o",
    linewidth=2,
    color="#287FB8"
)

ax.axvline(
    k_optimo,
    color="#E49A32",
    linestyle="--",
    label=f"Selected k = {k_optimo}"
)

ax.set(
    title="02 | Optimal Cluster Selection",
    xlabel="Number of clusters",
    ylabel="Silhouette score"
)

ax.legend()
guardar("02_Silhouette")

interpretar(
    "03. Selección de clusters",
    f"""
Se evaluaron soluciones entre
{K_MIN} y {K_MAX} grupos.

El número seleccionado fue:
{k_optimo} clusters.

El índice silhouette fue:
{max(silhouettes):.3f}.

INTERPRETACIÓN

El algoritmo busca grupos
de tickets con características
semejantes al momento de
su recepción.

Esto permite explorar
diferencias en completitud,
ambigüedad, complejidad
y canales de entrada.

La segmentación podría
revelar que determinados
tickets requieren controles
de validación diferentes.

No obstante, los clusters
no representan automáticamente
categorías de calidad.

La organización debe verificar
si los segmentos son estables
y comprensibles.

RECOMENDACIÓN

Utilizar los perfiles para
diseñar procedimientos de
recepción diferenciados.

La clasificación debe
validarse antes de utilizarse
para automatizar decisiones.
"""
)


# ============================================================
# 5. K-MEANS
# ============================================================

modelo_kmeans = KMeans(
    n_clusters=k_optimo,
    random_state=SEMILLA,
    n_init=20
)

base["cluster"] = modelo_kmeans.fit_predict(X)

pca = PCA(
    n_components=2,
    random_state=SEMILLA
)

X_pca = pca.fit_transform(X)

fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=base["cluster"].astype(str),
    palette="tab10",
    alpha=0.7,
    s=35,
    ax=ax
)

ax.set(
    title="03 | IT Support Ticket Operational Clusters",
    xlabel="Principal component 1",
    ylabel="Principal component 2"
)

ax.legend(title="Cluster")
guardar("03_KMeans_Clusters")


# ============================================================
# 6. CARACTERIZACIÓN DE CLUSTERS
# ============================================================

perfiles = base.groupby("cluster")[VARIABLES].mean()

media = base[VARIABLES].mean()

desviacion = base[VARIABLES].std().replace(0, 1)

perfiles_z = (
    perfiles - media
) / desviacion

fig, ax = plt.subplots(
    figsize=(11, max(4, k_optimo * 1.2))
)

sns.heatmap(
    perfiles_z,
    annot=True,
    fmt=".2f",
    cmap="RdYlBu_r",
    center=0,
    linewidths=0.5,
    ax=ax
)

ax.set(
    title="04 | IT Ticket Cluster Profiles",
    xlabel="Operational variables",
    ylabel="Cluster"
)

guardar("04_Cluster_Heatmap")

resumen = base.groupby("cluster").agg(
    tickets=("ticket_id", "size"),
    mttr=("tiempo_resolucion_h", "mean"),
    cumplimiento=("cumple_sla", "mean"),
    completitud=("completitud_pct", "mean"),
    ambiguedad=("ambiguedad_pct", "mean"),
    contactos=("contactos_adicionales", "mean"),
    reasignaciones=("reasignaciones", "mean")
)

resumen["cumplimiento_pct"] = (
    100 * resumen["cumplimiento"]
)

print("\nRESUMEN POR CLUSTER")
print(resumen.round(2).to_string())

peor_cluster = int(
    resumen["mttr"].idxmax()
)

mejor_cluster = int(
    resumen["mttr"].idxmin()
)

for cluster in perfiles.index:

    dominantes = (
        perfiles_z.loc[cluster]
        .sort_values(ascending=False)
        .head(3)
        .index.tolist()
    )

    interpretar(
        f"04. Interpretación del cluster {cluster}",
        f"""
El cluster {cluster} contiene
{int(resumen.loc[cluster, 'tickets'])}
tickets.

El MTTR promedio es:
{resumen.loc[cluster, 'mttr']:.2f} horas.

El cumplimiento del SLA es:
{resumen.loc[cluster, 'cumplimiento_pct']:.1f}%.

La completitud promedio alcanza:
{resumen.loc[cluster, 'completitud']:.1f}%.

La ambigüedad promedio es:
{resumen.loc[cluster, 'ambiguedad']:.1f}%.

Los contactos adicionales
promedio son:
{resumen.loc[cluster, 'contactos']:.2f}.

Las características relativamente
más elevadas son:

{', '.join(dominantes)}.

INTERPRETACIÓN GERENCIAL

El segmento permite estudiar
si determinadas condiciones
de entrada se asocian con
mayores tiempos de resolución.

Si la completitud es baja,
podría ser conveniente
rediseñar los formularios.

Si la ambigüedad es elevada,
podrían incorporarse preguntas
guiadas o ejemplos de descripción.

Si existen diferencias entre
canales, se debería investigar
si todos solicitan los mismos
datos mínimos.

Sin embargo, las asociaciones
no demuestran causalidad.

RECOMENDACIÓN

Evaluar un piloto de validación
preventiva para el segmento.

Medir si disminuyen los
contactos adicionales y
las reasignaciones.

También verificar que los
usuarios puedan registrar
sus solicitudes fácilmente.

El objetivo no es aumentar
el número de campos sin
justificación, sino mejorar
la información útil para
el diagnóstico inicial.
"""
    )


# ============================================================
# 7. ISOLATION FOREST
# ============================================================

modelo_iso = IsolationForest(
    n_estimators=250,
    contamination=CONTAMINACION,
    random_state=SEMILLA
)

base["anomalia"] = (
    modelo_iso.fit_predict(X) == -1
)

base["anomaly_score"] = (
    -modelo_iso.decision_function(X)
)

fig, ax = plt.subplots(figsize=(10, 6))

normales = ~base["anomalia"]

ax.scatter(
    X_pca[normales, 0],
    X_pca[normales, 1],
    s=25,
    alpha=0.5,
    color="#287FB8",
    label="Typical"
)

ax.scatter(
    X_pca[~normales, 0],
    X_pca[~normales, 1],
    s=65,
    facecolors="none",
    edgecolors="#D64B48",
    linewidth=1.5,
    label="Anomaly"
)

ax.set(
    title="05 | Isolation Forest: Unusual IT Tickets",
    xlabel="Principal component 1",
    ylabel="Principal component 2"
)

ax.legend()
guardar("05_Isolation_Forest")

n_anomalias = int(
    base["anomalia"].sum()
)

interpretar(
    "05. Interpretación de anomalías",
    f"""
Isolation Forest señaló
{n_anomalias} tickets atípicos.

Esto representa el
{100 * base.anomalia.mean():.1f}%
del conjunto.

El porcentaje depende del
parámetro de contaminación
utilizado.

INTERPRETACIÓN GERENCIAL

Un ticket atípico puede
combinar información muy
incompleta, complejidad
elevada o una descripción
inusual.

Sin embargo, también podría
corresponder a un incidente
extraordinario correctamente
registrado.

No debe asumirse que una
anomalía sea un error.

RECOMENDACIÓN

Revisar los tickets con
mayor puntuación de anomalía.

Determinar si requieren
información adicional,
validación especializada
o un procedimiento diferente.

Si se identifican patrones
recurrentes, podrían diseñarse
controles preventivos.

La revisión humana sigue
siendo necesaria.

No se recomienda rechazar
automáticamente tickets
por ser atípicos.
"""
)

print("\n15 TICKETS MÁS ATÍPICOS")

print(
    base.sort_values(
        "anomaly_score",
        ascending=False
    )
    .head(15)
    .to_string(index=False)
)


# ============================================================
# 8. EVALUACIÓN EXTERNA DEL MTTR
# ============================================================

fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(
    data=base,
    x="cluster",
    y="tiempo_resolucion_h",
    estimator="mean",
    errorbar=None,
    color="#368AC1",
    ax=ax
)

ax.axhline(
    SLA_HORAS,
    color="#D64B48",
    linestyle="--",
    linewidth=2,
    label="Illustrative SLA"
)

ax.set(
    title="06 | Mean Resolution Time by Cluster",
    xlabel="Cluster",
    ylabel="Average resolution time (hours)"
)

ax.legend()
guardar("06_MTTR_By_Cluster")

interpretar(
    "06. Interpretación del desempeño",
    f"""
El cluster con mayor MTTR
es el {peor_cluster}.

Su tiempo promedio es:
{resumen.loc[peor_cluster, 'mttr']:.2f} horas.

El cluster con menor MTTR
es el {mejor_cluster}.

Su tiempo promedio es:
{resumen.loc[mejor_cluster, 'mttr']:.2f} horas.

INTERPRETACIÓN GERENCIAL

El tiempo de resolución
no se utilizó para formar
los clusters.

Por tanto, esta comparación
permite evaluar externamente
si los perfiles descubiertos
presentan diferencias de
desempeño.

Si un segmento concentra
tickets con información
deficiente y tiempos altos,
sería razonable investigar
la relación entre ambos.

No se puede concluir
automáticamente que la
información incompleta sea
la única causa del retraso.

También pueden intervenir
la complejidad, la capacidad
del equipo y la necesidad
de especialistas.

RECOMENDACIÓN

Construir un VSM del flujo
de atención de tickets.

Separar tiempo de diagnóstico,
esperas, aclaraciones,
transferencias y resolución.

Identificar actividades
que no agregan valor.

Posteriormente, evaluar
si la prevención de errores
al registrar solicitudes
reduce el MTTR sin afectar
la experiencia del usuario.
"""
)


# ============================================================
# 9. EVOLUCIÓN TEMPORAL
# ============================================================

participacion = (
    pd.crosstab(
        base["mes"],
        base["cluster"],
        normalize="index"
    ) * 100
)

fig, ax = plt.subplots(figsize=(10, 5))

for c in participacion.columns:

    ax.plot(
        participacion.index,
        participacion[c],
        marker="o",
        linewidth=2,
        label=f"Cluster {c}"
    )

ax.set(
    title="07 | Monthly IT Ticket Cluster Evolution",
    xlabel="Month",
    ylabel="Share of tickets (%)",
    xticks=range(1, MESES + 1)
)

ax.legend(ncol=3)
guardar("07_Cluster_Evolution")

anomalias_mes = (
    base.groupby("mes")["anomalia"].mean() * 100
)

fig, ax = plt.subplots(figsize=(10, 4.5))

ax.plot(
    anomalias_mes.index,
    anomalias_mes.values,
    marker="o",
    color="#C66A35",
    linewidth=2
)

ax.set(
    title="08 | Monthly IT Ticket Anomaly Rate",
    xlabel="Month",
    ylabel="Flagged tickets (%)",
    xticks=range(1, MESES + 1)
)

guardar("08_Anomaly_Evolution")

reciente = (
    participacion[peor_cluster]
    .tail(3)
    .mean()
)

anterior = (
    participacion[peor_cluster]
    .iloc[-6:-3]
    .mean()
)

interpretar(
    "07. Análisis temporal",
    f"""
El segmento con mayor MTTR
representa {reciente:.1f}%
de los tickets en los
últimos tres meses.

En los tres meses anteriores
representaba {anterior:.1f}%.

La variación fue de
{reciente - anterior:+.1f}
puntos porcentuales.

INTERPRETACIÓN GERENCIAL

La evolución temporal permite
observar si cambian los
perfiles de tickets recibidos.

Un aumento de solicitudes
con información incompleta
podría justificar una
investigación sobre el
diseño del portal, cambios
de usuarios o nuevos
procedimientos.

Sin embargo, una fluctuación
mensual no demuestra una
tendencia permanente.

RECOMENDACIÓN

Crear un tablero mensual
con completitud, ambigüedad,
MTTR y contactos adicionales.

También convendría analizar
los indicadores por canal
y tipo de incidente.

Las alertas deberían
calibrarse con información
histórica suficiente.

El propósito sería detectar
oportunamente cambios en
la calidad de las solicitudes.
"""
)


# ============================================================
# 10. ALERTA ANTICIPATORIA
# ============================================================

"""
Se analiza un ticket nuevo con información
disponible al momento de su creación.

No se incluyen:
- Contactos posteriores.
- Reasignaciones posteriores.
- Tiempo final de resolución.

La alerta se basa en similitud con
perfiles históricos.

NO predice individualmente el MTTR.
"""

nuevo_ticket = pd.DataFrame([{
    "complejidad_1_5": 4,
    "completitud_pct": 42,
    "ambiguedad_pct": 78,
    "evidencia_adjunta": 0,
    "calidad_categoria_pct": 45,
    "experiencia_portal_1_5": 2,
    "canal": "Email"
}])

# Preparar variables indicadoras para el canal.
nuevo_dummy = pd.get_dummies(
    nuevo_ticket["canal"],
    prefix="canal",
    dtype=int
)

nuevo_X = pd.concat(
    [
        nuevo_ticket[VARIABLES],
        nuevo_dummy
    ],
    axis=1
)

# Garantizar mismas columnas y mismo orden.
nuevo_X = nuevo_X.reindex(
    columns=COLUMNAS_MODELO,
    fill_value=0
)

X_nuevo = escalador.transform(
    nuevo_X
)

cluster_nuevo = int(
    modelo_kmeans.predict(X_nuevo)[0]
)

anomalia_nueva = bool(
    modelo_iso.predict(X_nuevo)[0] == -1
)

referencia = base.loc[
    base["cluster"] == cluster_nuevo
]

interpretar(
    "08. Alerta anticipatoria",
    f"""
El ticket nuevo pertenece
al cluster {cluster_nuevo}.

El MTTR promedio histórico
sintético de ese grupo es:
{referencia.tiempo_resolucion_h.mean():.2f} horas.

El cumplimiento del SLA
en ese segmento es:
{100 * referencia.cumple_sla.mean():.1f}%.

El ticket
{'SÍ' if anomalia_nueva else 'NO'}
fue marcado como atípico.

INTERPRETACIÓN GERENCIAL

El sistema identifica
características semejantes
a las de un segmento
histórico de solicitudes.

Esto puede utilizarse
como alerta para verificar
la calidad de la información.

No significa que el modelo
haya predicho exactamente
el tiempo de resolución.

RECOMENDACIÓN

Antes de asignar el ticket,
verificar si contiene
la información mínima
necesaria para diagnosticar
el incidente.

Si faltan datos, solicitar
una aclaración específica.

Si la categoría parece
inadecuada, ofrecer
orientación al usuario.

La validación preventiva
debería evitar contactos
repetidos durante la atención.

La efectividad de estas
alertas debe evaluarse
con tickets posteriores.
"""
)


# ============================================================
# 11. MONTE CARLO: ALTERNATIVAS GERENCIALES
# ============================================================

"""
Se comparan las alternativas del problema:

1. Formularios inteligentes.
2. Asistente virtual con IA.
3. Capacitación y autoservicio.
4. Canal único.
5. Enrutamiento automático.
6. Rediseño Lean.
7. Validación preventiva.

Los efectos son SUPUESTOS.

Los escenarios NO constituyen una
evaluación causal ni financiera real.

Los porcentajes representan reducciones
hipotéticas en componentes del tiempo.

El costo de implementación se compara
con un presupuesto ilustrativo.
"""

ESCENARIOS = {

    "Current process": {
        "costo": 0,
        "incompletitud": 0,
        "ambiguedad": 0,
        "contactos": 0,
        "reasignaciones": 0,
        "evidencia": 0
    },

    "Smart intake forms": {
        "costo": 45000000,
        "incompletitud": 0.55,
        "ambiguedad": 0.35,
        "contactos": 0.35,
        "reasignaciones": 0.10,
        "evidencia": 0.30
    },

    "AI virtual assistant": {
        "costo": 75000000,
        "incompletitud": 0.45,
        "ambiguedad": 0.55,
        "contactos": 0.45,
        "reasignaciones": 0.15,
        "evidencia": 0.25
    },

    "Training and self-service": {
        "costo": 28000000,
        "incompletitud": 0.25,
        "ambiguedad": 0.30,
        "contactos": 0.20,
        "reasignaciones": 0.05,
        "evidencia": 0.15
    },

    "Single intake channel": {
        "costo": 35000000,
        "incompletitud": 0.20,
        "ambiguedad": 0.15,
        "contactos": 0.10,
        "reasignaciones": 0.20,
        "evidencia": 0.10
    },

    "Automatic ticket routing": {
        "costo": 60000000,
        "incompletitud": 0.05,
        "ambiguedad": 0.05,
        "contactos": 0.10,
        "reasignaciones": 0.60,
        "evidencia": 0
    },

    "Lean support redesign": {
        "costo": 55000000,
        "incompletitud": 0.25,
        "ambiguedad": 0.20,
        "contactos": 0.40,
        "reasignaciones": 0.40,
        "evidencia": 0.15
    },

    "Preventive validation": {
        "costo": 40000000,
        "incompletitud": 0.65,
        "ambiguedad": 0.35,
        "contactos": 0.50,
        "reasignaciones": 0.10,
        "evidencia": 0.45
    },

    "Integrated solution": {
        "costo": 110000000,
        "incompletitud": 0.75,
        "ambiguedad": 0.65,
        "contactos": 0.65,
        "reasignaciones": 0.60,
        "evidencia": 0.55
    }
}

# Selección de tickets con reemplazo.
indices = rng.integers(
    0,
    len(base),
    size=N_SIMULACIONES
)

muestra = base.iloc[indices]

# Ruido común para comparar alternativas.
ruido_mc = rng.normal(
    0,
    2.5,
    N_SIMULACIONES
)


def simular_escenario(parametros):
    """
    Simula el tiempo de resolución.

    Se modifica el impacto hipotético
    de cada componente operacional.

    No se estima causalidad.
    """

    p = parametros

    tiempo = (
        2.0

        + 2.5
        * muestra["complejidad_1_5"].to_numpy()

        + 0.08
        * (
            100
            - muestra["completitud_pct"].to_numpy()
        )
        * (1 - p["incompletitud"])

        + 0.055
        * muestra["ambiguedad_pct"].to_numpy()
        * (1 - p["ambiguedad"])

        + 2.2
        * muestra["contactos_adicionales"].to_numpy()
        * (1 - p["contactos"])

        + 2.6
        * muestra["reasignaciones"].to_numpy()
        * (1 - p["reasignaciones"])

        + 1.5
        * (
            1
            - muestra["evidencia_adjunta"].to_numpy()
        )
        * (1 - p["evidencia"])

        + ruido_mc
    )

    return np.maximum(1, tiempo)


resultados = []

for nombre, parametros in ESCENARIOS.items():

    tiempos = simular_escenario(parametros)

    resultados.append({
        "alternativa": nombre,
        "costo_cop": parametros["costo"],
        "mttr_simulado_h": tiempos.mean(),
        "percentil_90_h": np.percentile(tiempos, 90),
        "cumplimiento_pct": 100 * np.mean(
            tiempos <= SLA_HORAS
        ),
        "exceso_medio_sla_h": np.maximum(
            0,
            tiempos - SLA_HORAS
        ).mean()
    })

escenarios = pd.DataFrame(resultados)

base_simulada = escenarios.loc[
    escenarios["alternativa"] == "Current process"
].iloc[0]

escenarios["reduccion_mttr_h"] = (
    base_simulada["mttr_simulado_h"]
    - escenarios["mttr_simulado_h"]
)

# El siguiente cálculo representa un potencial
# monetario ilustrativo, NO un ahorro real de caja.
#
# El MTTR incluye esperas y no necesariamente
# equivale a horas efectivas del personal.
#
# Por eso, este valor no debe interpretarse
# como un ROI validado.

escenarios["valor_tiempo_ilustrativo_cop"] = (
    escenarios["reduccion_mttr_h"]
    * N_TICKETS
    * COSTO_HORA_SOPORTE_COP
)

escenarios["beneficio_neto_hipotetico_cop"] = (
    escenarios["valor_tiempo_ilustrativo_cop"]
    - escenarios["costo_cop"]
)

escenarios["factible"] = (
    escenarios["costo_cop"]
    <= PRESUPUESTO_MAX_COP
)

escenarios.to_csv(
    os.path.join(
        DIRECTORIO,
        "02_Escenarios_Monte_Carlo.csv"
    ),
    index=False
)


# ============================================================
# 12. COMPARACIÓN DE MTTR
# ============================================================

fig, ax = plt.subplots(figsize=(11, 6))

colores = [
    "#287FB8" if x else "#D5A04A"
    for x in escenarios["factible"]
]

ax.barh(
    escenarios["alternativa"],
    escenarios["mttr_simulado_h"],
    color=colores
)

ax.axvline(
    SLA_HORAS,
    color="#D64B48",
    linestyle="--",
    label="Illustrative SLA"
)

ax.invert_yaxis()

ax.set(
    title="09 | Simulated MTTR by Intervention",
    xlabel="Mean resolution time (hours)",
    ylabel=""
)

ax.legend()
guardar("09_Scenario_Comparison")


# ============================================================
# 13. TRADE-OFF COSTO VS CUMPLIMIENTO
# ============================================================

fig, ax = plt.subplots(figsize=(11, 6))

for _, fila in escenarios.iterrows():

    ax.scatter(
        fila["costo_cop"] / 1e6,
        fila["cumplimiento_pct"],
        s=140,
        color=(
            "#287FB8"
            if fila["factible"]
            else "#D99A39"
        )
    )

    ax.annotate(
        fila["alternativa"],
        (
            fila["costo_cop"] / 1e6,
            fila["cumplimiento_pct"]
        ),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=8
    )

ax.axvline(
    PRESUPUESTO_MAX_COP / 1e6,
    linestyle="--",
    color="#D64B48",
    label="Budget limit"
)

ax.set(
    title="10 | Cost–Service Trade-off",
    xlabel="Implementation cost (million COP)",
    ylabel="Simulated SLA compliance (%)"
)

ax.legend()
guardar("10_Cost_Service_Tradeoff")


# ============================================================
# 14. DECISIÓN GERENCIAL
# ============================================================

"""
Se selecciona la alternativa factible
con mayor cumplimiento simulado del SLA.

El menor MTTR se utiliza como segundo criterio.

Esta regla es ilustrativa.
No reemplaza el AHP del equipo.
"""

factibles = (
    escenarios.loc[escenarios["factible"]]
    .sort_values(
        [
            "cumplimiento_pct",
            "mttr_simulado_h"
        ],
        ascending=[False, True]
    )
)

recomendada = factibles.iloc[0]

print("\nRESULTADOS DE MONTE CARLO")

print(
    escenarios.round(2).to_string(index=False)
)

interpretar(
    "09. Interpretación prescriptiva",
    f"""
Con un presupuesto máximo
hipotético de COP
{PRESUPUESTO_MAX_COP:,.0f},
la alternativa factible
mejor posicionada es:

{recomendada.alternativa}

El MTTR simulado sería:
{recomendada.mttr_simulado_h:.2f} horas.

La reducción frente al
proceso actual sería:
{recomendada.reduccion_mttr_h:.2f} horas.

El cumplimiento del SLA
sería:
{recomendada.cumplimiento_pct:.1f}%.

INTERPRETACIÓN GERENCIAL

La simulación permite
comparar intervenciones
bajo supuestos comunes.

Sin embargo, la alternativa
seleccionada depende de
los efectos hipotéticos,
el presupuesto y la
regla de decisión.

No demuestra que la
inversión produzca esos
resultados en la realidad.

RECOMENDACIÓN ESTRATÉGICA

Contrastar la alternativa
con los criterios del AHP:

- Calidad de la información.
- Reducción del MTTR.
- Reducción de reprocesos.
- Facilidad de uso.
- Costo de implementación.
- Integración tecnológica.
- Escalabilidad.

RECOMENDACIÓN TÁCTICA

Realizar un piloto de
validación preventiva o
mejora del registro.

Comparar tickets de
características semejantes.

RECOMENDACIÓN OPERACIONAL

Medir contactos adicionales,
reasignaciones, completitud
y tiempo de resolución.

También evaluar satisfacción
y facilidad de uso.

La decisión final no debe
basarse exclusivamente
en reducir el MTTR.
"""
)


# ============================================================
# 15. SENSIBILIDAD DEL SLA
# ============================================================

parametros_rec = ESCENARIOS[
    recomendada["alternativa"]
]

tiempos_rec = simular_escenario(
    parametros_rec
)

print("\nSENSIBILIDAD DEL SLA")

for sla in [12, 24, 36, 48]:

    cumplimiento = 100 * np.mean(
        tiempos_rec <= sla
    )

    print(
        f"SLA {sla:>2} horas: "
        f"{cumplimiento:.1f}% de cumplimiento"
    )

interpretar(
    "10. Sensibilidad y robustez",
    """
El cumplimiento depende
del plazo de resolución
establecido.

Un SLA más amplio puede
aumentar matemáticamente
el cumplimiento sin mejorar
el servicio.

Por ello, no debería
modificarse únicamente
para presentar resultados
más favorables.

La organización debería
evaluar escenarios como:

1. Mayor volumen de tickets.
2. Menor disponibilidad de analistas.
3. Incidentes más complejos.
4. Usuarios nuevos.
5. Cambios en canales de entrada.
6. Mayor demanda de soporte remoto.
7. Integraciones tecnológicas fallidas.

También debería modificarse
la efectividad supuesta
de cada alternativa.

Si la recomendación cambia
ante pequeñas variaciones,
la decisión requiere
mayor evidencia.

Una solución robusta
debería mantener la
calidad de las solicitudes
sin generar barreras
innecesarias para el usuario.
"""
)


# ============================================================
# 16. CONCLUSIONES EJECUTIVAS
# ============================================================

interpretar(
    "11. Conclusiones estratégicas",
    f"""
RESULTADOS DEL DIAGNÓSTICO

Se identificaron {k_optimo}
perfiles operacionales.

Se señalaron {n_anomalias}
tickets atípicos.

El segmento con mayor MTTR
fue el cluster {peor_cluster}.

Los resultados permiten
formular hipótesis sobre
la relación entre calidad
de entrada y desempeño.

DECISIONES ESTRATÉGICAS

Evaluar una estrategia
de mejora de la calidad
de información desde
la creación del ticket.

Considerar formularios
inteligentes y mecanismos
de validación preventiva.

Priorizar soluciones que
puedan integrarse con
la plataforma existente.

DECISIONES TÁCTICAS

Revisar categorías,
campos obligatorios
y reglas de asignación.

Evaluar capacitación
y herramientas de
autoservicio.

Realizar pilotos antes
de escalar las soluciones.

DECISIONES OPERACIONALES

Monitorear completitud
y ambigüedad al ingreso.

Identificar solicitudes
que requieren aclaraciones.

Revisar tickets atípicos.

Medir contactos adicionales,
reasignaciones y reaperturas.

PERSPECTIVA PREDICTIVA

Utilizar la similitud
con perfiles históricos
para generar alertas
anticipatorias.

Validar estas alertas
con tickets posteriores.

No interpretar la
pertenencia a un cluster
como una predicción exacta.

PERSPECTIVA PRESCRIPTIVA

Comparar intervenciones
mediante Monte Carlo.

Analizar costos,
cumplimiento y MTTR.

Contrastar las alternativas
con la priorización AHP.

CONEXIÓN CON EXCELENCIA OPERACIONAL

El ejercicio puede
complementarse con:

- Lean.
- Poka-yoke.
- VSM.
- Pareto.
- Ishikawa.
- DMAIC.
- FMEA.
- AHP.

LIMITACIONES

Todos los datos son
sintéticos.

Los efectos de las
intervenciones son
hipotéticos.

No se demuestra
causalidad ni ahorro
financiero real.

CONCLUSIÓN GENERAL

La analítica no supervisada
permite descubrir diferencias
entre los tickets que
ingresan al sistema.

Estos hallazgos pueden
orientar el rediseño
del proceso de recepción.

El objetivo final no
es solamente resolver
más rápido.

Es recibir información
de mejor calidad,
reducir reprocesos
y mejorar la experiencia
del usuario y la
productividad del soporte.
"""
)


# ============================================================
# 17. EXPORTACIÓN FINAL
# ============================================================

base.to_csv(
    os.path.join(
        DIRECTORIO,
        "03_Tickets_Clusters_Anomalias.csv"
    ),
    index=False
)

resumen.to_csv(
    os.path.join(
        DIRECTORIO,
        "04_Resumen_Clusters.csv"
    )
)

print("\n" + "=" * 70)
print("LABORATORIO CASO 5 FINALIZADO")
print("=" * 70)

print("Resultados guardados en:")
print(os.path.abspath(DIRECTORIO))

print("\nArchivos generados:")
print("- 10 gráficas PNG individuales")
print("- Base sintética CSV")
print("- Escenarios Monte Carlo CSV")
print("- Tickets con clusters y anomalías")
print("- Resumen de desempeño por cluster")
```
