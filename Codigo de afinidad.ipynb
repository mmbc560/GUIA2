```python
# ==========================================================
# ANALIZADOR DE PERFILES - TREND TECH - GRUPO 12
# VERSIÓN CORREGIDA Y COMPLETA
# Compatible con Google Colab / Jupyter
# ==========================================================

import matplotlib.pyplot as plt
from collections import Counter

print("🤖 ANALIZADOR DE PERFILES - TREND TECH")
print("Grupo 12 cargado correctamente.")

# ==========================================================
# 1. BASE DE DATOS DEL GRUPO 12
# ==========================================================

estudiantes = [
    {
        "nombre": "Samuel Alejandro Yepes Calderon",
        "carrera": "Contaduria Publica",
        "interes_personal": ["deportes", "mundo automotriz", "videojuegos"],
        "habilidades": ["Pensamiento logico", "Organizacion y planificacion"],
        "aprender": ["Automatizacion de procesos", "Analisis de datos"],
        "aportes": ["Organizacion del trabajo", "Apoyar en todo lo necesario"],
        "proyecto": "Mejorar procesos con tecnologia",
        "frase": "Pensar diferente es mi ventaja."
    },
    {
        "nombre": "Gabriela Verdeza Suarez",
        "carrera": "Mercadeo y Estrategias Comerciales",
        "interes_personal": ["publicidad"],
        "habilidades": ["Diseño", "Trabajo en equipo"],
        "aprender": ["Innovacion y emprendimiento", "Marketing digital"],
        "aportes": ["Diseñar presentaciones", "Motivar al equipo"],
        "proyecto": "Ayudar a las personas en temas de publicidad",
        "frase": "Crear valor a traves de la innovacion."
    },
    {
        "nombre": "Juan Diego Vega Ponguta",
        "carrera": "Ingenieria de Sistemas",
        "interes_personal": ["tecnologia", "programacion"],
        "habilidades": ["Pensamiento logico", "Organizacion y planificacion"],
        "aprender": ["Desarrollo de apps o software", "Inteligencia artificial"],
        "aportes": ["Resolver problemas", "Apoyar en todo lo necesario"],
        "proyecto": "Diseñar una app o herramienta digital",
        "frase": "Pensar fuera de la caja."
    },
    {
        "nombre": "Juan Felipe Triana Cortes",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["comunicacion"],
        "habilidades": ["Analisis de datos", "Trabajo en equipo", "Comunicacion", "Pensamiento logico"],
        "aprender": ["Inteligencia artificial", "Analisis de datos"],
        "aportes": ["Generacion de ideas", "Analizar informacion"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Pensar diferente es mi ventaja."
    },
    {
        "nombre": "Angelica Maria Sufia Jimenez",
        "carrera": "Ingenieria Industrial",
        "interes_personal": ["optimizacion"],
        "habilidades": ["Liderazgo", "Comunicacion"],
        "aprender": ["Automatizacion de procesos"],
        "aportes": ["Coordinar tareas", "Apoyar en todo lo necesario"],
        "proyecto": "Mejorar procesos con tecnologia",
        "frase": "Los datos cuentan historias."
    },
    {
        "nombre": "Maria Paula Ruiz Pinzon",
        "carrera": "Lenguas Modernas",
        "interes_personal": ["idiomas"],
        "habilidades": ["Diseño", "Trabajo en equipo"],
        "aprender": ["Marketing digital", "Inteligencia artificial"],
        "aportes": ["Apoyar en todo lo necesario", "Motivar al equipo"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "La tecnologia es mi herramienta para cambiar el mundo."
    },
    {
        "nombre": "Andres Fernando Ribero Barajas",
        "carrera": "Administracion de Empresas",
        "interes_personal": ["cadenas productivas"],
        "habilidades": ["Creatividad e ideas innovadoras", "Resolucion de problemas"],
        "aprender": ["Innovacion y emprendimiento", "Analisis de datos"],
        "aportes": ["Generacion de ideas", "Investigar informacion"],
        "proyecto": "Mejorar procesos con tecnologia",
        "frase": "Siempre en busqueda de nuevas ideas."
    },
    {
        "nombre": "Vincenzo Razzini Peña",
        "carrera": "Turismo Sostenible",
        "interes_personal": ["crear contactos"],
        "habilidades": ["Creatividad e ideas innovadoras", "Trabajo en equipo", "Comunicacion", "Liderazgo", "Diseño", "Pensamiento logico"],
        "aprender": ["Inteligencia artificial", "Automatizacion de procesos", "Marketing digital", "Sostenibilidad y tecnologia"],
        "aportes": ["Generacion de ideas", "Liderar el equipo", "Investigar informacion", "Apoyar en todo lo necesario", "Motivar al equipo"],
        "proyecto": "Crear un proyecto sostenible con tecnologia",
        "frase": "Innovar hoy, sostener mañana."
    },
    {
        "nombre": "Nicol Stefany Porras Acosta",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["mercadeo", "marketing"],
        "habilidades": ["Creatividad e ideas innovadoras", "Diseño"],
        "aprender": ["Marketing digital", "Analisis de datos"],
        "aportes": ["Diseñar presentaciones", "Investigar informacion"],
        "proyecto": "Crear una solucion para mejorar procesos con tecnologia",
        "frase": "Aprender, innovar y crear."
    },
    {
        "nombre": "Julian Camilo Pira Diaz",
        "carrera": "Negocios y Administracion",
        "interes_personal": ["diseño"],
        "habilidades": ["Liderazgo", "Trabajo en equipo", "Comunicacion"],
        "aprender": ["Marketing digital", "Innovacion en edicion"],
        "aportes": ["Coordinar tareas"],
        "proyecto": "Crear una solucion tecnologica para mi empresa",
        "frase": "Esto es onegrow."
    },
    {
        "nombre": "Lia Isabela Paez Moreno",
        "carrera": "Lenguas Modernas",
        "interes_personal": ["turismo"],
        "habilidades": ["Creatividad e ideas innovadoras", "Trabajo en equipo"],
        "aprender": ["Transformacion digital", "Marketing digital"],
        "aportes": ["Generacion de ideas", "Diseñar presentaciones"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Siempre en busqueda de nuevas ideas."
    },
    {
        "nombre": "Laura Alejandra Ojeda Romero",
        "carrera": "Economia",
        "interes_personal": ["microeconomia", "emprender"],
        "habilidades": ["Creatividad e innovacion", "Planeacion"],
        "aprender": ["Marketing digital", "Sostenibilidad y tecnologia"],
        "aportes": ["Aportar ideas clave"],
        "proyecto": "Crear un proyecto sostenible con tecnologia",
        "frase": "Aprender, innovar y crear."
    },
    {
        "nombre": "Juana Valentina Moya Nossa",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["comercio exterior"],
        "habilidades": ["Trabajo en equipo", "Organizacion", "Planificacion"],
        "aprender": ["Inteligencia artificial", "Automatizacion de procesos", "Innovacion", "Emprendimientos"],
        "aportes": ["Organizacion del trabajo", "Motivar al equipo", "Apoyar en todo lo necesario"],
        "proyecto": "Crear una solucion tecnologica para una empresa y desarrollar una idea innovadora de negocio",
        "frase": "Aprender, innovar y crear."
    },
    {
        "nombre": "Juan Sebastian Monsalvo Nieto",
        "carrera": "Lenguas Modernas",
        "interes_personal": ["tecnologia"],
        "habilidades": ["Resolucion de problemas", "Pensamiento logico"],
        "aprender": ["Marketing digital", "Automatizacion de procesos"],
        "aportes": ["Organizacion del trabajo", "Resolver problemas"],
        "proyecto": "Mejorar procesos con tecnologia",
        "frase": "Pensar diferente es mi ventaja."
    },
    {
        "nombre": "Carlos Andres Mongui Rios",
        "carrera": "Ingenieria de Sistemas",
        "interes_personal": ["automatizacion de procesos", "desarrollo de software"],
        "habilidades": ["Resolucion de problemas", "Pensamiento logico"],
        "aprender": ["Desarrollo de apps o software", "Internet de las cosas"],
        "aportes": ["Resolver problemas"],
        "proyecto": "Mejorar procesos con tecnologia",
        "frase": "Conectar conocimiento con accion."
    },
    {
        "nombre": "Valery Sofia Imbrecht Beltran",
        "carrera": "Lenguas Modernas",
        "interes_personal": ["traduccion"],
        "habilidades": ["Organizacion y planificacion", "Diseño"],
        "aprender": ["Inteligencia artificial", "Sostenibilidad y tecnologia"],
        "aportes": ["Diseñar presentaciones", "Investigar informacion"],
        "proyecto": "Mejorar procesos con tecnologia",
        "frase": "Crear valor a traves de la innovacion."
    },
    {
        "nombre": "Valeria Escobar Pulido",
        "carrera": "Negocios Internacionales / Economia",
        "interes_personal": ["emprender", "ayudar a las personas"],
        "habilidades": ["Trabajo en equipo", "Liderazgo", "Organizacion y planificacion", "Comunicacion"],
        "aprender": ["Inteligencia artificial", "Internet de las cosas", "Automatizacion de procesos", "Analisis de datos", "Marketing digital", "Innovacion y emprendimiento", "Sostenibilidad y tecnologia"],
        "aportes": ["Organizacion del trabajo", "Liderar el equipo", "Coordinar tareas", "Generacion de ideas"],
        "proyecto": "Desarrollar una idea innovadora de negocio / crear un proyecto sostenible con tecnologia / resolver un problema real con tecnologia",
        "frase": "Conectar conocimiento con accion. / Aprender, innovar y crear."
    },
    {
        "nombre": "Juan Pablo Garzon Mogollon",
        "carrera": "Ingenieria de Sistemas",
        "interes_personal": ["tecnologia"],
        "habilidades": ["Creatividad e ideas innovadoras", "Resolucion de problemas"],
        "aprender": ["Automatizacion de procesos", "Transformacion digital"],
        "aportes": ["Generacion de ideas", "Analizar informacion"],
        "proyecto": "Mejorar procesos con tecnologia",
        "frase": "La tecnologia es el motor de la innovacion."
    },
    {
        "nombre": "Miguel Angel Funeme Chavez",
        "carrera": "Administracion de Empresas",
        "interes_personal": ["gestion de empresas"],
        "habilidades": ["Creatividad e ideas innovadoras", "Trabajo en equipo"],
        "aprender": ["Innovacion y emprendimiento", "Marketing digital"],
        "aportes": ["Investigar informacion", "Coordinar tareas"],
        "proyecto": "Desarrollar una idea creativa de negocio",
        "frase": "Aprender, innovar y crear."
    },
    {
        "nombre": "Julian Santiago Forero Moreno",
        "carrera": "Ingenieria de Sistemas",
        "interes_personal": ["tecnologia", "datos"],
        "habilidades": ["Resolucion de problemas", "Uso basico de herramientas digitales"],
        "aprender": ["Inteligencia artificial", "Automatizacion", "Analisis de datos"],
        "aportes": ["Analisis de informacion", "Investigar informacion", "Resolucion de problemas"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": ""
    },
    {
        "nombre": "Samuel Jacobo Fonseca Orjuela",
        "carrera": "Lenguas Modernas",
        "interes_personal": ["idiomas"],
        "habilidades": ["Comunicacion", "Creatividad"],
        "aprender": ["Inteligencia artificial", "Marketing digital"],
        "aportes": ["Comunicacion"],
        "proyecto": "Crear una solucion tecnologica para una empresa / desarrollar una idea innovadora de negocio",
        "frase": "Aprender, innovar y crear."
    },
    {
        "nombre": "Maria Paula Fernandez Ramirez",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["innovacion", "mercados globales"],
        "habilidades": ["Trabajo en equipo", "Comunicar mis ideas"],
        "aprender": ["Marketing digital", "Analisis de datos"],
        "aportes": ["Organizacion", "Ideas nuevas"],
        "proyecto": "Desarrollar una idea de negocio con enfoque internacional",
        "frase": "Pienso en grande, conecto ideas y las llevo mas alla de las fronteras."
    },
    {
        "nombre": "Maria Fernanda Erazo Gomez",
        "carrera": "Administracion de Empresas",
        "interes_personal": ["negocios con IA"],
        "habilidades": ["Liderazgo", "Organizacion y planificacion"],
        "aprender": ["Inteligencia artificial"],
        "aportes": ["Liderar el equipo", "Coordinar tareas"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Aprender, innovar y crear."
    },
    {
        "nombre": "Eliana Cardenas Restrepo",
        "carrera": "Administracion de Empresas",
        "interes_personal": ["marketing digital"],
        "habilidades": ["Creatividad e ideas innovadoras", "Comunicacion"],
        "aprender": ["Transformacion digital", "Marketing digital"],
        "aportes": ["Generacion de ideas", "Diseñar presentaciones"],
        "proyecto": "Resolver un problema real con tecnologia",
        "frase": "Aprender, innovar y crear."
    },
    {
        "nombre": "Brayan Nicolas Camacho Suaza",
        "carrera": "Economia",
        "interes_personal": ["analisis de datos"],
        "habilidades": ["Trabajo en equipo", "Diseño"],
        "aprender": ["Inteligencia artificial", "Analisis de datos"],
        "aportes": ["Investigar informacion", "Apoyar en todo lo necesario"],
        "proyecto": "Diseñar una app o herramienta digital",
        "frase": "Aprender, innovar y crear."
    },
    {
        "nombre": "Dina Sellenic Becerra Urrego",
        "carrera": "Comunicacion",
        "interes_personal": [],
        "habilidades": ["Organizacion", "Planificacion", "Creatividad e ideas innovadoras"],
        "aprender": ["Marketing digital", "Sostenibilidad y tecnologia"],
        "aportes": ["Diseñar presentaciones", "Liderazgo"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Crear valor a traves de innovacion."
    },
    {
        "nombre": "Maria Jose Alvarez Castillo",
        "carrera": "Negocios Internacionales",
        "interes_personal": ["sector de belleza"],
        "habilidades": ["Creatividad e ideas innovadoras", "Diseño"],
        "aprender": ["Inteligencia artificial", "Marketing digital"],
        "aportes": ["Diseñar presentaciones", "Apoyar en todo lo necesario"],
        "proyecto": "Desarrollar una idea innovadora de negocio",
        "frase": "Aprender, innovar y crear."
    },
    {
        "nombre": "Samuel Alvarado Currea",
        "carrera": "Ingenieria de Sistemas",
        "interes_personal": ["tecnologia"],
        "habilidades": ["Uso basico de herramientas digitales", "Comunicacion"],
        "aprender": ["Inteligencia artificial", "Automatizacion de procesos"],
        "aportes": ["Analisis de informacion", "Motivacion al equipo"],
        "proyecto": "Desarrollo de ideas innovadoras de negocio",
        "frase": "Crear valor a traves de la innovacion."
    },

    # ===== ESTUDIANTES SIN DATOS, SOLO NOMBRE =====
    {
        "nombre": "Samuel Valles Villate",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Laura Camila Suarez Rincon",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Santiago Romero Cruz",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Camila Restrepo Santofimio",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Mariana Ramirez Ortiz",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Angel Santiago Pulido Morales",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Sara Valentina Navarrete Aponte",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Juan Carlos Moreno Rubio",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Kina Sahely Hernandez Gonzalez",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Juan Miguel Delgado Gomez",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Mateo Mateo Cruz Perlaza",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Julian David Castro Pineda",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Zahira Giselle Avila Castro",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    },
    {
        "nombre": "Diego Sebastian Acosta Rozo",
        "carrera": "",
        "interes_personal": [],
        "habilidades": [],
        "aprender": [],
        "aportes": [],
        "proyecto": "",
        "frase": ""
    }
]

# ==========================================================
# 2. FUNCIONES AUXILIARES
# ==========================================================

def texto_seguro(valor):
    if isinstance(valor, list):
        if len(valor) == 0:
            return "Sin informacion"
        return ", ".join(valor)
    if valor == "":
        return "Sin informacion"
    return valor

def imprimir_separador():
    print("\n" + "=" * 80)

def mostrar_estudiante(est):
    print(f"\n👤 Nombre: {est['nombre']}")
    print(f"🎓 Carrera: {texto_seguro(est['carrera'])}")
    print(f"🌟 Interes personal: {texto_seguro(est['interes_personal'])}")
    print(f"🧠 Habilidades: {texto_seguro(est['habilidades'])}")
    print(f"🚀 Quiere aprender: {texto_seguro(est['aprender'])}")
    print(f"🤝 Puede aportar: {texto_seguro(est['aportes'])}")
    print(f"📌 Proyecto de interes: {texto_seguro(est['proyecto'])}")
    print(f"💬 Frase: {texto_seguro(est['frase'])}")

def listar_nombres():
    print("\n📋 LISTA DE ESTUDIANTES")
    for i, est in enumerate(estudiantes, start=1):
        print(f"{i}. {est['nombre']}")

def buscar_por_nombre():
    texto = input("\nEscribe una parte del nombre: ").strip().lower()
    encontrados = [e for e in estudiantes if texto in e["nombre"].lower()]

    if encontrados:
        print(f"\n✅ Se encontraron {len(encontrados)} estudiante(s).")
        for est in encontrados:
            imprimir_separador()
            mostrar_estudiante(est)
    else:
        print("\n⚠️ No se encontraron coincidencias.")

def ver_todos_los_perfiles():
    print("\n📚 TODOS LOS PERFILES DEL GRUPO")
    for est in estudiantes:
        imprimir_separador()
        mostrar_estudiante(est)

def filtrar_por_carrera():
    texto = input("\nEscribe la carrera o una parte: ").strip().lower()
    encontrados = [e for e in estudiantes if texto in e["carrera"].lower()]

    if encontrados:
        print(f"\n✅ Se encontraron {len(encontrados)} estudiante(s).")
        for est in encontrados:
            print(f"- {est['nombre']} ({texto_seguro(est['carrera'])})")
    else:
        print("\n⚠️ No se encontraron estudiantes con esa carrera.")

def filtrar_por_habilidad():
    texto = input("\nEscribe la habilidad o una parte: ").strip().lower()
    encontrados = []

    for est in estudiantes:
        for h in est["habilidades"]:
            if texto in h.lower():
                encontrados.append(est)
                break

    if encontrados:
        print(f"\n✅ Se encontraron {len(encontrados)} estudiante(s).")
        for est in encontrados:
            print(f"- {est['nombre']} | {texto_seguro(est['habilidades'])}")
    else:
        print("\n⚠️ No se encontraron estudiantes con esa habilidad.")

def filtrar_por_tecnologia():
    texto = input("\nEscribe la tecnologia o tema: ").strip().lower()
    encontrados = []

    for est in estudiantes:
        for t in est["aprender"]:
            if texto in t.lower():
                encontrados.append(est)
                break

    if encontrados:
        print(f"\n✅ Se encontraron {len(encontrados)} estudiante(s).")
        for est in encontrados:
            print(f"- {est['nombre']} | {texto_seguro(est['aprender'])}")
    else:
        print("\n⚠️ No se encontraron estudiantes con ese interes.")

def filtrar_por_aporte():
    texto = input("\nEscribe el aporte o una parte: ").strip().lower()
    encontrados = []

    for est in estudiantes:
        for a in est["aportes"]:
            if texto in a.lower():
                encontrados.append(est)
                break

    if encontrados:
        print(f"\n✅ Se encontraron {len(encontrados)} estudiante(s).")
        for est in encontrados:
            print(f"- {est['nombre']} | {texto_seguro(est['aportes'])}")
    else:
        print("\n⚠️ No se encontraron estudiantes con ese aporte.")

def ver_proyectos_interes():
    print("\n📌 PROYECTOS DE INTERES")
    for est in estudiantes:
        print(f"- {est['nombre']}: {texto_seguro(est['proyecto'])}")

def resumen_general():
    print("\n📊 RESUMEN GENERAL DEL GRUPO")
    print(f"Total de estudiantes registrados: {len(estudiantes)}")

    carreras = Counter([e["carrera"] for e in estudiantes if e["carrera"] != ""])
    habilidades = Counter([h for e in estudiantes for h in e["habilidades"]])
    aprender = Counter([a for e in estudiantes for a in e["aprender"]])
    aportes = Counter([a for e in estudiantes for a in e["aportes"]])

    print("\n🎓 Carreras representadas:")
    for carrera, cantidad in carreras.most_common():
        print(f"- {carrera}: {cantidad}")

    print("\n🧠 Habilidades mas frecuentes:")
    for habilidad, cantidad in habilidades.most_common():
        print(f"- {habilidad}: {cantidad}")

    print("\n🚀 Temas que quieren aprender:")
    for tema, cantidad in aprender.most_common():
        print(f"- {tema}: {cantidad}")

    print("\n🤝 Aportes mas frecuentes:")
    for aporte, cantidad in aportes.most_common():
        print(f"- {aporte}: {cantidad}")

def encontrar_afinidades():
    print("\n🔎 AFINIDADES ENTRE ESTUDIANTES")
    encontrados = 0

    for i in range(len(estudiantes)):
        for j in range(i + 1, len(estudiantes)):
            e1 = estudiantes[i]
            e2 = estudiantes[j]

            comunes_h = set(e1["habilidades"]) & set(e2["habilidades"])
            comunes_a = set(e1["aprender"]) & set(e2["aprender"])
            comunes_p = set(e1["aportes"]) & set(e2["aportes"])

            puntaje = len(comunes_h) + len(comunes_a) + len(comunes_p)

            if puntaje >= 2:
                encontrados += 1
                print("-" * 80)
                print(f"{e1['nombre']}  <-->  {e2['nombre']}")
                if comunes_h:
                    print("Habilidades en comun:", ", ".join(sorted(comunes_h)))
                if comunes_a:
                    print("Aprendizajes en comun:", ", ".join(sorted(comunes_a)))
                if comunes_p:
                    print("Aportes en comun:", ", ".join(sorted(comunes_p)))
                print("Puntaje de afinidad:", puntaje)

    if encontrados == 0:
        print("No se encontraron afinidades fuertes.")

def sugerir_perfiles_complementarios():
    print("\n🧩 PERFILES COMPLEMENTARIOS")
    encontrados = 0

    for i in range(len(estudiantes)):
        for j in range(i + 1, len(estudiantes)):
            e1 = estudiantes[i]
            e2 = estudiantes[j]

            h1 = set(e1["habilidades"])
            h2 = set(e2["habilidades"])

            comunes = h1 & h2
            diferentes = (h1 | h2) - comunes

            if len(comunes) >= 1 and len(diferentes) >= 2:
                encontrados += 1
                print("-" * 80)
                print(f"{e1['nombre']} + {e2['nombre']}")
                print("Coinciden en:", ", ".join(sorted(comunes)))
                print("Se complementan con:", ", ".join(sorted(diferentes)))

    if encontrados == 0:
        print("No se encontraron perfiles complementarios.")

def analizar_estudiante_individual():
    listar_nombres()
    opcion = input("\nEscribe el numero del estudiante: ").strip()

    if not opcion.isdigit():
        print("\n⚠️ Debes escribir un numero valido.")
        return

    indice = int(opcion) - 1

    if indice < 0 or indice >= len(estudiantes):
        print("\n⚠️ Numero fuera de rango.")
        return

    est = estudiantes[indice]
    imprimir_separador()
    mostrar_estudiante(est)

# ==========================================================
# 3. GRAFICAS
# ==========================================================

def grafica_carreras():
    datos = Counter([e["carrera"] for e in estudiantes if e["carrera"] != ""])
    if len(datos) == 0:
        print("No hay datos para graficar.")
        return

    etiquetas = list(datos.keys())
    valores = list(datos.values())

    plt.figure(figsize=(12, 6))
    plt.bar(etiquetas, valores, color=[
        "royalblue", "orange", "green", "red", "purple", "cyan",
        "gold", "teal", "pink", "brown"
    ])
    plt.title("Cantidad de estudiantes por carrera")
    plt.xlabel("Carreras")
    plt.ylabel("Cantidad")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

def grafica_habilidades():
    datos = Counter([h for e in estudiantes for h in e["habilidades"]])
    if len(datos) == 0:
        print("No hay datos para graficar.")
        return

    etiquetas = list(datos.keys())
    valores = list(datos.values())

    plt.figure(figsize=(12, 6))
    plt.bar(etiquetas, valores, color=[
        "magenta", "dodgerblue", "limegreen", "darkorange", "crimson",
        "mediumpurple", "gray", "olive", "coral"
    ])
    plt.title("Habilidades mas frecuentes")
    plt.xlabel("Habilidades")
    plt.ylabel("Frecuencia")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

def grafica_aprendizajes():
    datos = Counter([a for e in estudiantes for a in e["aprender"]])
    if len(datos) == 0:
        print("No hay datos para graficar.")
        return

    etiquetas = list(datos.keys())
    valores = list(datos.values())

    plt.figure(figsize=(12, 6))
    plt.bar(etiquetas, valores, color=[
        "slateblue", "tomato", "gold", "mediumseagreen", "orchid",
        "deepskyblue", "sandybrown", "lightcoral"
    ])
    plt.title("Temas o tecnologias que quieren aprender")
    plt.xlabel("Temas")
    plt.ylabel("Frecuencia")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

def grafica_aportes():
    datos = Counter([a for e in estudiantes for a in e["aportes"]])
    if len(datos) == 0:
        print("No hay datos para graficar.")
        return

    etiquetas = list(datos.keys())
    valores = list(datos.values())

    plt.figure(figsize=(12, 6))
    plt.bar(etiquetas, valores, color=[
        "navy", "skyblue", "salmon", "khaki", "turquoise",
        "plum", "lightgreen", "peru"
    ])
    plt.title("Aportes mas frecuentes al equipo")
    plt.xlabel("Aportes")
    plt.ylabel("Frecuencia")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

def menu_graficas():
    while True:
        print("\n" + "=" * 80)
        print("📈 MENU DE GRAFICAS")
        print("=" * 80)
        print("1. Grafica de carreras")
        print("2. Grafica de habilidades")
        print("3. Grafica de aprendizajes")
        print("4. Grafica de aportes")
        print("5. Volver al menu principal")
        print("=" * 80)

        opcion = input("Selecciona una opcion (1-5): ").strip()

        if opcion == "1":
            grafica_carreras()
        elif opcion == "2":
            grafica_habilidades()
        elif opcion == "3":
            grafica_aprendizajes()
        elif opcion == "4":
            grafica_aportes()
        elif opcion == "5":
            print("Volviendo al menu principal...")
            break
        else:
            print("⚠️ Opcion no valida.")

# ==========================================================
# 4. MENU PRINCIPAL
# ==========================================================

def mostrar_menu():
    print("\n" + "=" * 80)
    print("MENU DE ANALISIS DEL GRUPO 12 - TREND TECH")
    print("=" * 80)
    print("1. Ver todos los perfiles")
    print("2. Buscar estudiante por nombre")
    print("3. Analizar un estudiante individual")
    print("4. Filtrar por carrera")
    print("5. Filtrar por habilidad")
    print("6. Filtrar por tecnologia o tema que quieren aprender")
    print("7. Filtrar por aporte al equipo")
    print("8. Ver proyectos de interes")
    print("9. Ver resumen general del grupo")
    print("10. Encontrar afinidades entre estudiantes")
    print("11. Sugerir perfiles complementarios")
    print("12. Menu de graficas")
    print("13. Listar solo estudiantes sin datos")
    print("14. Salir")
    print("=" * 80)

def listar_sin_datos():
    print("\n📝 ESTUDIANTES SIN DATOS")
    vacios = []

    for est in estudiantes:
        if (
            est["carrera"] == "" and
            len(est["interes_personal"]) == 0 and
            len(est["habilidades"]) == 0 and
            len(est["aprender"]) == 0 and
            len(est["aportes"]) == 0 and
            est["proyecto"] == "" and
            est["frase"] == ""
        ):
            vacios.append(est["nombre"])

    if len(vacios) == 0:
        print("Todos tienen informacion cargada.")
    else:
        for nombre in vacios:
            print("-", nombre)

def ejecutar_menu():
    while True:
        mostrar_menu()
        opcion = input("👉 Selecciona una opcion (1-14): ").strip()

        if opcion == "1":
            ver_todos_los_perfiles()
        elif opcion == "2":
            buscar_por_nombre()
        elif opcion == "3":
            analizar_estudiante_individual()
        elif opcion == "4":
            filtrar_por_carrera()
        elif opcion == "5":
            filtrar_por_habilidad()
        elif opcion == "6":
            filtrar_por_tecnologia()
        elif opcion == "7":
            filtrar_por_aporte()
        elif opcion == "8":
            ver_proyectos_interes()
        elif opcion == "9":
            resumen_general()
        elif opcion == "10":
            encontrar_afinidades()
        elif opcion == "11":
            sugerir_perfiles_complementarios()
        elif opcion == "12":
            menu_graficas()
        elif opcion == "13":
            listar_sin_datos()
        elif opcion == "14":
            print("\n👋 Fin del analisis. Exitos con Trend Tech.")
            break
        else:
            print("\n⚠️ Opcion no valida. Intenta de nuevo.")

# ==========================================================
# 5. INICIO
# ==========================================================

ejecutar_menu()
```
