```python
import pandas as pd
import matplotlib.pyplot as plt

# ==================================================
# 1. ESCRIBA AQUÍ EL NOMBRE EXACTO DEL ARCHIVO
# ==================================================

nombre_archivo = "Base_Datos_Dashboard_TrendTech_SmartRouteAI.xlsx"

# ==================================================
# 2. CARGAR HOJAS DEL EXCEL
# ==================================================

datos = pd.read_excel(nombre_archivo, sheet_name="Datos_KPI")
procesos = pd.read_excel(nombre_archivo, sheet_name="Procesos")
riesgos = pd.read_excel(nombre_archivo, sheet_name="Riesgos")

# ==================================================
# 3. ZONA EDITABLE POR LOS ESTUDIANTES
# ==================================================

nombre_proyecto = "Nombre del proyecto"
nombre_empresa = "Nombre de la empresa"

indicador_positivo = "Entregas a tiempo"
indicador_negativo = "Retrasos"
indicador_tiempo = "Tiempo promedio de entrega"
indicador_error = "Errores de registro"
indicador_trazabilidad = "Trazabilidad de pedidos"
indicador_satisfaccion = "Satisfacción del cliente"
indicador_productividad = "Productividad del equipo"

orden_meses = ["Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio"]

# ==================================================
# 4. PREPARAR DATOS
# ==================================================

kpi_actual = datos.pivot_table(
    index="Mes",
    columns="Indicador",
    values="Estado actual",
    aggfunc="mean"
).reset_index()

kpi_actual["Mes"] = pd.Categorical(
    kpi_actual["Mes"],
    categories=orden_meses,
    ordered=True
)

kpi_actual = kpi_actual.sort_values("Mes")

# ==================================================
# 5. CREAR DASHBOARD DEL ESTADO ACTUAL
# ==================================================

plt.figure(figsize=(22, 13))

plt.suptitle(
    f"Dashboard Estado Actual - {nombre_proyecto}",
    fontsize=22,
    fontweight="bold"
)

plt.subplot(2, 3, 1)
plt.plot(kpi_actual["Mes"], kpi_actual[indicador_positivo] * 100, marker="o", linewidth=3, color="#2E86C1", label=indicador_positivo)
plt.plot(kpi_actual["Mes"], kpi_actual[indicador_negativo] * 100, marker="o", linewidth=3, color="#C0392B", label=indicador_negativo)
plt.title("Indicadores principales del problema", fontweight="bold")
plt.ylabel("%")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.legend()

plt.subplot(2, 3, 2)
plt.bar(kpi_actual["Mes"], kpi_actual[indicador_tiempo], color="#F39C12")
plt.title("Tiempo promedio del proceso", fontweight="bold")
plt.ylabel("Valor")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.subplot(2, 3, 3)
plt.bar(kpi_actual["Mes"], kpi_actual[indicador_error] * 100, color="#E74C3C")
plt.title("Errores o fallas del proceso", fontweight="bold")
plt.ylabel("%")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.subplot(2, 3, 4)
plt.barh(procesos["Proceso"], procesos["Nivel actual (1-5)"], color="#8E44AD")
plt.title("Nivel actual de los procesos", fontweight="bold")
plt.xlabel("Nivel 1 a 5")
plt.xlim(0, 5)
plt.grid(axis="x", alpha=0.3)

plt.subplot(2, 3, 5)
riesgos_ordenados = riesgos.sort_values("Nivel riesgo actual", ascending=True)
plt.barh(riesgos_ordenados["Riesgo/Causa crítica"], riesgos_ordenados["Nivel riesgo actual"], color="#D35400")
plt.title("Causas críticas o riesgos actuales", fontweight="bold")
plt.xlabel("Probabilidad x Impacto")
plt.grid(axis="x", alpha=0.3)

plt.subplot(2, 3, 6)
valores_resumen = [
    kpi_actual[indicador_trazabilidad].mean() * 100,
    kpi_actual[indicador_satisfaccion].mean() * 20,
    kpi_actual[indicador_productividad].mean()
]

etiquetas_resumen = [
    "Trazabilidad",
    "Satisfacción",
    "Productividad"
]

plt.bar(etiquetas_resumen, valores_resumen, color=["#16A085", "#3498DB", "#27AE60"])
plt.title("Resumen del estado actual", fontweight="bold")
plt.ylabel("Valor promedio")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
